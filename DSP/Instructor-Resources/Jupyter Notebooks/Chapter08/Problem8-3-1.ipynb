```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

# ==============================================================================
# PROBLEM 1: Symbolic Z-Transform and Interactive Pole-Zero / ROC Visualization
# Signal: x[n] = (1 + n) * u[n]
# ==============================================================================

# Symbolic verification and derivation
n, z = sp.symbols('n z', integer=True)
x_n = (1 + n)
X_z_expr = sp.simplify(sp.summation(x_n * z**(-n), (n, 0, sp.oo)))

# Solution overview using a clean static representation for the equation as requested
explanation_text = f"""
<div style="background-color: #f8f9fa; padding: 10px; border-radius: 5px; border: 1px solid #dee2e6; font-size: 13px;">
<b>Solution Overview</b><br>
* <b>Signal:</b> x[n] = (1 + n)u[n]<br>
* <b>Z-Transform Derivation:</b> X(z) = z^2 / (z-1)^2<br>
* <b>Poles & Zeros:</b> Double zero at z = 0, Double pole at p = 1.<br>
* <b>ROC (Region of Convergence):</b> |z| > 1. Since the pole is on the unit circle boundary, the ROC lies strictly outside it.<br>
* <b>Note:</b> No tunable parameters exist for this specific signal, so the pole and ROC remain fixed.
</div>
"""
display(widgets.HTML(explanation_text))

# --- Interactive Plotting Setup ---
fig, (ax_pz, ax_time) = plt.subplots(1, 2, figsize=(16, 5), gridspec_kw={'width_ratios': [1, 2]})
plt.subplots_adjust(wspace=0.25)

# --- 1. Pole-Zero Map & ROC ---
ax_pz.set_aspect('equal')
ax_pz.set_xlim(-2.0, 2.0)
ax_pz.set_ylim(-2.0, 2.0)
ax_pz.axhline(0, color='black', linewidth=1)
ax_pz.axvline(0, color='black', linewidth=1)
ax_pz.grid(True, linestyle=':', alpha=0.7)

# Unit Circle
theta = np.linspace(0, 2*np.pi, 100)
ax_pz.plot(np.cos(theta), np.sin(theta), 'k--', alpha=0.5)

# Region of Convergence (ROC) shading for |z| > 1
roc_theta = np.linspace(0, 2*np.pi, 200)
r_roc = np.linspace(1.0, 2.5, 50)
R, T = np.meshgrid(r_roc, roc_theta)
X_roc = R * np.cos(T)
Y_roc = R * np.sin(T)
ax_pz.contourf(X_roc, Y_roc, R, levels=20, cmap='Greens', alpha=0.2)

# Double Pole at p = 1
ax_pz.scatter([1], [0], s=140, color='r', marker='x', linewidths=3)
# Double Zero at origin
ax_pz.scatter([0], [0], s=120, facecolors='none', edgecolors='b', linewidths=2, marker='o')

ax_pz.set_title('Pole-Zero Map & ROC (|z| > 1)', fontsize=10, fontweight='bold')
ax_pz.set_xlabel('Real Part', fontsize=9)
ax_pz.set_ylabel('Imaginary Part', fontsize=9)

# Legend handles in a single straight line
unit_circle_handle = plt.Line2D([0], [0], color='k', linestyle='--', alpha=0.5, label='Unit Circle')
pole_handle = plt.Line2D([0], [0], marker='x', color='r', markersize=8, markeredgewidth=3, linestyle='None', label='Double Pole at p = 1')
zero_handle = plt.Line2D([0], [0], marker='o', markerfacecolor='none', markeredgecolor='b', markersize=8, markeredgewidth=2, linestyle='None', label='Double Zero at z = 0')

ax_pz.legend(handles=[unit_circle_handle, pole_handle, zero_handle], loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=3, fontsize=8)

# --- 2. Time Domain Plot ---
n_samples = 25
n_vec = np.arange(n_samples)
x_n_vals = 1 + n_vec

ax_time.stem(n_vec, x_n_vals, linefmt='r-', markerfmt='ro', basefmt='k-')
ax_time.set_title('Temporal Evolution: x[n] = (1 + n)u[n]', fontsize=10, fontweight='bold')
ax_time.set_xlabel('Time index n', fontsize=9)
ax_time.set_ylabel('x[n]', fontsize=9)
ax_time.set_xlim(-1, n_samples)
ax_time.set_ylim(0, np.max(x_n_vals) * 1.2)
ax_time.grid(True, linestyle=':', alpha=0.7)

plt.show()
```
