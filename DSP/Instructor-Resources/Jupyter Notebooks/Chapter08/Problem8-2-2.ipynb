```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output

n, z = sp.symbols('n z', complex=True)
alpha = sp.Symbol('alpha', real=True)
M_sym = sp.Symbol('M', integer=True)

x_n = alpha**n
X_z_sum = sp.summation(x_n * z**(-n), (n, 0, M_sym - 1))
X_z_clean = sp.simplify(X_z_sum)

print("=== SYMBOLIC Z-TRANSFORM ===")
display(X_z_clean)

out = widgets.Output()

def plot_pole_zero_finite(alpha_val, M_val):
    M = int(M_val)
    with out:
        clear_output(wait=True)
        
        plt.figure(figsize=(6, 6))
        ax = plt.subplot(111)
        
        theta = np.linspace(0, 2*np.pi, 200)
        r_max = 2.0
        r_min = 0.05
        
        r_fill = np.linspace(r_min, r_max, 100)
        R, Theta = np.meshgrid(r_fill, theta)
        ax.contourf(R * np.cos(Theta), R * np.sin(Theta), R, levels=[r_min, r_max], colors=['lightgreen'], alpha=0.3)
        
        k_vals = np.arange(M)
        zeros = alpha_val * np.exp(1j * 2 * np.pi * k_vals / M)
        r_alpha = abs(alpha_val)
        
        ax.plot(r_alpha * np.cos(theta), r_alpha * np.sin(theta), 'b:', linewidth=1, label=f'Circle radius $|\\alpha| = {r_alpha:.2f}$')
        ax.scatter([0], [0], s=150, color='r', marker='x', linewidths=3, label=f'Pole at $z=0$ (mult. {M-1})')
        
        if M > 0:
            ax.scatter(np.real(zeros), np.imag(zeros), s=100, facecolors='none', edgecolors='b', linewidths=2, marker='o', label=f'Zeros ($M={M}$)')
        
        ax.plot(np.cos(theta), np.sin(theta), 'k--', alpha=0.5, label='Unit Circle')
        
        ax.axhline(0, color='black', linewidth=0.5)
        ax.axvline(0, color='black', linewidth=0.5)
        ax.set_xlim(-1.8, 1.8)
        ax.set_ylim(-1.8, 1.8)
        ax.set_aspect('equal')
        ax.grid(True, linestyle=':', alpha=0.7)
        ax.set_title(f'Poles & Zeros ($M={M}, \\alpha={alpha_val:.3f}$)', fontweight='bold')
        ax.set_xlabel('Real Part')
        ax.set_ylabel('Imaginary Part')
        
        ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.12), ncol=2, fontsize=8)
        plt.tight_layout()
        plt.show()

alpha_slider = widgets.FloatSlider(value=0.8, min=-1.5, max=1.5, step=0.01, description='Alpha:', style={'description_width': 'initial'})
M_slider = widgets.IntSlider(value=8, min=2, max=16, step=1, description='M:', style={'description_width': 'initial'})

interactive_plot = widgets.interactive(plot_pole_zero_finite, alpha_val=alpha_slider, M_val=M_slider)

display(widgets.VBox([interactive_plot, out]))
```
