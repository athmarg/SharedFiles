```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt

from ipywidgets import (
    FloatSlider,
    IntSlider,
    HTML,
    HTMLMath,
    VBox,
    HBox,
    Layout
)

from IPython.display import display

# ============================================================
# SECOND-ORDER HOMOGENEOUS DIFFERENCE EQUATIONS
#
# y[n+2] + a1 y[n+1] + a2 y[n] = 0
# ============================================================

plt.ioff()

# ============================================================
# CLASSIC JUPYTER + JUPYTERLAB / NOTEBOOK 7 / BINDER
# ============================================================

display(HTML("""
<style>

.container {
    width: 98% !important;
    max-width: none !important;
}

.output_area,
.output_subarea {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

.output_scroll {
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
    box-shadow: none !important;
}

.jp-Cell-outputWrapper,
.jp-OutputArea,
.jp-OutputArea-child,
.jp-OutputArea-output {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

.widget-output,
.jupyter-widgets-output-area,
.widget-box {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

.jupyter-matplotlib,
.jupyter-matplotlib-figure {
    overflow: visible !important;
    resize: none !important;
}

.jupyter-matplotlib::-webkit-resizer,
.jupyter-matplotlib-figure::-webkit-resizer {
    display: none !important;
}

.de2-title {
    font-family:Arial, sans-serif;
    font-size:20px;
    font-weight:bold;
    color:#6f3fa0;
}

.de2-label {
    font-family:Arial, sans-serif;
    font-size:14px;
    font-weight:bold;
}

.de2-value {
    font-family:Arial, sans-serif;
    font-size:14px;
    font-weight:bold;
    color:#0b3d91;
}

</style>
"""))

# ============================================================
# DOCUMENTATION
# ============================================================

documentation = HTML("""
<div style="
    width:1260px;
    font-family:Arial, sans-serif;
    font-size:15px;
    line-height:1.48;
    margin-bottom:10px;
">

<div class="de2-title" style="margin-bottom:8px;">
Second-Order Homogeneous Difference Equations
</div>

<div style="margin-bottom:5px;">
A homogeneous second-order linear difference equation with constant
coefficients has the form
</div>

<div style="
    text-align:center;
    font-size:17px;
    margin:7px 0px;
">
<b>
y[n+2] + α₁y[n+1] + α₂y[n] = 0.
</b>
</div>

<div style="margin-bottom:5px;">
Assuming an exponential solution y[n] = λⁿ leads to the characteristic
equation λ² + α₁λ + α₂ = 0.
</div>

<div>
<b>This notebook:</b> calculates the characteristic roots, identifies
their type, constructs the corresponding closed-form solution, determines
its constants from y[0] and y[1], and compares it with the sequence
generated recursively.
</div>

</div>
""")

# ============================================================
# SLIDER STYLES
# ============================================================

slider_style = {
    'description_width': '0px'
}

slider_layout = Layout(
    width='230px'
)

label_layout = Layout(
    width='135px',
    min_width='135px'
)

value_layout = Layout(
    width='70px',
    min_width='70px',
    margin='0px 0px 0px 6px'
)

row_layout = Layout(
    width='460px',
    height='38px',
    align_items='center'
)

# ============================================================
# α1
# ============================================================

a1_slider = FloatSlider(
    min=-4.0,
    max=4.0,
    step=0.25,
    value=-1.0,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

a1_label = HTML(
    '<div class="de2-label">Coefficient α₁:</div>',
    layout=label_layout
)

a1_value = HTML(
    '<div class="de2-value">-1.00</div>',
    layout=value_layout
)

a1_row = HBox(
    [
        a1_label,
        a1_slider,
        a1_value
    ],
    layout=row_layout
)

# ============================================================
# α2
# ============================================================

a2_slider = FloatSlider(
    min=-12.0,
    max=12.0,
    step=0.25,
    value=-12.0,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

a2_label = HTML(
    '<div class="de2-label">Coefficient α₂:</div>',
    layout=label_layout
)

a2_value = HTML(
    '<div class="de2-value">-12.00</div>',
    layout=value_layout
)

a2_row = HBox(
    [
        a2_label,
        a2_slider,
        a2_value
    ],
    layout=row_layout
)

# ============================================================
# y[0]
# ============================================================

y0_slider = FloatSlider(
    min=-5.0,
    max=5.0,
    step=0.25,
    value=1.0,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

y0_label = HTML(
    '<div class="de2-label">Initial value y[0]:</div>',
    layout=label_layout
)

y0_value = HTML(
    '<div class="de2-value">1.00</div>',
    layout=value_layout
)

y0_row = HBox(
    [
        y0_label,
        y0_slider,
        y0_value
    ],
    layout=row_layout
)

# ============================================================
# y[1]
# ============================================================

y1_slider = FloatSlider(
    min=-5.0,
    max=5.0,
    step=0.25,
    value=1.0,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

y1_label = HTML(
    '<div class="de2-label">Initial value y[1]:</div>',
    layout=label_layout
)

y1_value = HTML(
    '<div class="de2-value">1.00</div>',
    layout=value_layout
)

y1_row = HBox(
    [
        y1_label,
        y1_slider,
        y1_value
    ],
    layout=row_layout
)

# ============================================================
# MAXIMUM n
# ============================================================

N_slider = IntSlider(
    min=4,
    max=12,
    step=1,
    value=8,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

N_label = HTML(
    '<div class="de2-label">Maximum n:</div>',
    layout=label_layout
)

N_value = HTML(
    '<div class="de2-value">8</div>',
    layout=value_layout
)

N_row = HBox(
    [
        N_label,
        N_slider,
        N_value
    ],
    layout=row_layout
)

# ============================================================
# PARAMETERS PANEL
# ============================================================

parameters_panel = VBox(
    [
        HTML("""
        <div class="de2-title" style="margin-bottom:10px;">
            Equation Parameters
        </div>
        """),

        a1_row,
        a2_row,
        y0_row,
        y1_row,
        N_row
    ],
    layout=Layout(
        width='490px',
        padding='10px 14px',
        border='1px solid #d2c2df',
        overflow='visible'
    )
)

# ============================================================
# SYMBOLIC / ANALYTICAL RESULT WIDGETS
# ============================================================

equation_math = HTMLMath()
characteristic_math = HTMLMath()
discriminant_math = HTMLMath()

root1_math = HTMLMath()
root2_math = HTMLMath()

root_type_math = HTMLMath()
solution_math = HTMLMath()

constant1_math = HTMLMath()
constant2_math = HTMLMath()

extra_math = HTMLMath()

# ============================================================
# SMALLER FIXED WIDTHS
#
# λ1 and λ2 appear close together.
# C1 and C2 appear close together.
# ============================================================

root_item_layout = Layout(
    width='185px',
    min_width='185px'
)

constant_item_layout = Layout(
    width='185px',
    min_width='185px'
)

root1_math.layout = root_item_layout
root2_math.layout = root_item_layout

constant1_math.layout = constant_item_layout
constant2_math.layout = constant_item_layout

equation_math.layout = Layout(
    width='690px'
)

characteristic_math.layout = Layout(
    width='690px'
)

discriminant_math.layout = Layout(
    width='690px'
)

root_type_math.layout = Layout(
    width='690px'
)

solution_math.layout = Layout(
    width='690px'
)

extra_math.layout = Layout(
    width='690px'
)

# ============================================================
# ROOT ROW
# ============================================================

roots_row = HBox(
    [
        root1_math,
        root2_math
    ],
    layout=Layout(
        width='410px',
        gap='12px',
        align_items='center',
        overflow='visible'
    )
)

# ============================================================
# CONSTANTS ROW
# ============================================================

constants_row = HBox(
    [
        constant1_math,
        constant2_math
    ],
    layout=Layout(
        width='410px',
        gap='12px',
        align_items='center',
        overflow='visible'
    )
)

# ============================================================
# SYMBOLIC RESULTS PANEL
# ============================================================

symbolic_panel = VBox(
    [
        HTML("""
        <div class="de2-title" style="margin-bottom:8px;">
            Characteristic Equation and Solution
        </div>
        """),

        equation_math,
        characteristic_math,
        discriminant_math,

        roots_row,

        root_type_math,
        solution_math,

        constants_row,

        extra_math
    ],
    layout=Layout(
        width='750px',
        padding='10px 14px',
        border='1px solid #d2c2df',
        overflow='visible'
    )
)

# ============================================================
# TOP ROW
# ============================================================

top_row = HBox(
    [
        parameters_panel,
        symbolic_panel
    ],
    layout=Layout(
        width='1260px',
        gap='16px',
        align_items='stretch',
        overflow='visible'
    )
)

# ============================================================
# FIGURE 1 — SOLUTION
# ============================================================

fig_solution, ax_solution = plt.subplots(
    figsize=(7.0, 4.8)
)

fig_solution.canvas.header_visible = False
fig_solution.canvas.footer_visible = False
fig_solution.canvas.toolbar_visible = False

fig_solution.canvas.layout = Layout(
    width='700px',
    height='480px',
    overflow='visible'
)

ax_solution.set_title(
    'Recursive and Closed-Form Solutions',
    fontsize=14,
    fontweight='bold',
    color='#6f3fa0'
)

ax_solution.set_xlabel(
    'n',
    fontsize=11
)

ax_solution.set_ylabel(
    'y[n]',
    fontsize=11
)

ax_solution.grid(
    True,
    linestyle=':',
    alpha=0.40
)

ax_solution.set_xlim(
    -0.5,
    12.5
)

ax_solution.set_ylim(
    -15000,
    15000
)

# ============================================================
# CURVES CREATED ONCE
# ============================================================

line_recursive, = ax_solution.plot(
    [],
    [],
    marker='o',
    markersize=5,
    linewidth=1.7,
    label='Recursive solution'
)

line_closed, = ax_solution.plot(
    [],
    [],
    marker='s',
    markersize=4,
    linewidth=1.3,
    linestyle='--',
    label='Closed-form solution'
)

ax_solution.legend(
    loc='upper left',
    fontsize=9
)

fig_solution.subplots_adjust(
    left=0.12,
    right=0.97,
    top=0.90,
    bottom=0.13
)

# ============================================================
# FIGURE 2 — CHARACTERISTIC ROOTS
# ============================================================

fig_roots, ax_roots = plt.subplots(
    figsize=(4.8, 4.8)
)

fig_roots.canvas.header_visible = False
fig_roots.canvas.footer_visible = False
fig_roots.canvas.toolbar_visible = False

fig_roots.canvas.layout = Layout(
    width='480px',
    height='480px',
    overflow='visible'
)

ax_roots.set_title(
    'Characteristic Roots',
    fontsize=14,
    fontweight='bold',
    color='#0b3d91'
)

ax_roots.set_xlabel(
    'Re(λ)',
    fontsize=11
)

ax_roots.set_ylabel(
    'Im(λ)',
    fontsize=11
)

ax_roots.axhline(
    0,
    color='black',
    linewidth=1.0
)

ax_roots.axvline(
    0,
    color='black',
    linewidth=1.0
)

ax_roots.grid(
    True,
    linestyle=':',
    alpha=0.40
)

ax_roots.set_aspect(
    'equal',
    adjustable='box'
)

ax_roots.set_xlim(
    -6.5,
    6.5
)

ax_roots.set_ylim(
    -6.5,
    6.5
)

# ============================================================
# ROOT MARKERS CREATED ONCE
# ============================================================

root_points, = ax_roots.plot(
    [],
    [],
    linestyle='None',
    marker='x',
    markersize=10,
    markeredgewidth=2
)

# ============================================================
# UNIT CIRCLE
# ============================================================

unit_circle_angle = np.linspace(
    0.0,
    2.0 * np.pi,
    500
)

ax_roots.plot(
    np.cos(unit_circle_angle),
    np.sin(unit_circle_angle),
    linestyle='--',
    linewidth=1.0,
    alpha=0.60,
    label='Unit circle'
)

ax_roots.legend(
    loc='upper right',
    fontsize=8
)

fig_roots.subplots_adjust(
    left=0.15,
    right=0.96,
    top=0.90,
    bottom=0.13
)

# ============================================================
# NUMERICAL VERIFICATION PANEL
# ============================================================

verification_html = HTML()

verification_panel = VBox(
    [
        HTML("""
        <div class="de2-title" style="margin-bottom:8px;">
            Numerical Verification
        </div>
        """),

        verification_html
    ],
    layout=Layout(
        width='340px',
        padding='10px 14px',
        border='1px solid #d2c2df',
        overflow='visible'
    )
)

# ============================================================
# VISUAL ROW
#
# NOW:
#
# [ solution figure ] [ root figure ] [ numerical verification ]
#
# No bottom verification panel.
# ============================================================

visual_row = HBox(
    [
        fig_solution.canvas,
        fig_roots.canvas,
        verification_panel
    ],
    layout=Layout(
        width='1560px',
        gap='16px',
        align_items='flex-start',
        overflow='visible'
    )
)

# ============================================================
# RECURSIVE SOLUTION
# ============================================================

def recursive_solution(
    a1,
    a2,
    y0,
    y1,
    N
):

    n_values = np.arange(
        0,
        N + 1
    )

    values = np.zeros(
        N + 1,
        dtype=float
    )

    values[0] = y0

    if N >= 1:

        values[1] = y1

    for k in range(
        0,
        N - 1
    ):

        values[k + 2] = (
            -a1 * values[k + 1]
            -
            a2 * values[k]
        )

    return (
        n_values,
        values
    )

# ============================================================
# CHARACTERISTIC ROOTS
# ============================================================

def characteristic_roots(
    a1,
    a2
):

    discriminant = (
        a1**2
        -
        4.0 * a2
    )

    sqrt_discriminant = np.sqrt(
        complex(
            discriminant
        )
    )

    lambda1 = (
        -a1
        +
        sqrt_discriminant
    ) / 2.0

    lambda2 = (
        -a1
        -
        sqrt_discriminant
    ) / 2.0

    return (
        discriminant,
        lambda1,
        lambda2
    )

# ============================================================
# CLOSED-FORM SOLUTION
# ============================================================

def closed_form_solution(
    a1,
    a2,
    y0,
    y1,
    N
):

    (
        discriminant,
        lambda1,
        lambda2
    ) = characteristic_roots(
        a1,
        a2
    )

    n_values = np.arange(
        0,
        N + 1
    )

    tolerance = 1e-10

    # ========================================================
    # CASE 1 — DISTINCT REAL ROOTS
    # ========================================================

    if discriminant > tolerance:

        C1 = (
            y1
            -
            y0 * lambda2
        ) / (
            lambda1
            -
            lambda2
        )

        C2 = (
            y0 * lambda1
            -
            y1
        ) / (
            lambda1
            -
            lambda2
        )

        values = (
            C1 * lambda1**n_values
            +
            C2 * lambda2**n_values
        )

        case = 'distinct_real'

        constants = (
            C1,
            C2
        )

    # ========================================================
    # CASE 2 — REPEATED REAL ROOT
    # ========================================================

    elif abs(discriminant) <= tolerance:

        lambda0 = (
            -a1 / 2.0
        )

        C1 = y0

        if abs(lambda0) > tolerance:

            C2 = (
                y1 / lambda0
                -
                C1
            )

            values = (
                C1
                +
                C2 * n_values
            ) * (
                lambda0**n_values
            )

        else:

            values = np.zeros(
                N + 1,
                dtype=float
            )

            values[0] = y0

            if N >= 1:

                values[1] = y1

            C2 = np.nan

        case = 'repeated_real'

        constants = (
            C1,
            C2
        )

    # ========================================================
    # CASE 3 — COMPLEX-CONJUGATE ROOTS
    # ========================================================

    else:

        r = abs(
            lambda1
        )

        theta = np.angle(
            lambda1
        )

        A = y0

        sin_theta = np.sin(
            theta
        )

        if abs(
            r * sin_theta
        ) > tolerance:

            B = (
                y1 / r
                -
                A * np.cos(theta)
            ) / sin_theta

        else:

            B = 0.0

        values = (
            r**n_values
            *
            (
                A
                *
                np.cos(
                    n_values * theta
                )
                +
                B
                *
                np.sin(
                    n_values * theta
                )
            )
        )

        case = 'complex'

        constants = (
            A,
            B,
            r,
            theta
        )

    values = np.real_if_close(
        values
    )

    return (
        np.asarray(
            values,
            dtype=float
        ),
        case,
        constants,
        discriminant,
        lambda1,
        lambda2
    )

# ============================================================
# COMPLEX NUMBER FORMAT
# ============================================================

def complex_string(z):

    z = complex(z)

    a = z.real
    b = z.imag

    if abs(a) < 1e-10:
        a = 0.0

    if abs(b) < 1e-10:
        b = 0.0

    if b == 0.0:

        return f'{a:.4g}'

    if a == 0.0:

        return f'{b:.4g}i'

    sign = '+' if b >= 0 else '-'

    return (
        f'{a:.4g}'
        f'{sign}'
        f'{abs(b):.4g}i'
    )

# ============================================================
# EXPAND Y AXIS ONLY WHEN NECESSARY
# ============================================================

def ensure_solution_visible(values):

    values = np.asarray(
        values,
        dtype=float
    )

    data_min = np.min(
        values
    )

    data_max = np.max(
        values
    )

    current_min, current_max = (
        ax_solution.get_ylim()
    )

    if (
        data_min >= current_min * 0.90
        and
        data_max <= current_max * 0.90
    ):

        return

    new_min = min(
        current_min,
        data_min
    )

    new_max = max(
        current_max,
        data_max
    )

    span = (
        new_max
        -
        new_min
    )

    if span < 1e-10:

        span = 2.0

    margin = (
        0.10 * span
    )

    ax_solution.set_ylim(
        new_min - margin,
        new_max + margin
    )

# ============================================================
# UPDATE FUNCTION
# ============================================================

def update_notebook(change=None):

    a1 = a1_slider.value
    a2 = a2_slider.value

    y0 = y0_slider.value
    y1 = y1_slider.value

    N = N_slider.value

    # ========================================================
    # VALUE LABELS
    # ========================================================

    a1_value.value = (
        f'<div class="de2-value">{a1:.2f}</div>'
    )

    a2_value.value = (
        f'<div class="de2-value">{a2:.2f}</div>'
    )

    y0_value.value = (
        f'<div class="de2-value">{y0:.2f}</div>'
    )

    y1_value.value = (
        f'<div class="de2-value">{y1:.2f}</div>'
    )

    N_value.value = (
        f'<div class="de2-value">{N}</div>'
    )

    # ========================================================
    # RECURSIVE SOLUTION
    # ========================================================

    (
        n_values,
        recursive_values
    ) = recursive_solution(
        a1,
        a2,
        y0,
        y1,
        N
    )

    # ========================================================
    # CLOSED-FORM SOLUTION
    # ========================================================

    (
        closed_values,
        case,
        constants,
        discriminant,
        lambda1,
        lambda2
    ) = closed_form_solution(
        a1,
        a2,
        y0,
        y1,
        N
    )

    # ========================================================
    # UPDATE CURVES ONLY
    # ========================================================

    line_recursive.set_data(
        n_values,
        recursive_values
    )

    line_closed.set_data(
        n_values,
        closed_values
    )

    ensure_solution_visible(
        np.concatenate(
            [
                recursive_values,
                closed_values
            ]
        )
    )

    # ========================================================
    # ROOT POINTS
    # ========================================================

    root_points.set_data(
        [
            lambda1.real,
            lambda2.real
        ],
        [
            lambda1.imag,
            lambda2.imag
        ]
    )

    # ========================================================
    # MATHEMATICAL TEXT
    #
    # NO clear_output()
    # NO \qquad
    # ========================================================

    equation_math.value = (
        r'\('
        r'y[n+2]'
        +
        f'{a1:+.2f}'
        +
        r'y[n+1]'
        +
        f'{a2:+.2f}'
        +
        r'y[n]=0'
        r'\)'
    )

    characteristic_math.value = (
        r'\('
        r'\lambda^2'
        +
        f'{a1:+.2f}'
        +
        r'\lambda'
        +
        f'{a2:+.2f}'
        +
        r'=0'
        r'\)'
    )

    discriminant_math.value = (
        r'\('
        r'\Delta=\alpha_1^2-4\alpha_2='
        +
        f'{discriminant:.4f}'
        +
        r'\)'
    )

    root1_math.value = (
        r'\('
        r'\lambda_1='
        +
        complex_string(
            lambda1
        )
        +
        r'\)'
    )

    root2_math.value = (
        r'\('
        r'\lambda_2='
        +
        complex_string(
            lambda2
        )
        +
        r'\)'
    )

    # ========================================================
    # DISTINCT REAL ROOTS
    # ========================================================

    if case == 'distinct_real':

        C1, C2 = constants

        root_type_math.value = (
            r'\('
            r'\lambda_1\neq\lambda_2'
            r'\)'
        )

        solution_math.value = (
            r'\('
            r'y[n]'
            r'='
            r'C_1\lambda_1^n'
            r'+'
            r'C_2\lambda_2^n'
            r'\)'
        )

        constant1_math.value = (
            r'\('
            r'C_1='
            +
            f'{np.real(C1):.6f}'
            +
            r'\)'
        )

        constant2_math.value = (
            r'\('
            r'C_2='
            +
            f'{np.real(C2):.6f}'
            +
            r'\)'
        )

        extra_math.value = ''

        case_text = (
            'Two distinct real roots'
        )

    # ========================================================
    # REPEATED ROOT
    # ========================================================

    elif case == 'repeated_real':

        C1, C2 = constants

        root_type_math.value = (
            r'\('
            r'\lambda_1=\lambda_2=\lambda'
            r'\)'
        )

        solution_math.value = (
            r'\('
            r'y[n]'
            r'='
            r'(C_1+nC_2)\lambda^n'
            r'\)'
        )

        constant1_math.value = (
            r'\('
            r'C_1='
            +
            f'{C1:.6f}'
            +
            r'\)'
        )

        if np.isnan(C2):

            constant2_math.value = ''

        else:

            constant2_math.value = (
                r'\('
                r'C_2='
                +
                f'{C2:.6f}'
                +
                r'\)'
            )

        extra_math.value = ''

        case_text = (
            'Repeated real root'
        )

    # ========================================================
    # COMPLEX-CONJUGATE ROOTS
    # ========================================================

    else:

        A, B, r, theta = constants

        root_type_math.value = (
            r'\('
            r'\lambda_{1,2}=re^{\pm i\theta}'
            r'\)'
        )

        solution_math.value = (
            r'\('
            r'y[n]'
            r'='
            r'r^n'
            r'\left['
            r'A\cos(n\theta)'
            r'+'
            r'B\sin(n\theta)'
            r'\right]'
            r'\)'
        )

        constant1_math.value = (
            r'\('
            r'A='
            +
            f'{A:.6f}'
            +
            r'\)'
        )

        constant2_math.value = (
            r'\('
            r'B='
            +
            f'{B:.6f}'
            +
            r'\)'
        )

        extra_math.value = (
            r'\('
            r'r='
            +
            f'{r:.6f}'
            +
            r'\quad'
            r'\theta='
            +
            f'{theta:.6f}'
            +
            r'\ {\rm rad}'
            r'\)'
        )

        case_text = (
            'Complex-conjugate roots'
        )

    # ========================================================
    # NUMERICAL VERIFICATION
    # ========================================================

    difference = (
        recursive_values
        -
        closed_values
    )

    max_difference = np.max(
        np.abs(
            difference
        )
    )

    table_N = min(
        len(n_values),
        10
    )

    rows = ""

    for index in range(
        table_N
    ):

        rows += f"""
        <tr>

            <td style="
                padding:5px 8px;
                text-align:center;
                border-bottom:1px solid #eeeeee;
            ">
                {n_values[index]}
            </td>

            <td style="
                padding:5px 8px;
                text-align:center;
                border-bottom:1px solid #eeeeee;
            ">
                {recursive_values[index]:.5f}
            </td>

            <td style="
                padding:5px 8px;
                text-align:center;
                border-bottom:1px solid #eeeeee;
            ">
                {closed_values[index]:.5f}
            </td>

        </tr>
        """

    verification_html.value = f"""
    <div style="
        font-family:Arial, sans-serif;
        font-size:13px;
        line-height:1.42;
    ">

        <div style="
            color:#6f3fa0;
            font-weight:bold;
            margin-bottom:8px;
        ">
            Root type: {case_text}
        </div>

        <table style="
            width:100%;
            border-collapse:collapse;
        ">

            <tr style="
                background:#f4eff8;
                font-weight:bold;
            ">

                <td style="padding:5px;">
                    n
                </td>

                <td style="padding:5px;">
                    Recursive
                </td>

                <td style="padding:5px;">
                    Closed form
                </td>

            </tr>

            {rows}

        </table>

        <hr style="margin:8px 0px;">

        Maximum absolute difference:

        <br>

        <b>{max_difference:.3e}</b>

    </div>
    """

    # ========================================================
    # REDRAW ONLY
    # ========================================================

    fig_solution.canvas.draw_idle()
    fig_roots.canvas.draw_idle()

# ============================================================
# CONNECT CONTROLS
# ============================================================

a1_slider.observe(
    update_notebook,
    names='value'
)

a2_slider.observe(
    update_notebook,
    names='value'
)

y0_slider.observe(
    update_notebook,
    names='value'
)

y1_slider.observe(
    update_notebook,
    names='value'
)

N_slider.observe(
    update_notebook,
    names='value'
)

# ============================================================
# INITIAL UPDATE
# ============================================================

update_notebook()

# ============================================================
# COMPLETE NOTEBOOK
# ============================================================

main_layout = VBox(
    [
        documentation,
        top_row,
        visual_row
    ],
    layout=Layout(
        width='1580px',
        gap='10px',
        overflow='visible'
    )
)

display(
    main_layout
)
```
