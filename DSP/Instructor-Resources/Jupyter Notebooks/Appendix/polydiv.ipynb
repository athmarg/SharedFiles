```python
import numpy as np

from IPython.display import (
    display,
    HTML,
    Math
)

# ============================================================
# POLYNOMIAL LONG DIVISION WITH NUMPY
# ============================================================

# ============================================================
# DOCUMENTATION
# ============================================================

display(
    HTML("""
    <div style="
        width:900px;
        font-family:Arial, sans-serif;
        font-size:15px;
        line-height:1.50;
        margin-bottom:14px;
    ">

        <div style="
            font-size:21px;
            font-weight:bold;
            color:#6f3fa0;
            margin-bottom:8px;
        ">
            Polynomial Division with NumPy
        </div>

        <div style="margin-bottom:6px;">
            NumPy provides the function <b>numpy.polydiv()</b> for the
            division of two polynomials.
        </div>

        <div style="margin-bottom:6px;">
            The dividend and divisor are represented by arrays containing
            their coefficients in descending order of powers.
        </div>

        <div style="margin-bottom:6px;">
            The function returns two arrays containing the coefficients of
            the <b>quotient</b> and the <b>remainder</b>.
        </div>

        <div>
            The result satisfies the polynomial identity
        </div>

    </div>
    """)
)

display(
    Math(
        r'f(x)=d(x)q(x)+r(x)'
    )
)

# ============================================================
# EXAMPLE
#
#        3x^3 - 2x^2 + 4x - 3
# ------------------------------------------------
#             x^2 + 3x + 3
#
# ============================================================

display(
    HTML("""
    <div style="
        font-family:Arial, sans-serif;
        font-size:18px;
        font-weight:bold;
        color:#0b3d91;
        margin-top:18px;
        margin-bottom:8px;
    ">
        Example
    </div>
    """)
)

display(
    Math(
        r'\frac{3x^3-2x^2+4x-3}{x^2+3x+3}'
    )
)

# ============================================================
# POLYNOMIAL COEFFICIENTS
# ============================================================

# Dividend:
# 3x^3 - 2x^2 + 4x - 3

f = np.array(
    [3, -2, 4, -3],
    dtype=float
)

# Divisor:
# x^2 + 3x + 3

d = np.array(
    [1, 3, 3],
    dtype=float
)

# ============================================================
# POLYNOMIAL DIVISION
# ============================================================

quotient, remainder = np.polydiv(
    f,
    d
)

# ============================================================
# REMOVE NUMERICAL ROUND-OFF NOISE
# ============================================================

quotient = np.real_if_close(
    quotient
)

remainder = np.real_if_close(
    remainder
)

quotient[np.abs(quotient) < 1e-12] = 0
remainder[np.abs(remainder) < 1e-12] = 0

# ============================================================
# DISPLAY COEFFICIENT ARRAYS
# ============================================================

display(
    HTML("""
    <div style="
        font-family:Arial;
        font-size:15px;
        margin-top:15px;
        margin-bottom:5px;
    ">
        Coefficient representation:
    </div>
    """)
)

display(
    Math(
        r'f=[3,\,-2,\,4,\,-3]'
    )
)

display(
    Math(
        r'd=[1,\,3,\,3]'
    )
)

# ============================================================
# DISPLAY QUOTIENT AND REMAINDER
# ============================================================

display(
    HTML("""
    <div style="
        font-family:Arial;
        font-size:15px;
        margin-top:15px;
        margin-bottom:5px;
    ">
        NumPy returns:
    </div>
    """)
)

display(
    Math(
        r'q=[3,\,-11]'
    )
)

display(
    Math(
        r'r=[28,\,30]'
    )
)

display(
    HTML("""
    <div style="
        font-family:Arial;
        font-size:15px;
        margin-top:15px;
        margin-bottom:5px;
    ">
        Therefore:
    </div>
    """)
)

display(
    Math(
        r'q(x)=3x-11'
    )
)

display(
    Math(
        r'r(x)=28x+30'
    )
)

# ============================================================
# FINAL RESULT
# ============================================================

display(
    Math(
        r'\frac{3x^3-2x^2+4x-3}{x^2+3x+3}'
        r'='
        r'(3x-11)'
        r'+'
        r'\frac{28x+30}{x^2+3x+3}'
    )
)

# ============================================================
# VERIFICATION
# ============================================================

# Polynomial multiplication:
# d(x)q(x)

product = np.polymul(
    d,
    quotient
)

# Add the remainder.
# Pad the remainder on the left so that polynomial degrees match.

remainder_padded = np.pad(
    remainder,
    (
        len(product) - len(remainder),
        0
    )
)

reconstructed = (
    product
    +
    remainder_padded
)

correct = np.allclose(
    reconstructed,
    f
)

display(
    HTML("""
    <div style="
        font-family:Arial;
        font-size:15px;
        margin-top:18px;
        margin-bottom:5px;
    ">
        Verification:
    </div>
    """)
)

display(
    Math(
        r'(x^2+3x+3)(3x-11)+(28x+30)'
        r'='
        r'3x^3-2x^2+4x-3'
    )
)

if correct:

    display(
        Math(
            r'f(x)=d(x)q(x)+r(x)\qquad\checkmark'
        )
    )
```
