```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import IntSlider, Dropdown, interactive, VBox
from IPython.display import display, Markdown

# ------------------------------------------------------------
# Relative Frequency and Probability
# ------------------------------------------------------------

def probability_experiment(experiment, N):
    
    # Fixed seed: the same N always produces the same experiment
    rng = np.random.default_rng(12345)

    if experiment == 'Coin: Heads':
        samples = rng.integers(0, 2, N)
        success = (samples == 1)
        theoretical_probability = 0.5
        event_text = 'Heads'

    elif experiment == 'Die: Even number':
        samples = rng.integers(1, 7, N)
        success = np.isin(samples, [2, 4, 6])
        theoretical_probability = 3 / 6
        event_text = '{2, 4, 6}'

    elif experiment == 'Die: Six':
        samples = rng.integers(1, 7, N)
        success = (samples == 6)
        theoretical_probability = 1 / 6
        event_text = '{6}'

    elif experiment == 'Die: Number > 4':
        samples = rng.integers(1, 7, N)
        success = np.isin(samples, [5, 6])
        theoretical_probability = 2 / 6
        event_text = '{5, 6}'

    cumulative_successes = np.cumsum(success)
    trials = np.arange(1, N + 1)
    relative_frequency = cumulative_successes / trials

    fig, ax = plt.subplots(figsize=(10, 5))

    ax.plot(
        trials,
        relative_frequency,
        linewidth=1.8,
        label='Experimental relative frequency'
    )

    ax.axhline(
        theoretical_probability,
        linestyle='--',
        linewidth=2,
        label=f'Theoretical probability = {theoretical_probability:.4f}'
    )

    ax.set_xlabel('Number of trials N')
    ax.set_ylabel('Relative frequency')
    ax.set_title(f'Convergence of Relative Frequency — Event {event_text}')
    ax.set_ylim(0, 1)
    ax.grid(True, alpha=0.3)
    ax.legend()

    plt.show()
    plt.close(fig)

    print(f'Number of trials                 : {N}')
    print(f'Number of event occurrences      : {cumulative_successes[-1]}')
    print(f'Experimental relative frequency  : {relative_frequency[-1]:.6f}')
    print(f'Theoretical probability          : {theoretical_probability:.6f}')
    print(f'Absolute error                   : {abs(relative_frequency[-1] - theoretical_probability):.6f}')


experiment_widget = Dropdown(
    options=[
        'Coin: Heads',
        'Die: Even number',
        'Die: Six',
        'Die: Number > 4'
    ],
    value='Coin: Heads',
    description='Experiment:'
)

N_widget = IntSlider(
    value=100,
    min=10,
    max=5000,
    step=10,
    description='N:',
    continuous_update=True,
    readout=True
)

display(Markdown("## Relative Frequency and Theoretical Probability"))

interactive_plot = interactive(
    probability_experiment,
    experiment=experiment_widget,
    N=N_widget
)

display(interactive_plot)
```
