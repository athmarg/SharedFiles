```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt

from ipywidgets import (
    FloatSlider,
    IntSlider,
    HTML,
    HTMLMath,
    VBox,
    HBox,
    Layout
)
from IPython.display import display

# ============================================================
# DISCRETE TO CONTINUOUS EIGENVALUE SPECTRUM
#
# Finite interval:
#
# lambda_n = n*pi/L
#
# Delta lambda = pi/L
#
# Continuous sine spectrum:
#
# D(lambda) =
# (2/pi) integral_0^1 f(x) sin(lambda*x) dx
#
# For the chosen initial condition:
#
# f(x) = 1 - |2x - 1|,   0 <= x <= 1
#
# and f(x) = 0 elsewhere.
#
# As L -> infinity:
#
# Delta lambda -> 0
#
# Fourier series -> Fourier integral
# ============================================================

plt.ioff()

# ============================================================
# JUPYTER / BINDER DISPLAY SETTINGS
# ============================================================

display(HTML("""
<style>

.container {
    width:98% !important;
    max-width:none !important;
}

.output_area,
.output_subarea,
.jp-Cell-outputWrapper,
.jp-OutputArea,
.jp-OutputArea-child,
.jp-OutputArea-output,
.widget-output,
.jupyter-widgets-output-area,
.widget-box {
    max-width:none !important;
    height:auto !important;
    max-height:none !important;
    overflow:visible !important;
}

.output_scroll {
    height:auto !important;
    max-height:none !important;
    overflow:visible !important;
    box-shadow:none !important;
}

.jupyter-matplotlib,
.jupyter-matplotlib-figure {
    overflow:visible !important;
    resize:none !important;
}

.jupyter-matplotlib::-webkit-resizer,
.jupyter-matplotlib-figure::-webkit-resizer {
    display:none !important;
}

.spec-title {
    font-family:Arial, sans-serif;
    font-size:20px;
    font-weight:bold;
    color:#6f3fa0;
}

.spec-label {
    font-family:Arial, sans-serif;
    font-size:14px;
    font-weight:bold;
}

.spec-value {
    font-family:Arial, sans-serif;
    font-size:14px;
    font-weight:bold;
    color:#0b3d91;
}

</style>
"""))

# ============================================================
# DOCUMENTATION
# ============================================================

documentation = HTML("""
<div style="
    width:1160px;
    font-family:Arial, sans-serif;
    font-size:15px;
    line-height:1.48;
    margin-bottom:10px;
">

<div class="spec-title" style="margin-bottom:8px;">
From a Discrete to a Continuous Eigenvalue Spectrum
</div>

<div style="margin-bottom:5px;">
For the heat equation on a finite interval of length L, the allowed
spatial eigenvalues are λₙ=nπ/L and their constant spacing is
Δλ=π/L.
</div>

<div style="margin-bottom:5px;">
As L increases, Δλ decreases and the discrete eigenvalues become
progressively denser. In the limiting case L→∞, the discrete
eigenvalue spectrum becomes continuous and the Fourier series is
replaced by a Fourier integral.
</div>

<div style="margin-bottom:5px;">
To make this transition physically meaningful, the right-hand plot
does not use an arbitrary spectral envelope. It displays the actual
Fourier sine spectrum of the triangular initial temperature profile
f(x)=1−|2x−1| for 0≤x≤1, with f(x)=0 elsewhere.
</div>

<div>
<b>This notebook:</b> shows both the increasing density of the
eigenvalues and how the discrete spectral values sample the
corresponding continuous Fourier sine spectrum.
</div>

</div>
""")

# ============================================================
# MATHEMATICAL RELATIONS
# ============================================================

relation1 = HTMLMath(
    value=(
        r'\('
        r'\lambda_n=\frac{n\pi}{L}'
        r',\;'
        r'\Delta\lambda=\frac{\pi}{L}'
        r',\;'
        r'L\to\infty'
        r'\Longrightarrow'
        r'\Delta\lambda\to0'
        r'\)'
    )
)

relation2 = HTMLMath(
    value=(
        r'\('
        r'D(\lambda)='
        r'\frac{2}{\pi}'
        r'\displaystyle\int_0^1'
        r'f(x)\sin(\lambda x)\,dx'
        r'\)'
    )
)

relation3 = HTMLMath(
    value=(
        r'\('
        r'C_n='
        r'\Delta\lambda\,D(\lambda_n)'
        r'\)'
    )
)

relations = VBox(
    [
        relation1,
        relation2,
        relation3
    ],
    layout=Layout(
        width='1050px',
        gap='2px'
    )
)

# ============================================================
# CONTROLS
# ============================================================

slider_style = {
    'description_width': '0px'
}

L_slider = FloatSlider(
    min=2.0,
    max=40.0,
    step=0.5,
    value=5.0,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=Layout(
        width='260px'
    )
)

N_slider = IntSlider(
    min=5,
    max=80,
    step=1,
    value=25,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=Layout(
        width='260px'
    )
)

L_value = HTML(
    '<div class="spec-value">5.0</div>'
)

N_value = HTML(
    '<div class="spec-value">25</div>'
)

def row(label, slider, value):

    return HBox(
        [
            HTML(
                f'<div class="spec-label">{label}</div>',
                layout=Layout(
                    width='130px',
                    min_width='130px'
                )
            ),

            slider,

            value
        ],
        layout=Layout(
            width='485px',
            height='40px',
            align_items='center'
        )
    )

# ============================================================
# PARAMETERS PANEL
# ============================================================

controls = VBox(
    [
        HTML("""
        <div class="spec-title" style="margin-bottom:7px;">
            Parameters
        </div>
        """),

        row(
            'Domain length L:',
            L_slider,
            L_value
        ),

        row(
            'Number of modes:',
            N_slider,
            N_value
        )
    ],
    layout=Layout(
        width='505px',
        padding='10px 14px',
        border='1px solid #d2c2df'
    )
)

# ============================================================
# CURRENT SPECTRUM PANEL
# ============================================================

current_math = HTMLMath()

current_panel = VBox(
    [
        HTML("""
        <div class="spec-title" style="margin-bottom:7px;">
            Current Spectrum
        </div>
        """),

        current_math
    ],
    layout=Layout(
        width='590px',
        padding='10px 14px',
        border='1px solid #d2c2df'
    )
)

# ============================================================
# TOP ROW
# ============================================================

top_row = HBox(
    [
        controls,
        current_panel
    ],
    layout=Layout(
        width='1120px',
        gap='15px',
        align_items='stretch',
        justify_content='flex-start',
        margin='0px',
        padding='0px'
    )
)

# ============================================================
# TRIANGULAR INITIAL TEMPERATURE PROFILE
#
# f(x) = 1 - |2x - 1|,   0 <= x <= 1
#
# and f(x) = 0 elsewhere.
# ============================================================

NX_TRANSFORM = 2500

x_transform = np.linspace(
    0.0,
    1.0,
    NX_TRANSFORM
)

f_transform = (
    1.0
    -
    np.abs(
        2.0*x_transform
        -
        1.0
    )
)

# ============================================================
# FOURIER SINE TRANSFORM
#
# D(lambda) =
# (2/pi) integral_0^1 f(x) sin(lambda*x) dx
#
# The integral is evaluated numerically.
# ============================================================

def sine_fourier_transform(
    lambda_values
):

    lambda_values = np.atleast_1d(
        lambda_values
    )

    sine_matrix = np.sin(
        lambda_values[:, None]
        *
        x_transform[None, :]
    )

    integrands = (
        sine_matrix
        *
        f_transform[None, :]
    )

    integral_values = np.trapezoid(
        integrands,
        x_transform,
        axis=1
    )

    return (
        2.0
        /
        np.pi
        *
        integral_values
    )

# ============================================================
# CONTINUOUS SPECTRUM
# ============================================================

LAMBDA_MAX = 8.0

lambda_cont = np.linspace(
    0.0,
    LAMBDA_MAX,
    1000
)

D_cont = sine_fourier_transform(
    lambda_cont
)

D_abs_cont = np.abs(
    D_cont
)

# ============================================================
# FIGURE 1
# DISCRETE EIGENVALUE LOCATIONS
# ============================================================

fig_spec, ax_spec = plt.subplots(
    figsize=(7.0, 4.2)
)

fig_spec.canvas.header_visible = False
fig_spec.canvas.footer_visible = False
fig_spec.canvas.toolbar_visible = False

fig_spec.canvas.layout = Layout(
    width='700px',
    height='420px',
    margin='0px',
    padding='0px'
)

ax_spec.set_title(
    'Discrete Eigenvalue Locations',
    fontsize=14,
    fontweight='bold',
    color='#6f3fa0'
)

ax_spec.set_xlabel(
    'Eigenvalue λ'
)

ax_spec.set_yticks(
    []
)

ax_spec.set_xlim(
    0.0,
    25.0
)

ax_spec.set_ylim(
    0.0,
    1.2
)

ax_spec.grid(
    True,
    axis='x',
    linestyle=':',
    alpha=0.35
)

eigen_points, = ax_spec.plot(
    [],
    [],
    linestyle='None',
    marker='|',
    markersize=18,
    markeredgewidth=2
)

fig_spec.subplots_adjust(
    left=0,
    right=0.97,
    top=0.88,
    bottom=0.14
)

# ============================================================
# FIGURE 2
# FOURIER SINE SPECTRUM
# ============================================================

fig_env, ax_env = plt.subplots(
    figsize=(4.3, 4.2)
)

fig_env.canvas.header_visible = False
fig_env.canvas.footer_visible = False
fig_env.canvas.toolbar_visible = False

fig_env.canvas.layout = Layout(
    width='430px',
    height='420px',
    margin='0px',
    padding='0px'
)

ax_env.set_title(
    'Fourier Sine Spectrum',
    fontsize=13,
    fontweight='bold',
    color='#0b3d91'
)

ax_env.set_xlabel(
    'λ'
)

ax_env.set_ylabel(
    '|D(λ)|'
)

ax_env.set_xlim(
    0.0,
    LAMBDA_MAX
)

spectrum_max = np.max(
    D_abs_cont
)

ax_env.set_ylim(
    0.0,
    1.08 * spectrum_max
)

ax_env.grid(
    True,
    linestyle=':',
    alpha=0.35
)

# ------------------------------------------------------------
# Continuous Fourier sine spectrum
# ------------------------------------------------------------

continuous_spectrum_line, = ax_env.plot(
    lambda_cont,
    D_abs_cont,
    linewidth=2.0,
    label='Continuous |D(λ)|'
)

# ------------------------------------------------------------
# Samples at discrete eigenvalues
# ------------------------------------------------------------

sample_points, = ax_env.plot(
    [],
    [],
    linestyle='None',
    marker='o',
    markersize=5,
    label='Samples at λₙ'
)

ax_env.legend(
    loc='upper right',
    fontsize=8
)

fig_env.subplots_adjust(
    left=0.15,
    right=0.97,
    top=0.88,
    bottom=0.14
)

# ============================================================
# FIGURES ROW
# ============================================================

figures_row = HBox(
    [
        fig_spec.canvas,
        fig_env.canvas
    ],
    layout=Layout(
        width='1140px',
        gap='10px',
        align_items='flex-start',
        justify_content='flex-start',
        margin='0px',
        padding='0px'
    )
)

# ============================================================
# INTERPRETATION PANEL
# ============================================================

interpretation_panel = HTML("""
<div style="
    width:1120px;
    padding:10px 13px;
    border:1px solid #d7c7e5;
    font-family:Arial, sans-serif;
    font-size:14px;
    line-height:1.55;
    box-sizing:border-box;
    margin-top:4px;
">

<div style="
    color:#6f3fa0;
    font-size:17px;
    font-weight:bold;
    margin-bottom:6px;
">
Interpretation
</div>

<div style="margin-bottom:6px;">
The left-hand plot shows the discrete eigenvalues
<b>λₙ=nπ/L</b> of the finite-domain problem. Their spacing is
<b>Δλ=π/L</b>. Increasing L therefore makes the eigenvalues
progressively more densely packed.
</div>

<div style="margin-bottom:6px;">
The continuous curve in the right-hand plot is not an arbitrary
envelope. It is the magnitude of the actual Fourier sine transform
<b>D(λ)</b> of the triangular initial temperature distribution
used in the calculation.
</div>

<div style="margin-bottom:6px;">
The discrete points are the values of this continuous spectrum at
the allowed finite-domain eigenvalues <b>λₙ</b>. Thus, the finite
problem samples the continuous Fourier spectrum only at discrete
spectral positions.
</div>

<div style="margin-bottom:6px;">
For the present normalization, the finite-domain Fourier coefficient
and the continuous Fourier spectrum are related by
<b>Cₙ = Δλ D(λₙ)</b>. As L increases, Δλ becomes smaller and the
sampling points become increasingly dense.
</div>

<div>
In the limiting case <b>L → ∞</b>, we have
<b>Δλ → 0</b>. The discrete set of eigenvalues then becomes a
continuous variable λ, and the Fourier-series sum passes naturally
to the corresponding Fourier integral. This is the spectral reason
for the transition from the finite-domain heat-equation solution
to the infinite-domain solution.
</div>

</div>
""")

# ============================================================
# UPDATE FUNCTION
# ============================================================

def update_notebook(change=None):

    L = L_slider.value
    N = N_slider.value

    # --------------------------------------------------------
    # Discrete mode numbers
    # --------------------------------------------------------

    n = np.arange(
        1,
        N + 1
    )

    # --------------------------------------------------------
    # Discrete eigenvalues
    # --------------------------------------------------------

    lambdas = (
        n
        *
        np.pi
        /
        L
    )

    # --------------------------------------------------------
    # Discrete eigenvalue locations
    # --------------------------------------------------------

    eigen_points.set_data(
        lambdas,
        np.ones_like(
            lambdas
        )
        *
        0.60
    )

    # --------------------------------------------------------
    # Eigenvalues visible in right-hand spectrum
    # --------------------------------------------------------

    visible = (
        lambdas
        <=
        LAMBDA_MAX
    )

    visible_lambdas = (
        lambdas[
            visible
        ]
    )

    # --------------------------------------------------------
    # Spectrum samples
    # --------------------------------------------------------

    if len(
        visible_lambdas
    ) > 0:

        D_samples = sine_fourier_transform(
            visible_lambdas
        )

        sample_points.set_data(
            visible_lambdas,
            np.abs(
                D_samples
            )
        )

    else:

        sample_points.set_data(
            [],
            []
        )

    # --------------------------------------------------------
    # Eigenvalue spacing
    # --------------------------------------------------------

    spacing = (
        np.pi
        /
        L
    )

    # --------------------------------------------------------
    # Slider values
    # --------------------------------------------------------

    L_value.value = (
        f'<div class="spec-value">{L:.1f}</div>'
    )

    N_value.value = (
        f'<div class="spec-value">{N}</div>'
    )

    # --------------------------------------------------------
    # Current spectrum
    # --------------------------------------------------------

    current_math.value = (
        r'\('
        r'\Delta\lambda='
        +
        f'{spacing:.6f}'
        +
        r',\;'
        r'\lambda_1='
        +
        f'{lambdas[0]:.6f}'
        +
        r',\;'
        r'\lambda_N='
        +
        f'{lambdas[-1]:.6f}'
        +
        r'\)'
    )

    # --------------------------------------------------------
    # Redraw only
    # --------------------------------------------------------

    fig_spec.canvas.draw_idle()
    fig_env.canvas.draw_idle()

# ============================================================
# CONNECT CONTROLS
# ============================================================

L_slider.observe(
    update_notebook,
    names='value'
)

N_slider.observe(
    update_notebook,
    names='value'
)

# ============================================================
# INITIAL UPDATE
# ============================================================

update_notebook()

# ============================================================
# COMPLETE NOTEBOOK
# ============================================================

main_layout = VBox(
    [
        documentation,
        relations,
        top_row,
        figures_row,
        interpretation_panel
    ],
    layout=Layout(
        width='1160px',
        gap='10px',
        align_items='flex-start',
        margin='0px',
        padding='0px'
    )
)

display(
    main_layout
)
```
