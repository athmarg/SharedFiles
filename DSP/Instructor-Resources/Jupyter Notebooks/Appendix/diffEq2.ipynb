```python
%matplotlib widget

import numpy as np
import sympy as sp
import matplotlib.pyplot as plt

from ipywidgets import (
    FloatSlider,
    IntSlider,
    HTML,
    Output,
    VBox,
    HBox,
    Layout
)

from IPython.display import (
    display,
    Math
)

# ============================================================
# FIRST-ORDER LINEAR DIFFERENCE EQUATION
#
# y[n+1] = n/(n+1) y[n] + n
# ============================================================

plt.ioff()

# ============================================================
# CLASSIC JUPYTER + JUPYTERLAB / NOTEBOOK 7 / BINDER
#
# - no forced scrollbars
# - no trimming
# - no resize triangle
# - output may extend horizontally if necessary
# ============================================================

display(HTML("""
<style>

.container {
    width: 98% !important;
    max-width: none !important;
}

.output_area,
.output_subarea {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

.output_scroll {
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
    box-shadow: none !important;
}

.jp-Cell-outputWrapper,
.jp-OutputArea,
.jp-OutputArea-child,
.jp-OutputArea-output {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

.widget-output,
.jupyter-widgets-output-area,
.widget-box {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

.jupyter-matplotlib,
.jupyter-matplotlib-figure {
    overflow: visible !important;
    resize: none !important;
}

.jupyter-matplotlib::-webkit-resizer,
.jupyter-matplotlib-figure::-webkit-resizer {
    display: none !important;
}

.de-title {
    font-family:Arial, sans-serif;
    font-size:20px;
    font-weight:bold;
    color:#6f3fa0;
}

.de-label {
    font-family:Arial, sans-serif;
    font-size:14px;
    font-weight:bold;
}

.de-value {
    font-family:Arial, sans-serif;
    font-size:14px;
    font-weight:bold;
    color:#0b3d91;
}

</style>
"""))

# ============================================================
# SYMBOLIC VARIABLES
# ============================================================

n = sp.symbols(
    'n',
    integer=True,
    positive=True
)

j = sp.symbols(
    'j',
    integer=True,
    positive=True
)

k = sp.symbols(
    'k',
    integer=True,
    positive=True
)

y1_symbol = sp.symbols(
    'y_1',
    real=True
)

# ============================================================
# DOCUMENTATION
# ============================================================

documentation = HTML("""
<div style="
    width:1040px;
    font-family:Arial, sans-serif;
    font-size:15px;
    line-height:1.48;
    margin-bottom:10px;
">

<div class="de-title" style="margin-bottom:8px;">
First-Order Linear Difference Equation
</div>

<div style="margin-bottom:5px;">
A first-order linear difference equation has the general form
</div>

<div style="
    text-align:center;
    font-size:17px;
    margin:7px 0px;
">
<b>
y[n+1] = p(n)y[n] + q(n).
</b>
</div>

<div style="margin-bottom:5px;">
For the example considered here,
p(n) = n/(n+1) and q(n) = n.
</div>

<div>
<b>This notebook:</b> derives the closed-form solution symbolically,
computes the sequence recursively, and compares the two results.
</div>

</div>
""")

# ============================================================
# SYMBOLIC SOLUTION
#
# General formula:
#
# y[n] =
#
# ( Π p(j) ) y[1]
#
# +
#
# Σ [ ( Π p(j) ) q(k) ]
# ============================================================

p_j = (
    j
    /
    (j + 1)
)

q_k = k

# ------------------------------------------------------------
# First product
#
# Π_{j=1}^{n-1} p(j)
# ------------------------------------------------------------

first_product = sp.product(
    p_j,
    (
        j,
        1,
        n - 1
    )
)

first_product = sp.simplify(
    first_product
)

# ------------------------------------------------------------
# Inner product
#
# Π_{j=k+1}^{n-1} p(j)
# ------------------------------------------------------------

inner_product = sp.product(
    p_j,
    (
        j,
        k + 1,
        n - 1
    )
)

inner_product = sp.simplify(
    inner_product
)

# ------------------------------------------------------------
# Sum term
#
# Σ_{k=1}^{n-1}
# [ Π p(j) ] q(k)
# ------------------------------------------------------------

sum_term = sp.summation(
    inner_product * q_k,
    (
        k,
        1,
        n - 1
    )
)

sum_term = sp.simplify(
    sum_term
)

# ------------------------------------------------------------
# Closed-form solution
# ------------------------------------------------------------

closed_form_symbolic = sp.simplify(
    first_product * y1_symbol
    +
    sum_term
)

# ============================================================
# SLIDER SETTINGS
# ============================================================

slider_style = {
    'description_width': '0px'
}

slider_layout = Layout(
    width='230px'
)

label_layout = Layout(
    width='130px',
    min_width='130px'
)

value_layout = Layout(
    width='70px',
    min_width='70px',
    margin='0px 0px 0px 6px'
)

row_layout = Layout(
    width='450px',
    height='40px',
    align_items='center'
)

# ============================================================
# INITIAL VALUE y[1]
# ============================================================

y1_slider = FloatSlider(
    min=-5.0,
    max=5.0,
    step=0.25,
    value=1.0,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

y1_label = HTML(
    '<div class="de-label">Initial value y[1]:</div>',
    layout=label_layout
)

y1_value = HTML(
    '<div class="de-value">1.00</div>',
    layout=value_layout
)

y1_row = HBox(
    [
        y1_label,
        y1_slider,
        y1_value
    ],
    layout=row_layout
)

# ============================================================
# MAXIMUM n
# ============================================================

N_slider = IntSlider(
    min=5,
    max=25,
    step=1,
    value=15,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

N_label = HTML(
    '<div class="de-label">Maximum n:</div>',
    layout=label_layout
)

N_value = HTML(
    '<div class="de-value">15</div>',
    layout=value_layout
)

N_row = HBox(
    [
        N_label,
        N_slider,
        N_value
    ],
    layout=row_layout
)

# ============================================================
# PARAMETERS PANEL
# ============================================================

parameters_panel = VBox(
    [
        HTML("""
        <div class="de-title" style="margin-bottom:10px;">
            Parameters
        </div>
        """),

        y1_row,
        N_row
    ],
    layout=Layout(
        width='470px',
        padding='10px 14px',
        border='1px solid #d2c2df',
        overflow='visible'
    )
)

# ============================================================
# SYMBOLIC SOLUTION PANEL
# ============================================================

symbolic_output = Output(
    layout=Layout(
        width='560px'
    )
)

symbolic_panel = VBox(
    [
        HTML("""
        <div class="de-title" style="margin-bottom:8px;">
            Symbolic Solution
        </div>
        """),

        symbolic_output
    ],
    layout=Layout(
        width='570px',
        padding='10px 14px',
        border='1px solid #d2c2df',
        overflow='visible'
    )
)

# ============================================================
# DISPLAY SYMBOLIC RESULTS
# ============================================================

with symbolic_output:

    # --------------------------------------------------------
    # Original equation
    # --------------------------------------------------------

    display(
        Math(
            r'y[n+1]'
            r'='
            r'\frac{n}{n+1}y[n]+n'
        )
    )

    # --------------------------------------------------------
    # p(n) and q(n)
    # --------------------------------------------------------

    display(
    Math(
        r'p(n)=\frac{n}{n+1}\quad q(n)=n'
    )
)

    # --------------------------------------------------------
    # First product
    # --------------------------------------------------------

    display(
        Math(
            r'\prod_{j=1}^{n-1}p(j)'
            r'='
            +
            sp.latex(
                first_product
            )
        )
    )

    # --------------------------------------------------------
    # Inner product
    # --------------------------------------------------------

    display(
        Math(
            r'\prod_{j=k+1}^{n-1}p(j)'
            r'='
            +
            sp.latex(
                inner_product
            )
        )
    )

    # --------------------------------------------------------
    # Sum
    # --------------------------------------------------------

    display(
        Math(
            r'\sum_{k=1}^{n-1}'
            r'\left('
            r'\prod_{j=k+1}^{n-1}p(j)'
            r'\right)'
            r'q(k)'
            r'='
            +
            sp.latex(
                sum_term
            )
        )
    )

    # --------------------------------------------------------
    # Final symbolic solution
    # --------------------------------------------------------

    display(
        Math(
            r'y[n]'
            r'='
            +
            sp.latex(
                closed_form_symbolic
            )
        )
    )

# ============================================================
# TOP ROW
# ============================================================

top_row = HBox(
    [
        parameters_panel,
        symbolic_panel
    ],
    layout=Layout(
        width='1070px',
        gap='16px',
        align_items='stretch',
        overflow='visible'
    )
)

# ============================================================
# FIGURE
# ============================================================

fig, ax = plt.subplots(
    figsize=(7.0, 4.8)
)

fig.canvas.header_visible = False
fig.canvas.footer_visible = False
fig.canvas.toolbar_visible = False

fig.canvas.layout = Layout(
    width='700px',
    height='480px',
    overflow='visible'
)

ax.set_title(
    'Recursive and Closed-Form Solutions',
    fontsize=14,
    fontweight='bold',
    color='#6f3fa0'
)

ax.set_xlabel(
    'n',
    fontsize=11
)

ax.set_ylabel(
    'y[n]',
    fontsize=11
)

ax.grid(
    True,
    linestyle=':',
    alpha=0.40
)

# ============================================================
# FIXED AXIS RANGES
#
# Slider movement does NOT change the axes.
# ============================================================

ax.set_xlim(
    0,
    25
)

ax.set_ylim(
    -10,
    230
)

# ============================================================
# CREATE CURVES ONCE
# ============================================================

line_recursive, = ax.plot(
    [],
    [],
    marker='o',
    markersize=5,
    linewidth=1.7,
    label='Recursive solution'
)

line_closed, = ax.plot(
    [],
    [],
    marker='s',
    markersize=4,
    linewidth=1.4,
    linestyle='--',
    label='Closed-form solution'
)

ax.legend(
    loc='upper left',
    fontsize=9
)

fig.subplots_adjust(
    left=0.12,
    right=0.97,
    top=0.90,
    bottom=0.13
)

# ============================================================
# NUMERICAL RESULTS PANEL
# ============================================================

results_html = HTML()

results_panel = VBox(
    [
        HTML("""
        <div class="de-title" style="margin-bottom:8px;">
            Numerical Verification
        </div>
        """),

        results_html
    ],
    layout=Layout(
        width='340px',
        padding='10px 14px',
        border='1px solid #d2c2df',
        overflow='visible'
    )
)

# ============================================================
# FIGURE + RESULTS
# ============================================================

visual_row = HBox(
    [
        fig.canvas,
        results_panel
    ],
    layout=Layout(
        width='1060px',
        gap='14px',
        align_items='flex-start',
        overflow='visible'
    )
)

# ============================================================
# RECURSIVE CALCULATION
# ============================================================

def recursive_solution(
    y1,
    N
):

    # --------------------------------------------------------
    # n = 1,2,...,N
    # --------------------------------------------------------

    n_values = np.arange(
        1,
        N + 1
    )

    y_values = np.zeros(
        N,
        dtype=float
    )

    # --------------------------------------------------------
    # Initial condition
    # --------------------------------------------------------

    y_values[0] = y1

    # --------------------------------------------------------
    # Recursion
    #
    # y[n+1] = n/(n+1) y[n] + n
    # --------------------------------------------------------

    for index in range(
        N - 1
    ):

        current_n = (
            index + 1
        )

        y_values[index + 1] = (
            current_n
            /
            (current_n + 1)
            *
            y_values[index]
            +
            current_n
        )

    return (
        n_values,
        y_values
    )

# ============================================================
# CLOSED-FORM NUMERICAL FUNCTION
# ============================================================

closed_form_function = sp.lambdify(
    (
        n,
        y1_symbol
    ),
    closed_form_symbolic,
    modules='numpy'
)

# ============================================================
# UPDATE FUNCTION
#
# No figure recreation.
# No axis rescaling.
#
# Only existing curve data are updated.
# ============================================================

def update_notebook(change=None):

    y1 = (
        y1_slider.value
    )

    N = (
        N_slider.value
    )

    # ========================================================
    # VALUE LABELS
    # ========================================================

    y1_value.value = (
        f'<div class="de-value">'
        f'{y1:.2f}'
        f'</div>'
    )

    N_value.value = (
        f'<div class="de-value">'
        f'{N}'
        f'</div>'
    )

    # ========================================================
    # RECURSIVE SOLUTION
    # ========================================================

    n_values, recursive_values = recursive_solution(
        y1,
        N
    )

    # ========================================================
    # CLOSED-FORM SOLUTION
    # ========================================================

    closed_values = np.asarray(
        closed_form_function(
            n_values,
            y1
        ),
        dtype=float
    )

    # ========================================================
    # DIFFERENCE BETWEEN THE TWO METHODS
    # ========================================================

    difference = (
        recursive_values
        -
        closed_values
    )

    max_difference = np.max(
        np.abs(
            difference
        )
    )

    # ========================================================
    # UPDATE CURVES ONLY
    # ========================================================

    line_recursive.set_data(
        n_values,
        recursive_values
    )

    line_closed.set_data(
        n_values,
        closed_values
    )

    # ========================================================
    # NUMERICAL TABLE
    # ========================================================

    table_N = min(
        N,
        10
    )

    rows = ""

    for index in range(
        table_N
    ):

        rows += f"""
        <tr>

            <td style="
                padding:5px 8px;
                text-align:center;
                border-bottom:1px solid #eeeeee;
            ">
                {n_values[index]}
            </td>

            <td style="
                padding:5px 8px;
                text-align:center;
                border-bottom:1px solid #eeeeee;
            ">
                {recursive_values[index]:.6f}
            </td>

            <td style="
                padding:5px 8px;
                text-align:center;
                border-bottom:1px solid #eeeeee;
            ">
                {closed_values[index]:.6f}
            </td>

        </tr>
        """

    results_html.value = f"""
    <div style="
        font-family:Arial, sans-serif;
        font-size:13px;
        line-height:1.45;
    ">

        <div style="margin-bottom:8px;">
            Initial condition:
            <b>y[1] = {y1:.2f}</b>
        </div>

        <table style="
            width:100%;
            border-collapse:collapse;
            text-align:center;
        ">

            <tr style="
                background:#f4eff8;
                font-weight:bold;
            ">

                <td style="padding:5px;">
                    n
                </td>

                <td style="padding:5px;">
                    Recursive
                </td>

                <td style="padding:5px;">
                    Closed form
                </td>

            </tr>

            {rows}

        </table>

        <hr style="margin:9px 0px;">

        Maximum absolute difference:

        <br>

        <b>{max_difference:.3e}</b>

    </div>
    """

    # ========================================================
    # REDRAW ONLY
    # ========================================================

    fig.canvas.draw_idle()

# ============================================================
# CONNECT CONTROLS
# ============================================================

y1_slider.observe(
    update_notebook,
    names='value'
)

N_slider.observe(
    update_notebook,
    names='value'
)

# ============================================================
# INITIAL UPDATE
# ============================================================

update_notebook()

# ============================================================
# COMPLETE NOTEBOOK
# ============================================================

main_layout = VBox(
    [
        documentation,
        top_row,
        visual_row
    ],
    layout=Layout(
        width='1080px',
        gap='10px',
        overflow='visible'
    )
)

display(
    main_layout
)
```
