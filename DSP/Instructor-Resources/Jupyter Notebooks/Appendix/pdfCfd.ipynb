```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, uniform, expon
from ipywidgets import Dropdown, FloatSlider, interactive
from IPython.display import display, Markdown

# ------------------------------------------------------------
# PDF, CDF and Interval Probability
# ------------------------------------------------------------

def pdf_cdf_demo(distribution, a, b):

    if a > b:
        a, b = b, a

    if distribution == 'Gaussian':
        x = np.linspace(-5, 5, 1200)
        pdf = norm.pdf(x, loc=0, scale=1)
        cdf = norm.cdf(x, loc=0, scale=1)

        Fa = norm.cdf(a)
        Fb = norm.cdf(b)

    elif distribution == 'Uniform':
        x = np.linspace(-2, 2, 1200)
        pdf = uniform.pdf(x, loc=-1, scale=2)
        cdf = uniform.cdf(x, loc=-1, scale=2)

        Fa = uniform.cdf(a, loc=-1, scale=2)
        Fb = uniform.cdf(b, loc=-1, scale=2)

    elif distribution == 'Exponential':
        x = np.linspace(-1, 6, 1200)
        pdf = expon.pdf(x, scale=1)
        cdf = expon.cdf(x, scale=1)

        Fa = expon.cdf(a, scale=1)
        Fb = expon.cdf(b, scale=1)

    probability = Fb - Fa

    fig, axes = plt.subplots(1, 2, figsize=(12, 4.5))

    # PDF
    axes[0].plot(x, pdf, linewidth=2)
    mask = (x >= a) & (x <= b)
    axes[0].fill_between(x[mask], pdf[mask], alpha=0.35)

    axes[0].axvline(a, linestyle='--')
    axes[0].axvline(b, linestyle='--')

    axes[0].set_title('Probability Density Function')
    axes[0].set_xlabel('x')
    axes[0].set_ylabel('$f_X(x)$')
    axes[0].grid(True, alpha=0.3)

    # CDF
    axes[1].plot(x, cdf, linewidth=2)
    axes[1].axvline(a, linestyle='--')
    axes[1].axvline(b, linestyle='--')
    axes[1].axhline(Fa, linestyle=':')
    axes[1].axhline(Fb, linestyle=':')

    axes[1].set_title('Cumulative Distribution Function')
    axes[1].set_xlabel('x')
    axes[1].set_ylabel('$F_X(x)$')
    axes[1].set_ylim(-0.05, 1.05)
    axes[1].grid(True, alpha=0.3)

    plt.tight_layout()
    plt.show()
    plt.close(fig)

    print(f'a = {a:.2f}')
    print(f'b = {b:.2f}')
    print()
    print(f'F(a) = {Fa:.6f}')
    print(f'F(b) = {Fb:.6f}')
    print()
    print(f'P(a < X < b) = F(b) - F(a) = {probability:.6f}')


distribution_widget = Dropdown(
    options=['Gaussian', 'Uniform', 'Exponential'],
    value='Gaussian',
    description='Distribution:'
)

a_widget = FloatSlider(
    value=-1,
    min=-4,
    max=4,
    step=0.1,
    description='a:',
    continuous_update=True
)

b_widget = FloatSlider(
    value=1,
    min=-4,
    max=4,
    step=0.1,
    description='b:',
    continuous_update=True
)

display(Markdown("## PDF, CDF and Interval Probability"))

display(
    interactive(
        pdf_cdf_demo,
        distribution=distribution_widget,
        a=a_widget,
        b=b_widget
    )
)
```
