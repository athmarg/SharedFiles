```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom, norm

from ipywidgets import (
    IntSlider,
    FloatSlider,
    VBox,
    HBox,
    Output,
    Layout
)

from IPython.display import display, Markdown, clear_output

# ------------------------------------------------------------
# Discrete versus Continuous Random Variables
# ------------------------------------------------------------

# ============================================================
# CONTROLS
# ============================================================

n_slider = IntSlider(
    value=10,
    min=1,
    max=30,
    step=1,
    description='n:',
    continuous_update=True
)

p_slider = FloatSlider(
    value=0.5,
    min=0.05,
    max=0.95,
    step=0.05,
    description='p:',
    continuous_update=True
)

mu_slider = FloatSlider(
    value=0,
    min=-3,
    max=3,
    step=0.1,
    description='μ:',
    continuous_update=True
)

sigma_slider = FloatSlider(
    value=1,
    min=0.2,
    max=3,
    step=0.1,
    description='σ:',
    continuous_update=True
)

# ------------------------------------------------------------
# Controls in two columns
# ------------------------------------------------------------

left_controls = VBox(
    [
        n_slider,
        p_slider
    ],
    layout=Layout(
        width='360px'
    )
)

right_controls = VBox(
    [
        mu_slider,
        sigma_slider
    ],
    layout=Layout(
        width='360px'
    )
)

controls_row = HBox(
    [
        left_controls,
        right_controls
    ],
    layout=Layout(
        width='760px',
        gap='40px',
        align_items='flex-start'
    )
)

# ============================================================
# OUTPUT AREAS
# ============================================================

plot_output = Output()

discrete_output = Output(
    layout=Layout(
        width='360px'
    )
)

continuous_output = Output(
    layout=Layout(
        width='420px'
    )
)

results_row = HBox(
    [
        discrete_output,
        continuous_output
    ],
    layout=Layout(
        width='820px',
        gap='40px',
        align_items='flex-start'
    )
)

# ============================================================
# FIXED AXIS FOR THE CONTINUOUS CASE
# ============================================================

X_MIN = -10.0
X_MAX = 10.0

x = np.linspace(
    X_MIN,
    X_MAX,
    1600
)

# ============================================================
# UPDATE FUNCTION
# ============================================================

def discrete_continuous_demo(change=None):

    n = n_slider.value
    p = p_slider.value
    mu = mu_slider.value
    sigma = sigma_slider.value

    # --------------------------------------------------------
    # Discrete random variable: Binomial
    # --------------------------------------------------------

    k = np.arange(
        0,
        n + 1
    )

    pmf = binom.pmf(
        k,
        n,
        p
    )

    cdf_discrete = binom.cdf(
        k,
        n,
        p
    )

    # --------------------------------------------------------
    # Continuous random variable: Gaussian
    #
    # IMPORTANT:
    # x is fixed. Only the Gaussian changes.
    # --------------------------------------------------------

    pdf = norm.pdf(
        x,
        loc=mu,
        scale=sigma
    )

    cdf_continuous = norm.cdf(
        x,
        loc=mu,
        scale=sigma
    )

    # ========================================================
    # PLOTS
    # ========================================================

    with plot_output:

        clear_output(wait=True)

        fig, axes = plt.subplots(
            2,
            2,
            figsize=(12, 7)
        )

        # ----------------------------------------------------
        # PMF
        # ----------------------------------------------------

        axes[0, 0].stem(
            k,
            pmf
        )

        axes[0, 0].set_title(
            'Discrete Random Variable — PMF'
        )

        axes[0, 0].set_xlabel('k')

        axes[0, 0].set_ylabel(
            '$P(X=k)$'
        )

        axes[0, 0].grid(
            True,
            alpha=0.3
        )

        # ----------------------------------------------------
        # Discrete CDF
        # ----------------------------------------------------

        axes[0, 1].step(
            k,
            cdf_discrete,
            where='post',
            linewidth=2
        )

        axes[0, 1].set_title(
            'Discrete Random Variable — CDF'
        )

        axes[0, 1].set_xlabel('k')

        axes[0, 1].set_ylabel(
            '$F_X(k)$'
        )

        axes[0, 1].set_ylim(
            -0.05,
            1.05
        )

        axes[0, 1].grid(
            True,
            alpha=0.3
        )

        # ----------------------------------------------------
        # Continuous PDF
        # ----------------------------------------------------

        axes[1, 0].plot(
            x,
            pdf,
            linewidth=2
        )

        axes[1, 0].set_title(
            'Continuous Random Variable — PDF'
        )

        axes[1, 0].set_xlabel('x')

        axes[1, 0].set_ylabel(
            '$f_X(x)$'
        )

        # Fixed horizontal axis
        axes[1, 0].set_xlim(
            X_MIN,
            X_MAX
        )

        # Fixed vertical axis
        axes[1, 0].set_ylim(
            0.0,
            2.1
        )

        # Mean position
        axes[1, 0].axvline(
            mu,
            linestyle='--',
            linewidth=1.2,
            alpha=0.7
        )

        axes[1, 0].grid(
            True,
            alpha=0.3
        )

        # ----------------------------------------------------
        # Continuous CDF
        # ----------------------------------------------------

        axes[1, 1].plot(
            x,
            cdf_continuous,
            linewidth=2
        )

        axes[1, 1].set_title(
            'Continuous Random Variable — CDF'
        )

        axes[1, 1].set_xlabel('x')

        axes[1, 1].set_ylabel(
            '$F_X(x)$'
        )

        # Fixed axes
        axes[1, 1].set_xlim(
            X_MIN,
            X_MAX
        )

        axes[1, 1].set_ylim(
            -0.05,
            1.05
        )

        # Mean position
        axes[1, 1].axvline(
            mu,
            linestyle='--',
            linewidth=1.2,
            alpha=0.7
        )

        axes[1, 1].grid(
            True,
            alpha=0.3
        )

        plt.tight_layout()
        plt.show()
        plt.close(fig)

    # ========================================================
    # DISCRETE RESULTS
    # ========================================================

    with discrete_output:

        clear_output(wait=True)

        print('DISCRETE CASE')
        print(
            f'Binomial: n = {n}, p = {p:.2f}'
        )
        print(
            f'Sum of PMF values = {np.sum(pmf):.8f}'
        )

    # ========================================================
    # CONTINUOUS RESULTS
    # ========================================================

    with continuous_output:

        clear_output(wait=True)

        print('CONTINUOUS CASE')
        print(
            f'Gaussian: μ = {mu:.2f}, σ = {sigma:.2f}'
        )

        print(
            f'Variance: σ² = {sigma**2:.4f}'
        )

        print(
            f'Numerical integral of PDF ≈ '
            f'{np.trapezoid(pdf, x):.8f}'
        )

# ============================================================
# CONNECT CONTROLS
# ============================================================

for widget in [
    n_slider,
    p_slider,
    mu_slider,
    sigma_slider
]:

    widget.observe(
        discrete_continuous_demo,
        names='value'
    )

# ============================================================
# DISPLAY
# ============================================================

display(
    Markdown(
        "## Discrete and Continuous Random Variables"
    )
)

display(
    controls_row
)

display(
    plot_output
)

display(
    results_row
)

# ============================================================
# INITIAL UPDATE
# ============================================================

discrete_continuous_demo()
```
