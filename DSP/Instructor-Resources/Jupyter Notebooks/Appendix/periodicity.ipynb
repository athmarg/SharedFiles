```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt
import mpmath as mp

from ipywidgets import (
    FloatSlider,
    HTML,
    HTMLMath,
    VBox,
    HBox,
    Layout
)

from IPython.display import display

# ============================================================
# DOUBLE PERIODICITY OF THE JACOBI ELLIPTIC SINE
#
# sn(u,k),  u = x + i y
#
# Real period:
#     4K
#
# Imaginary period:
#     2iK'
#
# where
#
#     K  = K(k)
#     K' = K(k')
#     k' = sqrt(1-k^2)
#
# mpmath uses the parameter m = k^2.
# ============================================================

plt.ioff()

mp.mp.dps = 20

# ============================================================
# JUPYTER / BINDER DISPLAY SETTINGS
# ============================================================

display(HTML("""
<style>

.container {
    width: 98% !important;
    max-width: none !important;
}

.output_area,
.output_subarea {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

.output_scroll {
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
    box-shadow: none !important;
}

.jp-Cell-outputWrapper,
.jp-OutputArea,
.jp-OutputArea-child,
.jp-OutputArea-output {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

.widget-output,
.jupyter-widgets-output-area,
.widget-box {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

.jupyter-matplotlib,
.jupyter-matplotlib-figure {
    overflow: visible !important;
    resize: none !important;
}

.jupyter-matplotlib::-webkit-resizer,
.jupyter-matplotlib-figure::-webkit-resizer {
    display: none !important;
}

.dp-title {
    font-family: Arial, sans-serif;
    font-size: 20px;
    font-weight: bold;
    color: #6f3fa0;
}

.dp-label {
    font-family: Arial, sans-serif;
    font-size: 14px;
    font-weight: bold;
}

.dp-value {
    font-family: Arial, sans-serif;
    font-size: 14px;
    font-weight: bold;
    color: #0b3d91;
}

</style>
"""))

# ============================================================
# DOCUMENTATION
# ============================================================

documentation = HTML("""
<div style="
    width:1180px;
    font-family:Arial, sans-serif;
    font-size:15px;
    line-height:1.48;
    margin-bottom:10px;
">

<div class="dp-title" style="margin-bottom:8px;">
Double Periodicity of the Jacobi Elliptic Sine
</div>

<div style="margin-bottom:5px;">
The Jacobi elliptic sine sn(u,k) is a doubly periodic function
of the complex variable u=x+iy.
</div>

<div style="margin-bottom:5px;">
Its real period is 4K and its imaginary period is 2iK′.
The axes of the complex plane are normalized as Re(u)/K
and Im(u)/K′, so the periodic structure remains visible
for every value of the elliptic modulus k.
</div>

<div>
<b>This notebook:</b> displays the magnitude pattern of sn(u,k)
on the complex plane and numerically verifies both periods
at a selected complex point u.
</div>

<div style="
    margin-top:10px;
    padding:8px 12px;
    border:1px solid #cc0000;
    font-family:Arial, sans-serif;
    font-size:15px;
    font-weight:bold;
    color:#cc0000;
">
Note: Due to the high computational requirements of evaluating the
Jacobi elliptic function over a complex two-dimensional grid, the
graph may require a few seconds to update after changing the elliptic
modulus k. Please allow the computation to complete.
</div>

</div>
""")

# ============================================================
# MATHEMATICAL DEFINITIONS
# ============================================================

real_period_math = HTMLMath(
    value=(
        r'\('
        r'\operatorname{sn}(u+4K,k)'
        r'='
        r'\operatorname{sn}(u,k)'
        r'\)'
    )
)

imag_period_math = HTMLMath(
    value=(
        r'\('
        r'\operatorname{sn}(u+2iK^{\prime},k)'
        r'='
        r'\operatorname{sn}(u,k)'
        r'\)'
    )
)

period_definitions = HBox(
    [
        real_period_math,
        imag_period_math
    ],
    layout=Layout(
        width='900px',
        gap='45px',
        align_items='center',
        overflow='visible'
    )
)

# ============================================================
# SLIDER STYLE
# ============================================================

slider_style = {
    'description_width': '0px'
}

slider_layout = Layout(
    width='230px'
)

label_layout = Layout(
    width='170px',
    min_width='170px'
)

value_layout = Layout(
    width='70px',
    min_width='70px',
    margin='0px 0px 0px 6px'
)

row_layout = Layout(
    width='490px',
    height='39px',
    align_items='center'
)

# ============================================================
# k SLIDER
# ============================================================

k_slider = FloatSlider(
    min=0.10,
    max=0.95,
    step=0.05,
    value=0.70,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

k_label = HTML(
    '<div class="dp-label">Elliptic modulus k:</div>',
    layout=label_layout
)

k_value = HTML(
    '<div class="dp-value">0.70</div>',
    layout=value_layout
)

k_row = HBox(
    [
        k_label,
        k_slider,
        k_value
    ],
    layout=row_layout
)

# ============================================================
# NORMALIZED REAL PART
#
# a = Re(u)/K
# ============================================================

a_slider = FloatSlider(
    min=-1.50,
    max=1.50,
    step=0.05,
    value=0.50,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

a_label = HTML(
    '<div class="dp-label">Re(u) / K:</div>',
    layout=label_layout
)

a_value = HTML(
    '<div class="dp-value">0.50</div>',
    layout=value_layout
)

a_row = HBox(
    [
        a_label,
        a_slider,
        a_value
    ],
    layout=row_layout
)

# ============================================================
# NORMALIZED IMAGINARY PART
#
# b = Im(u)/K'
# ============================================================

b_slider = FloatSlider(
    min=-0.75,
    max=0.75,
    step=0.05,
    value=0.30,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

b_label = HTML(
    '<div class="dp-label">Im(u) / K′:</div>',
    layout=label_layout
)

b_value = HTML(
    '<div class="dp-value">0.30</div>',
    layout=value_layout
)

b_row = HBox(
    [
        b_label,
        b_slider,
        b_value
    ],
    layout=row_layout
)

# ============================================================
# PARAMETERS PANEL
# ============================================================

parameters_panel = VBox(
    [
        HTML("""
        <div class="dp-title" style="margin-bottom:8px;">
            Parameters
        </div>
        """),

        k_row,
        a_row,
        b_row
    ],
    layout=Layout(
        width='520px',
        padding='10px 14px',
        border='1px solid #d2c2df',
        overflow='visible'
    )
)

# ============================================================
# CURRENT VALUES WIDGETS
# ============================================================

K_math = HTMLMath()
Kp_math = HTMLMath()

u_real_math = HTMLMath()
u_imag_math = HTMLMath()

sn_real_math = HTMLMath()
sn_imag_math = HTMLMath()

# ============================================================
# CURRENT VALUES PANEL
# ============================================================

current_values_panel = VBox(
    [
        HTML("""
        <div class="dp-title" style="margin-bottom:8px;">
            Current Values
        </div>
        """),

        HBox(
            [
                K_math,
                Kp_math
            ],
            layout=Layout(
                width='500px',
                gap='30px',
                overflow='visible'
            )
        ),

        HBox(
            [
                u_real_math,
                u_imag_math
            ],
            layout=Layout(
                width='540px',
                gap='30px',
                overflow='visible'
            )
        ),

        HBox(
            [
                sn_real_math,
                sn_imag_math
            ],
            layout=Layout(
                width='570px',
                gap='30px',
                overflow='visible'
            )
        )
    ],
    layout=Layout(
        width='620px',
        padding='10px 14px',
        border='1px solid #d2c2df',
        overflow='visible'
    )
)

# ============================================================
# TOP ROW
# ============================================================

top_row = HBox(
    [
        parameters_panel,
        current_values_panel
    ],
    layout=Layout(
        width='1160px',
        gap='15px',
        align_items='stretch',
        overflow='visible'
    )
)

# ============================================================
# NORMALIZED COMPLEX-PLANE GRID
#
# Horizontal:
#
#     X = Re(u)/K
#
# Vertical:
#
#     Y = Im(u)/K'
#
# The displayed area contains:
#
#     two real periods
#     two imaginary periods
# ============================================================

NX = 70
NY = 55

X_axis = np.linspace(
    -4.0,
    4.0,
    NX
)

Y_axis = np.linspace(
    -2.0,
    2.0,
    NY
)

# ============================================================
# JACOBI sn
# ============================================================

def jacobi_sn(z, m):

    return mp.ellipfun(
        'sn',
        z,
        m
    )

# ============================================================
# COMPLETE ELLIPTIC INTEGRALS
# ============================================================

def elliptic_constants(k):

    m = (
        k**2
    )

    kp = np.sqrt(
        1.0
        -
        k**2
    )

    K = float(
        mp.ellipk(
            m
        )
    )

    Kp = float(
        mp.ellipk(
            1.0 - m
        )
    )

    return (
        m,
        kp,
        K,
        Kp
    )

# ============================================================
# HEATMAP CALCULATION
#
# We display
#
# log10(1 + |sn(u,k)|)
#
# instead of |sn| itself because sn has poles.
#
# The display is clipped only for visualization.
# ============================================================

def calculate_heatmap(k):

    m, kp, K, Kp = elliptic_constants(
        k
    )

    data = np.zeros(
        (
            NY,
            NX
        ),
        dtype=float
    )

    for iy, Y in enumerate(
        Y_axis
    ):

        for ix, X in enumerate(
            X_axis
        ):

            u = (
                X * K
                +
                1j * Y * Kp
            )

            try:

                sn_value = jacobi_sn(
                    u,
                    m
                )

                magnitude = float(
                    abs(
                        sn_value
                    )
                )

                value = np.log10(
                    1.0
                    +
                    magnitude
                )

                if not np.isfinite(
                    value
                ):

                    value = 2.5

            except Exception:

                value = 2.5

            data[iy, ix] = min(
                value,
                2.5
            )

    return (
        data,
        m,
        K,
        Kp
    )

# ============================================================
# INITIAL HEATMAP
# ============================================================

(
    heatmap_initial,
    m_initial,
    K_initial,
    Kp_initial
) = calculate_heatmap(
    k_slider.value
)

# ============================================================
# HEATMAP FIGURE
# ============================================================

fig_map, ax_map = plt.subplots(
    figsize=(7.4, 5.4)
)

fig_map.canvas.header_visible = False
fig_map.canvas.footer_visible = False
fig_map.canvas.toolbar_visible = False

fig_map.canvas.layout = Layout(
    width='740px',
    height='540px',
    overflow='visible'
)

ax_map.set_title(
    'Double Periodicity of |sn(u,k)|',
    fontsize=14,
    fontweight='bold',
    color='#6f3fa0'
)

ax_map.set_xlabel(
    'Re(u) / K',
    fontsize=11
)

ax_map.set_ylabel(
    'Im(u) / K′',
    fontsize=11
)

# ============================================================
# IMAGE CREATED ONCE
# ============================================================

heatmap_image = ax_map.imshow(
    heatmap_initial,
    extent=[
        -4.0,
        4.0,
        -2.0,
        2.0
    ],
    origin='lower',
    aspect='auto',
    vmin=0.0,
    vmax=2.5,
    interpolation='bilinear'
)

# ============================================================
# COLORBAR
# ============================================================

colorbar = fig_map.colorbar(
    heatmap_image,
    ax=ax_map,
    fraction=0.046,
    pad=0.04
)

colorbar.set_label(
    'log₁₀(1 + |sn(u,k)|)',
    fontsize=10
)

# ============================================================
# PERIOD BOUNDARIES
#
# Real period = 4K
#
# Imaginary period = 2K'
#
# In normalized coordinates these are fixed.
# ============================================================

for X_line in [
    -4.0,
    0.0,
    4.0
]:

    ax_map.axvline(
        X_line,
        linestyle='--',
        linewidth=1.2,
        alpha=0.75
    )

for Y_line in [
    -2.0,
    0.0,
    2.0
]:

    ax_map.axhline(
        Y_line,
        linestyle='--',
        linewidth=1.2,
        alpha=0.75
    )

# ============================================================
# SELECTED COMPLEX POINT
# ============================================================

selected_point, = ax_map.plot(
    [],
    [],
    linestyle='None',
    marker='o',
    markersize=9,
    markerfacecolor='none',
    markeredgewidth=2.0
)

ax_map.set_xlim(
    -4.0,
    4.0
)

ax_map.set_ylim(
    -2.0,
    2.0
)

fig_map.subplots_adjust(
    left=0.11,
    right=0.91,
    top=0.90,
    bottom=0.13
)

# ============================================================
# PERIOD VERIFICATION WIDGETS
# ============================================================

original_math = HTMLMath()

real_shift_math = HTMLMath()
real_error_math = HTMLMath()

imag_shift_math = HTMLMath()
imag_error_math = HTMLMath()

# ============================================================
# PERIOD VERIFICATION PANEL
# ============================================================

verification_panel = VBox(
    [
        HTML("""
        <div class="dp-title" style="margin-bottom:8px;">
            Period Verification
        </div>
        """),

        original_math,

        HTML("""
        <div style="
            font-family:Arial;
            font-size:13px;
            font-weight:bold;
            color:#6f3fa0;
            margin-top:7px;
            margin-bottom:3px;
        ">
            Real period
        </div>
        """),

        real_shift_math,
        real_error_math,

        HTML("""
        <div style="
            font-family:Arial;
            font-size:13px;
            font-weight:bold;
            color:#6f3fa0;
            margin-top:8px;
            margin-bottom:3px;
        ">
            Imaginary period
        </div>
        """),

        imag_shift_math,
        imag_error_math
    ],
    layout=Layout(
        width='390px',
        padding='10px 14px',
        border='1px solid #d2c2df',
        overflow='visible'
    )
)

# ============================================================
# INTERPRETATION PANEL
# ============================================================

interpretation_panel = HTML("""
<div style="
    width:360px;
    padding:9px 12px;
    border:1px solid #d7c7e5;
    font-family:Arial, sans-serif;
    font-size:14px;
    line-height:1.52;
    box-sizing:border-box;
    margin-top:10px;
">

<div style="
    color:#6f3fa0;
    font-size:17px;
    font-weight:bold;
    margin-bottom:6px;
">
Interpretation
</div>

<div style="margin-bottom:6px;">
The repeated pattern along the horizontal direction corresponds
to the real period 4K.
</div>

<div style="margin-bottom:6px;">
The repeated pattern along the vertical direction corresponds
to the imaginary period 2iK′.
</div>

<div>
The bright regions are associated with large values of |sn(u,k)|
near its poles. A logarithmic magnitude scale is used so that both
regular regions and neighborhoods of poles remain visible.
</div>

</div>
""")

# ============================================================
# RIGHT COLUMN
# ============================================================

right_column = VBox(
    [
        verification_panel,
        interpretation_panel
    ],
    layout=Layout(
        width='400px',
        overflow='visible'
    )
)

# ============================================================
# VISUAL ROW
# ============================================================

visual_row = HBox(
    [
        fig_map.canvas,
        right_column
    ],
    layout=Layout(
        width='1160px',
        gap='15px',
        align_items='flex-start',
        overflow='visible'
    )
)

# ============================================================
# COMPLEX NUMBER FORMAT FOR LATEX
# ============================================================

def complex_components(z):

    z = complex(
        z
    )

    real_part = z.real
    imag_part = z.imag

    if abs(
        real_part
    ) < 1e-12:

        real_part = 0.0

    if abs(
        imag_part
    ) < 1e-12:

        imag_part = 0.0

    return (
        real_part,
        imag_part
    )

# ============================================================
# UPDATE SELECTED POINT AND PERIOD CHECKS
#
# This function does NOT recompute the heatmap.
# ============================================================

def update_selected_point(change=None):

    k = (
        k_slider.value
    )

    a = (
        a_slider.value
    )

    b = (
        b_slider.value
    )

    (
        m,
        kp,
        K,
        Kp
    ) = elliptic_constants(
        k
    )

    # ========================================================
    # SELECTED COMPLEX ARGUMENT
    # ========================================================

    u = (
        a * K
        +
        1j * b * Kp
    )

    # ========================================================
    # VALUES AT u AND PERIOD-SHIFTED POINTS
    # ========================================================

    sn_u = jacobi_sn(
        u,
        m
    )

    sn_real_shift = jacobi_sn(
        u + 4.0*K,
        m
    )

    sn_imag_shift = jacobi_sn(
        u + 2.0j*Kp,
        m
    )

    # ========================================================
    # PERIOD ERRORS
    # ========================================================

    real_error = float(
        abs(
            sn_real_shift
            -
            sn_u
        )
    )

    imag_error = float(
        abs(
            sn_imag_shift
            -
            sn_u
        )
    )

    # ========================================================
    # COMPLEX COMPONENTS
    # ========================================================

    u_re, u_im = complex_components(
        u
    )

    sn_re, sn_im = complex_components(
        sn_u
    )

    rs_re, rs_im = complex_components(
        sn_real_shift
    )

    is_re, is_im = complex_components(
        sn_imag_shift
    )

    # ========================================================
    # SLIDER VALUES
    # ========================================================

    k_value.value = (
        f'<div class="dp-value">{k:.2f}</div>'
    )

    a_value.value = (
        f'<div class="dp-value">{a:.2f}</div>'
    )

    b_value.value = (
        f'<div class="dp-value">{b:.2f}</div>'
    )

    # ========================================================
    # K AND K'
    # ========================================================

    K_math.value = (
        r'\('
        r'K='
        +
        f'{K:.6f}'
        +
        r'\)'
    )

    Kp_math.value = (
        r'\('
        r'K^{\prime}='
        +
        f'{Kp:.6f}'
        +
        r'\)'
    )

    # ========================================================
    # u
    # ========================================================

    u_real_math.value = (
        r'\('
        r'\operatorname{Re}(u)='
        +
        f'{u_re:.6f}'
        +
        r'\)'
    )

    u_imag_math.value = (
        r'\('
        r'\operatorname{Im}(u)='
        +
        f'{u_im:.6f}'
        +
        r'\)'
    )

    # ========================================================
    # sn(u,k)
    # ========================================================

    sn_real_math.value = (
        r'\('
        r'\operatorname{Re}[\operatorname{sn}(u,k)]='
        +
        f'{sn_re:.6f}'
        +
        r'\)'
    )

    sn_imag_math.value = (
        r'\('
        r'\operatorname{Im}[\operatorname{sn}(u,k)]='
        +
        f'{sn_im:.6f}'
        +
        r'\)'
    )

    # ========================================================
    # ORIGINAL VALUE
    # ========================================================

    original_math.value = (
        r'\('
        r'\operatorname{sn}(u,k)='
        +
        f'{sn_re:.6f}'
        +
        (
            '+'
            if sn_im >= 0
            else '-'
        )
        +
        f'{abs(sn_im):.6f}'
        +
        r'i'
        r'\)'
    )

    # ========================================================
    # REAL PERIOD SHIFT
    # ========================================================

    real_shift_math.value = (
        r'\('
        r'\operatorname{sn}(u+4K,k)='
        +
        f'{rs_re:.6f}'
        +
        (
            '+'
            if rs_im >= 0
            else '-'
        )
        +
        f'{abs(rs_im):.6f}'
        +
        r'i'
        r'\)'
    )

    real_error_math.value = (
        r'\('
        r'|\Delta_{\mathrm{real}}|='
        +
        f'{real_error:.3e}'
        +
        r'\)'
    )

    # ========================================================
    # IMAGINARY PERIOD SHIFT
    # ========================================================

    imag_shift_math.value = (
        r'\('
        r'\operatorname{sn}(u+2iK^{\prime},k)='
        +
        f'{is_re:.6f}'
        +
        (
            '+'
            if is_im >= 0
            else '-'
        )
        +
        f'{abs(is_im):.6f}'
        +
        r'i'
        r'\)'
    )

    imag_error_math.value = (
        r'\('
        r'|\Delta_{\mathrm{imag}}|='
        +
        f'{imag_error:.3e}'
        +
        r'\)'
    )

    # ========================================================
    # SELECTED POINT ON NORMALIZED PLANE
    # ========================================================

    selected_point.set_data(
        [a],
        [b]
    )

    fig_map.canvas.draw_idle()

# ============================================================
# UPDATE k
#
# Recompute the heatmap only when k changes.
# ============================================================

def update_k(change=None):

    k = (
        k_slider.value
    )

    (
        new_heatmap,
        m,
        K,
        Kp
    ) = calculate_heatmap(
        k
    )

    # ========================================================
    # UPDATE IMAGE DATA ONLY
    # ========================================================

    heatmap_image.set_data(
        new_heatmap
    )

    # ========================================================
    # UPDATE SELECTED POINT AND NUMERICAL VALUES
    # ========================================================

    update_selected_point()

    # ========================================================
    # REDRAW
    # ========================================================

    fig_map.canvas.draw_idle()

# ============================================================
# CONNECT CONTROLS
# ============================================================

k_slider.observe(
    update_k,
    names='value'
)

a_slider.observe(
    update_selected_point,
    names='value'
)

b_slider.observe(
    update_selected_point,
    names='value'
)

# ============================================================
# INITIAL UPDATE
# ============================================================

update_selected_point()

# ============================================================
# COMPLETE NOTEBOOK
# ============================================================

main_layout = VBox(
    [
        documentation,
        period_definitions,
        top_row,
        visual_row
    ],
    layout=Layout(
        width='1180px',
        gap='10px',
        overflow='visible'
    )
)

display(
    main_layout
)
```
