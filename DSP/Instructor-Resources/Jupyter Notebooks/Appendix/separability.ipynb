```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt

from ipywidgets import (
    FloatSlider,
    RadioButtons,
    HTML,
    HTMLMath,
    VBox,
    HBox,
    Layout
)

from IPython.display import display

# ============================================================
# SEPARABILITY OF THE TWO-DIMENSIONAL FOURIER TRANSFORM
#
# f(x,y) = f1(x) f2(y)
#
# then
#
# F(fx,fy) = F1(fx) F2(fy)
# ============================================================

plt.ioff()

# ============================================================
# JUPYTER / BINDER DISPLAY SETTINGS
# ============================================================

display(HTML("""
<style>

.container {
    width:98% !important;
    max-width:none !important;
}

.output_area,
.output_subarea {
    max-width:none !important;
    height:auto !important;
    max-height:none !important;
    overflow:visible !important;
}

.output_scroll {
    height:auto !important;
    max-height:none !important;
    overflow:visible !important;
    box-shadow:none !important;
}

.jp-Cell-outputWrapper,
.jp-OutputArea,
.jp-OutputArea-child,
.jp-OutputArea-output {
    max-width:none !important;
    height:auto !important;
    max-height:none !important;
    overflow:visible !important;
}

.widget-output,
.jupyter-widgets-output-area,
.widget-box {
    max-width:none !important;
    height:auto !important;
    max-height:none !important;
    overflow:visible !important;
}

.jupyter-matplotlib,
.jupyter-matplotlib-figure {
    overflow:visible !important;
    resize:none !important;
}

.jupyter-matplotlib::-webkit-resizer,
.jupyter-matplotlib-figure::-webkit-resizer {
    display:none !important;
}

.sep-title {
    font-family:Arial, sans-serif;
    font-size:20px;
    font-weight:bold;
    color:#6f3fa0;
}

.sep-label {
    font-family:Arial, sans-serif;
    font-size:14px;
    font-weight:bold;
}

.sep-value {
    font-family:Arial, sans-serif;
    font-size:14px;
    font-weight:bold;
    color:#0b3d91;
}

.sep-radio .widget-radio-box {
    display:flex !important;
    flex-direction:row !important;
    flex-wrap:nowrap !important;
    gap:18px !important;
    align-items:center !important;
}

.sep-radio .widget-radio-box label {
    margin:0 !important;
    white-space:nowrap !important;
}

.sep-radio > label {
    display:none !important;
}

</style>
"""))

# ============================================================
# DOCUMENTATION
# ============================================================

documentation = HTML("""
<div style="
    width:1200px;
    font-family:Arial, sans-serif;
    font-size:15px;
    line-height:1.48;
    margin-bottom:10px;
">

<div class="sep-title" style="margin-bottom:8px;">
Separability of the Two-Dimensional Fourier Transform
</div>

<div style="margin-bottom:5px;">
A two-dimensional function is separable when it can be written
as the product f(x,y)=f₁(x)f₂(y).
</div>

<div style="margin-bottom:5px;">
For such a function, the two-dimensional Fourier transform also
separates into the product F(fₓ,fᵧ)=F₁(fₓ)F₂(fᵧ).
</div>

<div style="margin-bottom:5px;">
The two function types used in this notebook are separable because
their two-dimensional forms can be expressed as products of
one-dimensional functions.
</div>

<div style="margin-bottom:5px;">
For a Gaussian function,
</div>

<div style="
    margin-left:20px;
    margin-bottom:5px;
">
f(x,y)=exp[−(x²/aₓ²+y²/aᵧ²)]
=exp(−x²/aₓ²) exp(−y²/aᵧ²)
=f₁(x)f₂(y).
</div>

<div style="margin-bottom:5px;">
For a rectangular function,
</div>

<div style="
    margin-left:20px;
    margin-bottom:5px;
">
f(x,y)=rect(x/2aₓ) rect(y/2aᵧ)
=f₁(x)f₂(y).
</div>

<div style="margin-bottom:5px;">
The one-dimensional factors f₁ and f₂ do not need to be identical.
Different widths may be used in the x and y directions, and the
notebook also allows mixed separable cases such as a Gaussian
function in one direction and a rectangular function in the other.
</div>

<div style="margin-bottom:5px;">
The sliders aₓ and aᵧ control the spatial extent of f₁(x) and f₂(y),
respectively.
</div>

<div style="margin-bottom:5px;">
For a Gaussian function, aₓ and aᵧ determine the Gaussian widths
in the x and y directions. Increasing one of these parameters makes
the corresponding Gaussian broader in space and therefore narrower
in the corresponding frequency direction.
</div>

<div style="margin-bottom:5px;">
For a rectangular function, aₓ and aᵧ are the half-widths of the
rectangular pulse. Therefore, the total spatial widths are 2aₓ and
2aᵧ in the x and y directions, respectively.
</div>

<div>
<b>This notebook:</b> computes the complete 2D Fourier transform
directly and compares it with the product of two independently
computed one-dimensional Fourier transforms.
</div>

</div>
""")

# ============================================================
# MATHEMATICAL RELATIONS
# ============================================================

relation1 = HTMLMath(
    value=(
        r'\('
        r'f(x,y)=f_1(x)f_2(y)'
        r'\)'
    )
)

relation2 = HTMLMath(
    value=(
        r'\('
        r'F(f_x,f_y)=F_1(f_x)F_2(f_y)'
        r'\)'
    )
)

relations_row = HBox(
    [
        relation1,
        relation2
    ],
    layout=Layout(
        width='850px',
        gap='45px',
        align_items='center',
        overflow='visible'
    )
)

# ============================================================
# FUNCTION SELECTORS
# ============================================================

x_function_selector = RadioButtons(
    options=[
        ('Gaussian', 'gaussian'),
        ('Rectangle', 'rectangle')
    ],
    value='gaussian',
    description='',
    layout=Layout(width='260px')
)

x_function_selector.add_class(
    'sep-radio'
)

y_function_selector = RadioButtons(
    options=[
        ('Gaussian', 'gaussian'),
        ('Rectangle', 'rectangle')
    ],
    value='rectangle',
    description='',
    layout=Layout(width='260px')
)

y_function_selector.add_class(
    'sep-radio'
)

# ============================================================
# SLIDER SETTINGS
# ============================================================

slider_style = {
    'description_width': '0px'
}

slider_layout = Layout(
    width='220px'
)

label_layout = Layout(
    width='115px',
    min_width='115px'
)

value_layout = Layout(
    width='65px',
    min_width='65px',
    margin='0px 0px 0px 6px'
)

row_layout = Layout(
    width='420px',
    height='40px',
    align_items='center'
)

# ============================================================
# WIDTH ax
# ============================================================

ax_slider = FloatSlider(
    min=0.20,
    max=2.00,
    step=0.05,
    value=0.70,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

ax_label = HTML(
    '<div class="sep-label">Width aₓ:</div>',
    layout=label_layout
)

ax_value = HTML(
    '<div class="sep-value">0.70</div>',
    layout=value_layout
)

ax_row = HBox(
    [
        ax_label,
        ax_slider,
        ax_value
    ],
    layout=row_layout
)

# ============================================================
# WIDTH ay
# ============================================================

ay_slider = FloatSlider(
    min=0.20,
    max=2.00,
    step=0.05,
    value=1.00,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

ay_label = HTML(
    '<div class="sep-label">Width aᵧ:</div>',
    layout=label_layout
)

ay_value = HTML(
    '<div class="sep-value">1.00</div>',
    layout=value_layout
)

ay_row = HBox(
    [
        ay_label,
        ay_slider,
        ay_value
    ],
    layout=row_layout
)

# ============================================================
# CONTROLS PANEL
# ============================================================

controls_panel = VBox(
    [
        HTML("""
        <div class="sep-title" style="margin-bottom:8px;">
            Parameters
        </div>
        """),

        HTML("""
        <div class="sep-label" style="margin-bottom:4px;">
            Function f₁(x)
        </div>
        """),

        x_function_selector,

        HTML("""
        <div class="sep-label" style="margin-top:7px; margin-bottom:4px;">
            Function f₂(y)
        </div>
        """),

        y_function_selector,

        ax_row,
        ay_row
    ],
    layout=Layout(
        width='470px',
        padding='10px 14px',
        border='1px solid #d2c2df',
        overflow='visible'
    )
)

# ============================================================
# NUMERICAL VERIFICATION PANEL
# ============================================================

error_math = HTMLMath()
relative_error_math = HTMLMath()

verification_panel = VBox(
    [
        HTML("""
        <div class="sep-title" style="margin-bottom:8px;">
            Numerical Verification
        </div>
        """),

        HTML("""
        <div style="
            font-family:Arial;
            font-size:14px;
            line-height:1.5;
            margin-bottom:7px;
        ">
        The direct two-dimensional transform and the separable
        product are calculated independently.
        </div>
        """),

        error_math,
        relative_error_math
    ],
    layout=Layout(
        width='650px',
        padding='10px 14px',
        border='1px solid #d2c2df',
        overflow='visible'
    )
)

# ============================================================
# TOP ROW
# ============================================================

top_row = HBox(
    [
        controls_panel,
        verification_panel
    ],
    layout=Layout(
        width='1140px',
        gap='15px',
        align_items='stretch',
        overflow='visible'
    )
)

# ============================================================
# SPATIAL GRID
# ============================================================

N = 256

x_min = -4.0
x_max = 4.0

x_axis = np.linspace(
    x_min,
    x_max,
    N
)

y_axis = np.linspace(
    x_min,
    x_max,
    N
)

dx = (
    x_axis[1]
    -
    x_axis[0]
)

dy = (
    y_axis[1]
    -
    y_axis[0]
)

# ============================================================
# FREQUENCY AXES
# ============================================================

fx_axis = np.fft.fftshift(
    np.fft.fftfreq(
        N,
        d=dx
    )
)

fy_axis = np.fft.fftshift(
    np.fft.fftfreq(
        N,
        d=dy
    )
)

# ============================================================
# ONE-DIMENSIONAL FUNCTIONS
# ============================================================

def create_1d_function(
    axis,
    mode,
    width
):

    if mode == 'gaussian':

        values = np.exp(
            -(axis**2) / width**2
        )

    else:

        values = (
            np.abs(axis)
            <=
            width
        ).astype(float)

    return values

# ============================================================
# COMPLETE CALCULATION
# ============================================================

def calculate_all():

    mode_x = (
        x_function_selector.value
    )

    mode_y = (
        y_function_selector.value
    )

    width_x = (
        ax_slider.value
    )

    width_y = (
        ay_slider.value
    )

    # --------------------------------------------------------
    # One-dimensional spatial functions
    # --------------------------------------------------------

    f1 = create_1d_function(
        x_axis,
        mode_x,
        width_x
    )

    f2 = create_1d_function(
        y_axis,
        mode_y,
        width_y
    )

    # --------------------------------------------------------
    # Separable 2D function
    #
    # rows -> y
    # columns -> x
    # --------------------------------------------------------

    spatial_2d = np.outer(
        f2,
        f1
    )

    # --------------------------------------------------------
    # DIRECT 2D FOURIER TRANSFORM
    # --------------------------------------------------------

    F_direct = np.fft.fftshift(
        np.fft.fft2(
            np.fft.ifftshift(
                spatial_2d
            )
        )
    )

    F_direct = (
        F_direct
        *
        dx
        *
        dy
    )

    # --------------------------------------------------------
    # INDEPENDENT 1D FOURIER TRANSFORMS
    # --------------------------------------------------------

    F1 = np.fft.fftshift(
        np.fft.fft(
            np.fft.ifftshift(
                f1
            )
        )
    )

    F1 = (
        F1
        *
        dx
    )

    F2 = np.fft.fftshift(
        np.fft.fft(
            np.fft.ifftshift(
                f2
            )
        )
    )

    F2 = (
        F2
        *
        dy
    )

    # --------------------------------------------------------
    # SEPARABLE PRODUCT
    #
    # F(fx,fy) = F1(fx) F2(fy)
    # --------------------------------------------------------

    F_separable = np.outer(
        F2,
        F1
    )

    # --------------------------------------------------------
    # Numerical difference
    # --------------------------------------------------------

    difference = (
        F_direct
        -
        F_separable
    )

    max_error = np.max(
        np.abs(
            difference
        )
    )

    reference = np.max(
        np.abs(
            F_direct
        )
    )

    if reference > 0:

        relative_error = (
            max_error
            /
            reference
        )

    else:

        relative_error = 0.0

    # --------------------------------------------------------
    # Normalize magnitudes only for visualization
    # --------------------------------------------------------

    spatial_display = spatial_2d.copy()

    direct_display = np.abs(
        F_direct
    )

    separable_display = np.abs(
        F_separable
    )

    if np.max(
        direct_display
    ) > 0:

        direct_display = (
            direct_display
            /
            np.max(
                direct_display
            )
        )

    if np.max(
        separable_display
    ) > 0:

        separable_display = (
            separable_display
            /
            np.max(
                separable_display
            )
        )

    return (
        spatial_display,
        direct_display,
        separable_display,
        max_error,
        relative_error
    )

# ============================================================
# INITIAL DATA
# ============================================================

(
    spatial_initial,
    direct_initial,
    separable_initial,
    error_initial,
    relative_initial
) = calculate_all()

# ============================================================
# FIGURE 1
# SPATIAL DOMAIN
# ============================================================

fig_space, ax_space = plt.subplots(
    figsize=(3.8, 3.8)
)

fig_space.canvas.header_visible = False
fig_space.canvas.footer_visible = False
fig_space.canvas.toolbar_visible = False

fig_space.canvas.layout = Layout(
    width='380px',
    height='380px',
    overflow='visible'
)

ax_space.set_title(
    'Separable Function f(x,y)',
    fontsize=13,
    fontweight='bold',
    color='#6f3fa0'
)

ax_space.set_xlabel(
    'x',
    fontsize=10
)

ax_space.set_ylabel(
    'y',
    fontsize=10
)

space_image = ax_space.imshow(
    spatial_initial,
    extent=[
        x_min,
        x_max,
        x_min,
        x_max
    ],
    origin='lower',
    aspect='equal',
    vmin=0.0,
    vmax=1.0,
    interpolation='bilinear'
)

fig_space.subplots_adjust(
    left=0.15,
    right=0.97,
    top=0.88,
    bottom=0.13
)

# ============================================================
# FIGURE 2
# DIRECT 2D FOURIER TRANSFORM
# ============================================================

fig_direct, ax_direct = plt.subplots(
    figsize=(3.8, 3.8)
)

fig_direct.canvas.header_visible = False
fig_direct.canvas.footer_visible = False
fig_direct.canvas.toolbar_visible = False

fig_direct.canvas.layout = Layout(
    width='380px',
    height='380px',
    overflow='visible'
)

ax_direct.set_title(
    'Direct 2D Fourier Transform',
    fontsize=13,
    fontweight='bold',
    color='#0b3d91'
)

ax_direct.set_xlabel(
    'fₓ',
    fontsize=10
)

ax_direct.set_ylabel(
    'fᵧ',
    fontsize=10
)

direct_image = ax_direct.imshow(
    direct_initial,
    extent=[
        fx_axis[0],
        fx_axis[-1],
        fy_axis[0],
        fy_axis[-1]
    ],
    origin='lower',
    aspect='equal',
    vmin=0.0,
    vmax=1.0,
    interpolation='bilinear'
)

frequency_limit = 4.0

ax_direct.set_xlim(
    -frequency_limit,
    frequency_limit
)

ax_direct.set_ylim(
    -frequency_limit,
    frequency_limit
)

fig_direct.subplots_adjust(
    left=0.15,
    right=0.97,
    top=0.88,
    bottom=0.13
)

# ============================================================
# FIGURE 3
# PRODUCT OF 1D FOURIER TRANSFORMS
# ============================================================

fig_sep, ax_sep = plt.subplots(
    figsize=(3.8, 3.8)
)

fig_sep.canvas.header_visible = False
fig_sep.canvas.footer_visible = False
fig_sep.canvas.toolbar_visible = False

fig_sep.canvas.layout = Layout(
    width='380px',
    height='380px',
    overflow='visible'
)

ax_sep.set_title(
    'Product F₁(fₓ)F₂(fᵧ)',
    fontsize=13,
    fontweight='bold',
    color='#16802b'
)

ax_sep.set_xlabel(
    'fₓ',
    fontsize=10
)

ax_sep.set_ylabel(
    'fᵧ',
    fontsize=10
)

separable_image = ax_sep.imshow(
    separable_initial,
    extent=[
        fx_axis[0],
        fx_axis[-1],
        fy_axis[0],
        fy_axis[-1]
    ],
    origin='lower',
    aspect='equal',
    vmin=0.0,
    vmax=1.0,
    interpolation='bilinear'
)

ax_sep.set_xlim(
    -frequency_limit,
    frequency_limit
)

ax_sep.set_ylim(
    -frequency_limit,
    frequency_limit
)

fig_sep.subplots_adjust(
    left=0.15,
    right=0.97,
    top=0.88,
    bottom=0.13
)

# ============================================================
# FIGURES ROW
# ============================================================

figures_row = HBox(
    [
        fig_space.canvas,
        fig_direct.canvas,
        fig_sep.canvas
    ],
    layout=Layout(
        width='1160px',
        gap='8px',
        align_items='flex-start',
        overflow='visible'
    )
)

# ============================================================
# CONCLUSION PANEL
# ============================================================

conclusion_panel = HTML("""
<div style="
    width:1140px;
    padding:9px 12px;
    border:1px solid #d7c7e5;
    font-family:Arial, sans-serif;
    font-size:14px;
    line-height:1.52;
    box-sizing:border-box;
">

<b style="color:#6f3fa0;">
Key observation:
</b>

The second and third plots should be visually identical.
The numerical verification confirms that the complete 2D Fourier
transform of a separable function is equal, within numerical precision,
to the product of its two one-dimensional Fourier transforms.

</div>
""")

# ============================================================
# UPDATE FUNCTION
#
# No clear_output()
# No figure recreation
# No axis rescaling
# ============================================================

def update_notebook(change=None):

    width_x = (
        ax_slider.value
    )

    width_y = (
        ay_slider.value
    )

    ax_value.value = (
        f'<div class="sep-value">{width_x:.2f}</div>'
    )

    ay_value.value = (
        f'<div class="sep-value">{width_y:.2f}</div>'
    )

    (
        spatial_data,
        direct_data,
        separable_data,
        max_error,
        relative_error
    ) = calculate_all()

    # --------------------------------------------------------
    # Update existing image objects only
    # --------------------------------------------------------

    space_image.set_data(
        spatial_data
    )

    direct_image.set_data(
        direct_data
    )

    separable_image.set_data(
        separable_data
    )

    # --------------------------------------------------------
    # Numerical verification
    # --------------------------------------------------------

    error_math.value = (
        r'\('
        r'\max|F_{\mathrm{2D}}-F_1F_2|='
        +
        f'{max_error:.3e}'
        +
        r'\)'
    )

    relative_error_math.value = (
        r'\('
        r'\mathrm{relative\ error}='
        +
        f'{relative_error:.3e}'
        +
        r'\)'
    )

    # --------------------------------------------------------
    # Redraw only
    # --------------------------------------------------------

    fig_space.canvas.draw_idle()
    fig_direct.canvas.draw_idle()
    fig_sep.canvas.draw_idle()

# ============================================================
# CONNECT CONTROLS
# ============================================================

x_function_selector.observe(
    update_notebook,
    names='value'
)

y_function_selector.observe(
    update_notebook,
    names='value'
)

ax_slider.observe(
    update_notebook,
    names='value'
)

ay_slider.observe(
    update_notebook,
    names='value'
)

# ============================================================
# INITIAL UPDATE
# ============================================================

update_notebook()

# ============================================================
# COMPLETE NOTEBOOK
# ============================================================

main_layout = VBox(
    [
        documentation,
        relations_row,
        top_row,
        figures_row,
        conclusion_panel
    ],
    layout=Layout(
        width='1180px',
        gap='10px',
        overflow='visible'
    )
)

display(
    main_layout
)
```
