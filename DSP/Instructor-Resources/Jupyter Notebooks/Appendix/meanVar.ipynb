```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, expon, uniform
from ipywidgets import Dropdown, FloatSlider, IntSlider, interactive
from IPython.display import display, Markdown

# ------------------------------------------------------------
# Mean, Variance and Moments
# ------------------------------------------------------------

def moments_demo(distribution, p1, p2, k):

    # ========================================================
    # GAUSSIAN
    # ========================================================

    if distribution == 'Gaussian':

        mu = p1
        sigma = max(p2, 0.1)

        dist = norm(
            loc=mu,
            scale=sigma
        )

        # Fixed horizontal axis
        x = np.linspace(
            -10,
            10,
            3000
        )

        x_limits = (-10, 10)

    # ========================================================
    # UNIFORM
    # ========================================================

    elif distribution == 'Uniform':

        a = p1
        width = max(p2, 0.1)

        dist = uniform(
            loc=a,
            scale=width
        )

        # Fixed horizontal axis
        x = np.linspace(
            -5,
            10,
            3000
        )

        x_limits = (-5, 10)

    # ========================================================
    # EXPONENTIAL
    # ========================================================

    elif distribution == 'Exponential':

        lam = max(p1, 0.1)

        dist = expon(
            scale=1/lam
        )

        # Fixed horizontal axis
        x = np.linspace(
            0,
            20,
            3000
        )

        x_limits = (0, 20)

    # ========================================================
    # PDF
    # ========================================================

    pdf = dist.pdf(x)

    # ========================================================
    # NUMERICAL MOMENTS
    # ========================================================

    EX = np.trapezoid(
        x * pdf,
        x
    )

    EX2 = np.trapezoid(
        x**2 * pdf,
        x
    )

    EXk = np.trapezoid(
        x**k * pdf,
        x
    )

    variance_from_moments = (
        EX2 - EX**2
    )

    # ========================================================
    # THEORETICAL VALUES
    # ========================================================

    mean_theoretical = dist.mean()

    variance_theoretical = dist.var()

    sigma_theoretical = np.sqrt(
        variance_theoretical
    )

    # ========================================================
    # FIGURE
    # ========================================================

    fig, ax = plt.subplots(
        figsize=(10, 5)
    )

    # --------------------------------------------------------
    # PDF
    # --------------------------------------------------------

    ax.plot(
        x,
        pdf,
        linewidth=2,
        label='$f_X(x)$'
    )

    # --------------------------------------------------------
    # Mean
    # --------------------------------------------------------

    ax.axvline(
        mean_theoretical,
        linestyle='--',
        linewidth=2,
        label='$\\mu$'
    )

    # --------------------------------------------------------
    # μ - σ
    # --------------------------------------------------------

    ax.axvline(
        mean_theoretical - sigma_theoretical,
        linestyle=':',
        linewidth=1.5,
        label='$\\mu-\\sigma$'
    )

    # --------------------------------------------------------
    # μ + σ
    # --------------------------------------------------------

    ax.axvline(
        mean_theoretical + sigma_theoretical,
        linestyle=':',
        linewidth=1.5,
        label='$\\mu+\\sigma$'
    )

    # --------------------------------------------------------
    # Region μ - σ <= x <= μ + σ
    # --------------------------------------------------------

    mask = (
        (x >= mean_theoretical - sigma_theoretical)
        &
        (x <= mean_theoretical + sigma_theoretical)
    )

    ax.fill_between(
        x[mask],
        pdf[mask],
        alpha=0.3
    )

    # ========================================================
    # FIXED HORIZONTAL AXIS
    # ========================================================

    ax.set_xlim(
        x_limits
    )

    # ========================================================
    # VERTICAL AXIS
    #
    # Automatically adapted only to the height of the PDF.
    # ========================================================

    ymax = np.max(pdf)

    ax.set_ylim(
        0,
        1.15 * ymax
    )

    # ========================================================
    # LABELS
    # ========================================================

    ax.set_title(
        f'{distribution} Distribution — Mean and Variance'
    )

    ax.set_xlabel(
        'x'
    )

    ax.set_ylabel(
        '$f_X(x)$'
    )

    ax.grid(
        True,
        alpha=0.3
    )

    ax.legend()

    plt.tight_layout()
    plt.show()
    plt.close(fig)

    # ========================================================
    # RESULTS
    # ========================================================

    print(
        f'E[X]                 = {EX:.6f}'
    )

    print(
        f'E[X²]                = {EX2:.6f}'
    )

    print(
        f'E[X^{k}]               = {EXk:.6f}'
    )

    print()

    print(
        f'E[X²] - E[X]²        = '
        f'{variance_from_moments:.6f}'
    )

    print(
        f'Theoretical Var(X)    = '
        f'{variance_theoretical:.6f}'
    )

    print()

    print(
        'Verification:'
    )

    print(
        'Var(X) = E[X²] - E[X]²'
    )


# ============================================================
# DISPLAY
# ============================================================

display(
    Markdown(
        "## Mean, Variance and Moments"
    )
)

display(
    interactive(
        moments_demo,

        distribution=Dropdown(
            options=[
                'Gaussian',
                'Uniform',
                'Exponential'
            ],
            value='Gaussian',
            description='Distribution:'
        ),

        p1=FloatSlider(
            value=0,
            min=-3,
            max=3,
            step=0.1,
            description='Parameter 1:',
            continuous_update=True
        ),

        p2=FloatSlider(
            value=1,
            min=0.1,
            max=3,
            step=0.1,
            description='Parameter 2:',
            continuous_update=True
        ),

        k=IntSlider(
            value=3,
            min=1,
            max=6,
            step=1,
            description='Moment k:',
            continuous_update=True
        )
    )
)
```
