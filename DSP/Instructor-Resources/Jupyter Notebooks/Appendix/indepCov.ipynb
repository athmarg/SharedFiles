```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt

from ipywidgets import (
    Dropdown,
    IntSlider,
    interactive,
    HBox,
    VBox,
    Output,
    Layout
)

from IPython.display import display, Markdown


# ------------------------------------------------------------
# Independence, Covariance and Correlation
# ------------------------------------------------------------

display(Markdown(
    "## Independence, Covariance and Correlation"
))


# ------------------------------------------------------------
# Widgets
# ------------------------------------------------------------

example_widget = Dropdown(
    options=[
        'Independent',
        'Linear dependence',
        'Nonlinear dependence: Y = X²'
    ],
    value='Independent',
    description='Example:',
    layout=Layout(width='360px')
)

N_widget = IntSlider(
    value=1000,
    min=100,
    max=5000,
    step=100,
    description='N:',
    continuous_update=True,
    layout=Layout(width='360px')
)


# ------------------------------------------------------------
# Main function
# ------------------------------------------------------------

def correlation_demo(example, N):

    rng = np.random.default_rng(1234)

    X = rng.normal(
        0,
        1,
        N
    )

    # --------------------------------------------------------
    # Examples
    # --------------------------------------------------------

    if example == 'Independent':

        Y = rng.normal(
            0,
            1,
            N
        )

        description = (
            'X and Y are statistically independent.'
        )

    elif example == 'Linear dependence':

        Y = (
            2*X
            +
            0.3*rng.normal(
                0,
                1,
                N
            )
        )

        description = (
            'Y is approximately a linear function of X.'
        )

    elif example == 'Nonlinear dependence: Y = X²':

        Y = X**2

        description = (
            'Y is completely determined by X, but their '
            'linear correlation is approximately zero.'
        )

    # --------------------------------------------------------
    # Statistical quantities
    # --------------------------------------------------------

    EX = np.mean(X)
    EY = np.mean(Y)
    EXY = np.mean(X*Y)

    covariance = np.mean(
        (X - EX)*(Y - EY)
    )

    rho = np.corrcoef(
        X,
        Y
    )[0, 1]

    # --------------------------------------------------------
    # Separate output areas
    # --------------------------------------------------------

    figure_output = Output(
        layout=Layout(
            width='620px'
        )
    )

    text_output = Output(
        layout=Layout(
            width='430px',
            padding='55px 0 0 20px'
        )
    )

    # --------------------------------------------------------
    # Figure
    # --------------------------------------------------------

    with figure_output:

        fig, ax = plt.subplots(
            figsize=(7, 6)
        )

        ax.scatter(
            X,
            Y,
            s=15,
            alpha=0.5
        )

        ax.set_title(
            example
        )

        ax.set_xlabel(
            'X'
        )

        ax.set_ylabel(
            'Y'
        )

        ax.grid(
            True,
            alpha=0.3
        )

        plt.show()
        plt.close(fig)

    # --------------------------------------------------------
    # Text output
    # --------------------------------------------------------

    with text_output:

        print(description)

        print()

        print(
            f'E[X]              = {EX:.6f}'
        )

        print(
            f'E[Y]              = {EY:.6f}'
        )

        print(
            f'E[XY]             = {EXY:.6f}'
        )

        print(
            f'E[X]E[Y]          = {EX*EY:.6f}'
        )

        print(
            f'Cov(X,Y)          = {covariance:.6f}'
        )

        print(
            f'ρ(X,Y)            = {rho:.6f}'
        )

    # --------------------------------------------------------
    # Figure left — results right
    # --------------------------------------------------------

    display(
        HBox(
            [
                figure_output,
                text_output
            ],
            layout=Layout(
                width='1080px',
                align_items='flex-start'
            )
        )
    )


# ------------------------------------------------------------
# Interactive notebook
# ------------------------------------------------------------

interactive_widget = interactive(
    correlation_demo,

    example=example_widget,
    N=N_widget
)

interactive_output = interactive_widget.children[-1]


# ------------------------------------------------------------
# Display controls and output
# ------------------------------------------------------------

controls = VBox(
    [
        example_widget,
        N_widget
    ]
)

display(
    controls,
    interactive_output
)
```
