```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm, laplace, uniform
from scipy.signal import fftconvolve
from ipywidgets import Dropdown, interactive
from IPython.display import display, Markdown

# ------------------------------------------------------------
# Characteristic Function and Convolution
# ------------------------------------------------------------

def characteristic_function_demo(distribution):

    # ========================================================
    # FIXED X-AXIS
    # ========================================================

    x = np.linspace(
        -8,
        8,
        4096
    )

    dx = x[1] - x[0]

    # ========================================================
    # SELECT PDF
    # ========================================================

    if distribution == 'Gaussian':

        pdf = norm.pdf(
            x,
            loc=0,
            scale=1
        )

    elif distribution == 'Laplace':

        pdf = laplace.pdf(
            x,
            loc=0,
            scale=1
        )

    elif distribution == 'Uniform':

        pdf = uniform.pdf(
            x,
            loc=-1,
            scale=2
        )

    # ========================================================
    # CHARACTERISTIC FUNCTION
    #
    # Phi(w) = integral f(x) exp(j w x) dx
    # ========================================================

    omega = np.linspace(
        -10,
        10,
        500
    )

    Phi = np.array([
        np.trapezoid(
            pdf * np.exp(1j*w*x),
            x
        )
        for w in omega
    ])

    # ========================================================
    # SUM OF TWO INDEPENDENT IDENTICALLY DISTRIBUTED
    # RANDOM VARIABLES
    #
    # Z = X1 + X2
    #
    # f_Z = f_X * f_X
    # Phi_Z = Phi_X Phi_X
    # ========================================================

    pdf_z = (
        fftconvolve(
            pdf,
            pdf,
            mode='same'
        )
        *
        dx
    )

    Phi_product = (
        Phi * Phi
    )

    # ========================================================
    # FIGURE
    # ========================================================

    fig, axes = plt.subplots(
        2,
        2,
        figsize=(12, 8)
    )

    # --------------------------------------------------------
    # Input PDF
    # --------------------------------------------------------

    axes[0, 0].plot(
        x,
        pdf,
        linewidth=2
    )

    axes[0, 0].set_title(
        '$f_X(x)$'
    )

    axes[0, 0].set_xlabel(
        'x'
    )

    axes[0, 0].set_ylabel(
        'PDF'
    )

    axes[0, 0].set_xlim(
        -8,
        8
    )

    axes[0, 0].grid(
        True,
        alpha=0.3
    )

    # --------------------------------------------------------
    # Characteristic function
    # --------------------------------------------------------

    axes[0, 1].plot(
        omega,
        np.real(Phi),
        linewidth=2,
        label='Re{$\\Phi_X(\\omega)$}'
    )

    axes[0, 1].plot(
        omega,
        np.imag(Phi),
        linestyle='--',
        linewidth=1.5,
        label='Im{$\\Phi_X(\\omega)$}'
    )

    axes[0, 1].set_title(
        'Characteristic Function'
    )

    axes[0, 1].set_xlabel(
        '$\\omega$'
    )

    axes[0, 1].set_ylabel(
        '$\\Phi_X(\\omega)$'
    )

    axes[0, 1].set_xlim(
        -10,
        10
    )

    axes[0, 1].grid(
        True,
        alpha=0.3
    )

    axes[0, 1].legend()

    # --------------------------------------------------------
    # Convolution PDF
    # --------------------------------------------------------

    axes[1, 0].plot(
        x,
        pdf_z,
        linewidth=2
    )

    axes[1, 0].set_title(
        '$f_Z(z)=f_X*f_X$'
    )

    axes[1, 0].set_xlabel(
        'z'
    )

    axes[1, 0].set_ylabel(
        '$f_Z(z)$'
    )

    axes[1, 0].set_xlim(
        -8,
        8
    )

    axes[1, 0].grid(
        True,
        alpha=0.3
    )

    # --------------------------------------------------------
    # Product of characteristic functions
    # --------------------------------------------------------

    axes[1, 1].plot(
        omega,
        np.real(Phi_product),
        linewidth=2
    )

    axes[1, 1].set_title(
        '$\\Phi_Z(\\omega)=\\Phi_X^2(\\omega)$'
    )

    axes[1, 1].set_xlabel(
        '$\\omega$'
    )

    axes[1, 1].set_ylabel(
        '$\\Phi_Z(\\omega)$'
    )

    axes[1, 1].set_xlim(
        -10,
        10
    )

    axes[1, 1].grid(
        True,
        alpha=0.3
    )

    plt.tight_layout()
    plt.show()
    plt.close(fig)

    # ========================================================
    # RESULTS
    # ========================================================

    print(
        f'Distribution: {distribution}'
    )

    print()

    print(
        'For two independent random variables X₁ and X₂:'
    )

    print()

    print(
        'Z = X₁ + X₂'
    )

    print()

    print(
        'f_Z(z) = f_X * f_X'
    )

    print(
        'Φ_Z(ω) = Φ_X(ω) Φ_X(ω)'
    )


# ============================================================
# TITLE
# ============================================================

display(
    Markdown(
        "## Characteristic Function — Fourier Transform and Convolution"
    )
)

# ============================================================
# SHORT INTRODUCTION
# ============================================================

display(
    Markdown(
        r"""
The **characteristic function** of a random variable is the Fourier
transform of its probability density function,
\(\Phi_X(\omega)=E[e^{i\omega X}]\).
For independent random variables, addition in the random-variable
domain corresponds to convolution of their PDFs and multiplication
of their characteristic functions.
"""
    )
)

# ============================================================
# INTERACTIVE CONTROL
# ============================================================

display(
    interactive(
        characteristic_function_demo,

        distribution=Dropdown(
            options=[
                'Gaussian',
                'Laplace',
                'Uniform'
            ],
            value='Gaussian',
            description='Distribution:'
        )
    )
)
```
