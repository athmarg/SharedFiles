```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt

from ipywidgets import (
    FloatSlider,
    RadioButtons,
    HTML,
    VBox,
    HBox,
    Layout
)
from IPython.display import display

# ============================================================
# COMPLEX VARIABLE FUNCTIONS
# MAPPING w = f(z) AND CAUCHY-RIEMANN CONDITIONS
# ============================================================

plt.ioff()

# ============================================================
# CLASSIC JUPYTER + JUPYTERLAB / NOTEBOOK 7 / BINDER
#
# - no horizontal scrollbar
# - no vertical scrollbar
# - no trimming
# - no resize triangle
# ============================================================

display(HTML("""
<style>

/* ============================================================
   CLASSIC JUPYTER
   ============================================================ */

.container {
    width: 98% !important;
    max-width: none !important;
}

.output_area,
.output_subarea {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

.output_scroll {
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
    box-shadow: none !important;
}

/* ============================================================
   JUPYTERLAB / NOTEBOOK 7
   ============================================================ */

.jp-Cell-outputWrapper,
.jp-OutputArea,
.jp-OutputArea-child,
.jp-OutputArea-output {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

/* ============================================================
   IPYWIDGETS
   ============================================================ */

.widget-output,
.jupyter-widgets-output-area,
.widget-box {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

/* ============================================================
   IPYMPL
   ============================================================ */

.jupyter-matplotlib,
.jupyter-matplotlib-figure {
    overflow: visible !important;
    resize: none !important;
}

.jupyter-matplotlib::-webkit-resizer,
.jupyter-matplotlib-figure::-webkit-resizer {
    display: none !important;
}

/* ============================================================
   STYLES
   ============================================================ */

.cvf-title {
    font-family: Arial, sans-serif;
    font-size: 21px;
    font-weight: bold;
    color: #6f3fa0;
    margin-bottom: 7px;
}

.cvf-label {
    font-family: Arial, sans-serif;
    font-size: 14px;
    font-weight: bold;
}

.cvf-value {
    font-family: Arial, sans-serif;
    font-size: 14px;
    font-weight: bold;
    color: #0b3d91;
}

.cvf-radio .widget-radio-box {
    display: flex !important;
    flex-direction: row !important;
    flex-wrap: nowrap !important;
    gap: 16px !important;
    align-items: center !important;
}

.cvf-radio .widget-radio-box label {
    margin: 0 !important;
    white-space: nowrap !important;
}

.cvf-radio > label {
    display: none !important;
}

</style>
"""))

# ============================================================
# DOCUMENTATION
# ============================================================

documentation = HTML("""
<div style="
    width:950px;
    font-family:Arial, sans-serif;
    font-size:15px;
    line-height:1.45;
    margin-bottom:8px;
">

<div class="cvf-title">
Complex Variable Functions and the Cauchy-Riemann Conditions
</div>

<div style="margin-bottom:4px;">
For z = x + iy, a complex function w = f(z) can be written as
w = u(x,y) + iv(x,y), where u and v are real-valued functions.
</div>

<div style="margin-bottom:4px;">
For an analytic function, the Cauchy-Riemann conditions are
u<sub>x</sub> = v<sub>y</sub> and
u<sub>y</sub> = −v<sub>x</sub>.
</div>

<div>
<b>This notebook:</b> maps a point z from the z-plane to w = f(z)
in the w-plane and checks the Cauchy-Riemann conditions at the selected point.
</div>

</div>
""")

# ============================================================
# FUNCTION SELECTOR
# ============================================================

function_selector = RadioButtons(
    options=[
        ('z²', 'square'),
        ('exp(z)', 'exp'),
        ('(1+z)/(1−z)', 'mobius'),
        ('sin(z)', 'sin'),
        ('conj(z)', 'conjugate')
    ],
    value='square',
    description='',
    layout=Layout(width='700px')
)

function_selector.add_class('cvf-radio')

# ============================================================
# SLIDER SETTINGS
# ============================================================

slider_style = {
    'description_width': '0px'
}

slider_layout = Layout(
    width='180px'
)

value_layout = Layout(
    width='52px',
    min_width='52px',
    margin='0px 0px 0px 5px'
)

label_layout = Layout(
    width='70px',
    min_width='70px'
)

# ============================================================
# SLIDERS
# ============================================================

x_slider = FloatSlider(
    min=-2.5,
    max=2.5,
    step=0.05,
    value=1.0,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

y_slider = FloatSlider(
    min=-2.5,
    max=2.5,
    step=0.05,
    value=1.0,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

# ============================================================
# VALUES
# ============================================================

x_value = HTML(
    '<div class="cvf-value">1.00</div>',
    layout=value_layout
)

y_value = HTML(
    '<div class="cvf-value">1.00</div>',
    layout=value_layout
)

# ============================================================
# LABELS
# ============================================================

x_label = HTML(
    '<div class="cvf-label">Re(z):</div>',
    layout=label_layout
)

y_label = HTML(
    '<div class="cvf-label">Im(z):</div>',
    layout=label_layout
)

# ============================================================
# ROWS
# ============================================================

x_row = HBox(
    [
        x_label,
        x_slider,
        x_value
    ],
    layout=Layout(
        width='315px',
        align_items='center'
    )
)

y_row = HBox(
    [
        y_label,
        y_slider,
        y_value
    ],
    layout=Layout(
        width='315px',
        align_items='center'
    )
)

xy_row = HBox(
    [
        x_row,
        y_row
    ],
    layout=Layout(
        width='645px',
        gap='15px',
        align_items='center'
    )
)

# ============================================================
# CONTROLS PANEL
# ============================================================

controls_panel = VBox(
    [
        HTML("""
        <div style="
            font-family:Arial;
            font-size:17px;
            font-weight:bold;
            color:#6f3fa0;
            margin-bottom:7px;
        ">
        Function and Input Point
        </div>
        """),

        function_selector,

        HTML("""
        <div style="
            font-family:Arial;
            font-size:14px;
            font-weight:bold;
            margin-top:7px;
            margin-bottom:3px;
        ">
        Input z = x + iy
        </div>
        """),

        xy_row
    ],
    layout=Layout(
        width='720px',
        padding='9px 12px',
        border='1px solid #d2c2df',
        overflow='visible'
    )
)

# ============================================================
# INPUT z-PLANE
# ============================================================

fig_z, ax_z = plt.subplots(
    figsize=(4.0, 3.9)
)

fig_z.canvas.header_visible = False

fig_z.canvas.layout = Layout(
    width='400px',
    height='390px'
)

ax_z.set_title(
    'Input Plane: z = x + iy',
    fontsize=14,
    fontweight='bold',
    color='#1555cc'
)

ax_z.set_xlabel(
    'Re(z)',
    fontsize=10
)

ax_z.set_ylabel(
    'Im(z)',
    fontsize=10
)

ax_z.set_aspect(
    'equal',
    adjustable='box'
)

ax_z.set_xlim(
    -3.0,
    3.0
)

ax_z.set_ylim(
    -3.0,
    3.0
)

ax_z.axhline(
    0,
    color='black',
    linewidth=1.0
)

ax_z.axvline(
    0,
    color='black',
    linewidth=1.0
)

ax_z.grid(
    True,
    linestyle=':',
    alpha=0.40
)

z_vector, = ax_z.plot(
    [],
    [],
    linewidth=2.0,
    color='blue',
    marker='o',
    markevery=[1]
)

z_text = ax_z.text(
    0,
    0,
    '',
    fontsize=10,
    color='blue'
)

fig_z.subplots_adjust(
    left=0.14,
    right=0.96,
    top=0.89,
    bottom=0.13
)

# ============================================================
# OUTPUT w-PLANE
# ============================================================

fig_w, ax_w = plt.subplots(
    figsize=(4.0, 3.9)
)

fig_w.canvas.header_visible = False

fig_w.canvas.layout = Layout(
    width='400px',
    height='390px'
)

ax_w.set_title(
    'Output Plane: w = f(z)',
    fontsize=14,
    fontweight='bold',
    color='#b22222'
)

ax_w.set_xlabel(
    'Re(w)',
    fontsize=10
)

ax_w.set_ylabel(
    'Im(w)',
    fontsize=10
)

ax_w.set_aspect(
    'equal',
    adjustable='box'
)

ax_w.axhline(
    0,
    color='black',
    linewidth=1.0
)

ax_w.axvline(
    0,
    color='black',
    linewidth=1.0
)

ax_w.grid(
    True,
    linestyle=':',
    alpha=0.40
)

w_vector, = ax_w.plot(
    [],
    [],
    linewidth=2.0,
    color='red',
    marker='o',
    markevery=[1]
)

w_text = ax_w.text(
    0,
    0,
    '',
    fontsize=10,
    color='red'
)

fig_w.subplots_adjust(
    left=0.14,
    right=0.96,
    top=0.89,
    bottom=0.13
)

# ============================================================
# NUMERICAL RESULTS PANEL
# ============================================================

results_html = HTML()

results_panel = VBox(
    [
        HTML("""
        <div style="
            font-family:Arial;
            font-size:16px;
            font-weight:bold;
            color:#6f3fa0;
            margin-bottom:6px;
        ">
        Mapping and Cauchy-Riemann Test
        </div>
        """),

        results_html
    ],
    layout=Layout(
        width='305px',
        padding='9px 10px',
        border='1px solid #d2c2df',
        overflow='visible'
    )
)

# ============================================================
# FUNCTIONS
# ============================================================

def evaluate_function(mode, z):

    if mode == 'square':
        return z**2

    elif mode == 'exp':
        return np.exp(z)

    elif mode == 'mobius':

        if abs(1.0 - z) < 1e-10:
            return np.nan + 1j * np.nan

        return (1.0 + z) / (1.0 - z)

    elif mode == 'sin':
        return np.sin(z)

    elif mode == 'conjugate':
        return np.conjugate(z)


def function_name(mode):

    names = {
        'square': 'f(z) = z²',
        'exp': 'f(z) = exp(z)',
        'mobius': 'f(z) = (1+z)/(1−z)',
        'sin': 'f(z) = sin(z)',
        'conjugate': 'f(z) = z*'
    }

    return names[mode]

# ============================================================
# NUMERICAL PARTIAL DERIVATIVES
# ============================================================

def partial_derivatives(mode, x, y):

    h = 1e-5

    z_x_plus = complex(
        x + h,
        y
    )

    z_x_minus = complex(
        x - h,
        y
    )

    z_y_plus = complex(
        x,
        y + h
    )

    z_y_minus = complex(
        x,
        y - h
    )

    f_x_plus = evaluate_function(
        mode,
        z_x_plus
    )

    f_x_minus = evaluate_function(
        mode,
        z_x_minus
    )

    f_y_plus = evaluate_function(
        mode,
        z_y_plus
    )

    f_y_minus = evaluate_function(
        mode,
        z_y_minus
    )

    values = [
        f_x_plus,
        f_x_minus,
        f_y_plus,
        f_y_minus
    ]

    if not all(
        np.isfinite(np.real(value))
        and
        np.isfinite(np.imag(value))
        for value in values
    ):

        return (
            np.nan,
            np.nan,
            np.nan,
            np.nan
        )

    ux = (
        np.real(f_x_plus)
        -
        np.real(f_x_minus)
    ) / (2.0 * h)

    vx = (
        np.imag(f_x_plus)
        -
        np.imag(f_x_minus)
    ) / (2.0 * h)

    uy = (
        np.real(f_y_plus)
        -
        np.real(f_y_minus)
    ) / (2.0 * h)

    vy = (
        np.imag(f_y_plus)
        -
        np.imag(f_y_minus)
    ) / (2.0 * h)

    return (
        ux,
        uy,
        vx,
        vy
    )

# ============================================================
# FIXED OUTPUT LIMITS FOR EACH FUNCTION
# ============================================================

W_LIMITS = {
    'square': 9.0,
    'exp': 13.0,
    'mobius': 10.0,
    'sin': 7.0,
    'conjugate': 3.0
}

# ============================================================
# UPDATE FUNCTION
# ============================================================

def update_notebook(change=None):

    x = x_slider.value
    y = y_slider.value

    mode = function_selector.value

    z = complex(
        x,
        y
    )

    # --------------------------------------------------------
    # Slider-value labels
    # --------------------------------------------------------

    x_value.value = (
        f'<div class="cvf-value">{x:.2f}</div>'
    )

    y_value.value = (
        f'<div class="cvf-value">{y:.2f}</div>'
    )

    # --------------------------------------------------------
    # Input plane
    # --------------------------------------------------------

    z_vector.set_data(
        [0.0, x],
        [0.0, y]
    )

    z_text.set_position(
        (
            x + 0.10,
            y + 0.10
        )
    )

    z_text.set_text(
        'z'
    )

    # --------------------------------------------------------
    # Function
    # --------------------------------------------------------

    w = evaluate_function(
        mode,
        z
    )

    w_limit = W_LIMITS[mode]

    ax_w.set_xlim(
        -w_limit,
        w_limit
    )

    ax_w.set_ylim(
        -w_limit,
        w_limit
    )

    # --------------------------------------------------------
    # Singular point
    # --------------------------------------------------------

    if (
        not np.isfinite(np.real(w))
        or
        not np.isfinite(np.imag(w))
    ):

        w_vector.set_data(
            [],
            []
        )

        w_text.set_text(
            ''
        )

        results_html.value = f"""
        <div style="
            font-family:Arial;
            font-size:13px;
            line-height:1.55;
        ">

        <b>{function_name(mode)}</b>

        <br><br>

        z =
        <b>{x:.4f} {'+' if y >= 0 else '−'} {abs(y):.4f}i</b>

        <br><br>

        <span style="
            color:#b00020;
            font-weight:bold;
        ">
        The function is undefined at this point.
        </span>

        <br><br>

        For f(z) = (1+z)/(1−z),
        z = 1 is a singular point.

        </div>
        """

        fig_z.canvas.draw_idle()
        fig_w.canvas.draw_idle()

        return

    # --------------------------------------------------------
    # Output plane
    # --------------------------------------------------------

    u = np.real(w)
    v = np.imag(w)

    w_vector.set_data(
        [0.0, u],
        [0.0, v]
    )

    w_text.set_position(
        (
            u + 0.05 * w_limit,
            v + 0.05 * w_limit
        )
    )

    w_text.set_text(
        'w'
    )

    # --------------------------------------------------------
    # Derivatives
    # --------------------------------------------------------

    ux, uy, vx, vy = partial_derivatives(
        mode,
        x,
        y
    )

    cr1_error = abs(
        ux - vy
    )

    cr2_error = abs(
        uy + vx
    )

    tolerance = 1e-5

    cr_satisfied = (
        np.isfinite(cr1_error)
        and
        np.isfinite(cr2_error)
        and
        cr1_error < tolerance
        and
        cr2_error < tolerance
    )

    derivative = complex(
        ux,
        vx
    )

    # --------------------------------------------------------
    # Status
    # --------------------------------------------------------

    if cr_satisfied:

        status = """
        <span style="
            color:#16802b;
            font-weight:bold;
        ">
        Cauchy-Riemann conditions satisfied
        </span>
        """

        conclusion = """
        The function is locally consistent with complex
        differentiability at the selected point.
        """

    else:

        status = """
        <span style="
            color:#b00020;
            font-weight:bold;
        ">
        Cauchy-Riemann conditions not satisfied
        </span>
        """

        conclusion = """
        The function is not complex differentiable
        at the selected point.
        """

    # --------------------------------------------------------
    # Numerical output
    # --------------------------------------------------------

    results_html.value = f"""
    <div style="
        font-family:Arial;
        font-size:13px;
        line-height:1.55;
    ">

    <b>{function_name(mode)}</b>

    <br><br>

    z =
    <b>{x:.4f} {'+' if y >= 0 else '−'} {abs(y):.4f}i</b>

    <br>

    w =
    <b>{u:.4f} {'+' if v >= 0 else '−'} {abs(v):.4f}i</b>

    <br><br>

    u(x,y) = <b>{u:.4f}</b><br>
    v(x,y) = <b>{v:.4f}</b>

    <hr style="margin:7px 0px;">

    <b>Cauchy-Riemann derivatives</b>

    <br><br>

    u<sub>x</sub> = <b>{ux:.6f}</b>
    &nbsp;&nbsp;
    v<sub>y</sub> = <b>{vy:.6f}</b>

    <br>

    u<sub>y</sub> = <b>{uy:.6f}</b>
    &nbsp;&nbsp;
    −v<sub>x</sub> = <b>{-vx:.6f}</b>

    <br><br>

    |u<sub>x</sub> − v<sub>y</sub>| =
    <b>{cr1_error:.2e}</b>

    <br>

    |u<sub>y</sub> + v<sub>x</sub>| =
    <b>{cr2_error:.2e}</b>

    <hr style="margin:7px 0px;">

    {status}

    <br><br>

    {conclusion}

    <br><br>

    <b>Derivative estimate:</b><br>

    f′(z) ≈
    <b>
    {np.real(derivative):.4f}
    {'+' if np.imag(derivative) >= 0 else '−'}
    {abs(np.imag(derivative)):.4f}i
    </b>

    </div>
    """

    fig_z.canvas.draw_idle()
    fig_w.canvas.draw_idle()

# ============================================================
# CONNECT CONTROLS
# ============================================================

x_slider.observe(
    update_notebook,
    names='value'
)

y_slider.observe(
    update_notebook,
    names='value'
)

function_selector.observe(
    update_notebook,
    names='value'
)

# ============================================================
# INITIAL UPDATE
# ============================================================

update_notebook()

# ============================================================
# TWO FIGURES
#
# Width = 810 px.
# No internal horizontal scrollbar.
# ============================================================

figures_row = HBox(
    [
        fig_z.canvas,
        fig_w.canvas
    ],
    layout=Layout(
        width='810px',
        gap='10px',
        align_items='flex-start',
        overflow='visible'
    )
)

# ============================================================
# INTERPRETATION
#
# NOW DIRECTLY UNDER THE TWO FIGURES.
# Shorter width as requested.
# ============================================================

interpretation = HTML("""
<div style="
    width:790px;
    padding:8px 10px;
    border:1px solid #d7c7e5;
    background:#fbf8fe;
    font-family:Arial, sans-serif;
    font-size:13px;
    line-height:1.45;
    box-sizing:border-box;
    margin-top:3px;
">

<b style="color:#6f3fa0;">Interpretation:</b>
A complex function maps z = x + iy in the input plane to
w = u + iv in the output plane. For analytic functions,
u and v satisfy the Cauchy-Riemann equations.
The conjugate function is a useful counterexample:continuous
but not complex differentiable.

</div>
""")

# ============================================================
# LEFT VISUAL COLUMN
#
# figures + interpretation
# ============================================================

visual_column = VBox(
    [
        figures_row,
        interpretation
    ],
    layout=Layout(
        width='810px',
        gap='2px',
        overflow='visible'
    )
)

# ============================================================
# MAIN VISUAL ROW
#
# Compact total width:
# 810 + 10 + 305 = 1125 px
#
# No enormous 1410 px dashboard.
# ============================================================

visual_results_row = HBox(
    [
        visual_column,
        results_panel
    ],
    layout=Layout(
        width='1125px',
        gap='10px',
        align_items='flex-start',
        overflow='visible'
    )
)

# ============================================================
# COMPLETE NOTEBOOK
# ============================================================

main_layout = VBox(
    [
        documentation,
        controls_panel,
        visual_results_row
    ],
    layout=Layout(
        width='1125px',
        overflow='visible'
    )
)

display(main_layout)
```
