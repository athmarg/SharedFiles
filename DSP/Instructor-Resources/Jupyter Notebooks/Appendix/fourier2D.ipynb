```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt

from ipywidgets import (
    FloatSlider,
    RadioButtons,
    HTML,
    VBox,
    HBox,
    Layout
)

from IPython.display import display

# ============================================================
# TWO-DIMENSIONAL FOURIER TRANSFORM
#
# f(x,y) <--> F(wx,wy)
#
# Three examples:
#
# 1. 2D Gaussian
# 2. Rectangular pulse
# 3. Rotated Gaussian
# ============================================================

plt.ioff()

# ============================================================
# JUPYTER / BINDER DISPLAY SETTINGS
# ============================================================

display(HTML("""
<style>

.container {
    width: 98% !important;
    max-width: none !important;
}

.output_area,
.output_subarea {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

.output_scroll {
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
    box-shadow: none !important;
}

.jp-Cell-outputWrapper,
.jp-OutputArea,
.jp-OutputArea-child,
.jp-OutputArea-output {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

.widget-output,
.jupyter-widgets-output-area,
.widget-box {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

.jupyter-matplotlib,
.jupyter-matplotlib-figure {
    overflow: visible !important;
    resize: none !important;
}

.jupyter-matplotlib::-webkit-resizer,
.jupyter-matplotlib-figure::-webkit-resizer {
    display: none !important;
}

.ft2-title {
    font-family: Arial, sans-serif;
    font-size: 20px;
    font-weight: bold;
    color: #6f3fa0;
}

.ft2-label {
    font-family: Arial, sans-serif;
    font-size: 14px;
    font-weight: bold;
}

.ft2-value {
    font-family: Arial, sans-serif;
    font-size: 14px;
    font-weight: bold;
    color: #0b3d91;
}

.ft2-radio .widget-radio-box {
    display: flex !important;
    flex-direction: row !important;
    flex-wrap: nowrap !important;
    gap: 20px !important;
    align-items: center !important;
}

.ft2-radio .widget-radio-box label {
    margin: 0 !important;
    white-space: nowrap !important;
}

.ft2-radio > label {
    display: none !important;
}

</style>
"""))

# ============================================================
# DOCUMENTATION
# ============================================================

documentation = HTML("""
<div style="
    width:1180px;
    font-family:Arial, sans-serif;
    font-size:15px;
    line-height:1.48;
    margin-bottom:10px;
">

<div class="ft2-title" style="margin-bottom:8px;">
Two-Dimensional Fourier Transform
</div>

<div style="margin-bottom:5px;">
The two-dimensional Fourier transform maps a function f(x,y)
defined in the spatial domain to a function F(ωₓ,ωᵧ) defined
in the two-dimensional frequency domain.
</div>

<div style="margin-bottom:5px;">
This notebook compares the spatial-domain function with the
magnitude of its two-dimensional Fourier transform.
</div>

<div>
<b>This notebook:</b> demonstrates the effects of spatial width,
anisotropy and rotation on the corresponding two-dimensional
frequency spectrum.
</div>

</div>
""")

# ============================================================
# FUNCTION SELECTOR
# ============================================================

function_selector = RadioButtons(
    options=[
        ('2D Gaussian', 'gaussian'),
        ('Rectangular pulse', 'rectangle'),
        ('Rotated Gaussian', 'rotated')
    ],
    value='gaussian',
    description='',
    layout=Layout(
        width='620px'
    )
)

function_selector.add_class(
    'ft2-radio'
)

# ============================================================
# SLIDER STYLE
# ============================================================

slider_style = {
    'description_width': '0px'
}

slider_layout = Layout(
    width='240px'
)

label_layout = Layout(
    width='135px',
    min_width='135px'
)

value_layout = Layout(
    width='70px',
    min_width='70px',
    margin='0px 0px 0px 6px'
)

row_layout = Layout(
    width='470px',
    height='40px',
    align_items='center'
)

# ============================================================
# PARAMETER a
# ============================================================

a_slider = FloatSlider(
    min=0.20,
    max=2.00,
    step=0.05,
    value=0.80,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

a_label = HTML(
    '<div class="ft2-label">Width a:</div>',
    layout=label_layout
)

a_value = HTML(
    '<div class="ft2-value">0.80</div>',
    layout=value_layout
)

a_row = HBox(
    [
        a_label,
        a_slider,
        a_value
    ],
    layout=row_layout
)

# ============================================================
# PARAMETER b
# ============================================================

b_slider = FloatSlider(
    min=0.20,
    max=2.00,
    step=0.05,
    value=1.20,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

b_label = HTML(
    '<div class="ft2-label">Width b:</div>',
    layout=label_layout
)

b_value = HTML(
    '<div class="ft2-value">1.20</div>',
    layout=value_layout
)

b_row = HBox(
    [
        b_label,
        b_slider,
        b_value
    ],
    layout=row_layout
)

# ============================================================
# ROTATION ANGLE
# ============================================================

angle_slider = FloatSlider(
    min=0.0,
    max=90.0,
    step=1.0,
    value=30.0,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

angle_label = HTML(
    '<div class="ft2-label">Rotation angle:</div>',
    layout=label_layout
)

angle_value = HTML(
    '<div class="ft2-value">30°</div>',
    layout=value_layout
)

angle_row = HBox(
    [
        angle_label,
        angle_slider,
        angle_value
    ],
    layout=row_layout
)

# ============================================================
# CONTROLS PANEL
# ============================================================

controls_panel = VBox(
    [
        HTML("""
        <div class="ft2-title" style="margin-bottom:8px;">
            Parameters
        </div>
        """),

        HTML("""
        <div class="ft2-label" style="margin-bottom:5px;">
            Spatial function
        </div>
        """),

        function_selector,

        HTML("""
        <div style="height:6px;"></div>
        """),

        a_row,
        b_row,
        angle_row
    ],
    layout=Layout(
        width='510px',
        padding='10px 14px',
        border='1px solid #d2c2df',
        overflow='visible'
    )
)

# ============================================================
# CURRENT INFORMATION
# ============================================================

current_info = HTML()

info_panel = VBox(
    [
        HTML("""
        <div class="ft2-title" style="margin-bottom:8px;">
            Interpretation
        </div>
        """),

        current_info
    ],
    layout=Layout(
        width='620px',
        padding='10px 14px',
        border='1px solid #d2c2df',
        overflow='visible'
    )
)

# ============================================================
# TOP ROW
# ============================================================

top_row = HBox(
    [
        controls_panel,
        info_panel
    ],
    layout=Layout(
        width='1150px',
        gap='15px',
        align_items='stretch',
        overflow='visible'
    )
)

# ============================================================
# SPATIAL GRID
# ============================================================

N = 256

x_min = -4.0
x_max = 4.0

x_axis = np.linspace(
    x_min,
    x_max,
    N
)

y_axis = np.linspace(
    x_min,
    x_max,
    N
)

X, Y = np.meshgrid(
    x_axis,
    y_axis
)

dx = (
    x_axis[1]
    -
    x_axis[0]
)

dy = (
    y_axis[1]
    -
    y_axis[0]
)

# ============================================================
# FREQUENCY AXES
#
# FFT frequencies in cycles per unit.
# ============================================================

fx_axis = np.fft.fftshift(
    np.fft.fftfreq(
        N,
        d=dx
    )
)

fy_axis = np.fft.fftshift(
    np.fft.fftfreq(
        N,
        d=dy
    )
)

# ============================================================
# FUNCTION GENERATOR
# ============================================================

def create_function(
    mode,
    a,
    b,
    angle_deg
):

    if mode == 'gaussian':

        Z = np.exp(
            -(
                X**2 / a**2
                +
                Y**2 / b**2
            )
        )

    elif mode == 'rectangle':

        Z = (
            (
                np.abs(X) <= a
            )
            &
            (
                np.abs(Y) <= b
            )
        ).astype(float)

    else:

        theta = np.deg2rad(
            angle_deg
        )

        Xr = (
            X * np.cos(theta)
            +
            Y * np.sin(theta)
        )

        Yr = (
            -X * np.sin(theta)
            +
            Y * np.cos(theta)
        )

        Z = np.exp(
            -(
                Xr**2 / a**2
                +
                Yr**2 / b**2
            )
        )

    return Z

# ============================================================
# FOURIER TRANSFORM
#
# Numerical approximation to the continuous 2D transform.
# ============================================================

def calculate_fft(Z):

    F = np.fft.fftshift(
        np.fft.fft2(
            np.fft.ifftshift(
                Z
            )
        )
    )

    F = (
        F
        *
        dx
        *
        dy
    )

    magnitude = np.abs(
        F
    )

    if np.max(
        magnitude
    ) > 0:

        magnitude = (
            magnitude
            /
            np.max(
                magnitude
            )
        )

    return magnitude

# ============================================================
# INITIAL DATA
# ============================================================

Z_initial = create_function(
    function_selector.value,
    a_slider.value,
    b_slider.value,
    angle_slider.value
)

F_initial = calculate_fft(
    Z_initial
)

# ============================================================
# FIGURE 1
# SPATIAL DOMAIN
# ============================================================

fig_space, ax_space = plt.subplots(
    figsize=(5.6, 4.8)
)

fig_space.canvas.header_visible = False
fig_space.canvas.footer_visible = False
fig_space.canvas.toolbar_visible = False

fig_space.canvas.layout = Layout(
    width='560px',
    height='480px',
    overflow='visible'
)

ax_space.set_title(
    'Spatial Domain f(x,y)',
    fontsize=14,
    fontweight='bold',
    color='#6f3fa0'
)

ax_space.set_xlabel(
    'x',
    fontsize=10
)

ax_space.set_ylabel(
    'y',
    fontsize=10
)

space_image = ax_space.imshow(
    Z_initial,
    extent=[
        x_min,
        x_max,
        x_min,
        x_max
    ],
    origin='lower',
    aspect='equal',
    vmin=0.0,
    vmax=1.0,
    interpolation='bilinear'
)

space_colorbar = fig_space.colorbar(
    space_image,
    ax=ax_space,
    fraction=0.046,
    pad=0.04
)

space_colorbar.set_label(
    'f(x,y)',
    fontsize=10
)

fig_space.subplots_adjust(
    left=0.12,
    right=0.90,
    top=0.90,
    bottom=0.13
)

# ============================================================
# FIGURE 2
# FREQUENCY DOMAIN
# ============================================================

fig_freq, ax_freq = plt.subplots(
    figsize=(5.6, 4.8)
)

fig_freq.canvas.header_visible = False
fig_freq.canvas.footer_visible = False
fig_freq.canvas.toolbar_visible = False

fig_freq.canvas.layout = Layout(
    width='560px',
    height='480px',
    overflow='visible'
)

ax_freq.set_title(
    'Magnitude of the 2D Fourier Transform',
    fontsize=14,
    fontweight='bold',
    color='#0b3d91'
)

ax_freq.set_xlabel(
    'Frequency fₓ',
    fontsize=10
)

ax_freq.set_ylabel(
    'Frequency fᵧ',
    fontsize=10
)

frequency_limit = 4.0

freq_image = ax_freq.imshow(
    F_initial,
    extent=[
        fx_axis[0],
        fx_axis[-1],
        fy_axis[0],
        fy_axis[-1]
    ],
    origin='lower',
    aspect='equal',
    vmin=0.0,
    vmax=1.0,
    interpolation='bilinear'
)

ax_freq.set_xlim(
    -frequency_limit,
    frequency_limit
)

ax_freq.set_ylim(
    -frequency_limit,
    frequency_limit
)

freq_colorbar = fig_freq.colorbar(
    freq_image,
    ax=ax_freq,
    fraction=0.046,
    pad=0.04
)

freq_colorbar.set_label(
    'Normalized |F(fₓ,fᵧ)|',
    fontsize=10
)

fig_freq.subplots_adjust(
    left=0.12,
    right=0.90,
    top=0.90,
    bottom=0.13
)

# ============================================================
# FIGURES ROW
# ============================================================

figures_row = HBox(
    [
        fig_space.canvas,
        fig_freq.canvas
    ],
    layout=Layout(
        width='1140px',
        gap='15px',
        align_items='flex-start',
        overflow='visible'
    )
)

# ============================================================
# BOTTOM INFORMATION
# ============================================================

bottom_panel = HTML("""
<div style="
    width:1120px;
    padding:9px 12px;
    border:1px solid #d7c7e5;
    font-family:Arial, sans-serif;
    font-size:14px;
    line-height:1.52;
    box-sizing:border-box;
">

<b style="color:#6f3fa0;">
Key observation:
</b>

A narrow spatial distribution produces a broad frequency-domain
distribution along the corresponding direction, while a broad spatial
distribution produces a narrow frequency-domain distribution.

</div>
""")

# ============================================================
# UPDATE FUNCTION
#
# No clear_output()
# No figure recreation
# No axis rescaling
#
# Only image data are updated.
# ============================================================

def update_notebook(change=None):

    mode = (
        function_selector.value
    )

    a = (
        a_slider.value
    )

    b = (
        b_slider.value
    )

    angle = (
        angle_slider.value
    )

    # --------------------------------------------------------
    # Current parameter labels
    # --------------------------------------------------------

    a_value.value = (
        f'<div class="ft2-value">{a:.2f}</div>'
    )

    b_value.value = (
        f'<div class="ft2-value">{b:.2f}</div>'
    )

    angle_value.value = (
        f'<div class="ft2-value">{angle:.0f}°</div>'
    )

    # --------------------------------------------------------
    # Function
    # --------------------------------------------------------

    Z = create_function(
        mode,
        a,
        b,
        angle
    )

    # --------------------------------------------------------
    # Fourier transform
    # --------------------------------------------------------

    F = calculate_fft(
        Z
    )

    # --------------------------------------------------------
    # Update images only
    # --------------------------------------------------------

    space_image.set_data(
        Z
    )

    freq_image.set_data(
        F
    )

    # --------------------------------------------------------
    # Interpretation
    # --------------------------------------------------------

    if mode == 'gaussian':

        current_info.value = f"""
        <div style="
            font-family:Arial;
            font-size:14px;
            line-height:1.55;
        ">

        <b>Selected function:</b>
        2D Gaussian

        <br><br>

        Width along x:
        <b>{a:.2f}</b>

        <br>

        Width along y:
        <b>{b:.2f}</b>

        <br><br>

        Increasing the spatial width in one direction narrows
        the Fourier spectrum in the corresponding frequency
        direction.

        </div>
        """

    elif mode == 'rectangle':

        current_info.value = f"""
        <div style="
            font-family:Arial;
            font-size:14px;
            line-height:1.55;
        ">

        <b>Selected function:</b>
        Rectangular pulse

        <br><br>

        Half-width along x:
        <b>{a:.2f}</b>

        <br>

        Half-width along y:
        <b>{b:.2f}</b>

        <br><br>

        The Fourier transform has the familiar two-dimensional
        sinc-like structure. Reducing the rectangle width broadens
        the corresponding frequency spectrum.

        </div>
        """

    else:

        current_info.value = f"""
        <div style="
            font-family:Arial;
            font-size:14px;
            line-height:1.55;
        ">

        <b>Selected function:</b>
        Rotated Gaussian

        <br><br>

        Principal width a:
        <b>{a:.2f}</b>

        <br>

        Principal width b:
        <b>{b:.2f}</b>

        <br>

        Rotation:
        <b>{angle:.0f}°</b>

        <br><br>

        Rotating the spatial function rotates the orientation
        of its Fourier-domain magnitude as well.

        </div>
        """

    # --------------------------------------------------------
    # Redraw only
    # --------------------------------------------------------

    fig_space.canvas.draw_idle()
    fig_freq.canvas.draw_idle()

# ============================================================
# CONNECT CONTROLS
# ============================================================

function_selector.observe(
    update_notebook,
    names='value'
)

a_slider.observe(
    update_notebook,
    names='value'
)

b_slider.observe(
    update_notebook,
    names='value'
)

angle_slider.observe(
    update_notebook,
    names='value'
)

# ============================================================
# INITIAL UPDATE
# ============================================================

update_notebook()

# ============================================================
# COMPLETE NOTEBOOK
# ============================================================

main_layout = VBox(
    [
        documentation,
        top_row,
        figures_row,
        bottom_panel
    ],
    layout=Layout(
        width='1180px',
        gap='10px',
        overflow='visible'
    )
)

display(
    main_layout
)
```
