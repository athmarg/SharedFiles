```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt

from ipywidgets import (
    FloatSlider,
    RadioButtons,
    HTML,
    VBox,
    HBox,
    Layout
)
from IPython.display import display

# ============================================================
# RESIDUES AND THE RESIDUE THEOREM
# ============================================================

plt.ioff()

# ============================================================
# CLASSIC JUPYTER + JUPYTERLAB / NOTEBOOK 7 / BINDER
#
# - no forced scrollbars
# - no trimming
# - no resize triangle
# - no matplotlib coordinate footer
# ============================================================

display(HTML("""
<style>

/* ============================================================
   CLASSIC JUPYTER NOTEBOOK
   ============================================================ */

.container {
    width: 98% !important;
    max-width: none !important;
}

.output_area,
.output_subarea {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

.output_scroll {
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
    box-shadow: none !important;
}

/* ============================================================
   JUPYTERLAB / NOTEBOOK 7
   ============================================================ */

.jp-Cell-outputWrapper,
.jp-OutputArea,
.jp-OutputArea-child,
.jp-OutputArea-output {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

/* ============================================================
   IPYWIDGETS
   ============================================================ */

.widget-output,
.jupyter-widgets-output-area,
.widget-box {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

/* ============================================================
   IPYMPL
   ============================================================ */

.jupyter-matplotlib,
.jupyter-matplotlib-figure {
    overflow: visible !important;
    resize: none !important;
}

.jupyter-matplotlib::-webkit-resizer,
.jupyter-matplotlib-figure::-webkit-resizer {
    display: none !important;
}

/* ============================================================
   NOTEBOOK STYLES
   ============================================================ */

.res-title {
    font-family:Arial, sans-serif;
    font-size:21px;
    font-weight:bold;
    color:#6f3fa0;
    margin-bottom:7px;
}

.res-label {
    font-family:Arial, sans-serif;
    font-size:14px;
    font-weight:bold;
}

.res-value {
    font-family:Arial, sans-serif;
    font-size:14px;
    font-weight:bold;
    color:#0b3d91;
}

.res-radio .widget-radio-box {
    display:flex !important;
    flex-direction:row !important;
    flex-wrap:nowrap !important;
    gap:18px !important;
    align-items:center !important;
}

.res-radio .widget-radio-box label {
    margin:0 !important;
    white-space:nowrap !important;
}

.res-radio > label {
    display:none !important;
}

</style>
"""))

# ============================================================
# DOCUMENTATION
# ============================================================

documentation = HTML("""
<div style="
    width:980px;
    font-family:Arial, sans-serif;
    font-size:15px;
    line-height:1.45;
    margin-bottom:8px;
">

<div class="res-title">
Residues and the Residue Theorem
</div>

<div style="margin-bottom:4px;">
If a complex function is analytic on and inside a closed contour C except
for isolated poles zₖ, then
∮<sub>C</sub> f(z)dz = 2πi Σ Res[f(zₖ)].
</div>

<div style="margin-bottom:4px;">
Only poles located inside the selected contour contribute to the integral.
Changing the contour radius may therefore change the value of the integral
when additional poles enter or leave the enclosed region.
</div>

<div>
<b>This notebook:</b> shows the poles of selected rational functions,
the circular contour |z| = R, the enclosed residues and the corresponding
value predicted by the residue theorem.
</div>

</div>
""")

# ============================================================
# FUNCTION SELECTOR
# ============================================================

function_selector = RadioButtons(
    options=[
        ('(1+z) / [z(2−z)]', 'f1'),
        ('e^z / [(z−1)(z+3)²]', 'f2')
    ],
    value='f1',
    description='',
    layout=Layout(
        width='620px'
    )
)

function_selector.add_class(
    'res-radio'
)

# ============================================================
# RADIUS SLIDER
# ============================================================

slider_style = {
    'description_width': '0px'
}

radius_slider = FloatSlider(
    min=0.25,
    max=6.0,
    step=0.05,
    value=1.0,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=Layout(
        width='220px'
    )
)

radius_value = HTML(
    '<div class="res-value">1.00</div>',
    layout=Layout(
        width='55px',
        min_width='55px',
        margin='0px 0px 0px 5px'
    )
)

radius_label = HTML(
    '<div class="res-label">Contour radius R:</div>',
    layout=Layout(
        width='145px',
        min_width='145px'
    )
)

radius_row = HBox(
    [
        radius_label,
        radius_slider,
        radius_value
    ],
    layout=Layout(
        width='430px',
        align_items='center'
    )
)

# ============================================================
# CONTROLS PANEL
# ============================================================

controls_panel = VBox(
    [
        HTML("""
        <div style="
            font-family:Arial;
            font-size:17px;
            font-weight:bold;
            color:#6f3fa0;
            margin-bottom:7px;
        ">
        Residue-Theorem Parameters
        </div>
        """),

        function_selector,

        radius_row
    ],
    layout=Layout(
        width='700px',
        padding='9px 12px',
        border='1px solid #d2c2df',
        overflow='visible'
    )
)

# ============================================================
# FUNCTION DATA
# ============================================================

def get_function_data(mode):

    if mode == 'f1':

        poles = np.array(
            [
                0.0 + 0.0j,
                2.0 + 0.0j
            ],
            dtype=complex
        )

        residues = np.array(
            [
                0.5 + 0.0j,
                -1.5 + 0.0j
            ],
            dtype=complex
        )

        name = (
            'f(z) = (1+z) / [z(2−z)]'
        )

        pole_orders = [
            'simple',
            'simple'
        ]

    else:

        poles = np.array(
            [
                1.0 + 0.0j,
                -3.0 + 0.0j
            ],
            dtype=complex
        )

        residues = np.array(
            [
                np.e / 16.0,
                -5.0 / (16.0 * np.e**3)
            ],
            dtype=complex
        )

        name = (
            'f(z) = e^z / [(z−1)(z+3)²]'
        )

        pole_orders = [
            'simple',
            'double'
        ]

    return (
        poles,
        residues,
        name,
        pole_orders
    )

# ============================================================
# COMPLEX STRING
# ============================================================

def complex_string(z):

    a = np.real(z)
    b = np.imag(z)

    if abs(a) < 1e-12:
        a = 0.0

    if abs(b) < 1e-12:
        b = 0.0

    sign = '+' if b >= 0 else '−'

    return (
        f'{a:.6f} '
        f'{sign} '
        f'{abs(b):.6f}i'
    )

# ============================================================
# COMPLEX PLANE FIGURE
# ============================================================

fig_plane, ax_plane = plt.subplots(
    figsize=(5.4, 5.0)
)

fig_plane.canvas.header_visible = False
fig_plane.canvas.footer_visible = False
fig_plane.canvas.toolbar_visible = False

fig_plane.canvas.layout = Layout(
    width='540px',
    height='500px'
)

ax_plane.set_title(
    'Poles and Circular Contour',
    fontsize=14,
    fontweight='bold',
    color='#6f3fa0'
)

ax_plane.set_xlabel(
    'Re(z)',
    fontsize=11
)

ax_plane.set_ylabel(
    'Im(z)',
    fontsize=11
)

ax_plane.set_aspect(
    'equal',
    adjustable='box'
)

ax_plane.set_xlim(
    -6.5,
    6.5
)

ax_plane.set_ylim(
    -6.5,
    6.5
)

ax_plane.axhline(
    0,
    color='black',
    linewidth=1.0
)

ax_plane.axvline(
    0,
    color='black',
    linewidth=1.0
)

ax_plane.grid(
    True,
    linestyle=':',
    alpha=0.40
)

# ============================================================
# CONTOUR
# ============================================================

theta = np.linspace(
    0.0,
    2.0 * np.pi,
    600
)

contour_line, = ax_plane.plot(
    [],
    [],
    linewidth=2.0,
    label='|z| = R'
)

# ============================================================
# POLE MARKERS
# ============================================================

pole_points, = ax_plane.plot(
    [],
    [],
    linestyle='None',
    marker='x',
    markersize=10,
    markeredgewidth=2
)

inside_points, = ax_plane.plot(
    [],
    [],
    linestyle='None',
    marker='o',
    markersize=14,
    fillstyle='none',
    markeredgewidth=2
)

pole_texts = [
    ax_plane.text(
        0,
        0,
        '',
        fontsize=10
    ),

    ax_plane.text(
        0,
        0,
        '',
        fontsize=10
    )
]

ax_plane.legend(
    loc='upper right',
    fontsize=9
)

fig_plane.subplots_adjust(
    left=0.12,
    right=0.97,
    top=0.90,
    bottom=0.11
)

# ============================================================
# RESIDUE CONTRIBUTION FIGURE
# ============================================================

fig_res, ax_res = plt.subplots(
    figsize=(5.2, 4.4)
)

fig_res.canvas.header_visible = False
fig_res.canvas.footer_visible = False
fig_res.canvas.toolbar_visible = False

fig_res.canvas.layout = Layout(
    width='520px',
    height='440px'
)

ax_res.set_title(
    'Residue Contributions',
    fontsize=14,
    fontweight='bold',
    color='#6f3fa0'
)

ax_res.set_xlabel(
    'Pole',
    fontsize=10
)

ax_res.set_ylabel(
    'Real part of residue',
    fontsize=10
)

ax_res.axhline(
    0,
    color='black',
    linewidth=0.9
)

ax_res.grid(
    True,
    axis='y',
    linestyle=':',
    alpha=0.40
)

residue_bars = ax_res.bar(
    [0, 1],
    [0, 0]
)

ax_res.set_xticks(
    [0, 1]
)

fig_res.subplots_adjust(
    left=0.14,
    right=0.97,
    top=0.89,
    bottom=0.15
)

# ============================================================
# RESULTS PANEL
# ============================================================

results_html = HTML()

results_panel = VBox(
    [
        HTML("""
        <div style="
            font-family:Arial;
            font-size:17px;
            font-weight:bold;
            color:#6f3fa0;
            margin-bottom:6px;
        ">
        Residue-Theorem Results
        </div>
        """),

        results_html
    ],
    layout=Layout(
        width='355px',
        padding='10px 12px',
        border='1px solid #d2c2df',
        overflow='visible'
    )
)

# ============================================================
# INTERPRETATION PANEL
# ============================================================

interpretation_html = HTML(
    """
    <div style="
        width:1045px;
        padding:9px 11px;
        border:1px solid #d7c7e5;
        background:#fbf8fe;
        font-family:Arial, sans-serif;
        font-size:13px;
        line-height:1.50;
        box-sizing:border-box;
        margin-top:8px;
    ">

        <div style="
            color:#6f3fa0;
            font-weight:bold;
            margin-bottom:5px;
        ">
            Interpretation
        </div>

        <div>
            Only poles enclosed by the contour contribute to the contour
            integral. When the radius R crosses the modulus of a pole,
            that pole enters or leaves the contour and the residue sum
            changes accordingly.
        </div>

    </div>
    """
)

# ============================================================
# UPDATE FUNCTION
# ============================================================

def update_notebook(change=None):

    mode = function_selector.value

    R = radius_slider.value

    radius_value.value = (
        f'<div class="res-value">{R:.2f}</div>'
    )

    poles, residues, name, pole_orders = get_function_data(
        mode
    )

    # ========================================================
    # CONTOUR
    # ========================================================

    contour_line.set_data(
        R * np.cos(theta),
        R * np.sin(theta)
    )

    # ========================================================
    # POLES
    # ========================================================

    pole_points.set_data(
        np.real(poles),
        np.imag(poles)
    )

    inside_mask = (
        np.abs(poles)
        <
        R
    )

    inside_poles = poles[
        inside_mask
    ]

    inside_points.set_data(
        np.real(inside_poles),
        np.imag(inside_poles)
    )

    # --------------------------------------------------------
    # Labels
    # --------------------------------------------------------

    for index in range(len(poles)):

        pole_texts[index].set_position(
            (
                np.real(poles[index]) + 0.18,
                np.imag(poles[index]) + 0.20
            )
        )

        pole_texts[index].set_text(
            f'z{index + 1}'
        )

    # ========================================================
    # RESIDUE SUM
    # ========================================================

    enclosed_residues = residues[
        inside_mask
    ]

    residue_sum = np.sum(
        enclosed_residues
    )

    contour_integral = (
        2.0
        *
        np.pi
        *
        1j
        *
        residue_sum
    )

    # ========================================================
    # BAR GRAPH
    #
    # Only residues corresponding to poles INSIDE the contour
    # contribute to the residue sum.
    #
    # A pole outside the contour therefore has ZERO
    # contribution and its bar is not displayed.
    # ========================================================

    ax_res.set_xticklabels(
        [
            f'z₁ = {np.real(poles[0]):.0f}',
            f'z₂ = {np.real(poles[1]):.0f}'
        ]
    )

    real_residues = np.real(
        residues
    )

    for index, bar in enumerate(residue_bars):

        if inside_mask[index]:

            # Pole inside C:
            # its residue contributes to the integral.

            bar.set_height(
                real_residues[index]
            )

            bar.set_alpha(
                1.0
            )

        else:

            # Pole outside C:
            # contribution to the residue sum is zero.

            bar.set_height(
                0.0
            )

            bar.set_alpha(
                1.0
            )

    max_bar = max(
        1.0,
        np.max(
            np.abs(real_residues)
        )
    )

    ax_res.set_ylim(
        -1.25 * max_bar,
        1.25 * max_bar
    )

    # ========================================================
    # RESULTS TEXT
    # ========================================================

    pole_lines = ""

    for index in range(len(poles)):

        location = (
            "inside"
            if inside_mask[index]
            else "outside"
        )

        pole_lines += f"""
        <div style="margin-bottom:6px;">
            <b>z{index+1} = {complex_string(poles[index])}</b>
            &nbsp; ({pole_orders[index]} pole)
            <br>
            Res[f(z{index+1})]
            =
            <b>{complex_string(residues[index])}</b>
            <br>
            Position:
            <b>{location} C</b>
        </div>
        """

    if len(enclosed_residues) == 0:

        enclosed_text = (
            "No poles are enclosed."
        )

    elif len(enclosed_residues) == 1:

        enclosed_text = (
            "One pole is enclosed."
        )

    else:

        enclosed_text = (
            "Two poles are enclosed."
        )

    results_html.value = f"""
    <div style="
        font-family:Arial, sans-serif;
        font-size:13px;
        line-height:1.52;
    ">

        <b>{name}</b>

        <br><br>

        Contour:
        <b>|z| = {R:.2f}</b>

        <br><br>

        {pole_lines}

        <hr style="margin:8px 0px;">

        <b>{enclosed_text}</b>

        <br><br>

        Σ Res =
        <b>{complex_string(residue_sum)}</b>

        <br><br>

        Residue theorem:

        <br>

        ∮<sub>C</sub> f(z)dz
        =
        2πi Σ Res

        <br>

        =
        <b>{complex_string(contour_integral)}</b>

    </div>
    """

    # ========================================================
    # REDRAW
    # ========================================================

    fig_plane.canvas.draw_idle()
    fig_res.canvas.draw_idle()

# ============================================================
# CONNECT CONTROLS
# ============================================================

function_selector.observe(
    update_notebook,
    names='value'
)

radius_slider.observe(
    update_notebook,
    names='value'
)

# ============================================================
# INITIAL UPDATE
# ============================================================

update_notebook()

# ============================================================
# FIGURES ROW
# ============================================================

figures_row = HBox(
    [
        fig_plane.canvas,
        fig_res.canvas
    ],
    layout=Layout(
        width='1070px',
        gap='10px',
        align_items='flex-start',
        overflow='visible'
    )
)

# ============================================================
# LEFT COLUMN
# ============================================================

left_column = VBox(
    [
        figures_row,
        interpretation_html
    ],
    layout=Layout(
        width='1070px',
        overflow='visible'
    )
)

# ============================================================
# MAIN VISUAL ROW
# ============================================================

visual_row = HBox(
    [
        left_column,
        results_panel
    ],
    layout=Layout(
        width='1440px',
        gap='12px',
        align_items='flex-start',
        overflow='visible'
    )
)

# ============================================================
# COMPLETE NOTEBOOK
# ============================================================

main_layout = VBox(
    [
        documentation,
        controls_panel,
        visual_row
    ],
    layout=Layout(
        width='1440px',
        overflow='visible'
    )
)

display(main_layout)
```
