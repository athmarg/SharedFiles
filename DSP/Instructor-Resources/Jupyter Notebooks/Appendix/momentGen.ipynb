```python
%matplotlib inline

import sympy as sp
from ipywidgets import Dropdown, IntSlider, interactive
from IPython.display import display, Markdown, Math

# ------------------------------------------------------------
# Moment-Generating Function
# ------------------------------------------------------------

s = sp.symbols('s', real=True)

def mgf_demo(distribution, k):

    if distribution == 'Standard Gaussian':
        M = sp.exp(s**2 / 2)

    elif distribution == 'Exponential λ=1':
        M = 1 / (1 - s)

    elif distribution == 'Bernoulli p=1/2':
        M = (
            sp.Rational(1, 2)
            + sp.Rational(1, 2) * sp.exp(s)
        )

    elif distribution == 'Poisson λ=2':
        M = sp.exp(
            2 * (sp.exp(s) - 1)
        )

    # --------------------------------------------------------
    # Symbolic derivative and raw moment
    # --------------------------------------------------------

    derivative = sp.diff(
        M,
        s,
        k
    )

    moment = sp.simplify(
        derivative.subs(s, 0)
    )

    # --------------------------------------------------------
    # Results
    # --------------------------------------------------------

    display(
        Markdown(
            f"### {distribution}"
        )
    )

    display(
        Math(
            r'M_X(s) = '
            + sp.latex(M)
        )
    )

    display(
        Math(
            r'\frac{d^{'
            + str(k)
            + r'}M_X(s)}{ds^{'
            + str(k)
            + r'}}'
            r' = '
            + sp.latex(derivative)
        )
    )

    display(
        Math(
            r'E[X^{'
            + str(k)
            + r'}]'
            r' = M_X^{('
            + str(k)
            + r')}(0)'
            r' = '
            + sp.latex(moment)
        )
    )

    print()

    print(
        f'Raw moment of order {k} = {moment}'
    )


# ============================================================
# TITLE AND INTRODUCTION
# ============================================================

display(
    Markdown(
        "## Moment-Generating Function and Moments"
    )
)

display(
    Markdown(
        r"""
The **moment-generating function (MGF)** of a random variable \(X\) is
defined as \(M_X(s)=E[e^{sX}]\), whenever this expectation exists in
a neighborhood of \(s=0\). Its derivatives evaluated at the origin
generate the raw moments of the distribution according to
\(E[X^k]=M_X^{(k)}(0)\). This notebook performs these differentiations
symbolically for several common probability distributions.
"""
    )
)

# ============================================================
# INTERACTIVE CONTROLS
# ============================================================

display(
    interactive(
        mgf_demo,

        distribution=Dropdown(
            options=[
                'Standard Gaussian',
                'Exponential λ=1',
                'Bernoulli p=1/2',
                'Poisson λ=2'
            ],
            value='Standard Gaussian',
            description='Distribution:'
        ),

        k=IntSlider(
            value=1,
            min=1,
            max=8,
            step=1,
            description='Order k:',
            continuous_update=True
        )
    )
)
```
