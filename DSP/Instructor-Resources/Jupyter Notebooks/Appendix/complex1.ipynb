```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt

from ipywidgets import (
    FloatSlider,
    RadioButtons,
    HTML,
    VBox,
    HBox,
    Layout
)
from IPython.display import display

# ============================================================
# CLASSIC JUPYTER + JUPYTERLAB / NOTEBOOK 7 / BINDER
# WIDTH AND OUTPUT SETTINGS
# ============================================================

display(HTML("""
<style>

/* ============================================================
   CLASSIC JUPYTER NOTEBOOK
   ============================================================ */

.container {
    width: 98% !important;
    max-width: none !important;
}

.output_area,
.output_subarea {
    width: 100% !important;
    max-width: none !important;
    overflow: visible !important;
}

.output_scroll {
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
    box-shadow: none !important;
}

/* ============================================================
   JUPYTERLAB / NOTEBOOK 7
   ============================================================ */

.jp-Notebook {
    width: 98% !important;
    max-width: none !important;
}

.jp-Cell,
.jp-CodeCell,
.jp-Cell-outputWrapper,
.jp-OutputArea,
.jp-OutputArea-child,
.jp-OutputArea-output {
    max-width: none !important;
    overflow: visible !important;
}

.jp-OutputArea-output {
    overflow: visible !important;
}

/* ============================================================
   IPYWIDGETS
   ============================================================ */

.widget-output,
.jupyter-widgets-output-area {
    max-width: none !important;
    overflow: visible !important;
}

/* ============================================================
   NOTEBOOK-SPECIFIC STYLES
   ============================================================ */

.complex-title {
    font-family: Arial, sans-serif;
    font-size: 21px;
    font-weight: bold;
    color: #714d95;
    margin-bottom: 7px;
}

.complex-label {
    font-family: Arial, sans-serif;
    font-size: 14px;
    font-weight: bold;
}

.complex-value {
    font-family: Arial, sans-serif;
    font-size: 14px;
    font-weight: bold;
    color: #0b3d91;
}

/* Radio buttons in one compact row */
.complex-radio .widget-radio-box {
    display: flex !important;
    flex-direction: row !important;
    flex-wrap: nowrap !important;
    gap: 14px !important;
    align-items: center !important;
}

.complex-radio .widget-radio-box label {
    margin: 0 !important;
    white-space: nowrap !important;
}

.complex-radio > label {
    display: none !important;
}

/* Prevent internal widget scrollbars */
.complex-controls,
.complex-results,
.complex-main {
    overflow: visible !important;
}

</style>
"""))

# ============================================================
# INITIAL MATPLOTLIB STATE
# ============================================================

plt.ioff()

# ============================================================
# SHORT DOCUMENTATION
# ============================================================

documentation = HTML("""
<div style="
    width:1040px;
    font-family:Arial, sans-serif;
    font-size:15px;
    line-height:1.45;
    margin-bottom:8px;
">

<div class="complex-title">
Complex Numbers: Representation and Basic Operations
</div>

A complex number z = a + ib is represented by the point (a,b) in the complex plane.
Its modulus and phase define the equivalent polar and exponential forms.
The conjugate is obtained by reflecting the point about the real axis.
This notebook illustrates these representations and the four elementary operations
between two complex numbers.

</div>
""")

# ============================================================
# SLIDER SETTINGS
# ============================================================

slider_style = {
    'description_width': '0px'
}

slider_layout = Layout(
    width='150px'
)

value_layout = Layout(
    width='50px',
    min_width='50px',
    margin='0px 0px 0px 5px'
)

label_layout = Layout(
    width='72px',
    min_width='72px'
)

# ============================================================
# SLIDERS
# ============================================================

a1_slider = FloatSlider(
    min=-3.0,
    max=3.0,
    step=0.25,
    value=2.0,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

b1_slider = FloatSlider(
    min=-3.0,
    max=3.0,
    step=0.25,
    value=2.0,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

a2_slider = FloatSlider(
    min=-3.0,
    max=3.0,
    step=0.25,
    value=-2.0,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

b2_slider = FloatSlider(
    min=-3.0,
    max=3.0,
    step=0.25,
    value=1.5,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

# ============================================================
# CURRENT VALUE LABELS
# ============================================================

a1_value = HTML(
    '<div class="complex-value">2.00</div>',
    layout=value_layout
)

b1_value = HTML(
    '<div class="complex-value">2.00</div>',
    layout=value_layout
)

a2_value = HTML(
    '<div class="complex-value">-2.00</div>',
    layout=value_layout
)

b2_value = HTML(
    '<div class="complex-value">1.50</div>',
    layout=value_layout
)

# ============================================================
# TEXT LABELS
# ============================================================

a1_label = HTML(
    '<div class="complex-label">Re(z₁):</div>',
    layout=label_layout
)

b1_label = HTML(
    '<div class="complex-label">Im(z₁):</div>',
    layout=label_layout
)

a2_label = HTML(
    '<div class="complex-label">Re(z₂):</div>',
    layout=label_layout
)

b2_label = HTML(
    '<div class="complex-label">Im(z₂):</div>',
    layout=label_layout
)

# ============================================================
# SLIDER ROWS
# ============================================================

a1_row = HBox(
    [
        a1_label,
        a1_slider,
        a1_value
    ],
    layout=Layout(
        width='280px',
        align_items='center'
    )
)

b1_row = HBox(
    [
        b1_label,
        b1_slider,
        b1_value
    ],
    layout=Layout(
        width='280px',
        align_items='center'
    )
)

a2_row = HBox(
    [
        a2_label,
        a2_slider,
        a2_value
    ],
    layout=Layout(
        width='280px',
        align_items='center'
    )
)

b2_row = HBox(
    [
        b2_label,
        b2_slider,
        b2_value
    ],
    layout=Layout(
        width='280px',
        align_items='center'
    )
)

slider_row_1 = HBox(
    [
        a1_row,
        b1_row
    ],
    layout=Layout(
        width='580px',
        gap='12px',
        align_items='center'
    )
)

slider_row_2 = HBox(
    [
        a2_row,
        b2_row
    ],
    layout=Layout(
        width='580px',
        gap='12px',
        align_items='center'
    )
)

# ============================================================
# OPERATION SELECTOR
# ============================================================

operation_selector = RadioButtons(
    options=[
        'Representation',
        'Conjugate',
        'Addition',
        'Subtraction',
        'Multiplication',
        'Division'
    ],
    value='Representation',
    description='',
    layout=Layout(
        width='590px'
    )
)

operation_selector.add_class(
    'complex-radio'
)

# ============================================================
# CONTROLS PANEL
# ============================================================

controls_panel = VBox(
    [
        HTML("""
        <div style="
            font-family:Arial;
            font-size:17px;
            font-weight:bold;
            color:#714d95;
            margin-bottom:7px;
        ">
        Complex Number Parameters
        </div>
        """),

        slider_row_1,
        slider_row_2,

        HTML("""
        <div style="
            font-family:Arial;
            font-size:14px;
            font-weight:bold;
            margin-top:7px;
            margin-bottom:3px;
        ">
        Display / Operation
        </div>
        """),

        operation_selector
    ],
    layout=Layout(
        width='620px',
        padding='9px 12px',
        border='1px solid #d1c1df',
        overflow='visible'
    )
)

controls_panel.add_class(
    'complex-controls'
)

# ============================================================
# COMPLEX PLANE FIGURE
# ============================================================

fig, ax = plt.subplots(
    figsize=(5.5, 5.2)
)

fig.canvas.header_visible = False

fig.canvas.layout = Layout(
    width='550px',
    height='520px'
)

ax.set_aspect(
    'equal',
    adjustable='box'
)

ax.axhline(
    0,
    color='black',
    linewidth=1.0
)

ax.axvline(
    0,
    color='black',
    linewidth=1.0
)

ax.grid(
    True,
    linestyle=':',
    alpha=0.40
)

ax.set_xlabel(
    'Real axis',
    fontsize=11
)

ax.set_ylabel(
    'Imaginary axis',
    fontsize=11
)

ax.set_title(
    'Complex Plane',
    fontsize=15,
    fontweight='bold',
    color='#714d95'
)

# ============================================================
# PRE-CREATED GRAPHICAL OBJECTS
# ============================================================

z1_line, = ax.plot(
    [],
    [],
    linewidth=2.0,
    color='blue',
    marker='o',
    markevery=[1],
    label='z₁'
)

z2_line, = ax.plot(
    [],
    [],
    linewidth=2.0,
    color='red',
    marker='o',
    markevery=[1],
    label='z₂'
)

result_line, = ax.plot(
    [],
    [],
    linewidth=2.2,
    color='green',
    marker='o',
    markevery=[1],
    label='result'
)

conjugate_line, = ax.plot(
    [],
    [],
    linewidth=2.0,
    color='purple',
    linestyle='--',
    marker='o',
    markevery=[1],
    label='z₁*'
)

angle_arc_1, = ax.plot(
    [],
    [],
    linewidth=1.3,
    color='blue',
    alpha=0.8
)

angle_arc_2, = ax.plot(
    [],
    [],
    linewidth=1.3,
    color='red',
    alpha=0.8
)

# ============================================================
# DYNAMIC TEXT LABELS
# ============================================================

z1_text = ax.text(
    0,
    0,
    '',
    fontsize=10,
    color='blue'
)

z2_text = ax.text(
    0,
    0,
    '',
    fontsize=10,
    color='red'
)

result_text = ax.text(
    0,
    0,
    '',
    fontsize=10,
    color='green'
)

conjugate_text = ax.text(
    0,
    0,
    '',
    fontsize=10,
    color='purple'
)

ax.legend(
    loc='upper left',
    fontsize=9
)

fig.subplots_adjust(
    left=0.12,
    right=0.97,
    top=0.91,
    bottom=0.11
)

# ============================================================
# NUMERICAL OUTPUT
# ============================================================

result_html = HTML()

result_panel = VBox(
    [
        HTML("""
        <div style="
            font-family:Arial;
            font-size:17px;
            font-weight:bold;
            color:#714d95;
            margin-bottom:5px;
        ">
        Numerical Results
        </div>
        """),

        result_html
    ],
    layout=Layout(
        width='430px',
        padding='10px 12px',
        border='1px solid #d1c1df',
        overflow='visible'
    )
)

result_panel.add_class(
    'complex-results'
)

# ============================================================
# FORMAT FUNCTIONS
# ============================================================

def rectangular_form(z):

    a = np.real(z)
    b = np.imag(z)

    sign = '+' if b >= 0 else '−'

    return f'{a:.4f} {sign} {abs(b):.4f}i'


def polar_data(z):

    r = np.abs(z)

    if r < 1e-12:
        return 0.0, 0.0, 0.0

    phi = np.angle(z)

    phi_deg = np.degrees(phi)

    if phi_deg < 0:
        phi_deg += 360.0

    phi_display = np.deg2rad(
        phi_deg
    )

    return r, phi_display, phi_deg


def polar_form(z):

    r, phi, phi_deg = polar_data(z)

    return (
        f'{r:.4f}'
        f'[cos({phi_deg:.2f}°) + i sin({phi_deg:.2f}°)]'
    )


def exponential_form(z):

    r, phi, phi_deg = polar_data(z)

    return (
        f'{r:.4f} e<sup>i {phi:.4f}</sup>'
        f' &nbsp; ({phi_deg:.2f}°)'
    )

# ============================================================
# ANGLE ARC GENERATOR
# ============================================================

def create_angle_arc(z, radius):

    if np.abs(z) < 1e-12:
        return np.array([]), np.array([])

    phi = np.angle(z)

    if phi < 0:
        phi += 2.0 * np.pi

    theta = np.linspace(
        0,
        phi,
        120
    )

    return (
        radius * np.cos(theta),
        radius * np.sin(theta)
    )

# ============================================================
# HIDE DYNAMIC OBJECT
# ============================================================

def hide_line(line):

    line.set_data(
        [],
        []
    )

# ============================================================
# DYNAMIC AXIS SCALING
#
# The displayed complex numbers now occupy a substantial
# fraction of the complex plane.
# ============================================================

def update_axis_limits(values):

    coords = [0.0]

    for z in values:

        if z is not None and np.isfinite(np.real(z)) and np.isfinite(np.imag(z)):

            coords.append(
                abs(np.real(z))
            )

            coords.append(
                abs(np.imag(z))
            )

    maximum = max(coords)

    # Minimum visible scale:
    # inputs in [-3,3] are now clearly visible.
    limit = max(
        4.0,
        1.30 * maximum
    )

    # Round upward to a convenient half-unit
    limit = np.ceil(
        2.0 * limit
    ) / 2.0

    ax.set_xlim(
        -limit,
        limit
    )

    ax.set_ylim(
        -limit,
        limit
    )

# ============================================================
# MAIN UPDATE FUNCTION
# ============================================================

def update_notebook(change=None):

    # --------------------------------------------------------
    # COMPLEX NUMBERS
    # --------------------------------------------------------

    a1 = a1_slider.value
    b1 = b1_slider.value
    a2 = a2_slider.value
    b2 = b2_slider.value

    z1 = complex(
        a1,
        b1
    )

    z2 = complex(
        a2,
        b2
    )

    operation = operation_selector.value

    # --------------------------------------------------------
    # UPDATE CURRENT SLIDER VALUES
    # --------------------------------------------------------

    a1_value.value = (
        f'<div class="complex-value">{a1:.2f}</div>'
    )

    b1_value.value = (
        f'<div class="complex-value">{b1:.2f}</div>'
    )

    a2_value.value = (
        f'<div class="complex-value">{a2:.2f}</div>'
    )

    b2_value.value = (
        f'<div class="complex-value">{b2:.2f}</div>'
    )

    # --------------------------------------------------------
    # SHOW z1 AND z2
    # --------------------------------------------------------

    z1_line.set_data(
        [0, a1],
        [0, b1]
    )

    z2_line.set_data(
        [0, a2],
        [0, b2]
    )

    z1_text.set_position(
        (
            a1 + 0.15,
            b1 + 0.15
        )
    )

    z1_text.set_text(
        'z₁'
    )

    z2_text.set_position(
        (
            a2 + 0.15,
            b2 + 0.15
        )
    )

    z2_text.set_text(
        'z₂'
    )

    # --------------------------------------------------------
    # HIDE OPTIONAL OBJECTS
    # --------------------------------------------------------

    hide_line(
        result_line
    )

    hide_line(
        conjugate_line
    )

    hide_line(
        angle_arc_1
    )

    hide_line(
        angle_arc_2
    )

    result_text.set_text(
        ''
    )

    conjugate_text.set_text(
        ''
    )

    # --------------------------------------------------------
    # BASIC POLAR DATA
    # --------------------------------------------------------

    r1, phi1, phi1_deg = polar_data(
        z1
    )

    r2, phi2, phi2_deg = polar_data(
        z2
    )

    displayed_values = [
        z1,
        z2
    ]

    # ========================================================
    # REPRESENTATION
    # ========================================================

    if operation == 'Representation':

        base_radius = max(
            0.55,
            0.22 * max(
                abs(z1),
                abs(z2),
                1.0
            )
        )

        arc_x1, arc_y1 = create_angle_arc(
            z1,
            radius=base_radius
        )

        arc_x2, arc_y2 = create_angle_arc(
            z2,
            radius=1.45 * base_radius
        )

        angle_arc_1.set_data(
            arc_x1,
            arc_y1
        )

        angle_arc_2.set_data(
            arc_x2,
            arc_y2
        )

        result_html.value = f"""
        <div style="
            font-family:Arial;
            font-size:14px;
            line-height:1.58;
        ">

        <b style="color:#1555cc;">z₁</b><br>

        Rectangular:
        <b>{rectangular_form(z1)}</b><br>

        Modulus:
        |z₁| = <b>{r1:.4f}</b><br>

        Phase:
        φ₁ = <b>{phi1:.4f} rad = {phi1_deg:.2f}°</b><br>

        Polar:
        <b>{polar_form(z1)}</b><br>

        Exponential:
        <b>{exponential_form(z1)}</b>

        <hr style="margin:9px 0px;">

        <b style="color:#d7191c;">z₂</b><br>

        Rectangular:
        <b>{rectangular_form(z2)}</b><br>

        Modulus:
        |z₂| = <b>{r2:.4f}</b><br>

        Phase:
        φ₂ = <b>{phi2:.4f} rad = {phi2_deg:.2f}°</b><br>

        Polar:
        <b>{polar_form(z2)}</b><br>

        Exponential:
        <b>{exponential_form(z2)}</b>

        </div>
        """

    # ========================================================
    # CONJUGATE
    # ========================================================

    elif operation == 'Conjugate':

        zc = np.conjugate(
            z1
        )

        displayed_values.append(
            zc
        )

        conjugate_line.set_data(
            [0, np.real(zc)],
            [0, np.imag(zc)]
        )

        conjugate_text.set_position(
            (
                np.real(zc) + 0.15,
                np.imag(zc) + 0.15
            )
        )

        conjugate_text.set_text(
            'z₁*'
        )

        product = (
            z1 * zc
        )

        result_html.value = f"""
        <div style="
            font-family:Arial;
            font-size:14px;
            line-height:1.62;
        ">

        <b>Original number</b><br>

        z₁ = <b>{rectangular_form(z1)}</b>

        <br><br>

        <b>Conjugate</b><br>

        z₁* = <b>{rectangular_form(zc)}</b>

        <br><br>

        |z₁| = <b>{abs(z1):.4f}</b><br>

        |z₁*| = <b>{abs(zc):.4f}</b>

        <br><br>

        z₁z₁* = <b>{np.real(product):.4f}</b><br>

        |z₁|² = <b>{abs(z1)**2:.4f}</b>

        <br><br>

        The conjugate is the reflection of z₁
        about the real axis.

        </div>
        """

    # ========================================================
    # ADDITION
    # ========================================================

    elif operation == 'Addition':

        result = (
            z1 + z2
        )

        displayed_values.append(
            result
        )

        result_line.set_data(
            [0, np.real(result)],
            [0, np.imag(result)]
        )

        result_text.set_position(
            (
                np.real(result) + 0.15,
                np.imag(result) + 0.15
            )
        )

        result_text.set_text(
            'z₁ + z₂'
        )

        result_html.value = f"""
        <div style="
            font-family:Arial;
            font-size:14px;
            line-height:1.62;
        ">

        z₁ = <b>{rectangular_form(z1)}</b><br>

        z₂ = <b>{rectangular_form(z2)}</b>

        <hr style="margin:8px 0px;">

        <b>Addition</b><br>

        z₁ + z₂ =
        <b>{rectangular_form(result)}</b>

        <br><br>

        Polar:
        <b>{polar_form(result)}</b><br>

        Exponential:
        <b>{exponential_form(result)}</b>

        </div>
        """

    # ========================================================
    # SUBTRACTION
    # ========================================================

    elif operation == 'Subtraction':

        result = (
            z1 - z2
        )

        displayed_values.append(
            result
        )

        result_line.set_data(
            [0, np.real(result)],
            [0, np.imag(result)]
        )

        result_text.set_position(
            (
                np.real(result) + 0.15,
                np.imag(result) + 0.15
            )
        )

        result_text.set_text(
            'z₁ − z₂'
        )

        result_html.value = f"""
        <div style="
            font-family:Arial;
            font-size:14px;
            line-height:1.62;
        ">

        z₁ = <b>{rectangular_form(z1)}</b><br>

        z₂ = <b>{rectangular_form(z2)}</b>

        <hr style="margin:8px 0px;">

        <b>Subtraction</b><br>

        z₁ − z₂ =
        <b>{rectangular_form(result)}</b>

        <br><br>

        Polar:
        <b>{polar_form(result)}</b><br>

        Exponential:
        <b>{exponential_form(result)}</b>

        </div>
        """

    # ========================================================
    # MULTIPLICATION
    # ========================================================

    elif operation == 'Multiplication':

        result = (
            z1 * z2
        )

        displayed_values.append(
            result
        )

        result_line.set_data(
            [0, np.real(result)],
            [0, np.imag(result)]
        )

        result_text.set_position(
            (
                np.real(result) + 0.15,
                np.imag(result) + 0.15
            )
        )

        result_text.set_text(
            'z₁z₂'
        )

        rr, pp, pp_deg = polar_data(
            result
        )

        result_html.value = f"""
        <div style="
            font-family:Arial;
            font-size:14px;
            line-height:1.60;
        ">

        z₁ = <b>{rectangular_form(z1)}</b><br>

        z₂ = <b>{rectangular_form(z2)}</b>

        <hr style="margin:8px 0px;">

        <b>Multiplication</b><br>

        z₁z₂ =
        <b>{rectangular_form(result)}</b>

        <br><br>

        |z₁z₂| = |z₁||z₂|

        <br>

        {rr:.4f}
        =
        {r1:.4f} × {r2:.4f}

        <br><br>

        arg(z₁z₂)
        =
        arg(z₁) + arg(z₂)

        <br>

        <b>{pp_deg:.2f}°</b>
        =
        ({phi1_deg:.2f}° + {phi2_deg:.2f}°)
        mod 360°

        <br><br>

        <b>Geometrically:</b>
        multiplication combines scaling and rotation.

        </div>
        """

    # ========================================================
    # DIVISION
    # ========================================================

    elif operation == 'Division':

        if abs(z2) < 1e-12:

            result_html.value = """
            <div style="
                font-family:Arial;
                font-size:14px;
                line-height:1.60;
                color:#b00020;
            ">

            <b>Division is undefined because z₂ = 0.</b>

            </div>
            """

        else:

            result = (
                z1 / z2
            )

            displayed_values.append(
                result
            )

            result_line.set_data(
                [0, np.real(result)],
                [0, np.imag(result)]
            )

            result_text.set_position(
                (
                    np.real(result) + 0.15,
                    np.imag(result) + 0.15
                )
            )

            result_text.set_text(
                'z₁ / z₂'
            )

            rr, pp, pp_deg = polar_data(
                result
            )

            result_html.value = f"""
            <div style="
                font-family:Arial;
                font-size:14px;
                line-height:1.60;
            ">

            z₁ = <b>{rectangular_form(z1)}</b><br>

            z₂ = <b>{rectangular_form(z2)}</b>

            <hr style="margin:8px 0px;">

            <b>Division</b><br>

            z₁ / z₂ =
            <b>{rectangular_form(result)}</b>

            <br><br>

            |z₁/z₂|
            =
            |z₁| / |z₂|

            <br>

            {rr:.4f}
            =
            {r1:.4f} / {r2:.4f}

            <br><br>

            arg(z₁/z₂)
            =
            arg(z₁) − arg(z₂)

            <br>

            <b>{pp_deg:.2f}°</b>
            =
            ({phi1_deg:.2f}° − {phi2_deg:.2f}°)
            mod 360°

            <br><br>

            <b>Geometrically:</b>
            division combines inverse scaling
            and subtraction of angles.

            </div>
            """

    # --------------------------------------------------------
    # DYNAMIC AXES
    # --------------------------------------------------------

    update_axis_limits(
        displayed_values
    )

    # --------------------------------------------------------
    # REDRAW
    # --------------------------------------------------------

    fig.canvas.draw_idle()

# ============================================================
# CONNECT CONTROLS
# ============================================================

a1_slider.observe(
    update_notebook,
    names='value'
)

b1_slider.observe(
    update_notebook,
    names='value'
)

a2_slider.observe(
    update_notebook,
    names='value'
)

b2_slider.observe(
    update_notebook,
    names='value'
)

operation_selector.observe(
    update_notebook,
    names='value'
)

# ============================================================
# INITIAL UPDATE
# ============================================================

update_notebook()

# ============================================================
# FIGURE + RESULTS
#
# No overflow hidden -> no vertical scrollbar.
# ============================================================

figure_results_row = HBox(
    [
        fig.canvas,
        result_panel
    ],
    layout=Layout(
        width='1000px',
        align_items='flex-start',
        gap='14px',
        overflow='visible'
    )
)

figure_results_row.add_class(
    'complex-main'
)

# ============================================================
# INTERPRETATION
# ============================================================

interpretation = HTML("""
<div style="
    width:1000px;
    padding:9px 12px;
    border:1px solid #d7c7e5;
    background:#fbf8fe;
    font-family:Arial, sans-serif;
    font-size:14px;
    line-height:1.45;
    box-sizing:border-box;
    margin-top:5px;
">

<b style="color:#714d95;">Interpretation:</b>
The rectangular form directly gives the coordinates in the complex plane,
whereas the polar form separates magnitude and phase.
Complex conjugation corresponds to reflection about the real axis.
Addition and subtraction act componentwise, while multiplication and division
are naturally interpreted in terms of changes in magnitude and phase.

</div>
""")

# ============================================================
# COMPLETE NOTEBOOK
# ============================================================

main_layout = VBox(
    [
        documentation,
        controls_panel,
        figure_results_row,
        interpretation
    ],
    layout=Layout(
        width='1040px',
        overflow='visible'
    )
)

display(main_layout)
```
