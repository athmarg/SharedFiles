```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt

from ipywidgets import (
    FloatSlider,
    IntSlider,
    RadioButtons,
    HTML,
    VBox,
    HBox,
    Layout
)
from IPython.display import display

plt.ioff()

# ============================================================
# CLASSIC JUPYTER + JUPYTERLAB / NOTEBOOK 7 / BINDER
# ============================================================

display(HTML("""
<style>

/* ---------- Classic Notebook ---------- */

.container {
    width: 98% !important;
    max-width: none !important;
}

.output_area,
.output_subarea {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

.output_scroll {
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
    box-shadow: none !important;
}

/* ---------- JupyterLab / Notebook 7 ---------- */

.jp-Cell-outputWrapper,
.jp-OutputArea,
.jp-OutputArea-child,
.jp-OutputArea-output {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

/* ---------- ipywidgets ---------- */

.widget-output,
.jupyter-widgets-output-area {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

/* ---------- ipympl ---------- */

.jupyter-matplotlib,
.jupyter-matplotlib-figure {
    overflow: visible !important;
    resize: none !important;
}

/* Remove browser resize handle */
.jupyter-matplotlib::-webkit-resizer,
.jupyter-matplotlib-figure::-webkit-resizer {
    display: none !important;
}

/* ---------- Notebook styles ---------- */

.cn2-title {
    font-family: Arial, sans-serif;
    font-size: 21px;
    font-weight: bold;
    color: #6f3fa0;
    margin-bottom: 7px;
}

.cn2-label {
    font-family: Arial, sans-serif;
    font-size: 14px;
    font-weight: bold;
}

.cn2-value {
    font-family: Arial, sans-serif;
    font-size: 14px;
    font-weight: bold;
    color: #0b3d91;
}

.cn2-radio .widget-radio-box {
    display: flex !important;
    flex-direction: row !important;
    flex-wrap: nowrap !important;
    gap: 16px !important;
    align-items: center !important;
}

.cn2-radio .widget-radio-box label {
    margin: 0 !important;
    white-space: nowrap !important;
}

.cn2-radio > label {
    display: none !important;
}

</style>
"""))

# ============================================================
# DOCUMENTATION
# ============================================================

documentation = HTML("""
<div style="
    width:1040px;
    font-family:Arial, sans-serif;
    font-size:15px;
    line-height:1.45;
    margin-bottom:8px;
">

<div class="cn2-title">
Complex Numbers II: Powers, Roots and Further Operations
</div>

<div style="margin-bottom:4px;">
For z = r e<sup>iφ</sup>, powers are obtained from
z<sup>k</sup> = r<sup>k</sup>e<sup>ikφ</sup>, which is De Moivre's rule.
</div>

<div style="margin-bottom:4px;">
The k-th roots have modulus r<sup>1/k</sup> and phases
(φ + 2πn)/k, n = 0,1,...,k−1.
</div>

<div>
<b>This notebook:</b> demonstrates powers, De Moivre's rule,
complex roots, inner and outer products, and the complex logarithm.
</div>

</div>
""")

# ============================================================
# SLIDER SETTINGS
# ============================================================

slider_style = {'description_width': '0px'}

slider_layout = Layout(width='165px')

value_layout = Layout(
    width='52px',
    min_width='52px',
    margin='0px 0px 0px 5px'
)

label_layout = Layout(
    width='80px',
    min_width='80px'
)

row_layout = Layout(
    width='300px',
    align_items='center'
)

# ============================================================
# SLIDERS
# ============================================================

a1_slider = FloatSlider(
    min=-4.0, max=4.0, step=0.25, value=2.0,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

b1_slider = FloatSlider(
    min=-4.0, max=4.0, step=0.25, value=2.0,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

a2_slider = FloatSlider(
    min=-4.0, max=4.0, step=0.25, value=-2.0,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

b2_slider = FloatSlider(
    min=-4.0, max=4.0, step=0.25, value=1.0,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

k_slider = IntSlider(
    min=2, max=8, step=1, value=3,
    readout=False,
    continuous_update=True,
    disabled=True,
    style=slider_style,
    layout=slider_layout
)

# ============================================================
# VALUES
# ============================================================

a1_value = HTML('<div class="cn2-value">2.00</div>', layout=value_layout)
b1_value = HTML('<div class="cn2-value">2.00</div>', layout=value_layout)
a2_value = HTML('<div class="cn2-value">-2.00</div>', layout=value_layout)
b2_value = HTML('<div class="cn2-value">1.00</div>', layout=value_layout)
k_value  = HTML('<div class="cn2-value">3</div>', layout=value_layout)

# ============================================================
# LABELS
# ============================================================

a1_label = HTML('<div class="cn2-label">Re(z₁):</div>', layout=label_layout)
b1_label = HTML('<div class="cn2-label">Im(z₁):</div>', layout=label_layout)
a2_label = HTML('<div class="cn2-label">Re(z₂):</div>', layout=label_layout)
b2_label = HTML('<div class="cn2-label">Im(z₂):</div>', layout=label_layout)
k_label  = HTML('<div class="cn2-label">Order k:</div>', layout=label_layout)

# ============================================================
# ROWS
# ============================================================

a1_row = HBox([a1_label, a1_slider, a1_value], layout=row_layout)
b1_row = HBox([b1_label, b1_slider, b1_value], layout=row_layout)

a2_row = HBox([a2_label, a2_slider, a2_value], layout=row_layout)
b2_row = HBox([b2_label, b2_slider, b2_value], layout=row_layout)

k_row = HBox([k_label, k_slider, k_value], layout=row_layout)

# ============================================================
# PARAMETERS GRID
# ============================================================

parameters_grid = HBox(
    [
        VBox(
            [
                HTML("""
                <div style="
                    font-family:Arial;
                    font-size:15px;
                    font-weight:bold;
                    color:#1555cc;
                    margin-bottom:3px;">
                First complex number z₁
                </div>
                """),
                a1_row,
                b1_row
            ]
        ),

        VBox(
            [
                HTML("""
                <div style="
                    font-family:Arial;
                    font-size:15px;
                    font-weight:bold;
                    color:#d7191c;
                    margin-bottom:3px;">
                Second complex number z₂
                </div>
                """),
                a2_row,
                b2_row
            ]
        ),

        VBox(
            [
                HTML("""
                <div style="
                    font-family:Arial;
                    font-size:15px;
                    font-weight:bold;
                    color:#16802b;
                    margin-bottom:3px;">
                Power / Root
                </div>
                """),
                k_row
            ]
        )
    ],
    layout=Layout(
        width='1010px',
        gap='12px',
        align_items='flex-start'
    )
)

# ============================================================
# MODE SELECTOR
# ============================================================

mode_selector = RadioButtons(
    options=[
        'Power / De Moivre',
        'k-th Roots',
        'Inner / Outer Product',
        'Complex Logarithm'
    ],
    value='Power / De Moivre',
    description='',
    layout=Layout(width='760px')
)

mode_selector.add_class('cn2-radio')

# ============================================================
# CONTROLS PANEL
# ============================================================

controls_panel = VBox(
    [
        HTML("""
        <div style="
            font-family:Arial;
            font-size:17px;
            font-weight:bold;
            color:#6f3fa0;
            margin-bottom:7px;">
        Parameters
        </div>
        """),

        parameters_grid,

        HTML("""
        <div style="
            font-family:Arial;
            font-size:14px;
            font-weight:bold;
            margin-top:8px;
            margin-bottom:4px;">
        Operation
        </div>
        """),

        mode_selector
    ],
    layout=Layout(
        width='1040px',
        padding='9px 12px',
        border='1px solid #d2c2df'
    )
)

# ============================================================
# COMPLEX PLANE
# ============================================================

fig, ax = plt.subplots(figsize=(5.6, 5.35))

fig.canvas.header_visible = False

fig.canvas.layout = Layout(
    width='560px',
    height='535px'
)

ax.set_aspect('equal', adjustable='box')

ax.axhline(0, color='black', linewidth=1.0)
ax.axvline(0, color='black', linewidth=1.0)

ax.grid(True, linestyle=':', alpha=0.40)

ax.set_xlabel('Real axis', fontsize=11)
ax.set_ylabel('Imaginary axis', fontsize=11)

ax.set_title(
    'Complex Plane',
    fontsize=15,
    fontweight='bold',
    color='#6f3fa0'
)

# ============================================================
# FIXED AXES
# ============================================================

AXIS_LIMIT = 5.0

ax.set_xlim(-AXIS_LIMIT, AXIS_LIMIT)
ax.set_ylim(-AXIS_LIMIT, AXIS_LIMIT)

fig.subplots_adjust(
    left=0.12,
    right=0.97,
    top=0.91,
    bottom=0.11
)

# ============================================================
# GRAPHICAL OBJECTS
# ============================================================

z1_line, = ax.plot(
    [], [],
    linewidth=2.0,
    color='blue',
    marker='o',
    markevery=[1],
    label='z₁'
)

z2_line, = ax.plot(
    [], [],
    linewidth=2.0,
    color='red',
    marker='o',
    markevery=[1],
    label='z₂'
)

result_line, = ax.plot(
    [], [],
    linewidth=2.1,
    color='green',
    marker='o',
    markevery=[1],
    label='result'
)

roots_points, = ax.plot(
    [], [],
    linestyle='None',
    marker='o',
    markersize=7,
    color='green'
)

root_circle, = ax.plot(
    [], [],
    linestyle='--',
    linewidth=1.2,
    color='gray'
)

# ============================================================
# TEXT LABELS
# ============================================================

z1_text = ax.text(0, 0, '', fontsize=10, color='blue')
z2_text = ax.text(0, 0, '', fontsize=10, color='red')
result_text = ax.text(0, 0, '', fontsize=10, color='green')

# ============================================================
# RESULTS PANEL
# ============================================================

results_html = HTML()

results_panel = VBox(
    [
        HTML("""
        <div style="
            font-family:Arial;
            font-size:17px;
            font-weight:bold;
            color:#6f3fa0;
            margin-bottom:6px;">
        Numerical Results
        </div>
        """),

        results_html
    ],
    layout=Layout(
        width='410px',
        padding='10px 12px',
        border='1px solid #d2c2df',
        overflow='visible'
    )
)

# ============================================================
# FUNCTIONS
# ============================================================

def rectangular_form(z):

    a = np.real(z)
    b = np.imag(z)

    if abs(a) < 1e-12:
        a = 0.0

    if abs(b) < 1e-12:
        b = 0.0

    sign = '+' if b >= 0 else '−'

    return f'{a:.4f} {sign} {abs(b):.4f}i'


def phase_data(z):

    if abs(z) < 1e-12:
        return 0.0, 0.0

    phi = np.angle(z)

    phi_deg = np.degrees(phi)

    if phi_deg < 0:
        phi_deg += 360.0

    return np.deg2rad(phi_deg), phi_deg


def reset_optional_graphics():

    result_line.set_data([], [])
    roots_points.set_data([], [])
    root_circle.set_data([], [])
    result_text.set_text('')

# ============================================================
# UPDATE
# ============================================================

def update_notebook(change=None):

    a1 = a1_slider.value
    b1 = b1_slider.value

    a2 = a2_slider.value
    b2 = b2_slider.value

    k = k_slider.value

    z1 = complex(a1, b1)
    z2 = complex(a2, b2)

    mode = mode_selector.value

    # --------------------------------------------------------
    # k ENABLED ONLY FOR ROOTS
    # --------------------------------------------------------

    k_slider.disabled = (mode != 'k-th Roots')

    # --------------------------------------------------------
    # VALUES
    # --------------------------------------------------------

    a1_value.value = f'<div class="cn2-value">{a1:.2f}</div>'
    b1_value.value = f'<div class="cn2-value">{b1:.2f}</div>'

    a2_value.value = f'<div class="cn2-value">{a2:.2f}</div>'
    b2_value.value = f'<div class="cn2-value">{b2:.2f}</div>'

    k_value.value = f'<div class="cn2-value">{k}</div>'

    # --------------------------------------------------------
    # BASE VECTORS
    # --------------------------------------------------------

    z1_line.set_data(
        [0, a1],
        [0, b1]
    )

    z2_line.set_data(
        [0, a2],
        [0, b2]
    )

    z1_text.set_position(
        (a1 + 0.12, b1 + 0.12)
    )

    z1_text.set_text('z₁')

    z2_text.set_position(
        (a2 + 0.12, b2 + 0.12)
    )

    z2_text.set_text('z₂')

    reset_optional_graphics()

    # ========================================================
    # POWER / DE MOIVRE
    # ========================================================

    if mode == 'Power / De Moivre':

        z_power = z1 ** k

        # Draw result only if it lies inside the fixed axes.
        if (
            abs(np.real(z_power)) <= AXIS_LIMIT
            and
            abs(np.imag(z_power)) <= AXIS_LIMIT
        ):

            result_line.set_data(
                [0, np.real(z_power)],
                [0, np.imag(z_power)]
            )

            result_text.set_position(
                (
                    np.real(z_power) + 0.12,
                    np.imag(z_power) + 0.12
                )
            )

            result_text.set_text(
                f'z₁^{k}'
            )

        r = abs(z1)

        phi, phi_deg = phase_data(z1)

        rp = abs(z_power)

        phip, phip_deg = phase_data(z_power)

        results_html.value = f"""
        <div style="
            font-family:Arial;
            font-size:14px;
            line-height:1.58;">

        <b>Original number</b><br>

        z₁ = <b>{rectangular_form(z1)}</b><br>

        |z₁| = <b>{r:.4f}</b><br>

        arg(z₁) = <b>{phi_deg:.2f}°</b>

        <hr style="margin:8px 0px;">

        <b>Power</b><br>

        z₁<sup>{k}</sup> =
        <b>{rectangular_form(z_power)}</b>

        <br><br>

        |z₁<sup>{k}</sup>|
        =
        |z₁|<sup>{k}</sup>
        =
        <b>{rp:.4f}</b>

        <br><br>

        arg(z₁<sup>{k}</sup>)
        =
        {k} arg(z₁)

        <br>

        =
        {k} × {phi_deg:.2f}°
        =
        <b>{phip_deg:.2f}° mod 360°</b>

        <br><br>

        <b>De Moivre:</b><br>

        [cos φ + i sin φ]<sup>{k}</sup><br>

        = cos({k}φ) + i sin({k}φ)

        </div>
        """

    # ========================================================
    # K-TH ROOTS
    # ========================================================

    elif mode == 'k-th Roots':

        z2_line.set_data([], [])
        z2_text.set_text('')

        r = abs(z1)
        phi = np.angle(z1)

        if r < 1e-12:

            roots = np.zeros(
                k,
                dtype=complex
            )

        else:

            n = np.arange(k)

            roots = (
                r ** (1.0 / k)
                *
                np.exp(
                    1j
                    *
                    (
                        phi
                        +
                        2.0 * np.pi * n
                    )
                    /
                    k
                )
            )

        roots_points.set_data(
            np.real(roots),
            np.imag(roots)
        )

        root_radius = (
            r ** (1.0 / k)
            if r > 0
            else 0.0
        )

        theta = np.linspace(
            0,
            2.0 * np.pi,
            400
        )

        root_circle.set_data(
            root_radius * np.cos(theta),
            root_radius * np.sin(theta)
        )

        roots_list = ""

        for idx, root in enumerate(roots):

            _, root_deg = phase_data(root)

            roots_list += f"""
            <div style="margin-bottom:3px;">
            z<sub>{idx}</sub>
            =
            <b>{rectangular_form(root)}</b>
            &nbsp;&nbsp;
            φ = {root_deg:.2f}°
            </div>
            """

        angular_separation = 360.0 / k

        results_html.value = f"""
        <div style="
            font-family:Arial;
            font-size:14px;
            line-height:1.55;">

        <b>Original number</b><br>

        z₁ = <b>{rectangular_form(z1)}</b><br>

        |z₁| = <b>{r:.4f}</b>

        <hr style="margin:8px 0px;">

        <b>{k}-th roots</b>

        <br><br>

        Root modulus:<br>

        |zₙ|
        =
        |z₁|<sup>1/{k}</sup>
        =
        <b>{root_radius:.4f}</b>

        <br><br>

        Successive roots are separated by<br>

        2π/{k}
        =
        <b>{angular_separation:.2f}°</b>

        <hr style="margin:8px 0px;">

        {roots_list}

        </div>
        """

    # ========================================================
    # INNER / OUTER PRODUCT
    # ========================================================

    elif mode == 'Inner / Outer Product':

        dot_product = (
            a1 * a2
            +
            b1 * b2
        )

        cross_product = (
            a1 * b2
            -
            a2 * b1
        )

        magnitude_product = (
            abs(z1)
            *
            abs(z2)
        )

        if (
            abs(z1) > 1e-12
            and
            abs(z2) > 1e-12
        ):

            cos_theta = np.clip(
                dot_product / magnitude_product,
                -1.0,
                1.0
            )

            theta_deg = np.degrees(
                np.arccos(cos_theta)
            )

        else:

            theta_deg = 0.0

        z1_conj_z2 = (
            np.conjugate(z1)
            *
            z2
        )

        results_html.value = f"""
        <div style="
            font-family:Arial;
            font-size:14px;
            line-height:1.58;">

        z₁ = <b>{rectangular_form(z1)}</b><br>

        z₂ = <b>{rectangular_form(z2)}</b>

        <hr style="margin:8px 0px;">

        <b>Inner product</b><br>

        z₁ ◦ z₂ = a₁a₂ + b₁b₂<br>

        = <b>{dot_product:.4f}</b>

        <br><br>

        = Re(z₁* z₂)

        <br><br>

        <b>Outer product</b><br>

        z₁ × z₂ = a₁b₂ − a₂b₁<br>

        = <b>{cross_product:.4f}</b>

        <br><br>

        = Im(z₁* z₂)

        <hr style="margin:8px 0px;">

        z₁* z₂ =
        <b>{rectangular_form(z1_conj_z2)}</b>

        <br><br>

        Angle between vectors:<br>

        θ = <b>{theta_deg:.2f}°</b>

        </div>
        """

    # ========================================================
    # COMPLEX LOGARITHM
    # ========================================================

    elif mode == 'Complex Logarithm':

        z2_line.set_data([], [])
        z2_text.set_text('')

        if abs(z1) < 1e-12:

            results_html.value = """
            <div style="
                font-family:Arial;
                font-size:14px;
                color:#b00020;
                line-height:1.55;">

            <b>The complex logarithm is undefined at z = 0.</b>

            </div>
            """

        else:

            r = abs(z1)

            principal_phi = np.angle(z1)

            principal_log = (
                np.log(r)
                +
                1j * principal_phi
            )

            principal_deg = np.degrees(
                principal_phi
            )

            results_html.value = f"""
            <div style="
                font-family:Arial;
                font-size:14px;
                line-height:1.58;">

            z₁ =
            <b>{rectangular_form(z1)}</b>

            <br><br>

            |z₁| =
            <b>{r:.4f}</b>

            <br>

            Principal argument =
            <b>{principal_phi:.4f} rad
            ({principal_deg:.2f}°)</b>

            <hr style="margin:8px 0px;">

            <b>Principal logarithm</b>

            <br><br>

            Log(z₁)
            =
            ln|z₁| + i Arg(z₁)

            <br><br>

            =
            ln({r:.4f})
            + i({principal_phi:.4f})

            <br>

            =
            <b>{rectangular_form(principal_log)}</b>

            <hr style="margin:8px 0px;">

            <b>Multivalued logarithm</b>

            <br><br>

            ln(z₁)
            =
            ln|z₁|
            +
            i(φ + 2πn)

            <br>

            n = 0, ±1, ±2, ...

            </div>
            """

    # ========================================================
    # FIXED AXES — NEVER CHANGE
    # ========================================================

    ax.set_xlim(
        -AXIS_LIMIT,
        AXIS_LIMIT
    )

    ax.set_ylim(
        -AXIS_LIMIT,
        AXIS_LIMIT
    )

    fig.canvas.draw_idle()

# ============================================================
# CONNECT CONTROLS
# ============================================================

a1_slider.observe(update_notebook, names='value')
b1_slider.observe(update_notebook, names='value')

a2_slider.observe(update_notebook, names='value')
b2_slider.observe(update_notebook, names='value')

k_slider.observe(update_notebook, names='value')

mode_selector.observe(update_notebook, names='value')

# ============================================================
# INITIAL UPDATE
# ============================================================

update_notebook()

# ============================================================
# FIGURE + RESULTS
#
# IMPORTANT:
# Compact fixed layout exactly as before.
# No width='100%' here.
# ============================================================

figure_results_row = HBox(
    [
        fig.canvas,
        results_panel
    ],
    layout=Layout(
        width='984px',
        align_items='flex-start',
        gap='14px',
        overflow='visible'
    )
)

# ============================================================
# INTERPRETATION
# ============================================================

interpretation = HTML("""
<div style="
    width:984px;
    padding:9px 12px;
    border:1px solid #d7c7e5;
    background:#fbf8fe;
    font-family:Arial, sans-serif;
    font-size:14px;
    line-height:1.45;
    box-sizing:border-box;
    margin-top:5px;
">

<b style="color:#6f3fa0;">Interpretation:</b>
Powers multiply the phase and raise the modulus to the corresponding power.
The k-th roots lie uniformly on a circle and are separated by 2π/k.
The inner and outer products are respectively the real and imaginary parts
of z₁* z₂. The complex logarithm separates the logarithm of the modulus
from the argument and is intrinsically multivalued.

</div>
""")

# ============================================================
# COMPLETE NOTEBOOK
# ============================================================

main_layout = VBox(
    [
        documentation,
        controls_panel,
        figure_results_row,
        interpretation
    ],
    layout=Layout(
        width='1040px',
        overflow='visible'
    )
)

display(main_layout)
```
