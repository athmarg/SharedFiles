```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import norm

from ipywidgets import (
    Dropdown,
    FloatSlider,
    interactive,
    VBox,
    HBox,
    Output,
    Layout
)

from IPython.display import display, Markdown


# ============================================================
# FUNCTION OF A RANDOM VARIABLE
# Y = g(X)
# ============================================================

display(
    Markdown(
        "## Function of a Random Variable — Y = g(X)"
    )
)

# ============================================================
# CONTROLS
# ============================================================

transformation_widget = Dropdown(
    options=[
        'Y = aX + b',
        'Y = X²',
        'Y = |X|',
        'Y = exp(aX)'
    ],
    value='Y = aX + b',
    description='Function:',
    layout=Layout(width='330px')
)

a_slider = FloatSlider(
    value=1,
    min=-2,
    max=2,
    step=0.1,
    description='a:',
    continuous_update=True,
    layout=Layout(width='330px')
)

b_slider = FloatSlider(
    value=0,
    min=-3,
    max=3,
    step=0.1,
    description='b:',
    continuous_update=True,
    layout=Layout(width='330px')
)

# ------------------------------------------------------------
# Left upper column
# ------------------------------------------------------------

controls_column = VBox(
    [
        transformation_widget,
        a_slider,
        b_slider
    ],
    layout=Layout(
        width='350px'
    )
)

# ============================================================
# TEXT OUTPUT
# ============================================================

text_output = Output(
    layout=Layout(
        width='430px',
        padding='5px 0px 0px 20px'
    )
)

# ============================================================
# MAIN FUNCTION
# ============================================================

def random_variable_transformation(
    transformation,
    a,
    b
):

    rng = np.random.default_rng(
        12345
    )

    N = 200000

    # --------------------------------------------------------
    # Input random variable
    # --------------------------------------------------------

    X = rng.normal(
        0,
        1,
        N
    )

    # --------------------------------------------------------
    # Transformation
    # --------------------------------------------------------

    if transformation == 'Y = aX + b':

        Y = (
            a * X
            +
            b
        )

        transformation_text = (
            f'Y = {a:.2f} X + {b:.2f}'
        )

    elif transformation == 'Y = X²':

        Y = X**2

        transformation_text = (
            'Y = X²'
        )

    elif transformation == 'Y = |X|':

        Y = np.abs(X)

        transformation_text = (
            'Y = |X|'
        )

    elif transformation == 'Y = exp(aX)':

        Y = np.exp(
            a * X
        )

        transformation_text = (
            f'Y = exp({a:.2f} X)'
        )

    # ========================================================
    # UPDATE TEXT OUTPUT
    # ========================================================

    with text_output:

        text_output.clear_output(
            wait=True
        )

        print(
            f'Transformation: {transformation_text}'
        )

        print()

        print(
            f'E[X]       ≈ {np.mean(X):.6f}'
        )

        print(
            f'Var(X)     ≈ {np.var(X):.6f}'
        )

        print()

        print(
            f'E[Y]       ≈ {np.mean(Y):.6f}'
        )

        print(
            f'Var(Y)     ≈ {np.var(Y):.6f}'
        )

    # ========================================================
    # INPUT PDF
    # ========================================================

    x = np.linspace(
        -4,
        4,
        1000
    )

    pdf_x = norm.pdf(
        x
    )

    # ========================================================
    # OUTPUT PDF ESTIMATED FROM SAMPLES
    # ========================================================

    low = np.percentile(
        Y,
        0.2
    )

    high = np.percentile(
        Y,
        99.8
    )

    bins = np.linspace(
        low,
        high,
        120
    )

    # ========================================================
    # FIGURE
    # ========================================================

    fig, axes = plt.subplots(
        1,
        3,
        figsize=(15, 4.5)
    )

    # --------------------------------------------------------
    # 1. INPUT PDF
    # --------------------------------------------------------

    axes[0].plot(
        x,
        pdf_x,
        linewidth=2
    )

    axes[0].set_title(
        'Input Random Variable X'
    )

    axes[0].set_xlabel(
        'x'
    )

    axes[0].set_ylabel(
        'f_X(x)'
    )

    axes[0].grid(
        True,
        alpha=0.3
    )

    # --------------------------------------------------------
    # 2. TRANSFORMATION
    # --------------------------------------------------------

    x_transform = np.linspace(
        -3,
        3,
        1000
    )

    if transformation == 'Y = aX + b':

        y_transform = (
            a * x_transform
            +
            b
        )

    elif transformation == 'Y = X²':

        y_transform = (
            x_transform**2
        )

    elif transformation == 'Y = |X|':

        y_transform = np.abs(
            x_transform
        )

    elif transformation == 'Y = exp(aX)':

        y_transform = np.exp(
            a * x_transform
        )

    axes[1].plot(
        x_transform,
        y_transform,
        linewidth=2
    )

    axes[1].set_title(
        f'Transformation\n{transformation_text}'
    )

    axes[1].set_xlabel(
        'x'
    )

    axes[1].set_ylabel(
        'y = g(x)'
    )

    axes[1].grid(
        True,
        alpha=0.3
    )

    # --------------------------------------------------------
    # 3. OUTPUT PDF
    # --------------------------------------------------------

    axes[2].hist(
        Y,
        bins=bins,
        density=True,
        alpha=0.6
    )

    axes[2].set_title(
        'Output Random Variable Y'
    )

    axes[2].set_xlabel(
        'y'
    )

    axes[2].set_ylabel(
        'f_Y(y)'
    )

    axes[2].grid(
        True,
        alpha=0.3
    )

    plt.tight_layout()
    plt.show()
    plt.close(fig)


# ============================================================
# INTERACTIVE OBJECT
# ============================================================

interactive_widget = interactive(
    random_variable_transformation,

    transformation=transformation_widget,
    a=a_slider,
    b=b_slider
)

plot_output = interactive_widget.children[-1]

# ============================================================
# TOP ROW
#
# Controls left — text output right
# ============================================================

top_row = HBox(
    [
        controls_column,
        text_output
    ],
    layout=Layout(
        width='850px',
        gap='35px',
        align_items='flex-start'
    )
)

# ============================================================
# DISPLAY
# ============================================================

display(
    top_row,
    plot_output
)
```
