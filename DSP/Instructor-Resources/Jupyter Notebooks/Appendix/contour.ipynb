```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt

from ipywidgets import (
    RadioButtons,
    HTML,
    VBox,
    HBox,
    Layout
)
from IPython.display import display

# ============================================================
# COMPLEX CONTOUR INTEGRALS AND PATH DEPENDENCE
# ============================================================

plt.ioff()

# ============================================================
# CLASSIC JUPYTER + JUPYTERLAB / NOTEBOOK 7 / BINDER
#
# - no forced scrollbars
# - no trimming of matplotlib canvases
# - output grows vertically when necessary
# - remove ipympl resize handle
# ============================================================

display(HTML("""
<style>

/* ============================================================
   CLASSIC JUPYTER NOTEBOOK
   ============================================================ */

.container {
    width: 98% !important;
    max-width: none !important;
}

.output_area,
.output_subarea {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

.output_scroll {
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
    box-shadow: none !important;
}

/* ============================================================
   JUPYTERLAB / NOTEBOOK 7
   ============================================================ */

.jp-Cell-outputWrapper,
.jp-OutputArea,
.jp-OutputArea-child,
.jp-OutputArea-output {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

/* ============================================================
   IPYWIDGETS
   ============================================================ */

.widget-output,
.jupyter-widgets-output-area,
.widget-box {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

/* ============================================================
   IPYMPL
   ============================================================ */

.jupyter-matplotlib,
.jupyter-matplotlib-figure {
    overflow: visible !important;
    resize: none !important;
}

.jupyter-matplotlib::-webkit-resizer,
.jupyter-matplotlib-figure::-webkit-resizer {
    display: none !important;
}

/* ============================================================
   NOTEBOOK STYLES
   ============================================================ */

.ci-title {
    font-family: Arial, sans-serif;
    font-size: 21px;
    font-weight: bold;
    color: #6f3fa0;
    margin-bottom: 7px;
}

.ci-radio .widget-radio-box {
    display: flex !important;
    flex-direction: row !important;
    flex-wrap: nowrap !important;
    gap: 18px !important;
    align-items: center !important;
}

.ci-radio .widget-radio-box label {
    margin: 0 !important;
    white-space: nowrap !important;
}

.ci-radio > label {
    display: none !important;
}

</style>
"""))

# ============================================================
# DOCUMENTATION
# ============================================================

documentation = HTML("""
<div style="
    width:980px;
    font-family:Arial, sans-serif;
    font-size:15px;
    line-height:1.45;
    margin-bottom:8px;
">

<div class="ci-title">
Complex Contour Integrals and Path Dependence
</div>

<div style="margin-bottom:4px;">
A complex contour integral is evaluated along a path C in the complex plane,
using z = z(t) and dz = z′(t)dt.
</div>

<div style="margin-bottom:4px;">
For a non-analytic function such as f(z) = z*, the integral between two fixed
points may depend on the selected path.
</div>

<div>
For an analytic function with an antiderivative, such as f(z) = z²,
the integral between the same two points is independent of the path.
</div>

</div>
""")

# ============================================================
# FUNCTION SELECTOR
# ============================================================

function_selector = RadioButtons(
    options=[
        ('f(z) = z*', 'conjugate'),
        ('f(z) = z²', 'square')
    ],
    value='conjugate',
    description='',
    layout=Layout(
        width='330px'
    )
)

function_selector.add_class(
    'ci-radio'
)

# ============================================================
# PATH SELECTOR
# ============================================================

path_selector = RadioButtons(
    options=[
        ('Path C₁', 'C1'),
        ('Path C₂', 'C2'),
        ('Path C₃', 'C3')
    ],
    value='C1',
    description='',
    layout=Layout(
        width='330px'
    )
)

path_selector.add_class(
    'ci-radio'
)

# ============================================================
# CONTROLS PANEL
# ============================================================

controls_panel = VBox(
    [
        HTML("""
        <div style="
            font-family:Arial;
            font-size:17px;
            font-weight:bold;
            color:#6f3fa0;
            margin-bottom:6px;
        ">
        Contour-Integral Parameters
        </div>
        """),

        HTML("""
        <div style="
            font-family:Arial;
            font-size:14px;
            font-weight:bold;
            margin-bottom:3px;
        ">
        Function
        </div>
        """),

        function_selector,

        HTML("""
        <div style="
            font-family:Arial;
            font-size:14px;
            font-weight:bold;
            margin-top:7px;
            margin-bottom:3px;
        ">
        Integration Path
        </div>
        """),

        path_selector
    ],
    layout=Layout(
        width='690px',
        padding='9px 12px',
        border='1px solid #d2c2df',
        overflow='visible'
    )
)

# ============================================================
# GEOMETRY OF THE THREE PATHS
# ============================================================

O = 0.0 + 0.0j
A1 = 0.0 + 2.0j
A2 = 4.0 + 0.0j
B = 4.0 + 2.0j

N_SEGMENT = 700

# ============================================================
# PATH GENERATORS
# ============================================================

def line_segment(z_start, z_end, N=N_SEGMENT):

    t = np.linspace(
        0.0,
        1.0,
        N
    )

    z = (
        z_start
        +
        t * (z_end - z_start)
    )

    return z


def create_path(path_name):

    if path_name == 'C1':

        first = line_segment(
            O,
            A1
        )

        second = line_segment(
            A1,
            B
        )

        return np.concatenate(
            [
                first,
                second[1:]
            ]
        )

    elif path_name == 'C2':

        first = line_segment(
            O,
            A2
        )

        second = line_segment(
            A2,
            B
        )

        return np.concatenate(
            [
                first,
                second[1:]
            ]
        )

    else:

        return line_segment(
            O,
            B,
            2 * N_SEGMENT
        )

# ============================================================
# FUNCTION DEFINITIONS
# ============================================================

def evaluate_function(mode, z):

    if mode == 'conjugate':

        return np.conjugate(
            z
        )

    elif mode == 'square':

        return z**2


def function_name(mode):

    if mode == 'conjugate':

        return 'f(z) = z*'

    else:

        return 'f(z) = z²'

# ============================================================
# NUMERICAL CONTOUR INTEGRAL
# ============================================================

def contour_integral(mode, z_path):

    f_values = evaluate_function(
        mode,
        z_path
    )

    dz = np.diff(
        z_path
    )

    f_mid = 0.5 * (
        f_values[:-1]
        +
        f_values[1:]
    )

    integral = np.sum(
        f_mid * dz
    )

    return integral

# ============================================================
# COMPUTE ALL THREE PATH INTEGRALS
# ============================================================

def all_path_integrals(mode):

    values = {}

    for name in [
        'C1',
        'C2',
        'C3'
    ]:

        path = create_path(
            name
        )

        values[name] = contour_integral(
            mode,
            path
        )

    return values

# ============================================================
# COMPLEX NUMBER FORMAT
# ============================================================

def complex_string(z):

    real = np.real(z)
    imag = np.imag(z)

    if abs(real) < 1e-10:
        real = 0.0

    if abs(imag) < 1e-10:
        imag = 0.0

    sign = '+' if imag >= 0 else '−'

    return (
        f'{real:.4f} '
        f'{sign} '
        f'{abs(imag):.4f}i'
    )

# ============================================================
# COMPLEX PLANE FIGURE
# ============================================================

fig, ax = plt.subplots(
    figsize=(6.0, 4.9)
)

fig.canvas.header_visible = False

fig.canvas.layout = Layout(
    width='600px',
    height='490px'
)

ax.set_title(
    'Integration Paths in the Complex Plane',
    fontsize=14,
    fontweight='bold',
    color='#6f3fa0'
)

ax.set_xlabel(
    'Re(z)',
    fontsize=11
)

ax.set_ylabel(
    'Im(z)',
    fontsize=11
)

ax.set_aspect(
    'equal',
    adjustable='box'
)

# ============================================================
# FIXED AXES
# ============================================================

ax.set_xlim(
    -0.7,
    4.7
)

ax.set_ylim(
    -0.7,
    2.7
)

ax.axhline(
    0,
    color='black',
    linewidth=1.0
)

ax.axvline(
    0,
    color='black',
    linewidth=1.0
)

ax.grid(
    True,
    linestyle=':',
    alpha=0.40
)

# ============================================================
# CREATE ALL THREE PATH CURVES ONCE
# ============================================================

path_C1 = create_path(
    'C1'
)

path_C2 = create_path(
    'C2'
)

path_C3 = create_path(
    'C3'
)

line_C1, = ax.plot(
    np.real(path_C1),
    np.imag(path_C1),
    linewidth=1.5,
    linestyle='--',
    alpha=0.35,
    label='C₁'
)

line_C2, = ax.plot(
    np.real(path_C2),
    np.imag(path_C2),
    linewidth=1.5,
    linestyle='--',
    alpha=0.35,
    label='C₂'
)

line_C3, = ax.plot(
    np.real(path_C3),
    np.imag(path_C3),
    linewidth=1.5,
    linestyle='--',
    alpha=0.35,
    label='C₃'
)

# ============================================================
# SELECTED PATH
# ============================================================

selected_path_line, = ax.plot(
    [],
    [],
    linewidth=3.0
)

# ============================================================
# IMPORTANT POINTS
# ============================================================

ax.plot(
    [0],
    [0],
    marker='o',
    markersize=7
)

ax.text(
    0.08,
    0.10,
    'O',
    fontsize=11,
    fontweight='bold'
)

ax.plot(
    [0],
    [2],
    marker='o',
    markersize=6
)

ax.text(
    0.08,
    2.08,
    'A₁',
    fontsize=10
)

ax.plot(
    [4],
    [0],
    marker='o',
    markersize=6
)

ax.text(
    4.05,
    0.08,
    'A₂',
    fontsize=10
)

ax.plot(
    [4],
    [2],
    marker='o',
    markersize=7
)

ax.text(
    4.05,
    2.08,
    'B',
    fontsize=11,
    fontweight='bold'
)

ax.legend(
    loc='upper left',
    fontsize=9
)

fig.subplots_adjust(
    left=0.11,
    right=0.97,
    top=0.90,
    bottom=0.13
)

# ============================================================
# RESULTS PANEL
# ============================================================

results_html = HTML()

results_panel = VBox(
    [
        HTML("""
        <div style="
            font-family:Arial;
            font-size:17px;
            font-weight:bold;
            color:#6f3fa0;
            margin-bottom:6px;
        ">
        Contour-Integral Results
        </div>
        """),

        results_html
    ],
    layout=Layout(
        width='350px',
        padding='10px 12px',
        border='1px solid #d2c2df',
        overflow='visible'
    )
)

# ============================================================
# INTERPRETATION PANEL
# ============================================================

interpretation_html = HTML()

interpretation_panel = VBox(
    [
        interpretation_html
    ],
    layout=Layout(
        width='590px',
        padding='8px 10px',
        border='1px solid #d7c7e5',
        margin='40px 0px 0px 0px',
        overflow='visible'
    )
)

# ============================================================
# UPDATE FUNCTION
# ============================================================

def update_notebook(change=None):

    mode = function_selector.value
    selected_path = path_selector.value

    # --------------------------------------------------------
    # All contour integrals
    # --------------------------------------------------------

    integrals = all_path_integrals(
        mode
    )

    I1 = integrals['C1']
    I2 = integrals['C2']
    I3 = integrals['C3']

    # --------------------------------------------------------
    # Selected path
    # --------------------------------------------------------

    z_path = create_path(
        selected_path
    )

    selected_path_line.set_data(
        np.real(z_path),
        np.imag(z_path)
    )

    # --------------------------------------------------------
    # Highlight selected path
    # --------------------------------------------------------

    line_C1.set_alpha(
        0.25 if selected_path != 'C1' else 0.55
    )

    line_C2.set_alpha(
        0.25 if selected_path != 'C2' else 0.55
    )

    line_C3.set_alpha(
        0.25 if selected_path != 'C3' else 0.55
    )

    # --------------------------------------------------------
    # Selected result
    # --------------------------------------------------------

    selected_integral = integrals[
        selected_path
    ]

    # --------------------------------------------------------
    # Path-independence comparison
    # --------------------------------------------------------

    max_difference = max(
        abs(I1 - I2),
        abs(I1 - I3),
        abs(I2 - I3)
    )

    # ========================================================
    # ANALYTIC FUNCTION
    # ========================================================

    if mode == 'square':

        antiderivative_value = (
            B**3 / 3.0
            -
            O**3 / 3.0
        )

        conclusion_text = (
            "The integral is path-independent."
        )

        conclusion_color = (
            "#16802b"
        )

        explanation = """
        Since f(z) = z² is analytic everywhere and has the
        antiderivative F(z) = z³/3, all three paths give the
        same result:
        ∫<sub>C</sub> z² dz = F(B) − F(O).
        """

        extra_result = f"""
        <hr style="margin:8px 0px;">

        <b>Antiderivative check</b>

        <br><br>

        F(z) = z³/3

        <br>

        F(B) − F(O)
        =
        <b>{complex_string(antiderivative_value)}</b>
        """

    # ========================================================
    # NON-ANALYTIC FUNCTION
    # ========================================================

    else:

        conclusion_text = (
            "The integral depends on the path."
        )

        conclusion_color = (
            "#b00020"
        )

        explanation = """
        The function f(z) = z* is not analytic.
        Therefore it does not possess a complex antiderivative
        in the region, and different paths between O and B
        produce different contour-integral values.
        """

        extra_result = ""

    # --------------------------------------------------------
    # RESULTS PANEL
    # --------------------------------------------------------

    results_html.value = f"""
    <div style="
        font-family:Arial;
        font-size:14px;
        line-height:1.58;
    ">

    <b>{function_name(mode)}</b>

    <br><br>

    Start:
    <b>O = 0</b>

    <br>

    End:
    <b>B = 4 + 2i</b>

    <hr style="margin:8px 0px;">

    <b>Selected path:</b>
    {selected_path}

    <br><br>

    ∫<sub>{selected_path}</sub> f(z) dz
    =
    <b>{complex_string(selected_integral)}</b>

    <hr style="margin:8px 0px;">

    <b>Comparison of all paths</b>

    <br><br>

    C₁:
    <b>{complex_string(I1)}</b>

    <br>

    C₂:
    <b>{complex_string(I2)}</b>

    <br>

    C₃:
    <b>{complex_string(I3)}</b>

    <br><br>

    Maximum difference:
    <b>{max_difference:.4e}</b>

    {extra_result}

    </div>
    """

    # --------------------------------------------------------
    # INTERPRETATION
    # --------------------------------------------------------

    interpretation_html.value = f"""
    <div style="
        font-family:Arial, sans-serif;
        font-size:13px;
        line-height:1.50;
    ">

        <div style="
            color:#6f3fa0;
            font-weight:bold;
            margin-bottom:5px;
        ">
            Interpretation:
        </div>

        <div style="
            color:{conclusion_color};
            font-weight:bold;
            margin-bottom:7px;
        ">
            {conclusion_text}
        </div>

        <div style="
            color:#222222;
            line-height:1.55;
        ">
            {explanation}
        </div>

    </div>
    """

    # --------------------------------------------------------
    # REDRAW ONLY
    # --------------------------------------------------------

    fig.canvas.draw_idle()

# ============================================================
# CONNECT CONTROLS
# ============================================================

function_selector.observe(
    update_notebook,
    names='value'
)

path_selector.observe(
    update_notebook,
    names='value'
)

# ============================================================
# INITIAL UPDATE
# ============================================================

update_notebook()

# ============================================================
# LEFT COLUMN
# ============================================================

left_column = VBox(
    [
        fig.canvas,
        interpretation_panel
    ],
    layout=Layout(
        width='610px',
        gap='2px',
        overflow='visible'
    )
)

# ============================================================
# MAIN VISUAL ROW
# ============================================================

visual_row = HBox(
    [
        left_column,
        results_panel
    ],
    layout=Layout(
        width='975px',
        gap='12px',
        align_items='flex-start',
        overflow='visible'
    )
)

# ============================================================
# COMPLETE NOTEBOOK
# ============================================================

main_layout = VBox(
    [
        documentation,
        controls_panel,
        visual_row
    ],
    layout=Layout(
        width='990px',
        overflow='visible'
    )
)

display(main_layout)
```
