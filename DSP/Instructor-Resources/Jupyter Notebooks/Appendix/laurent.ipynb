```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt
import math

from ipywidgets import (
    FloatSlider,
    IntSlider,
    RadioButtons,
    HTML,
    VBox,
    HBox,
    Layout
)
from IPython.display import display

# ============================================================
# TAYLOR AND LAURENT SERIES OF COMPLEX FUNCTIONS
# ============================================================

plt.ioff()

# ============================================================
# CLASSIC JUPYTER + JUPYTERLAB / NOTEBOOK 7 / BINDER
#
# - use available width
# - no horizontal / vertical output scrollbars
# - no trimming
# - no matplotlib coordinate footer
# - no resize triangle
# ============================================================

display(HTML("""
<style>

/* ============================================================
   CLASSIC JUPYTER NOTEBOOK
   ============================================================ */

.container {
    width: 98% !important;
    max-width: none !important;
}

.output_area,
.output_subarea {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

.output_scroll {
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
    box-shadow: none !important;
}

/* ============================================================
   JUPYTERLAB / NOTEBOOK 7
   ============================================================ */

.jp-Cell-outputWrapper,
.jp-OutputArea,
.jp-OutputArea-child,
.jp-OutputArea-output {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

/* ============================================================
   IPYWIDGETS
   ============================================================ */

.widget-output,
.jupyter-widgets-output-area,
.widget-box {
    max-width: none !important;
    height: auto !important;
    max-height: none !important;
    overflow: visible !important;
}

/* ============================================================
   IPYMPL
   ============================================================ */

.jupyter-matplotlib,
.jupyter-matplotlib-figure {
    overflow: visible !important;
    resize: none !important;
}

.jupyter-matplotlib::-webkit-resizer,
.jupyter-matplotlib-figure::-webkit-resizer {
    display: none !important;
}

/* ============================================================
   NOTEBOOK STYLES
   ============================================================ */

.tl-title {
    font-family:Arial, sans-serif;
    font-size:21px;
    font-weight:bold;
    color:#6f3fa0;
    margin-bottom:7px;
}

.tl-label {
    font-family:Arial, sans-serif;
    font-size:14px;
    font-weight:bold;
}

.tl-value {
    font-family:Arial, sans-serif;
    font-size:14px;
    font-weight:bold;
    color:#0b3d91;
}

.tl-radio .widget-radio-box {
    display:flex !important;
    flex-direction:row !important;
    flex-wrap:nowrap !important;
    gap:18px !important;
    align-items:center !important;
}

.tl-radio .widget-radio-box label {
    margin:0 !important;
    white-space:nowrap !important;
}

.tl-radio > label {
    display:none !important;
}

</style>
"""))

# ============================================================
# DOCUMENTATION
# ============================================================

documentation = HTML("""
<div style="
    width:980px;
    font-family:Arial, sans-serif;
    font-size:15px;
    line-height:1.45;
    margin-bottom:8px;
">

<div class="tl-title">
Taylor and Laurent Series of Complex Functions
</div>

<div style="margin-bottom:4px;">
An analytic complex function can be represented locally by a Taylor series,
while a Laurent series also contains negative powers and can therefore
describe the behavior of a function near an isolated singularity.
</div>

<div style="margin-bottom:4px;">
For the Taylor case we use sin(z) expanded around z₀ = π/4.
For the Laurent case we use e<sup>2z</sup>/(z−1)³ expanded around z₀ = 1.
</div>

<div>
<b>This notebook:</b> compares the exact complex function with a truncated
Taylor or Laurent expansion and shows how the approximation error changes
with the number of retained terms.
</div>

</div>
""")

# ============================================================
# MODE SELECTOR
# ============================================================

mode_selector = RadioButtons(
    options=[
        ('Taylor: sin(z) about π/4', 'taylor'),
        ('Laurent: e²ᶻ/(z−1)³ about 1', 'laurent')
    ],
    value='taylor',
    description='',
    layout=Layout(
        width='650px'
    )
)

mode_selector.add_class(
    'tl-radio'
)

# ============================================================
# SLIDER SETTINGS
# ============================================================

slider_style = {
    'description_width': '0px'
}

slider_layout = Layout(
    width='180px'
)

label_layout = Layout(
    width='105px',
    min_width='105px'
)

value_layout = Layout(
    width='55px',
    min_width='55px',
    margin='0px 0px 0px 5px'
)

row_layout = Layout(
    width='350px',
    align_items='center'
)

# ============================================================
# REAL PART OF SELECTED POINT
# ============================================================

x_slider = FloatSlider(
    min=-1.0,
    max=3.0,
    step=0.05,
    value=1.25,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

# ============================================================
# IMAGINARY PART
# ============================================================

y_slider = FloatSlider(
    min=-1.5,
    max=1.5,
    step=0.05,
    value=0.40,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

# ============================================================
# NUMBER OF TERMS
# ============================================================

N_slider = IntSlider(
    min=1,
    max=18,
    step=1,
    value=6,
    readout=False,
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)

# ============================================================
# VALUE LABELS
# ============================================================

x_value = HTML(
    '<div class="tl-value">1.25</div>',
    layout=value_layout
)

y_value = HTML(
    '<div class="tl-value">0.40</div>',
    layout=value_layout
)

N_value = HTML(
    '<div class="tl-value">6</div>',
    layout=value_layout
)

# ============================================================
# LABELS
# ============================================================

x_label = HTML(
    '<div class="tl-label">Re(z):</div>',
    layout=label_layout
)

y_label = HTML(
    '<div class="tl-label">Im(z):</div>',
    layout=label_layout
)

N_label = HTML(
    '<div class="tl-label">Terms N:</div>',
    layout=label_layout
)

# ============================================================
# ROWS
# ============================================================

x_row = HBox(
    [
        x_label,
        x_slider,
        x_value
    ],
    layout=row_layout
)

y_row = HBox(
    [
        y_label,
        y_slider,
        y_value
    ],
    layout=row_layout
)

N_row = HBox(
    [
        N_label,
        N_slider,
        N_value
    ],
    layout=row_layout
)

# ============================================================
# CONTROLS
# ============================================================

controls_panel = VBox(
    [
        HTML("""
        <div style="
            font-family:Arial;
            font-size:17px;
            font-weight:bold;
            color:#6f3fa0;
            margin-bottom:7px;
        ">
        Expansion Parameters
        </div>
        """),

        mode_selector,

        HBox(
            [
                VBox([x_row, y_row]),
                VBox([N_row])
            ],
            layout=Layout(
                width='760px',
                gap='18px',
                align_items='flex-start'
            )
        )
    ],
    layout=Layout(
        width='790px',
        padding='9px 12px',
        border='1px solid #d2c2df',
        overflow='visible'
    )
)

# ============================================================
# TAYLOR SERIES
#
# sin(z) around z0 = pi/4
# ============================================================

def taylor_sin(z, N):

    z0 = np.pi / 4.0

    total = 0.0 + 0.0j

    # derivative cycle:
    # sin, cos, -sin, -cos
    for n in range(N):

        phase = n % 4

        if phase == 0:
            derivative = np.sin(z0)

        elif phase == 1:
            derivative = np.cos(z0)

        elif phase == 2:
            derivative = -np.sin(z0)

        else:
            derivative = -np.cos(z0)

        total += (
            derivative
            *
            (z - z0)**n
            /
            math.factorial(n)
        )

    return total

# ============================================================
# LAURENT SERIES
#
# exp(2z)/(z-1)^3
#
# z - 1 = u
#
# exp(2z)/(z-1)^3
# =
# e^2/u^3 * exp(2u)
#
# =
# e^2 sum_{n=0}^{inf}
# (2^n / n!) u^(n-3)
# ============================================================

def laurent_exp(z, N):

    u = z - 1.0

    if abs(u) < 1e-12:
        return np.nan + 1j * np.nan

    total = 0.0 + 0.0j

    for n in range(N):

        total += (
            np.e**2
            *
            (2.0**n)
            /
            math.factorial(n)
            *
            u**(n - 3)
        )

    return total

# ============================================================
# EXACT FUNCTIONS
# ============================================================

def exact_function(mode, z):

    if mode == 'taylor':

        return np.sin(z)

    else:

        if abs(z - 1.0) < 1e-12:
            return np.nan + 1j * np.nan

        return (
            np.exp(2.0 * z)
            /
            (z - 1.0)**3
        )

# ============================================================
# APPROXIMATION
# ============================================================

def approximate_function(mode, z, N):

    if mode == 'taylor':

        return taylor_sin(
            z,
            N
        )

    else:

        return laurent_exp(
            z,
            N
        )

# ============================================================
# COMPLEX STRING
# ============================================================

def complex_string(z):

    if (
        not np.isfinite(np.real(z))
        or
        not np.isfinite(np.imag(z))
    ):
        return 'undefined'

    a = np.real(z)
    b = np.imag(z)

    if abs(a) < 1e-12:
        a = 0.0

    if abs(b) < 1e-12:
        b = 0.0

    sign = '+' if b >= 0 else '−'

    return (
        f'{a:.5f} '
        f'{sign} '
        f'{abs(b):.5f}i'
    )

# ============================================================
# FIGURE 1
# COMPLEX PLANE / EXPANSION GEOMETRY
# ============================================================

fig_plane, ax_plane = plt.subplots(
    figsize=(4.6, 4.35)
)

fig_plane.canvas.header_visible = False
fig_plane.canvas.footer_visible = False
fig_plane.canvas.toolbar_visible = False

fig_plane.canvas.layout = Layout(
    width='460px',
    height='435px'
)

ax_plane.set_title(
    'Expansion Point and Selected z',
    fontsize=14,
    fontweight='bold',
    color='#6f3fa0'
)

ax_plane.set_xlabel(
    'Re(z)',
    fontsize=10
)

ax_plane.set_ylabel(
    'Im(z)',
    fontsize=10
)

ax_plane.set_aspect(
    'equal',
    adjustable='box'
)

ax_plane.set_xlim(
    -1.2,
    3.2
)

ax_plane.set_ylim(
    -2.0,
    2.0
)

ax_plane.axhline(
    0,
    color='black',
    linewidth=1.0
)

ax_plane.axvline(
    0,
    color='black',
    linewidth=1.0
)

ax_plane.grid(
    True,
    linestyle=':',
    alpha=0.40
)

# ------------------------------------------------------------
# Expansion center
# ------------------------------------------------------------

center_point, = ax_plane.plot(
    [],
    [],
    marker='o',
    markersize=8,
    linestyle='None',
    label='Expansion center'
)

# ------------------------------------------------------------
# Selected point z
# ------------------------------------------------------------

selected_point, = ax_plane.plot(
    [],
    [],
    marker='o',
    markersize=8,
    linestyle='None',
    label='Selected z'
)

# ------------------------------------------------------------
# Radius from center to z
# ------------------------------------------------------------

radius_line, = ax_plane.plot(
    [],
    [],
    linestyle='--',
    linewidth=1.5
)

# ------------------------------------------------------------
# Singular point for Laurent case
# ------------------------------------------------------------

singular_point, = ax_plane.plot(
    [],
    [],
    marker='x',
    markersize=10,
    markeredgewidth=2,
    linestyle='None',
    label='Singularity'
)

center_text = ax_plane.text(
    0,
    0,
    '',
    fontsize=10
)

z_text = ax_plane.text(
    0,
    0,
    '',
    fontsize=10
)

singular_text = ax_plane.text(
    0,
    0,
    '',
    fontsize=10
)

ax_plane.legend(
    loc='upper left',
    fontsize=8
)

fig_plane.subplots_adjust(
    left=0.14,
    right=0.97,
    top=0.89,
    bottom=0.13
)

# ============================================================
# FIGURE 2
# ERROR VERSUS NUMBER OF TERMS
# ============================================================

fig_error, ax_error = plt.subplots(
    figsize=(5.0, 4.35)
)

fig_error.canvas.header_visible = False
fig_error.canvas.footer_visible = False
fig_error.canvas.toolbar_visible = False

fig_error.canvas.layout = Layout(
    width='500px',
    height='435px'
)

ax_error.set_title(
    'Approximation Error',
    fontsize=14,
    fontweight='bold',
    color='#6f3fa0'
)

ax_error.set_xlabel(
    'Number of retained terms',
    fontsize=10
)

ax_error.set_ylabel(
    '|f(z) − Sₙ(z)|',
    fontsize=10
)

ax_error.set_xlim(
    1,
    18
)

ax_error.set_yscale(
    'log'
)

ax_error.grid(
    True,
    linestyle=':',
    alpha=0.40
)

error_line, = ax_error.plot(
    [],
    [],
    marker='o',
    markersize=4,
    linewidth=1.5
)

current_N_point, = ax_error.plot(
    [],
    [],
    marker='o',
    markersize=8,
    linestyle='None'
)

fig_error.subplots_adjust(
    left=0.16,
    right=0.97,
    top=0.89,
    bottom=0.13
)

# ============================================================
# NUMERICAL RESULTS
# ============================================================

results_html = HTML()

results_panel = VBox(
    [
        HTML("""
        <div style="
            font-family:Arial;
            font-size:17px;
            font-weight:bold;
            color:#6f3fa0;
            margin-bottom:6px;
        ">
        Numerical Results
        </div>
        """),

        results_html
    ],
    layout=Layout(
        width='320px',
        padding='9px 11px',
        border='1px solid #d2c2df',
        overflow='visible'
    )
)

# ============================================================
# UPDATE FUNCTION
# ============================================================

def update_notebook(change=None):

    mode = mode_selector.value

    x = x_slider.value
    y = y_slider.value
    N = N_slider.value

    z = complex(
        x,
        y
    )

    # --------------------------------------------------------
    # Value labels
    # --------------------------------------------------------

    x_value.value = (
        f'<div class="tl-value">{x:.2f}</div>'
    )

    y_value.value = (
        f'<div class="tl-value">{y:.2f}</div>'
    )

    N_value.value = (
        f'<div class="tl-value">{N}</div>'
    )

    # ========================================================
    # EXPANSION CENTER
    # ========================================================

    if mode == 'taylor':

        z0 = np.pi / 4.0

        mode_name = (
            'Taylor expansion of sin(z)'
        )

        series_name = (
            'Sₙ(z)'
        )

        singular_point.set_data(
            [],
            []
        )

        singular_text.set_text(
            ''
        )

    else:

        z0 = 1.0

        mode_name = (
            'Laurent expansion of e²ᶻ/(z−1)³'
        )

        series_name = (
            'Lₙ(z)'
        )

        singular_point.set_data(
            [1.0],
            [0.0]
        )

        singular_text.set_position(
            (
                1.08,
                -0.20
            )
        )

        singular_text.set_text(
            'pole, z = 1'
        )

    # --------------------------------------------------------
    # Selected point and center
    # --------------------------------------------------------

    center_point.set_data(
        [np.real(z0)],
        [0.0]
    )

    selected_point.set_data(
        [x],
        [y]
    )

    radius_line.set_data(
        [
            np.real(z0),
            x
        ],
        [
            0.0,
            y
        ]
    )

    center_text.set_position(
        (
            np.real(z0) + 0.08,
            0.12
        )
    )

    center_text.set_text(
        'z₀'
    )

    z_text.set_position(
        (
            x + 0.08,
            y + 0.10
        )
    )

    z_text.set_text(
        'z'
    )

    # ========================================================
    # EXACT AND APPROXIMATE VALUES
    # ========================================================

    exact_value = exact_function(
        mode,
        z
    )

    approx_value = approximate_function(
        mode,
        z,
        N
    )

    if (
        np.isfinite(np.real(exact_value))
        and
        np.isfinite(np.imag(exact_value))
        and
        np.isfinite(np.real(approx_value))
        and
        np.isfinite(np.imag(approx_value))
    ):

        current_error = abs(
            exact_value
            -
            approx_value
        )

    else:

        current_error = np.nan

    # ========================================================
    # ERROR FOR ALL TERM COUNTS
    # ========================================================

    term_values = np.arange(
        1,
        19
    )

    errors = np.zeros(
        len(term_values)
    )

    for index, n_terms in enumerate(term_values):

        approximation = approximate_function(
            mode,
            z,
            int(n_terms)
        )

        if (
            np.isfinite(np.real(exact_value))
            and
            np.isfinite(np.imag(exact_value))
            and
            np.isfinite(np.real(approximation))
            and
            np.isfinite(np.imag(approximation))
        ):

            errors[index] = max(
                abs(
                    exact_value
                    -
                    approximation
                ),
                1e-16
            )

        else:

            errors[index] = np.nan

    error_line.set_data(
        term_values,
        errors
    )

    if np.isfinite(current_error):

        current_N_point.set_data(
            [N],
            [max(current_error, 1e-16)]
        )

    else:

        current_N_point.set_data(
            [],
            []
        )

    # --------------------------------------------------------
    # Error-axis limits
    # --------------------------------------------------------

    finite_errors = errors[
        np.isfinite(errors)
    ]

    if len(finite_errors) > 0:

        positive_errors = finite_errors[
            finite_errors > 0
        ]

        if len(positive_errors) > 0:

            ymin = max(
                np.min(positive_errors) / 5.0,
                1e-16
            )

            ymax = max(
                np.max(positive_errors) * 5.0,
                1e-14
            )

            if ymax <= ymin:
                ymax = ymin * 100.0

            ax_error.set_ylim(
                ymin,
                ymax
            )

    # ========================================================
    # RESULTS
    # ========================================================

    distance = abs(
        z - z0
    )

    if mode == 'taylor':

        extra_text = f"""
        Expansion center:
        <b>z₀ = π/4 ≈ {np.pi/4:.5f}</b>

        <br><br>

        Distance from center:
        <b>|z−z₀| = {distance:.5f}</b>
        """

    else:

        extra_text = f"""
        Expansion center / pole:
        <b>z₀ = 1</b>

        <br><br>

        Distance from pole:
        <b>|z−1| = {distance:.5f}</b>

        <br><br>

        The principal part contains the terms
        (z−1)<sup>−3</sup>,
        (z−1)<sup>−2</sup> and
        (z−1)<sup>−1</sup>.
        """

    results_html.value = f"""
    <div style="
        font-family:Arial, sans-serif;
        font-size:13px;
        line-height:1.55;
    ">

    <b>{mode_name}</b>

    <br><br>

    Selected point:

    <br>

    z =
    <b>{complex_string(z)}</b>

    <hr style="margin:8px 0px;">

    Exact value:

    <br>

    f(z) =
    <b>{complex_string(exact_value)}</b>

    <br><br>

    Approximation with N = {N}:

    <br>

    {series_name} =
    <b>{complex_string(approx_value)}</b>

    <br><br>

    Absolute error:

    <br>

    |f(z) − {series_name}|
    =
    <b>
    {'undefined' if not np.isfinite(current_error) else f'{current_error:.6e}'}
    </b>

    <hr style="margin:8px 0px;">

    {extra_text}

    </div>
    """

    # --------------------------------------------------------
    # REDRAW ONLY
    # --------------------------------------------------------

    fig_plane.canvas.draw_idle()
    fig_error.canvas.draw_idle()

# ============================================================
# CONNECT CONTROLS
# ============================================================

mode_selector.observe(
    update_notebook,
    names='value'
)

x_slider.observe(
    update_notebook,
    names='value'
)

y_slider.observe(
    update_notebook,
    names='value'
)

N_slider.observe(
    update_notebook,
    names='value'
)

# ============================================================
# INITIAL UPDATE
# ============================================================

update_notebook()

# ============================================================
# FIGURES ROW
# ============================================================

figures_row = HBox(
    [
        fig_plane.canvas,
        fig_error.canvas
    ],
    layout=Layout(
        width='970px',
        gap='10px',
        align_items='flex-start',
        overflow='visible'
    )
)

# ============================================================
# INTERPRETATION
#
# Placed BELOW the figures with its own spacing.
# No matplotlib coordinate footer exists.
# ============================================================

interpretation = HTML("""
<div style="
    width:950px;
    padding:9px 11px;
    border:1px solid #d7c7e5;
    background:#fbf8fe;
    font-family:Arial, sans-serif;
    font-size:13px;
    line-height:1.50;
    box-sizing:border-box;
    margin-top:10px;
">

<div style="
    color:#6f3fa0;
    font-weight:bold;
    margin-bottom:5px;
">
Interpretation
</div>

<div>
Increasing the number of retained terms generally improves the approximation
at the selected point. A Taylor series contains only non-negative powers of
(z−z₀), whereas a Laurent series may also contain negative powers. These
negative-power terms form the principal part and describe the behavior near
an isolated singularity.
</div>

</div>
""")

# ============================================================
# LEFT COLUMN
# ============================================================

left_column = VBox(
    [
        figures_row,
        interpretation
    ],
    layout=Layout(
        width='970px',
        overflow='visible'
    )
)

# ============================================================
# MAIN VISUAL ROW
# ============================================================

visual_row = HBox(
    [
        left_column,
        results_panel
    ],
    layout=Layout(
        width='1310px',
        gap='12px',
        align_items='flex-start',
        overflow='visible'
    )
)

# ============================================================
# COMPLETE NOTEBOOK
# ============================================================

main_layout = VBox(
    [
        documentation,
        controls_panel,
        visual_row
    ],
    layout=Layout(
        width='1310px',
        overflow='visible'
    )
)

display(main_layout)
```
