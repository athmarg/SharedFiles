```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
from ipywidgets import Dropdown, FloatSlider, interactive
from IPython.display import display, Markdown

# ------------------------------------------------------------
# Probability Distributions Gallery
#
# The plotting axes are kept fixed for each distribution.
# Therefore parameter changes modify the curves themselves
# rather than continuously rescaling the axes.
# ------------------------------------------------------------

def distribution_gallery(distribution, p1, p2):

    discrete = False

    # ========================================================
    # GAUSSIAN
    # ========================================================

    if distribution == 'Gaussian':

        mu = p1
        sigma = max(p2, 0.1)

        x = np.linspace(-15, 15, 1500)

        y1 = stats.norm.pdf(
            x,
            loc=mu,
            scale=sigma
        )

        y2 = stats.norm.cdf(
            x,
            loc=mu,
            scale=sigma
        )

        parameters = (
            f'μ = {mu:.2f}, σ = {sigma:.2f}'
        )

        x_limits = (-15, 15)
        y1_limits = (0, 4.2)

    # ========================================================
    # UNIFORM
    # ========================================================

    elif distribution == 'Uniform':

        a = p1
        width = max(p2, 0.1)

        x = np.linspace(-2, 12, 1500)

        y1 = stats.uniform.pdf(
            x,
            loc=a,
            scale=width
        )

        y2 = stats.uniform.cdf(
            x,
            loc=a,
            scale=width
        )

        parameters = (
            f'a = {a:.2f}, width = {width:.2f}'
        )

        x_limits = (-2, 12)
        y1_limits = (0, 10.5)

    # ========================================================
    # EXPONENTIAL
    # ========================================================

    elif distribution == 'Exponential':

        lam = max(p1, 0.1)

        x = np.linspace(0, 20, 1500)

        y1 = stats.expon.pdf(
            x,
            scale=1/lam
        )

        y2 = stats.expon.cdf(
            x,
            scale=1/lam
        )

        parameters = (
            f'λ = {lam:.2f}'
        )

        x_limits = (0, 20)
        y1_limits = (0, 5.3)

    # ========================================================
    # LAPLACE
    # ========================================================

    elif distribution == 'Laplace':

        mu = p1
        b = max(p2, 0.1)

        x = np.linspace(-15, 20, 1500)

        y1 = stats.laplace.pdf(
            x,
            loc=mu,
            scale=b
        )

        y2 = stats.laplace.cdf(
            x,
            loc=mu,
            scale=b
        )

        parameters = (
            f'μ = {mu:.2f}, b = {b:.2f}'
        )

        x_limits = (-15, 20)
        y1_limits = (0, 5.3)

    # ========================================================
    # RAYLEIGH
    # ========================================================

    elif distribution == 'Rayleigh':

        sigma = max(p1, 0.1)

        x = np.linspace(0, 20, 1500)

        y1 = stats.rayleigh.pdf(
            x,
            scale=sigma
        )

        y2 = stats.rayleigh.cdf(
            x,
            scale=sigma
        )

        parameters = (
            f'σ = {sigma:.2f}'
        )

        x_limits = (0, 20)
        y1_limits = (0, 6.5)

    # ========================================================
    # GAMMA
    # ========================================================

    elif distribution == 'Gamma':

        alpha = max(p1, 0.1)
        scale = max(p2, 0.1)

        x = np.linspace(0.001, 80, 1800)

        y1 = stats.gamma.pdf(
            x,
            a=alpha,
            scale=scale
        )

        y2 = stats.gamma.cdf(
            x,
            a=alpha,
            scale=scale
        )

        parameters = (
            f'α = {alpha:.2f}, '
            f'scale = {scale:.2f}'
        )

        x_limits = (0, 80)
        y1_limits = (0, 10.5)

    # ========================================================
    # BETA
    # ========================================================

    elif distribution == 'Beta':

        alpha = max(p1, 0.1)
        beta = max(p2, 0.1)

        x = np.linspace(
            0.001,
            0.999,
            1500
        )

        y1 = stats.beta.pdf(
            x,
            alpha,
            beta
        )

        y2 = stats.beta.cdf(
            x,
            alpha,
            beta
        )

        parameters = (
            f'α = {alpha:.2f}, β = {beta:.2f}'
        )

        x_limits = (0, 1)
        y1_limits = (0, 12)

    # ========================================================
    # CAUCHY
    # ========================================================

    elif distribution == 'Cauchy':

        x0 = p1
        gamma = max(p2, 0.1)

        x = np.linspace(-20, 25, 1800)

        y1 = stats.cauchy.pdf(
            x,
            loc=x0,
            scale=gamma
        )

        y2 = stats.cauchy.cdf(
            x,
            loc=x0,
            scale=gamma
        )

        parameters = (
            f'x₀ = {x0:.2f}, γ = {gamma:.2f}'
        )

        x_limits = (-20, 25)
        y1_limits = (0, 3.3)

    # ========================================================
    # POISSON
    # ========================================================

    elif distribution == 'Poisson':

        discrete = True

        lam = max(p1, 0.1)

        # Fixed discrete support for visualization
        x = np.arange(0, 31)

        y1 = stats.poisson.pmf(
            x,
            mu=lam
        )

        y2 = stats.poisson.cdf(
            x,
            mu=lam
        )

        parameters = (
            f'λ = {lam:.2f}'
        )

        x_limits = (-0.5, 30.5)
        y1_limits = (0, 1.05)

    # ========================================================
    # GEOMETRIC
    # ========================================================

    elif distribution == 'Geometric':

        discrete = True

        p = min(
            max(p1, 0.05),
            0.95
        )

        # Fixed discrete support for visualization
        x = np.arange(1, 31)

        y1 = stats.geom.pmf(
            x,
            p
        )

        y2 = stats.geom.cdf(
            x,
            p
        )

        parameters = (
            f'p = {p:.2f}'
        )

        x_limits = (0.5, 30.5)
        y1_limits = (0, 1.05)

    # ========================================================
    # FIGURE
    # ========================================================

    fig, axes = plt.subplots(
        1,
        2,
        figsize=(12, 4.5)
    )

    # --------------------------------------------------------
    # PDF / PMF
    # --------------------------------------------------------

    if discrete:

        axes[0].stem(
            x,
            y1
        )

        axes[0].set_ylabel(
            'PMF'
        )

    else:

        axes[0].plot(
            x,
            y1,
            linewidth=2
        )

        axes[0].set_ylabel(
            'PDF'
        )

    # --------------------------------------------------------
    # CDF
    # --------------------------------------------------------

    if discrete:

        axes[1].step(
            x,
            y2,
            where='post',
            linewidth=2
        )

    else:

        axes[1].plot(
            x,
            y2,
            linewidth=2
        )

    # ========================================================
    # FIXED AXES
    # ========================================================

    axes[0].set_xlim(
        x_limits
    )

    axes[0].set_ylim(0, 1.15 * np.max(y1))

    axes[1].set_xlim(
        x_limits
    )

    axes[1].set_ylim(
        -0.05,
        1.05
    )

    # ========================================================
    # LABELS
    # ========================================================

    axes[0].set_title(
        f'{distribution} — Density/Mass Function'
    )

    axes[0].set_xlabel(
        'x'
    )

    axes[0].grid(
        True,
        alpha=0.3
    )

    axes[1].set_title(
        f'{distribution} — CDF'
    )

    axes[1].set_xlabel(
        'x'
    )

    axes[1].set_ylabel(
        'CDF'
    )

    axes[1].grid(
        True,
        alpha=0.3
    )

    plt.tight_layout()
    plt.show()
    plt.close(fig)

    # ========================================================
    # CURRENT PARAMETERS
    # ========================================================

    print(distribution)
    print(parameters)


# ============================================================
# DISPLAY
# ============================================================

display(
    Markdown(
        "## Interactive Probability Distributions Gallery"
    )
)

display(
    interactive(
        distribution_gallery,

        distribution=Dropdown(
            options=[
                'Gaussian',
                'Uniform',
                'Exponential',
                'Laplace',
                'Rayleigh',
                'Gamma',
                'Beta',
                'Cauchy',
                'Poisson',
                'Geometric'
            ],
            value='Gaussian',
            description='Distribution:'
        ),

        p1=FloatSlider(
            value=1,
            min=0.1,
            max=5,
            step=0.1,
            description='Parameter 1:',
            continuous_update=True
        ),

        p2=FloatSlider(
            value=1,
            min=0.1,
            max=5,
            step=0.1,
            description='Parameter 2:',
            continuous_update=True
        )
    )
)
```
