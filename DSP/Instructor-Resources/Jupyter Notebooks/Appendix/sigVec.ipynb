```python
%matplotlib widget

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

from ipywidgets import (
    FloatSlider, HTML, HTMLMath,
    VBox, HBox, Layout
)
from IPython.display import display

plt.ioff()

# ============================================================
# DISPLAY SETTINGS
# ============================================================

display(HTML("""
<style>

.container {
    width:98% !important;
    max-width:none !important;
}

.output_area,
.output_subarea,
.jp-Cell-outputWrapper,
.jp-OutputArea,
.jp-OutputArea-child,
.jp-OutputArea-output,
.widget-output,
.jupyter-widgets-output-area,
.widget-box {
    max-width:none !important;
    height:auto !important;
    max-height:none !important;
    overflow:visible !important;
}

.output_scroll {
    height:auto !important;
    max-height:none !important;
    overflow:visible !important;
    box-shadow:none !important;
}

.jupyter-matplotlib,
.jupyter-matplotlib-figure {
    overflow:visible !important;
    resize:none !important;
}

.sig-title {
    font-family:Arial, sans-serif;
    font-size:20px;
    font-weight:bold;
    color:#6f3fa0;
}

.sig-label {
    font-family:Arial, sans-serif;
    font-size:14px;
    font-weight:bold;
}

.sig-value {
    font-family:Arial, sans-serif;
    font-size:14px;
    font-weight:bold;
    color:#0b3d91;
}

</style>
"""))

# ============================================================
# DOCUMENTATION
# ============================================================

documentation = HTML("""
<div style="
    width:1140px;
    padding:10px 14px;
    border:1px solid #d2c2df;
    font-family:Arial,sans-serif;
    font-size:15px;
    line-height:1.50;
    box-sizing:border-box;
">

<div class="sig-title" style="margin-bottom:8px;">
A Signal as a Vector in a Function Space
</div>

<div style="margin-bottom:6px;">
A signal can be regarded as a vector in a function space.
Once a basis of functions is chosen, the signal is represented
by the coefficients of its expansion with respect to that basis.
</div>

<div style="margin-bottom:6px;">
In this notebook the basis functions are
φ₀(t)=1, φ₁(t)=cos(t), and φ₂(t)=sin(t).
</div>

<div>
<b>This notebook:</b> shows how the coefficients a₀, a₁, a₂
act as the coordinates of the signal in the chosen function-space basis.
</div>

</div>
""")

# ============================================================
# SYMBOLIC BASIS
# ============================================================

t = sp.symbols('t', real=True)

phi0 = sp.Integer(1)
phi1 = sp.cos(t)
phi2 = sp.sin(t)

# ============================================================
# SLIDERS
# ============================================================

def make_slider(value):
    return FloatSlider(
        min=-3.0,
        max=3.0,
        step=0.1,
        value=value,
        readout=False,
        continuous_update=True,
        layout=Layout(width='300px')
    )

a0_slider = make_slider(-1.0)
a1_slider = make_slider(3.0)
a2_slider = make_slider(1.8)

a0_value = HTML()
a1_value = HTML()
a2_value = HTML()

def slider_row(label, slider, value_widget):
    return HBox(
        [
            HTML(
                f'<div class="sig-label">{label}</div>',
                layout=Layout(
                    width='55px',
                    min_width='55px'
                )
            ),

            HTML(
                '<div style="font-family:Arial;font-size:13px;">−3</div>',
                layout=Layout(width='25px')
            ),

            slider,

            HTML(
                '<div style="font-family:Arial;font-size:13px;">3</div>',
                layout=Layout(width='20px')
            ),

            value_widget
        ],
        layout=Layout(
            width='525px',
            height='40px',
            align_items='center'
        )
    )

# ============================================================
# COORDINATES PANEL
# ============================================================

coordinates_panel = VBox(
    [
        HTML(
            '<div class="sig-title" style="margin-bottom:7px;">'
            'Signal Coordinates'
            '</div>'
        ),

        slider_row('a₀:', a0_slider, a0_value),
        slider_row('a₁:', a1_slider, a1_value),
        slider_row('a₂:', a2_slider, a2_value)
    ],
    layout=Layout(
        width='565px',
        padding='10px 14px',
        border='1px solid #d2c2df'
    )
)

# ============================================================
# CURRENT SIGNAL PANEL
# ============================================================

basis_math = HTMLMath(
    value=(
        r'\('
        r'\phi_0(t)=1,\;\;'
        r'\phi_1(t)=\cos t,\;\;'
        r'\phi_2(t)=\sin t'
        r'\)'
        r'<div style="height:8px;"></div>'
    )
)

signal_math = HTMLMath()
coordinates_math = HTMLMath()

current_panel = VBox(
    [
        HTML(
            '<div class="sig-title" style="margin-bottom:7px;">'
            'Current Signal'
            '</div>'
        ),

        basis_math,

        signal_math,

        coordinates_math
    ],
    layout=Layout(
        width='565px',
        padding='10px 14px',
        border='1px solid #d2c2df'
    )
)

# ============================================================
# TOP ROW
# ============================================================

top_row = HBox(
    [
        coordinates_panel,
        current_panel
    ],
    layout=Layout(
        width='1140px',
        gap='0px',
        align_items='stretch'
    )
)

# ============================================================
# TIME AXIS
# ============================================================

t_values = np.linspace(
    -2*np.pi,
    2*np.pi,
    1500
)

phi0_num = np.ones_like(t_values)
phi1_num = np.cos(t_values)
phi2_num = np.sin(t_values)

# ============================================================
# FIGURE
# ============================================================

fig, ax = plt.subplots(
    figsize=(10.8, 5.1)
)

fig.canvas.header_visible = False
fig.canvas.footer_visible = False
fig.canvas.toolbar_visible = False

fig.canvas.layout = Layout(
    width='1120px',
    height='510px',
    margin='0px'
)

ax.set_title(
    'Signal Expansion in a Function-Space Basis',
    fontsize=14,
    fontweight='bold',
    color='#6f3fa0'
)

ax.set_xlabel(
    r'$t$',
    fontsize=12
)

ax.set_ylabel(
    'Amplitude'
)

ax.set_xlim(
    -2*np.pi,
    2*np.pi
)

ax.set_ylim(
    -6,
    6
)

ax.set_xticks(
    [
        -2*np.pi,
        -np.pi,
        0,
        np.pi,
        2*np.pi
    ]
)

ax.set_xticklabels(
    [
        r'$-2\pi$',
        r'$-\pi$',
        r'$0$',
        r'$\pi$',
        r'$2\pi$'
    ]
)

ax.axhline(
    0,
    linewidth=0.8
)

ax.axvline(
    0,
    linewidth=0.8
)

ax.grid(
    True,
    linestyle=':',
    alpha=0.40
)

# ============================================================
# INITIAL VALUES
# ============================================================

a0_initial = a0_slider.value
a1_initial = a1_slider.value
a2_initial = a2_slider.value

component0_line, = ax.plot(
    t_values,
    a0_initial * phi0_num,
    linestyle='--',
    linewidth=1.6
)

component1_line, = ax.plot(
    t_values,
    a1_initial * phi1_num,
    linestyle='--',
    linewidth=1.6
)

component2_line, = ax.plot(
    t_values,
    a2_initial * phi2_num,
    linestyle='--',
    linewidth=1.6
)

sum_line, = ax.plot(
    t_values,
    a0_initial * phi0_num
    +
    a1_initial * phi1_num
    +
    a2_initial * phi2_num,
    linewidth=2.4
)

# ============================================================
# LEGEND OUTSIDE RIGHT
# ============================================================

legend = ax.legend(
    [
        component0_line,
        component1_line,
        component2_line,
        sum_line
    ],

    [
        r'$a_0\phi_0(t)$',
        r'$a_1\phi_1(t)$',
        r'$a_2\phi_2(t)$',
        r'$x(t)=a_0\phi_0(t)+a_1\phi_1(t)+a_2\phi_2(t)$'
    ],

    loc='center left',
    bbox_to_anchor=(1.02, 0.50),
    fontsize=9,
    frameon=True
)

fig.subplots_adjust(
    left=0.08,
    right=0.76,
    top=0.88,
    bottom=0.14
)

# ============================================================
# INTERPRETATION
# ============================================================

interpretation = HTML("""
<div style="
    width:1140px;
    padding:11px 14px;
    border:1px solid #d7c7e5;
    font-family:Arial,sans-serif;
    font-size:14px;
    line-height:1.58;
    box-sizing:border-box;
    margin-top:6px;
">

<div style="
    color:#6f3fa0;
    font-size:17px;
    font-weight:bold;
    margin-bottom:7px;
">
Interpretation
</div>

<div style="margin-bottom:6px;">
The basis functions play the same role as basis vectors in ordinary
linear algebra. The signal x(t) is a linear combination of these basis functions.
</div>

<div style="margin-bottom:6px;">
The numbers a₀, a₁, a₂ are therefore the coordinates of the signal
with respect to the basis B={φ₀,φ₁,φ₂}. Changing one slider changes
one coordinate while the basis itself remains fixed.
</div>

<div>
This provides the geometric interpretation of Fourier-type expansions:
a signal is represented by its coordinates with respect to a selected
family of basis functions.
</div>

</div>
""")

# ============================================================
# UPDATE FUNCTION
# ============================================================

def update_notebook(change=None):

    a0 = a0_slider.value
    a1 = a1_slider.value
    a2 = a2_slider.value

    # --------------------------------------------------------
    # Slider values
    # --------------------------------------------------------

    a0_value.value = (
        f'<div class="sig-value">{a0:.1f}</div>'
    )

    a1_value.value = (
        f'<div class="sig-value">{a1:.1f}</div>'
    )

    a2_value.value = (
        f'<div class="sig-value">{a2:.1f}</div>'
    )

    # --------------------------------------------------------
    # Symbolic signal
    # --------------------------------------------------------

    a0_sym = sp.Float(
        round(a0, 1)
    )

    a1_sym = sp.Float(
        round(a1, 1)
    )

    a2_sym = sp.Float(
        round(a2, 1)
    )

    x_symbolic = sp.expand(
        a0_sym * phi0
        +
        a1_sym * phi1
        +
        a2_sym * phi2
    )

    signal_latex = sp.latex(
        x_symbolic
    )

    signal_latex = signal_latex.replace(
        r'\frac',
        r'\dfrac'
    )

    # --------------------------------------------------------
    # Extra vertical separation from basis row
    # --------------------------------------------------------

    signal_math.value = (
        r'<div style="height:6px;"></div>'
        r'\('
        r'x(t)='
        +
        signal_latex
        +
        r'\)'
    )

    # --------------------------------------------------------
    # Coordinate vector
    #
    # Large adaptive brackets + right alignment
    # --------------------------------------------------------

    coordinates_math.value = (
        r'\('
        r'[x]_{\mathcal B}='
        r'\left['
        r'\begin{array}{r}'
        +
        f'{a0:.1f}'
        +
        r'\\[5pt]'
        +
        f'{a1:.1f}'
        +
        r'\\[5pt]'
        +
        f'{a2:.1f}'
        +
        r'\end{array}'
        r'\right]'
        r'\)'
    )

    # --------------------------------------------------------
    # Curves
    # --------------------------------------------------------

    component0_line.set_ydata(
        a0 * phi0_num
    )

    component1_line.set_ydata(
        a1 * phi1_num
    )

    component2_line.set_ydata(
        a2 * phi2_num
    )

    sum_line.set_ydata(
        a0 * phi0_num
        +
        a1 * phi1_num
        +
        a2 * phi2_num
    )

    # --------------------------------------------------------
    # Dynamic legend text
    # --------------------------------------------------------

    legend_texts = [
        rf'$a_0\phi_0(t)=({a0:.1f})\cdot1$',
        rf'$a_1\phi_1(t)=({a1:.1f})\cdot\cos(t)$',
        rf'$a_2\phi_2(t)=({a2:.1f})\cdot\sin(t)$',
        r'$x(t)=a_0\phi_0(t)+a_1\phi_1(t)+a_2\phi_2(t)$'
    ]

    for text_object, new_text in zip(
        legend.get_texts(),
        legend_texts
    ):
        text_object.set_text(
            new_text
        )

    fig.canvas.draw_idle()

# ============================================================
# CONNECT SLIDERS
# ============================================================

for slider in [
    a0_slider,
    a1_slider,
    a2_slider
]:
    slider.observe(
        update_notebook,
        names='value'
    )

# ============================================================
# INITIAL UPDATE
# ============================================================

update_notebook()

# ============================================================
# FIGURE BLOCK
# ============================================================

figure_block = VBox(
    [
        fig.canvas
    ],
    layout=Layout(
        width='1140px',
        margin='0px 0px 4px 0px'
    )
)

# ============================================================
# FINAL DISPLAY
# ============================================================

display(
    VBox(
        [
            documentation,
            top_row,
            figure_block,
            interpretation
        ],
        layout=Layout(
            width='1180px',
            gap='7px',
            align_items='flex-start'
        )
    )
)
```
