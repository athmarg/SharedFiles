```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import multivariate_normal

from ipywidgets import (
    FloatSlider,
    interactive_output,
    HTML,
    HTMLMath,
    VBox,
    Layout
)

from IPython.display import display, Markdown

# ============================================================
# JOINT GAUSSIAN DISTRIBUTION AND CORRELATION
# ============================================================

# ------------------------------------------------------------
# Title
# ------------------------------------------------------------

display(
    Markdown(
        "## Joint Gaussian Distribution and Correlation"
    )
)

# ------------------------------------------------------------
# Short theoretical introduction
# ------------------------------------------------------------

introduction = HTML("""
<div style="
    width:900px;
    font-family:Arial, sans-serif;
    font-size:14px;
    line-height:1.45;
    margin-bottom:8px;
">
A bivariate Gaussian distribution describes the joint statistical
behavior of two Gaussian random variables <b>X</b> and <b>Y</b>.
The correlation coefficient ρ determines their linear dependence
and therefore the orientation and elongation of the joint
probability-density contours. For ρ=0 the contours are axis-aligned,
whereas positive and negative correlation orient them in opposite
directions.
</div>
""")

display(introduction)

# ============================================================
# SLIDER
# ============================================================

rho_slider = FloatSlider(
    value=0,
    min=-0.95,
    max=0.95,
    step=0.05,
    description='ρ:',
    continuous_update=True,
    layout=Layout(
        width='300px'
    )
)

display(rho_slider)

# ============================================================
# MATHEMATICAL OUTPUT
# ============================================================

math_output = HTMLMath()

# ============================================================
# DEMO FUNCTION
# ============================================================

def joint_gaussian_demo(rho):

    # --------------------------------------------------------
    # Standard deviations
    # --------------------------------------------------------

    sigma_x = 1.0
    sigma_y = 1.0

    # --------------------------------------------------------
    # Covariance matrix
    # --------------------------------------------------------

    covariance = [
        [
            sigma_x**2,
            rho * sigma_x * sigma_y
        ],
        [
            rho * sigma_x * sigma_y,
            sigma_y**2
        ]
    ]

    mean = [0, 0]

    # --------------------------------------------------------
    # Fixed coordinate grid
    # --------------------------------------------------------

    x = np.linspace(
        -4,
        4,
        300
    )

    y = np.linspace(
        -4,
        4,
        300
    )

    X, Y = np.meshgrid(
        x,
        y
    )

    positions = np.dstack(
        (X, Y)
    )

    # --------------------------------------------------------
    # Bivariate Gaussian distribution
    # --------------------------------------------------------

    distribution = multivariate_normal(
        mean=mean,
        cov=covariance
    )

    Z = distribution.pdf(
        positions
    )

    # ========================================================
    # FIGURES
    # ========================================================

    fig, axes = plt.subplots(
        1,
        2,
        figsize=(12, 5)
    )

    # --------------------------------------------------------
    # Contour plot
    # --------------------------------------------------------

    contour = axes[0].contour(
        X,
        Y,
        Z,
        levels=10
    )

    axes[0].clabel(
        contour,
        inline=True,
        fontsize=8
    )

    axes[0].set_title(
        f'Joint Gaussian PDF — ρ = {rho:.2f}'
    )

    axes[0].set_xlabel('x')
    axes[0].set_ylabel('y')

    axes[0].set_xlim(-4, 4)
    axes[0].set_ylim(-4, 4)

    axes[0].set_aspect(
        'equal'
    )

    axes[0].grid(
        True,
        alpha=0.3
    )

    # --------------------------------------------------------
    # Density map
    # --------------------------------------------------------

    density = axes[1].imshow(
        Z,
        extent=[
            x.min(),
            x.max(),
            y.min(),
            y.max()
        ],
        origin='lower',
        aspect='equal'
    )

    axes[1].set_title(
        'Joint Probability Density'
    )

    axes[1].set_xlabel('x')
    axes[1].set_ylabel('y')

    axes[1].set_xlim(-4, 4)
    axes[1].set_ylim(-4, 4)

    fig.colorbar(
        density,
        ax=axes[1],
        label='$f_{XY}(x,y)$'
    )

    plt.tight_layout()
    plt.show()
    plt.close(fig)

    # ========================================================
    # SYMBOLIC / MATHEMATICAL RESULTS
    # ========================================================

    math_output.value = (
        r'\('
        r'\mathbf{\Sigma}'
        r'='
        r'\begin{bmatrix}'
        r'\sigma_X^2 & \rho\sigma_X\sigma_Y \\'
        r'\rho\sigma_X\sigma_Y & \sigma_Y^2'
        r'\end{bmatrix}'
        r'='
        r'\begin{bmatrix}'
        f'{covariance[0][0]:.2f}'
        r' & '
        f'{covariance[0][1]:.2f}'
        r'\\'
        f'{covariance[1][0]:.2f}'
        r' & '
        f'{covariance[1][1]:.2f}'
        r'\end{bmatrix}'
        r',\qquad'
        r'\rho='
        f'{rho:.4f}'
        r'\)'
    )

# ============================================================
# INTERACTIVE OUTPUT
# ============================================================

plot_output = interactive_output(
    joint_gaussian_demo,
    {
        'rho': rho_slider
    }
)

# ============================================================
# DISPLAY
# ============================================================

display(
    VBox(
        [
            plot_output,
            math_output
        ]
    )
)
```
