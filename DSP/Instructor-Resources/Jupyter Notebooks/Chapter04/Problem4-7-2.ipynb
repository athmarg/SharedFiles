```python
# Notebook: Inverse Laplace Transform via Complex Integration (Bromwich Contour)
# Example: X(s) = exp(-3s) / (s^2 * (s - 1)) -> x(t) for t > 3 and t < 3

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# --- 1. SYMPY: Symbolic calculation & Residues ---
s = sp.Symbol('s')
t = sp.Symbol('t', real=True)

# Define the transform X(s) = exp(-3*s) / (s^2 * (s - 1))
X_s = sp.exp(-3 * s) / (s**2 * (s - 1))

print("1. Given Transform X(s):")
display(X_s)

# Function inside the Bromwich integral: X(s) * exp(s*t) = exp((t - 3)*s) / (s^2 * (s - 1))
integrand = X_s * sp.exp(s * t)
print("Integrand X(s) * exp(s*t):")
display(integrand)

# Find poles: double pole at s = 0 and simple pole at s = 1
poles = sp.solve(s**2 * (s - 1), s)
print("Poles of the function:", poles)

# Compute residue at double pole s = 0: Res = d/ds [ s^2 * integrand ] evaluated at s = 0
res_0 = sp.limit(sp.diff(integrand * s**2, s), s, 0)
print("\nResidue at double pole s = 0:")
display(res_0)

# Compute residue at simple pole s = 1: Res = limit (s - 1) * integrand as s -> 1
res_1 = sp.limit(integrand * (s - 1), s, 1)
print("\nResidue at simple pole s = 1:")
display(res_1)

# Total inverse transform for t > 3 (sum of residues inside left contour)
x_t_gt_3 = res_0 + res_1
print("\n2. Computed Inverse Transform x(t) for t > 3:")
display(x_t_gt_3)

# Inverse Laplace transform using SymPy's built-in engine
x_t_symbolic = sp.inverse_laplace_transform(X_s, s, t)
print("Computed via SymPy inverse_laplace_transform:")
display(x_t_symbolic)


# --- 2. MATPLOTLIB: Visualization of the Bromwich Contours (Figure 0407) ---
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 7))

delta = 2.0  # Abscissa of convergence Re(s) = delta > 1
R = 5.0      # Radius of semicircle

y_vals = np.linspace(-R, R, 200)

# --- Left Image (a): Case t > 3 (Closed to the left) ---
ax1.scatter([0, 1], [0, 0], color='red', s=120, zorder=5, label=r'Poles ($s = 0, s = 1$)')
ax1.plot([delta]*len(y_vals), y_vals, color='blue', lw=2.5, label=r'Bromwich Path ($\delta + j\omega$)')

# Left semicircle centered at (delta, 0) of radius R going through the left half-plane
theta_left = np.linspace(np.pi/2, 3*np.pi/2, 200)
semi_re_1 = delta + R * np.cos(theta_left)
semi_im_1 = R * np.sin(theta_left)
ax1.plot(semi_re_1, semi_im_1, color='blue', ls='--', lw=1.5, label=r'Left Semicircle $C_R$ ($t > 3$)')

ax1.annotate('', xy=(delta, R/2), xytext=(delta, R/2 - 0.1), arrowprops=dict(arrowstyle="->", color='blue', lw=2))
ax1.annotate('', xy=(delta, -R/2), xytext=(delta, -R/2 + 0.1), arrowprops=dict(arrowstyle="->", color='blue', lw=2))

ax1.axhline(0, color='black', linewidth=1)
ax1.axvline(0, color='black', linewidth=1)
ax1.set_xlim(-6, 4)
ax1.set_ylim(-6, 6)
ax1.set_xlabel(r'Real part ($\sigma$)', fontsize=12)
ax1.set_ylabel(r'Imaginary part ($j\omega$)', fontsize=12)
ax1.set_title(r'(a) Case $t > 3$ (Encloses Poles $s=0, 1$)', fontsize=13)
ax1.grid(True, linestyle=':', alpha=0.6)
ax1.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=2, frameon=True)


# --- Right Image (b): Case t < 3 (Closed to the right) ---
ax2.scatter([0, 1], [0, 0], color='red', s=120, zorder=5, label=r'Poles ($s = 0, s = 1$)')
ax2.plot([delta]*len(y_vals), y_vals, color='blue', lw=2.5, label=r'Bromwich Path ($\delta + j\omega$)')

# Right semicircle centered at (delta, 0) of radius R going through the right half-plane
theta_right = np.linspace(3*np.pi/2, 5*np.pi/2, 200)
semi_re_2 = delta + R * np.cos(theta_right)
semi_im_2 = R * np.sin(theta_right)
ax2.plot(semi_re_2, semi_im_2, color='blue', ls='--', lw=1.5, label=r'Right Semicircle $C_R$ ($t < 3$)')

ax2.annotate('', xy=(delta, R/2), xytext=(delta, R/2 - 0.1), arrowprops=dict(arrowstyle="->", color='blue', lw=2))
ax2.annotate('', xy=(delta, -R/2), xytext=(delta, -R/2 + 0.1), arrowprops=dict(arrowstyle="->", color='blue', lw=2))

ax2.axhline(0, color='black', linewidth=1)
ax2.axvline(0, color='black', linewidth=1)
ax2.set_xlim(-2, 8)
ax2.set_ylim(-6, 6)
ax2.set_xlabel(r'Real part ($\sigma$)', fontsize=12)
ax2.set_ylabel(r'Imaginary part ($j\omega$)', fontsize=12)
ax2.set_title(r'(b) Case $t < 3$ (No Poles Enclosed)', fontsize=13)
ax2.grid(True, linestyle=':', alpha=0.6)
ax2.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=2, frameon=True)

plt.suptitle(r'Contour Integration for $X(s) = \frac{e^{-3s}}{s^2(s-1)}$ (Figure 0407)', fontsize=15, y=0.98)
plt.tight_layout(rect=[0, 0.05, 1, 0.95])
plt.show()
```
