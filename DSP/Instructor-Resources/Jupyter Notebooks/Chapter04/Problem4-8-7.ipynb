```python
# Notebook: RC Circuit Analysis - Interactive Current and Voltage Response
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

t_vals = np.linspace(0, 10, 800)

def rc_circuit_responses(t, V, R, C, v0):
    tau = max(R * C, 1e-3)
    i_t = ((V - v0) / R) * np.exp(-t / tau)
    v_t = V * (1.0 - np.exp(-t / tau)) + v0 * np.exp(-t / tau)
    return i_t, v_t, tau

@interact(
    V=widgets.FloatSlider(value=5.0, min=1.0, max=20.0, step=0.5, description=r'$V$ (V):'),
    R=widgets.FloatSlider(value=2.0, min=0.1, max=10.0, step=0.1, description=r'$R$ ($\Omega$):'),
    C=widgets.FloatSlider(value=1.0, min=0.1, max=5.0, step=0.1, description=r'$C$ (F):'),
    v0=widgets.FloatSlider(value=1.0, min=-5.0, max=10.0, step=0.5, description=r'$v_0$ (V):')
)
def update_rc_plot(V, R, C, v0):
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 7))
    fig.subplots_adjust(right=0.75, hspace=0.35)
    
    i_vals, v_vals, tau = rc_circuit_responses(t_vals, V, R, C, v0)
    
    # --- Υπογράφημα 1: Τάση Πυκνωτή v_c(t) ---
    ax1.plot(t_vals, v_vals, color='royalblue', lw=2.5, label=r'Voltage $v_c(t)$')
    ax1.axhline(V, color='gray', linestyle='--', alpha=0.7, label=rf'Steady-state ($V$ = {V:.2f}V)')
    ax1.axhline(v0, color='crimson', linestyle=':', alpha=0.7, label=rf'Initial $v_0$ = {v0:.2f}V')
    ax1.axhline(0, color='black', linewidth=1)
    ax1.axvline(0, color='black', linewidth=1)
    ax1.set_xlim(0, 10)
    ax1.set_ylabel(r'Voltage $v_c(t)$ (V)', fontsize=11)
    # Διορθωμένο με :.2f για R και C
    ax1.set_title(rf'RC Circuit Responses ($R$={R:.2f} $\Omega$, $C$={C:.2f} F, $\tau$={tau:.3f}s)', fontsize=11)
    ax1.grid(True, linestyle=':', alpha=0.6)
    ax1.legend(loc='upper left', bbox_to_anchor=(1.02, 1.0), frameon=True, fontsize=9)
    
    # --- Υπογράφημα 2: Ρεύμα Κυκλώματος i(t) ---
    ax2.plot(t_vals, i_vals, color='darkorange', lw=2.5, label=r'Current $i(t)$')
    ax2.axhline(0, color='black', linewidth=1)
    ax2.axvline(0, color='black', linewidth=1)
    ax2.set_xlim(0, 10)
    ax2.set_xlabel(r'Time $t$ (s)', fontsize=11)
    ax2.set_ylabel(r'Current $i(t)$ (A)', fontsize=11)
    ax2.grid(True, linestyle=':', alpha=0.6)
    ax2.legend(loc='upper left', bbox_to_anchor=(1.02, 1.0), frameon=True, fontsize=9)
    
    plt.show()
    
    # Εκτύπωση δυναμικών μεγεθών κάτω από το γράφημα
    initial_current = (V - v0) / R
    settling_time = 4.0 * tau
    print(f"=== RC Circuit Metrics ===")
    print(f"• Time Constant (tau = RC): {tau:.3f} s  |  • Initial Current i(0): {initial_current:.3f} A")
    print(f"• Final Voltage vc(inf): {V:.2f} V     |  • Settling Time (4*tau): {settling_time:.3f} s")
```
