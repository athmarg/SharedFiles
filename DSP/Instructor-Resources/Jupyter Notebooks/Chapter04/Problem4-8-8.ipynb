```python
# Notebook: RLC Circuit Analysis - Interactive Output Voltage Response \upsilon_2(t)
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact

t_vals = np.linspace(0, 10, 800)

def rlc_circuit_response(t, V1):
    # Analytical solution for the specific conditions of Exercise 5.5.4:
    # \upsilon_2(t) = 0.5 * V1 * exp(-0.75 * t) * u(t)
    y = 0.5 * V1 * np.exp(-0.75 * t)
    return y

@interact(
    V1=widgets.FloatSlider(value=1.0, min=0.5, max=10.0, step=0.5, description=r'$V_1$ (V):')
)
def update_rlc_plot(V1):
    fig, ax = plt.subplots(figsize=(10, 4.5))
    fig.subplots_adjust(right=0.75)
    
    y_vals = rlc_circuit_response(t_vals, V1)
    
    # Σχεδίαση καμπύλης εξόδου \upsilon_2(t)
    ax.plot(t_vals, y_vals, color='crimson', lw=2.5, label=r'Output $\upsilon_2(t)$')
    ax.axhline(0, color='gray', linestyle='--', alpha=0.7, label=rf'Steady-state ($0.00$ V)')
    
    ax.axhline(0, color='black', linewidth=1)
    ax.axvline(0, color='black', linewidth=1)
    ax.set_xlim(0, 10)
    ax.set_ylim(-0.1, max(1.2 * V1 * 0.5, 1.2))
    
    ax.set_xlabel(r'Time $t$ (s)', fontsize=11)
    ax.set_ylabel(r'Voltage $\upsilon_2(t)$ (V)', fontsize=11)
    ax.set_title(rf'RLC Circuit Step Response ($\upsilon_1(t) = {V1:.2f}u(t)$)', fontsize=11)
    ax.grid(True, linestyle=':', alpha=0.6)
    
    # Υπόμνημα εκτός γραφήματος, επάνω δεξιά
    ax.legend(loc='upper left', bbox_to_anchor=(1.02, 1.0), frameon=True, fontsize=9)
    
    plt.show()
    
    # Εκτύπωση δυναμικών μεγεθών κάτω από το γράφημα (με χρήση rf"..." για αποφυγή Unicode escape errors)
    initial_val = 0.5 * V1
    time_constant = 1.0 / 0.75  # ~1.33 s
    print(rf"=== RLC Circuit Metrics ===")
    print(rf"• Initial Peak Voltage \upsilon_2(0): {initial_val:.2f} V")
    print(rf"• Exponential Decay Rate (3/4): 0.75 s^-1")
    print(rf"• Equivalent Time Constant (1/0.75): {time_constant:.2f} s")
    print(rf"• Final Steady-State Voltage: 0.00 V")
```
