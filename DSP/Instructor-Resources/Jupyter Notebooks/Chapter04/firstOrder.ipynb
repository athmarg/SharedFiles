```python
# Notebook: Continuous-Time First-Order Systems - Step and Impulse Responses
# Equations: 
#   y(t) = alpha * kappa * (1 - exp(-t / tau)) * u(t)
#   h(t) = (kappa / tau) * exp(-t / tau) * u(t)

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interactive

# --- 1. SYMPY: Symbolic verification of Step and Impulse responses ---
s = sp.Symbol('s')
t = sp.Symbol('t', real=True, positive=True)
tau, kappa, alpha = sp.symbols('tau kappa alpha', positive=True, real=True)

# Transfer function H(s) = kappa / (tau * s + 1)
H_s = kappa / (tau * s + 1)

# Step response transform: Y_s = H(s) * (alpha / s)
Y_s = H_s * (alpha / s)

print("1. Step Response Transform Y(s):")
display(Y_s)

# Inverse Laplace for step response
y_t = sp.inverse_laplace_transform(Y_s, s, t)
print("\n2. Computed Step Response y(t):")
display(y_t)

# Inverse Laplace for impulse response h(t) = L^{-1}{H(s)}
h_t = sp.inverse_laplace_transform(H_s, s, t)
print("\n3. Computed Impulse Response h(t):")
display(h_t)


# --- 2. NUMPY & INTERACTIVE MATPLOTLIB PLOT WITH WIDGETS ---
t_vals = np.linspace(0, 10, 500)

def update_first_order_system(alpha, tau, kappa):
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4.5))
    
    # Calculate responses numerically
    y_vals = alpha * kappa * (1.0 - np.exp(-t_vals / tau))
    h_vals = (kappa / tau) * np.exp(-t_vals / tau)
    
    # --- Plot 1: Step Response ---
    ax1.plot(t_vals, y_vals, color='royalblue', lw=2.5, label=r'$y(t)$ (Step Response)')
    ax1.axhline(alpha * kappa, color='gray', linestyle='--', alpha=0.7, label=r'Steady-state limit ($\alpha\kappa$ = {:.2f})'.format(alpha * kappa))
    ax1.axvline(tau, color='crimson', linestyle=':', alpha=0.8, label=r'Time constant $\tau$ = {:.2f}s ($63.2\%$)'.format(tau))
    ax1.axhline(0.632 * alpha * kappa, color='crimson', alpha=0.3, linestyle='-.')
    
    ax1.axhline(0, color='black', linewidth=1)
    ax1.axvline(0, color='black', linewidth=1)
    ax1.set_xlim(0, 10)
    ax1.set_ylim(bottom=0)
    ax1.set_xlabel(r'Time $t$', fontsize=11)
    ax1.set_ylabel(r'Output $y(t)$', fontsize=11)
    ax1.set_title(r'Step Response ($\alpha={:.1f}, \tau={:.1f}, \kappa={:.1f}$)'.format(alpha, tau, kappa), fontsize=12)
    ax1.grid(True, linestyle=':', alpha=0.6)
    ax1.legend(loc='lower right', frameon=True, fontsize=9)
    
    # --- Plot 2: Impulse Response ---
    ax2.plot(t_vals, h_vals, color='darkorange', lw=2.5, label=r'$h(t)$ (Impulse Response)')
    ax2.axvline(tau, color='crimson', linestyle=':', alpha=0.8, label=r'Time constant $\tau$ = {:.2f}s'.format(tau))
    
    ax2.axhline(0, color='black', linewidth=1)
    ax2.axvline(0, color='black', linewidth=1)
    ax2.set_xlim(0, 10)
    ax2.set_ylim(bottom=0)
    ax2.set_xlabel(r'Time $t$', fontsize=11)
    ax2.set_ylabel(r'Impulse Response $h(t)$', fontsize=11)
    ax2.set_title(r'Impulse Response ($\tau={:.1f}, \kappa={:.1f}$)'.format(tau, kappa), fontsize=12)
    ax2.grid(True, linestyle=':', alpha=0.6)
    ax2.legend(loc='upper right', frameon=True, fontsize=9)
    
    plt.tight_layout()
    plt.show()

# Create interactive widget controls for alpha, tau, and kappa
interactive_plot = interactive(
    update_first_order_system,
    alpha=widgets.FloatSlider(value=1.0, min=0.5, max=5.0, step=0.1, description=r'$\alpha$:'),
    tau=widgets.FloatSlider(value=1.5, min=0.2, max=5.0, step=0.1, description=r'$\tau$:'),
    kappa=widgets.FloatSlider(value=1.0, min=0.5, max=5.0, step=0.1, description=r'$\kappa$:')
)

display(interactive_plot)
```
