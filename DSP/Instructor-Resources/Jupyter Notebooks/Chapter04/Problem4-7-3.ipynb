```python
# Notebook: Inverse Laplace Transform via Complex Integration (Bromwich Contour - Infinite Poles)
# Example: X(s) = 1 / (s * sinh(alpha * s)) -> x(t)

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# --- 1. SYMPY: Symbolic calculation & Residues ---
s = sp.Symbol('s')
t = sp.Symbol('t', real=True, positive=True)
alpha = sp.Symbol('alpha', positive=True, real=True)

# Define the transform X(s) = 1 / (s * sinh(alpha * s))
X_s = 1 / (s * sp.sinh(alpha * s))

print("1. Given Transform X(s):")
display(X_s)

# Function inside the Bromwich integral: X(s) * exp(s*t) = exp(s*t) / (s * sinh(alpha * s))
integrand = X_s * sp.exp(s * t)
print("Integrand X(s) * exp(s*t):")
display(integrand)

# Compute residue at double pole s = 0 via L'Hopital / diff limit
res_0 = sp.limit(sp.diff(integrand * s**2, s), s, 0)
print("\nResidue at double pole s = 0 (gives t/alpha term):")
display(res_0)

# General Residue formula at simple poles s_n = n * pi * j / alpha
n = sp.Symbol('n', integer=True, positive=True)
s_n = n * sp.pi * sp.I / alpha
res_n_formula = sp.limit(integrand * (s - s_n), s, s_n)
print("\nGeneral Residue formula at simple poles s_n:")
display(res_n_formula)


# --- 2. MATPLOTLIB: Visualization of Bromwich Contour and Equidistant Infinite Poles ---
fig, ax = plt.subplots(figsize=(7, 7))

delta = 1.0  # Abscissa of convergence Re(s) = delta
R = 6.0      # Radius of semicircle

y_vals = np.linspace(-R, R, 400)

# 1. Bromwich path (vertical line)
ax.plot([delta]*len(y_vals), y_vals, color='blue', lw=2.5, label=r'Bromwich Path ($\delta + j\omega$)')

# 2. Left semicircle centered at (delta, 0) of radius R going through the left half-plane
theta_left = np.linspace(np.pi/2, 3*np.pi/2, 300)
semi_re = delta + R * np.cos(theta_left)
semi_im = R * np.sin(theta_left)
ax.plot(semi_re, semi_im, color='blue', ls='--', lw=1.5, label=r'Left Semicircle $C_R$ ($R \to \infty$)')

# Arrows for integration direction along the path
ax.annotate('', xy=(delta, R/2), xytext=(delta, R/2 - 0.1), arrowprops=dict(arrowstyle="->", color='blue', lw=2))
ax.annotate('', xy=(delta, -R/2), xytext=(delta, -R/2 + 0.1), arrowprops=dict(arrowstyle="->", color='blue', lw=2))

# 3. Equidistant Simple Poles along imaginary axis
step_vis = 1.2 
pos_poles = [i * step_vis for i in range(1, 5)]   # 4 poles above origin
neg_poles = [-i * step_vis for i in range(1, 5)]  # 4 poles below origin
all_simple_poles = pos_poles + neg_poles

# Plot simple poles (red 'x')
ax.scatter([0]*len(all_simple_poles), all_simple_poles, color='red', s=100, marker='x', linewidth=2, zorder=5, label=r'Simple Poles ($s_n = \pm \frac{n\pi j}{\alpha}$)')

# Plot double pole at origin
ax.scatter([0], [0], color='darkred', s=140, marker='x', linewidth=3, zorder=6, label=r'Double Pole ($s = 0$)')

# Dots at the top and bottom of the axis to indicate infinity
ax.text(0.5, 5.2, r'$\vdots$', fontsize=20, color='red', fontweight='bold')
ax.text(0.5, -5.7, r'$\vdots$', fontsize=20, color='red', fontweight='bold')

# Axes styling and layout
ax.axhline(0, color='black', linewidth=1)
ax.axvline(0, color='black', linewidth=1)
ax.set_xlim(-7.5, 3.0)
ax.set_ylim(-7.0, 7.0)
ax.set_xlabel(r'Real part ($\sigma$)', fontsize=11)
ax.set_ylabel(r'Imaginary part ($j\omega$)', fontsize=11)
ax.set_title(r'Bromwich Contour Enclosing Infinite Poles for $\frac{1}{s\sinh(\alpha s)}$', fontsize=12)
ax.grid(True, linestyle=':', alpha=0.6)

# Legend below plot
ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=2, frameon=True)

plt.tight_layout()
plt.show()
```
