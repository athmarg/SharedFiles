```python
# Notebook: Solving First-Order Differential Equation via Laplace Transform with Interactive Slider
# Example: x'(t) + alpha * x(t) = u(t), with x(0) = 0 -> x(t) = (1 / alpha) * (1 - exp(-alpha * t)) * u(t)

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interactive

# --- 1. SYMPY: Symbolic solution verification ---
s, alpha_sym = sp.symbols('s alpha', positive=True, real=True)
t = sp.Symbol('t', real=True, positive=True)

# Define transform X(s) = 1 / (s * (s + alpha))
X_s = 1 / (s * (s + alpha_sym))

print("1. Transformed Solution X(s):")
display(X_s)

# Perform Partial Fraction Expansion
X_s_pfe = sp.apart(X_s, s)
print("Partial Fraction Expansion of X(s):")
display(X_s_pfe)

# Compute Inverse Laplace Transform
x_t = sp.inverse_laplace_transform(X_s, s, t)
print("\n2. Computed Solution x(t):")
display(x_t)


# --- 2. NUMPY & INTERACTIVE MATPLOTLIB PLOT WITH WIDGETS ---
# Define analytical response function for numerical evaluation
def calculate_response(t_arr, alpha):
    # x(t) = (1 / alpha) * (1 - exp(-alpha * t))
    with np.errstate(divide='ignore', invalid='ignore'):
        res = (1.0 / alpha) * (1.0 - np.exp(-alpha * t_arr))
    return res

t_vals = np.linspace(0, 5, 400)

def update_plot(alpha):
    fig, ax = plt.subplots(figsize=(7, 4.5))
    x_vals = calculate_response(t_vals, alpha)
    
    ax.plot(t_vals, x_vals, color='royalblue', lw=2.5, label=r'$x(t) = \frac{1}{\alpha}(1 - e^{-\alpha t})u(t)$')
    
    ax.axhline(1/alpha, color='gray', linestyle='--', alpha=0.7, label=r'Steady-state limit ($1/\alpha$ = {:.2f})'.format(1/alpha))
    ax.axhline(0, color='black', linewidth=1)
    ax.axvline(0, color='black', linewidth=1)
    
    ax.set_xlim(0, 5)
    ax.set_ylim(0, max(2.5, 1.2 / max(alpha, 0.2)))
    ax.set_xlabel(r'Time $t$', fontsize=11)
    ax.set_ylabel(r'$x(t)$', fontsize=11)
    ax.set_title(r'First-Order Response for $\alpha = {:.2f}$'.format(alpha), fontsize=12)
    ax.grid(True, linestyle=':', alpha=0.6)
    ax.legend(loc='best', frameon=True)
    
    plt.tight_layout()
    plt.show()

# Create interactive slider widget for alpha ranging from 0.5 to 5.0
interactive_plot = interactive(update_plot, alpha=widgets.FloatSlider(value=1.0, min=0.5, max=5.0, step=0.1, description=r'$\alpha$:'))
display(interactive_plot)
```
