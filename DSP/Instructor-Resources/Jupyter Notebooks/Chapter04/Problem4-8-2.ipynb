```python
# Notebook: Solving Second-Order Differential Equation via Laplace Transform (Step Forcing)
# Example: x''(t) + 5x'(t) + 4x(t) = 10, with x(0) = 0, x'(0) = 0

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# --- 1. SYMPY: Differential Equation Setup & Laplace Solution ---
s = sp.Symbol('s')
t = sp.Symbol('t', real=True, positive=True)

# Define X(s) = 10 / (s * (s + 1) * (s + 4))
numerator = 10
denominator = s * (s + 1) * (s + 4)
X_s = numerator / denominator

print("1. Transformed Solution X(s):")
display(X_s)

# Perform Partial Fraction Expansion to verify coefficients
X_s_pfe = sp.apart(X_s, s)
print("Partial Fraction Expansion of X(s):")
display(X_s_pfe)

# Compute Inverse Laplace Transform to get x(t)
x_t = sp.inverse_laplace_transform(X_s, s, t)
print("\n2. Computed Solution x(t):")
display(x_t)


# --- 2. NUMPY & MATPLOTLIB: Time-domain response visualization ---
x_t_numeric = sp.lambdify(t, x_t, 'numpy')

t_vals = np.linspace(0, 4, 400)
x_vals = x_t_numeric(t_vals)

fig, ax = plt.subplots(figsize=(7, 4.5))
ax.plot(t_vals, x_vals, color='darkorange', lw=2.5, label=r'$x(t) = \left(\frac{5}{2} - \frac{10}{3}e^{-t} + \frac{5}{6}e^{-4t}\right)u(t)$')

ax.axhline(2.5, color='gray', linestyle='--', alpha=0.7, label=r'Steady-state limit ($5/2 = 2.5$)')
ax.axhline(0, color='black', linewidth=1)
ax.axvline(0, color='black', linewidth=1)
ax.set_xlabel(r'Time $t$', fontsize=11)
ax.set_ylabel(r'$x(t)$', fontsize=11)
ax.set_title(r'Step Response of Second-Order Differential Equation', fontsize=12)
ax.grid(True, linestyle=':', alpha=0.6)
ax.legend(loc='best', frameon=True)

plt.tight_layout()
plt.show()
```
