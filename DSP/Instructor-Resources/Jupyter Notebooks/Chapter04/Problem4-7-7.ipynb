```python
# Notebook: Inverse Laplace Transform via Partial Fraction Expansion
# Example: X(s) = (s^2 + 2s - 4) / (s^4 + 2s^3 + s^2) = (s^2 + 2s - 4) / (s^2 * (s + 1)^2)

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# --- 1. SYMPY: Symbolic calculation & Partial Fraction Expansion ---
s = sp.Symbol('s')
t = sp.Symbol('t', real=True, positive=True)

# Define the transform X(s)
numerator = s**2 + 2*s - 4
denominator = s**4 + 2*s**3 + s**2
X_s = numerator / denominator

print("1. Given Transform X(s):")
display(X_s)

# Factor the denominator to see poles explicitly
X_s_factored = sp.factor(X_s)
print("Factored form of X(s):")
display(X_s_factored)

# Perform Partial Fraction Expansion (PFE)
X_s_pfe = sp.apart(X_s, s)
print("Partial Fraction Expansion of X(s):")
display(X_s_pfe)

# Compute Inverse Laplace Transform term by term or via built-in engine
x_t = sp.inverse_laplace_transform(X_s, s, t)
print("\n2. Computed Inverse Transform x(t) for t > 0:")
display(x_t)


# --- 2. NUMPY & MATPLOTLIB: Time-domain response visualization ---
# Convert the SymPy expression of x(t) to a numerical function
# Note: inverse_laplace_transform returns a Heaviside function wrapper or piecewise expression
x_t_numeric = sp.lambdify(t, x_t, 'numpy')

t_vals = np.linspace(0, 5, 400)
x_vals = x_t_numeric(t_vals)

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(t_vals, x_vals, color='purple', lw=2.5, label=r'$x(t) = (10 - 4t - 10e^{-t} - 5t e^{-t})u(t)$')

ax.axhline(0, color='black', linewidth=1)
ax.axvline(0, color='black', linewidth=1)
ax.set_xlabel(r'Time $t$', fontsize=12)
ax.set_ylabel(r'$x(t)$', fontsize=12)
ax.set_title(r'Inverse Laplace Transform Response via Partial Fractions', fontsize=13)
ax.grid(True, linestyle=':', alpha=0.6)
ax.legend(loc='best', frameon=True)

plt.tight_layout()
plt.show()
```
