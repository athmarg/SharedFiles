```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Parameters definition
N = 16        # DFT length
alpha = 0.8   # Base parameter for the exponential signal (0 < alpha < 1)

# 2. Time-domain signal: x[n] = alpha^n for 0 <= n <= N-1
n = np.arange(N)
x = alpha ** n

# 3. Discrete Fourier Transform (DFT) computation based on the solution:
# X[k] = (1 - alpha^N) / (1 - alpha * exp(-j * 2*pi*k / N))
k = np.arange(N)
X_k = (1 - alpha**N) / (1 - alpha * np.exp(-1j * 2 * np.pi * k / N))

# 4. Plotting: Time Domain, DFT Magnitude, and DFT Phase
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(16, 4.5))
fig.suptitle(rf'Complex DFT of an Exponential Signal: $x[n] = {alpha}^n$ ($N={N}$)', fontsize=13, fontweight='bold')

# --- Plot 1: Time Domain Signal ---
ax1.stem(n, x, basefmt="k-", linefmt='C0-', markerfmt='C0o')
ax1.set_title(r'Time Domain: Exponential $x[n]$', fontsize=11, fontweight='bold')
ax1.set_xlabel('Time Index $n$', fontsize=10)
ax1.set_ylabel('Amplitude', fontsize=10)
ax1.set_xlim(-0.5, N - 0.5)
ax1.set_ylim(-0.1, 1.2)
ax1.grid(True, linestyle='--', alpha=0.5)

# --- Plot 2: DFT Magnitude ---
ax2.stem(k, np.abs(X_k), basefmt="k-", linefmt='C3-', markerfmt='C3o')
ax2.set_title(r'Frequency Domain: Magnitude $|\mathcal{X}[k]|$', fontsize=11, fontweight='bold')
ax2.set_xlabel('Frequency Index $k$', fontsize=10)
ax2.set_ylabel('Magnitude', fontsize=10)
ax2.set_xlim(-0.5, N - 0.5)
ax2.grid(True, linestyle='--', alpha=0.5)

# --- Plot 3: DFT Phase ---
ax3.stem(k, np.angle(X_k), basefmt="k-", linefmt='C2-', markerfmt='C2o')
ax3.set_title(r'Frequency Domain: Phase $\angle \mathcal{X}[k]$ (rad)', fontsize=11, fontweight='bold')
ax3.set_xlabel('Frequency Index $k$', fontsize=10)
ax3.set_ylabel('Phase (radians)', fontsize=10)
ax3.set_xlim(-0.5, N - 0.5)
ax3.set_ylim(-np.pi - 0.5, np.pi + 0.5)
ax3.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
