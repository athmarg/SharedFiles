```python
# ==============================================================================
# Frequency Response Analysis of a Discrete LTI System
# Target Publication: Springer
# ==============================================================================

import numpy as np
from scipy import signal
import matplotlib.pyplot as plt
import sympy as sp
from IPython.display import display, Markdown

# 1. Display Title and Symbolic Frequency Response Expression
display(Markdown("### **Analytical Frequency Response Function Equation**"))

omega = sp.Symbol('omega', real=True)
H_symbolic = (1 + 2 * sp.exp(-1j * omega) + sp.exp(-2j * omega)) / (1 - 0.5 * sp.exp(-1j * omega))

# Format nicely with H(e^{j\omega}) on the left-hand side
H_expr = sp.Eq(sp.Function('H')(sp.exp(1j * omega)), H_symbolic)
display(H_expr)

# 2. Define Numerical LTI System Coefficients for Analysis
# Difference equation: y[n] - 0.5*y[n-1] = x[n] + 2*x[n-1] + x[n-2]
b = [1.0, 2.0, 1.0]  # Numerator coefficients [b0, b1, b2]
a = [1.0, -0.5]      # Denominator coefficients [1, -a1]

# 3. Compute the Frequency Response (DTFT) numerically
omega_num = np.linspace(-np.pi, np.pi, 2048)
w, H = signal.freqz(b, a, worN=omega_num)

magnitude = np.abs(H)
phase = np.angle(H)

# 4. Plotting the Magnitude and Phase Responses
plt.rcParams.update({
    'font.size': 12,
    'axes.labelsize': 14,
    'axes.titlesize': 14,
    'xtick.labelsize': 12,
    'ytick.labelsize': 12,
    'figure.figsize': (10, 6)
})

fig, (ax1, ax2) = plt.subplots(2, 1, sharex=True)

# Magnitude Response Plot
ax1.plot(w / np.pi, magnitude, 'b-', linewidth=2)
ax1.set_title('Frequency Response of the Discrete LTI System')
ax1.set_ylabel(r'$|\mathcal{H}(e^{j\omega})|$')
ax1.grid(True, linestyle='--', alpha=0.6)
ax1.set_xlim(-1, 1)

# Phase Response Plot
ax2.plot(w / np.pi, phase, 'r-', linewidth=2)
ax2.set_xlabel(r'Normalized Frequency ($\omega / \pi$ radians/sample)')
ax2.set_ylabel(r'$\angle \mathcal{H}(e^{j\omega})$ (rad)')
ax2.grid(True, linestyle='--', alpha=0.6)
ax2.set_xlim(-1, 1)

plt.tight_layout()
plt.show()
```
