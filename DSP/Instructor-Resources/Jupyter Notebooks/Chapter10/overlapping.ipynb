```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import IntSlider, HBox, VBox, Output
from IPython.display import display, clear_output

# ============================================================
# Parameters
# ============================================================

L = 100
initial_windows = 10
window = np.hamming(L)
n = np.arange(L)

# ============================================================
# Slider
# ============================================================

overlap_slider = IntSlider(
    value=0,
    min=0,
    max=90,
    step=5,
    description='Overlap:',
    continuous_update=True,
    style={'description_width': 'initial'},
    layout={'width': '500px'}
)

# ============================================================
# Output
# ============================================================

plot_output = Output()

# ============================================================
# Update Function
# ============================================================

def update(overlap):

    hop = max(1, int(L * (1 - overlap / 100)))

    number_of_windows = int(np.ceil((initial_windows - 1) * L / hop)) + 1
    positions = np.arange(number_of_windows) * hop

    fig, ax = plt.subplots(figsize=(17.25, 5))

    for position in positions:
        ax.plot(position + n, window, linewidth=2)
        ax.fill_between(position + n, 0, window, alpha=0.12)

    ax.set_xlim(-10, positions[-1] + L + 10)
    ax.set_ylim(0, 1.1)

    ax.set_xlabel('Position')
    ax.set_ylabel('Amplitude')

    ax.set_title(
        f'Hamming Windows — Overlap = {overlap}%   |   '
        f'Window Length = {L}   |   '
        f'Number of Windows = {number_of_windows}',
        fontsize=12
    )

    ax.grid(True, alpha=0.25)

    plt.tight_layout()

    with plot_output:
        clear_output(wait=True)
        display(fig)

    plt.close(fig)

# ============================================================
# Slider Callback
# ============================================================

def slider_changed(change):
    if change['name'] == 'value':
        update(change['new'])

overlap_slider.observe(slider_changed, names='value')

# ============================================================
# Initial Display
# ============================================================

update(0)

display(
    VBox([
        HBox([overlap_slider]),
        plot_output
    ])
)
```
