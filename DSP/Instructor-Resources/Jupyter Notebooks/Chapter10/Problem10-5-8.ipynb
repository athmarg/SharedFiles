```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Parameters definition
N = 16           # DFT length
omega_0 = 1.2    # Frequency of the complex exponential signal (rad/rad)

# 2. Time-domain signal: x[n] = exp(j * omega_0 * n) for 0 <= n <= N-1
n = np.arange(N)
x = np.exp(1j * omega_0 * n)

# 3. Continuous DTFT computation over a fine frequency grid omega in [0, 2*pi]
omega_continuous = np.linspace(0, 2 * np.pi, 1000)
numerator = 1 - np.exp(-1j * (omega_continuous - omega_0) * N)
denominator = 1 - np.exp(-1j * (omega_continuous - omega_0))
dtft_magnitude = np.abs(numerator / denominator)
dtft_magnitude[np.abs(denominator) < 1e-10] = N

# 4. Discrete Fourier Transform (DFT) computation at frequencies omega_k = 2*pi*k / N
k = np.arange(N)
omega_k = 2 * np.pi * k / N
X_k = np.exp(-1j * (omega_k - omega_0) * (N - 1) / 2) * (np.sin((omega_k - omega_0) * N / 2) / np.sin((omega_k - omega_0) / 2))
for idx, freq in enumerate(omega_k):
    if np.isclose((freq - omega_0) % (2 * np.pi), 0, atol=1e-7):
        X_k[idx] = N

# 5. Plotting: Continuous DTFT vs. DFT Samples
fig, ax = plt.subplots(figsize=(10, 5))

# Plot continuous DTFT magnitude
ax.plot(omega_continuous, dtft_magnitude, color='C0', linewidth=2, label=r'Continuous DTFT $|\mathcal{X}(e^{j\omega})|$')

# Plot DFT magnitude samples as red stems/dots
ax.stem(omega_k, np.abs(X_k), basefmt="k-", linefmt='C3-', markerfmt='C3o', label=r'DFT Samples $|\mathcal{X}[k]|$ at $\omega_k = \frac{2\pi k}{N}$')

# Formatting
ax.set_title(r'Bridge Between DTFT and DFT: $x[n] = e^{j ' + str(omega_0) + r' n}$ ($N=' + str(N) + r'$)', fontsize=12, fontweight='bold')
ax.set_xlabel(r'Frequency $\omega$ (radians)', fontsize=11)
ax.set_ylabel(r'Magnitude', fontsize=11)
ax.set_xlim(0, 2 * np.pi)
ax.grid(True, linestyle='--', alpha=0.5)

# Place the legend outside the plot to the right
ax.legend(loc='upper left', bbox_to_anchor=(1.02, 1), fontsize=10)

plt.tight_layout()
plt.show()
```
