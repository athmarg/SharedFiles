```python
import numpy as np
from IPython.display import display, HTML

def get_dft_matrix(N):
    n = np.arange(N)
    k = n.reshape((N, 1))
    W_N = np.exp(-2j * np.pi * k * n / N)
    return W_N

html_content = ""

for N in range(1, 9):
    W_N = get_dft_matrix(N)
    
    html_content += f"<h3>DFT Matrix $\\mathbf{{W}}_{{{N}}}$ ($N = {N}$)</h3>"
    html_content += "<table style='border-collapse: collapse; text-align: center; margin-bottom: 20px; font-family: sans-serif;'>"
    
    for row in W_N:
        html_content += "<tr>"
        for val in row:
            re = np.round(val.real, 3)
            im = np.round(val.imag, 3)
            if abs(re) < 1e-10: re = 0.0
            if abs(im) < 1e-10: im = 0.0
            
            if im == 0:
                s = f"{re:g}"
            elif re == 0:
                if im == 1: s = "j"
                elif im == -1: s = "-j"
                else: s = f"{im:g}j"
            else:
                sign = "+" if im > 0 else "-"
                s = f"{re:g} {sign} {abs(im):g}j"
                
            html_content += f"<td style='border: 1px solid black; padding: 8px 12px;'>${s}$</td>"
        html_content += "</tr>"
        
    html_content += "</table>"

display(HTML(html_content))
```
