```python
import sympy as sp
from IPython.display import display, Math

# Define symbolic variables
n, k = sp.symbols('n k', integer=True)
N = 4

# 1. Symbolic definition of signals x[n] and y[n] for n = 0, 1, 2, 3
n_vals = list(range(N))
x_n = sp.cos(sp.pi * n / 2)
y_n = 2**n

# 2. Symbolic DFT calculation for X[k]
X_k = sum(x_n.subs(n, ni) * sp.exp(sp.I * 2 * sp.pi * k * ni / N) for ni in range(N))
X_k = sp.simplify(X_k)

# 3. Symbolic DFT calculation for Y[k]
Y_k = sum(y_n.subs(n, ni) * sp.exp(sp.I * 2 * sp.pi * k * ni / N) for ni in range(N))
Y_k = sp.simplify(Y_k)

# 4. Frequency-domain product Z[k] = X[k] * Y[k]
Z_k = sp.expand(X_k * Y_k)

# 5. Fully symbolic Inverse DFT (IDFT) to derive z[n] mathematically
z_evaluated = []
for ni in range(N):
    val = (1 / N) * sum(Z_k.subs(k, ki) * sp.exp(-sp.I * 2 * sp.pi * ki * ni / N) for ki in range(N))
    z_evaluated.append(sp.simplify(val))

# Displaying everything cleanly using IPython Math mode
display(Math(r"\text{--- Symbolic Analysis \& Derivation ---}"))
display(Math(r"x[n] = " + sp.latex(x_n)))
display(Math(r"y[n] = " + sp.latex(y_n)))

display(Math(r"\mathbf{1.}\ \text{Transform } \mathcal{X}[k]: \quad \mathcal{X}[k] = " + sp.latex(X_k)))
display(Math(r"\mathbf{2.}\ \text{Transform } \mathcal{Y}[k]: \quad \mathcal{Y}[k] = " + sp.latex(Y_k)))
display(Math(r"\mathbf{3.}\ \text{Product } \mathcal{Z}[k] = \mathcal{X}[k]\mathcal{Y}[k]: \quad \mathcal{Z}[k] = " + sp.latex(Z_k)))

# Constructing the final LaTeX representation for z[n]
z_latex_elements = [sp.latex(val) for val in z_evaluated]
z_vector_str = ", ".join(z_latex_elements)
display(Math(r"\mathbf{4.}\ \text{Final circular convolution result } z[n]:"))
display(Math(r"z[n] = \begin{bmatrix} " + z_vector_str + r" \end{bmatrix}"))
```
