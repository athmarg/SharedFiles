```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML

display(HTML("<style>.output_scroll { height: unset !important; }</style>"))


def symbolic_lti_sinusoidal_response():

    # ============================================================
    # 1. SYMBOLIC DERIVATION
    # ============================================================

    n = sp.symbols('n', integer=True, real=True)
    omega = sp.symbols('omega', real=True)
    A = sp.symbols('A', real=True, positive=True)
    H_mag = sp.symbols('H_mag', real=True, positive=True)
    H_phase = sp.symbols('H_phase', real=True)

    H = H_mag*sp.exp(sp.I*H_phase)

    kappa_1 = A*sp.exp(sp.I*omega*n)
    kappa_2 = A*sp.exp(-sp.I*omega*n)

    lambda_1 = H*kappa_1
    lambda_2 = H*kappa_2

    x_cos = sp.simplify(sp.expand_complex((kappa_1 + kappa_2)/2))
    y_cos = sp.simplify(sp.expand_complex((lambda_1 + lambda_2)/2))

    x_sin = sp.simplify(sp.expand_complex((kappa_1 - kappa_2)/(2*sp.I)))
    y_sin = sp.simplify(sp.expand_complex((lambda_1 - lambda_2)/(2*sp.I)))

    display(HTML(r"<b>Input cosine:</b>"))
    display(sp.Eq(sp.Symbol('x_1[n]'), x_cos))

    display(HTML(r"<b>Output cosine:</b>"))
    display(sp.Eq(sp.Symbol('y_1[n]'), sp.trigsimp(y_cos)))

    display(HTML(r"<b>Input sine:</b>"))
    display(sp.Eq(sp.Symbol('x_2[n]'), x_sin))

    display(HTML(r"<b>Output sine:</b>"))
    display(sp.Eq(sp.Symbol('y_2[n]'), sp.trigsimp(y_sin)))

    # ============================================================
    # 2. INTERACTIVE PLOTS
    # ============================================================

    def update_plot(A_val, omega_val, H_mag_val, H_phase_val, max_n):

        clear_output(wait=True)

        n_vals = np.arange(0, max_n + 1)

        x_vals = A_val*np.cos(omega_val*n_vals)

        y_vals = A_val*H_mag_val*np.cos(omega_val*n_vals + H_phase_val)

        # Fixed vertical scale
        y_limit = max(3.2, 1.15*A_val*3.0)

        fig, axes = plt.subplots(2, 1, figsize=(12, 5.5))

        # ========================================================
        # INPUT SIGNAL
        # ========================================================

        markerline1, stemlines1, baseline1 = axes[0].stem(n_vals, x_vals, linefmt='C0-', markerfmt='C0o', basefmt='k-')
        plt.setp(stemlines1, alpha=0.7)
        plt.setp(markerline1, alpha=0.7)

        axes[0].set_title(r'Input Signal $x_1[n]=A\cos(\omega n)$', fontsize=12, fontweight='bold')
        axes[0].set_xlabel(r'$n$', fontsize=10)
        axes[0].set_ylabel(r'$x_1[n]$', fontsize=10)
        axes[0].set_xlim(0, max_n)
        axes[0].set_ylim(-y_limit, y_limit)
        axes[0].grid(True, linestyle='--', alpha=0.6)

        # ========================================================
        # OUTPUT SIGNAL
        # ========================================================

        markerline2, stemlines2, baseline2 = axes[1].stem(n_vals, y_vals, linefmt='r-', markerfmt='ro', basefmt='k-')
        plt.setp(stemlines2, linewidth=1.5)

        axes[1].set_title(r'Output Signal $y_1[n]=A|H(e^{j\omega})|\cos(\omega n+\angle H(e^{j\omega}))$', fontsize=12, fontweight='bold')
        axes[1].set_xlabel(r'$n$', fontsize=10)
        axes[1].set_ylabel(r'$y_1[n]$', fontsize=10)
        axes[1].set_xlim(0, max_n)
        axes[1].set_ylim(-y_limit, y_limit)
        axes[1].grid(True, linestyle='--', alpha=0.6)

        plt.subplots_adjust(left=0.08, right=0.98, top=0.90, bottom=0.10, hspace=0.42)
        plt.show()

    # ============================================================
    # 3. CONTROLS
    # ============================================================

    A_slider = widgets.FloatSlider(value=1.0, min=0.2, max=3.0, step=0.1, description=r'Amplitude $A$:', style={'description_width':'initial'}, continuous_update=False)

    omega_slider = widgets.FloatSlider(value=0.5, min=0.1, max=2.5, step=0.05, description=r'Frequency $\omega$:', style={'description_width':'initial'}, continuous_update=False)

    H_mag_slider = widgets.FloatSlider(value=1.5, min=0.2, max=3.0, step=0.1, description=r'$|H(e^{j\omega})|$:', style={'description_width':'initial'}, continuous_update=False)

    H_phase_slider = widgets.FloatSlider(value=np.pi/4, min=-np.pi, max=np.pi, step=np.pi/20, description=r'$\angle H(e^{j\omega})$:', style={'description_width':'initial'}, continuous_update=False)

    n_slider = widgets.IntSlider(value=40, min=20, max=100, step=5, description=r'Max index $n$:', style={'description_width':'initial'}, continuous_update=False)

    ui = widgets.VBox([A_slider, omega_slider, H_mag_slider, H_phase_slider, n_slider])

    out = widgets.interactive_output(update_plot, {'A_val': A_slider, 'omega_val': omega_slider, 'H_mag_val': H_mag_slider, 'H_phase_val': H_phase_slider, 'max_n': n_slider})

    display(ui, out)


symbolic_lti_sinusoidal_response()
```
