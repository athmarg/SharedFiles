```python
# ==============================================================================
# Symbolic Difference Equation from Frequency Response
# Difference Equation:
# Derived automatically from the given frequency response
# Target Publication: Springer
# ==============================================================================

import sympy as sp
from IPython.display import display, Markdown

sp.init_printing(use_unicode=True)

# ------------------------------------------------------------------------------
# 1. Symbolic Definitions
# ------------------------------------------------------------------------------

omega = sp.Symbol('omega', real=True)
n = sp.Symbol('n', integer=True)
z = sp.Symbol('z')
x = sp.Function('x')
y = sp.Function('y')
H = sp.Function('H')

# ------------------------------------------------------------------------------
# 2. Define the Frequency Response Symbolically
# ------------------------------------------------------------------------------

num_coeffs = [sp.Rational(1, 1), -sp.Rational(1, 2), 0, sp.Rational(1, 1)]
den_coeffs = [sp.Rational(1, 1), sp.Rational(1, 2), sp.Rational(3, 4)]

num_z = sum(num_coeffs[k] * z**(-k) for k in range(len(num_coeffs)))
den_z = sum(den_coeffs[k] * z**(-k) for k in range(len(den_coeffs)))

H_z = sp.cancel(num_z / den_z)
H_omega = sp.simplify(H_z.subs(z, sp.exp(sp.I * omega)))

# ------------------------------------------------------------------------------
# 3. Display Frequency Response
# ------------------------------------------------------------------------------

display(Markdown("### **Frequency Response**"))
display(sp.Eq(H(sp.exp(sp.I * omega)), H_omega))

# ------------------------------------------------------------------------------
# 4. Construct Y(e^jw) and X(e^jw) Symbolically
#
# H(e^jw) = Y(e^jw) / X(e^jw)
#
# Therefore:
#
# Denominator * Y(e^jw) = Numerator * X(e^jw)
# ------------------------------------------------------------------------------

y_terms = [den_coeffs[k] * y(n - k) for k in range(len(den_coeffs))]
x_terms = [num_coeffs[k] * x(n - k) for k in range(len(num_coeffs))]

difference_equation = sp.Eq(sp.Add(*y_terms), sp.Add(*x_terms))

# ------------------------------------------------------------------------------
# 5. Display Difference Equation
# ------------------------------------------------------------------------------

display(Markdown("### **Difference Equation**"))
display(difference_equation)
```
