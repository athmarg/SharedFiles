```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import IntSlider, VBox, HBox, HTML, interactive_output
from IPython.display import display

# Base parameters
N_base = 16
m_freq = 3 

def plot_zero_padding_clean(L_total):
    plt.close('all')
    
    # 1. Time domain setup
    n_base = np.arange(N_base)
    x_base = np.cos(2 * np.pi * m_freq * n_base / N_base)
    x_padded = np.zeros(L_total)
    x_padded[:N_base] = x_base
    n_full = np.arange(L_total)
    
    # 2. Frequency domain setup
    X_dft = np.fft.fft(x_padded)
    freqs = np.arange(L_total) 
    n_fft = 4096
    x_dtft_buf = np.zeros(n_fft)
    x_dtft_buf[:N_base] = x_base
    X_dtft = np.fft.fft(x_dtft_buf, n_fft)
    dtft_freq_scaled = np.linspace(0, L_total, n_fft)
    
    # 3. Plotting layout
    fig, (ax_t, ax_f) = plt.subplots(1, 2, figsize=(13, 5))
    
    # --- TIME DOMAIN ---
    ax_t.plot(n_full, x_padded, 'r-', lw=1.2)
    ax_t.stem(n_full, x_padded, linefmt='C0-', markerfmt='C0o', basefmt='k-')
    ax_t.set_title(f"Input Signal ($N = {L_total}$)", fontsize=10, fontweight='bold')
    ax_t.set_xlabel("Time ($n$)")
    ax_t.grid(True, linestyle='--', alpha=0.6)
    
    # --- FREQUENCY DOMAIN ---
    ax_f.plot(dtft_freq_scaled, np.abs(X_dtft) * (N_base / L_total), 'r-', lw=1.2, alpha=0.7, label='Continuous Spectrum (DTFT)')
    ax_f.stem(freqs, np.abs(X_dft) * (N_base / L_total), 
              linefmt='C0-', markerfmt='C0o', basefmt='k-', label=f'DFT Points (L={L_total})')
    
    ax_f.set_title(f"DFT Magnitude (Total Points = {L_total})", fontsize=10, fontweight='bold')
    ax_f.set_xlabel("DFT Frequency Index ($k$)")
    ax_f.grid(True, linestyle='--', alpha=0.6)
    
    # ΜΕΤΑΦΟΡΑ ΥΠΟΜΝΗΜΑΤΟΣ: Έξω από το σχήμα (στο πάνω μέρος)
    ax_f.legend(loc='upper center', bbox_to_anchor=(0.5, 1.15), ncol=2, fontsize=9, frameon=False)
    
    plt.tight_layout()
    plt.show()

# Slider
len_slider = IntSlider(value=16, min=16, max=64, step=16, description='Length (L):')
out_plot = interactive_output(plot_zero_padding_clean, {'L_total': len_slider})

# HTML Guide
info_html = HTML("""<div style="padding: 15px; background-color: #f8f9fa; border: 1px solid #ddd; width: 260px;">
    <b>💡 Guide</b><ul><li>Left: Signal + Zeros</li><li>Right: DFT Bins on DTFT</li></ul>
    Zero padding interpolates the frequency grid.</div>""")

display(HBox([VBox([len_slider, out_plot]), info_html]))
```
