```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML

display(HTML("<style>.output_scroll { height: unset !important; }</style>"))

def symbolic_dtft_impulse_train():
    n, k, N = sp.symbols('n k N', integer=True, positive=True)
    omega = sp.symbols('omega', real=True)

    # ============================================================
    # 1. PERIODIC IMPULSE TRAIN
    # ============================================================

    x_one_period = sp.KroneckerDelta(n, 0)

    display(HTML(r"<b>Periodic impulse train:</b> $x[n]=\displaystyle\sum_{m=-\infty}^{\infty}\delta[n-mN]$"))
    display(HTML(r"<b>One period:</b> $x[n]=\delta[n],\qquad 0\leq n\leq N-1$"))

    # ============================================================
    # 2. DTFS COEFFICIENT
    # ============================================================

    alpha_k = sp.simplify(sp.KroneckerDelta(0, 0)/N)

    display(HTML(r"<b>DTFS coefficient:</b> $a_k=\frac{1}{N}\displaystyle\sum_{n=0}^{N-1}x[n]e^{-jk2\pi n/N}$"))
    display(HTML(r"<b>Symbolically calculated:</b> $a_k=$" + sp.latex(alpha_k) + "$"))

    # ============================================================
    # 3. DTFS -> DTFT
    # ============================================================

    dtft_coefficient = sp.simplify(2*sp.pi*alpha_k)

    display(HTML(r"<b>Impulse amplitude in frequency:</b> $2\pi a_k=$" + sp.latex(dtft_coefficient) + "$"))
    display(HTML(r"<b>Fourier transform:</b> $\boxed{X(e^{j\omega})=\frac{2\pi}{N}\displaystyle\sum_{k=-\infty}^{\infty}\delta\left(\omega-\frac{2\pi k}{N}\right)}$"))
    display(HTML(r"<b>Frequency spacing:</b> $\Delta\omega=\frac{2\pi}{N}$"))

    # ============================================================
    # 4. INTERACTIVE PLOTS
    # ============================================================

    def update_plots(N_val):
        clear_output(wait=True)

        n_min, n_max = -2*N_val, 2*N_val
        n_vals = np.arange(n_min, n_max+1)
        impulse_positions = n_vals[n_vals % N_val == 0]

        omega_min, omega_max = -2*np.pi, 2*np.pi
        k_min = int(np.ceil(omega_min*N_val/(2*np.pi)))
        k_max = int(np.floor(omega_max*N_val/(2*np.pi)))
        k_vals = np.arange(k_min, k_max+1)
        omega_impulses = 2*np.pi*k_vals/N_val
        amplitude = 2*np.pi/N_val

        fig, axes = plt.subplots(2, 1, figsize=(14, 5.5))

        # ========================================================
        # GRAPH 1: TIME DOMAIN
        # ========================================================

        axes[0].axhline(0, color='black', linewidth=0.8)

        for position in impulse_positions:
            axes[0].annotate('', xy=(position, 1), xytext=(position, 0), arrowprops=dict(arrowstyle='-|>', color='red', lw=2.0))

        axes[0].set_title(f'Periodic Impulse Train $x[n]$   (N = {N_val})', fontsize=12, fontweight='bold')
        axes[0].set_xlabel('n', fontsize=10)
        axes[0].set_ylabel('x[n]', fontsize=10)
        axes[0].set_xlim(n_min-1, n_max+1)
        axes[0].set_ylim(-0.05, 1.15)
        axes[0].grid(True, linestyle='--', alpha=0.6)
        axes[0].legend(['Impulse'], loc='center left', bbox_to_anchor=(1.02, 0.5))

        # ========================================================
        # GRAPH 2: FREQUENCY DOMAIN
        # ========================================================

        axes[1].axhline(0, color='black', linewidth=0.8)

        for position in omega_impulses:
            axes[1].annotate('', xy=(position/np.pi, amplitude), xytext=(position/np.pi, 0), arrowprops=dict(arrowstyle='-|>', color='red', lw=2.0))

        axes[1].set_title(r'Fourier Transform $X(e^{j\omega})$', fontsize=12, fontweight='bold')
        axes[1].set_xlabel(r'$\omega/\pi$', fontsize=10)
        axes[1].set_ylabel('Amplitude', fontsize=10)
        axes[1].set_xlim(-2, 2)
        axes[1].set_ylim(0, np.pi+0.2)
        axes[1].set_xticks([-2, -1.5, -1, -0.5, 0, 0.5, 1, 1.5, 2])
        axes[1].set_xticklabels(['-2π', '-3π/2', '-π', '-π/2', '0', 'π/2', 'π', '3π/2', '2π'])
        axes[1].grid(True, linestyle='--', alpha=0.6)
        axes[1].legend([r'$X(e^{j\omega})$'], loc='center left', bbox_to_anchor=(1.02, 0.5))

        plt.subplots_adjust(left=0.08, right=0.80, top=0.93, bottom=0.09, hspace=0.38)
        plt.show()

    # ============================================================
    # 5. SLIDER
    # ============================================================

    N_slider = widgets.IntSlider(value=8, min=2, max=20, step=1, description='Period N:', style={'description_width': 'initial'}, continuous_update=False)
    ui = widgets.VBox([N_slider])
    out = widgets.interactive_output(update_plots, {'N_val': N_slider})
    display(ui, out)

symbolic_dtft_impulse_train()
```
