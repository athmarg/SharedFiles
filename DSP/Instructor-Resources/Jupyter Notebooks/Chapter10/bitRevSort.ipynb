```python
import numpy as np
import matplotlib.pyplot as plt

# Sample data values (N = 16)
sample_values_16 = np.array([
    5.27, 8.48, 3.71, 0.33, 4.85, 1.70, 9.23, 0.45,
    5.17, 3.07, 1.09, 3.12, 8.16, 9.96, 1.12, 6.85
])

def bit_reversal_mapping(sample_vals):
    N = len(sample_vals)
    bits = int(np.log2(N))
    rev_indices = np.zeros(N, dtype=int)
    for i in range(N):
        rev = 0
        for j in range(bits):
            if (i & (1 << j)) != 0:
                rev |= (1 << (bits - 1 - j))
        rev_indices[i] = rev
    return rev_indices

# Calculations for N=8 (3 bits) and N=16 (4 bits)
N_8 = 8
samples_8 = sample_values_16[:N_8]
rev_8 = bit_reversal_mapping(samples_8)

N_16 = 16
samples_16 = sample_values_16
rev_16 = bit_reversal_mapping(samples_16)

# Create subplots (2 rows, 1 column) with side-by-side text
fig, axes = plt.subplots(2, 1, figsize=(14, 12))

# --- PLOT 1: 3 Bits (N = 8) ---
ax = axes[0]
for i in range(N_8):
    target = rev_8[i]
    ax.plot([0, 1], [i, target], 'o-', color='dodgerblue', lw=1.5, markersize=6)
    ax.text(-0.05, i, f"X[{i:02d}] = {samples_8[i]:.2f}", ha='right', va='center', fontsize=9)
    ax.text(1.05, target, f"X_rev[{target:02d}]", ha='left', va='center', fontsize=9, fontweight='bold', color='crimson')

ax.set_xlim(-0.5, 1.5)
ax.set_ylim(-1, N_8)
ax.axis('off')
ax.set_title("Bit-Reverse Sorting Flow: 3 Bits ($N = 8$)", fontsize=11, fontweight='bold', pad=10)

# Text output on the right for N=8
text_str_8 = "Rearranged Sequence for FFT Processing:\n"
for i in range(N_8):
    orig_idx = rev_8[i]
    text_str_8 += f"Index {orig_idx:02d} (Binary: {orig_idx:03b}) -> Value: {samples_8[orig_idx]:.2f}\n"

ax.text(1.35, N_8 - 0.5, text_str_8, fontsize=9, family='monospace', va='top', 
        bbox=dict(boxstyle='round,pad=0.5', facecolor='whitesmoke', edgecolor='lightgray', alpha=0.9))


# --- PLOT 2: 4 Bits (N = 16) ---
ax = axes[1]
for i in range(N_16):
    target = rev_16[i]
    ax.plot([0, 1], [i, target], 'o-', color='crimson', lw=1.5, markersize=5)
    ax.text(-0.05, i, f"X[{i:02d}] = {samples_16[i]:.2f}", ha='right', va='center', fontsize=9)
    ax.text(1.05, target, f"X_rev[{target:02d}]", ha='left', va='center', fontsize=9, fontweight='bold', color='dodgerblue')

ax.set_xlim(-0.5, 1.5)
ax.set_ylim(-1, N_16)
ax.axis('off')
ax.set_title("Bit-Reverse Sorting Flow: 4 Bits ($N = 16$)", fontsize=11, fontweight='bold', pad=10)

# Text output on the right for N=16
text_str_16 = "Rearranged Sequence for FFT Processing:\n"
for i in range(N_16):
    orig_idx = rev_16[i]
    text_str_16 += f"Index {orig_idx:02d} (Binary: {orig_idx:04b}) -> Value: {samples_16[orig_idx]:.2f}\n"

ax.text(1.35, N_16 - 0.5, text_str_16, fontsize=9, family='monospace', va='top', 
        bbox=dict(boxstyle='round,pad=0.5', facecolor='whitesmoke', edgecolor='lightgray', alpha=0.9))

plt.tight_layout()
plt.show()
```
