```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Parameters definition
N = 16        # DFT length
k0 = 3        # Frequency bin index (must be an integer: 0 <= k0 <= N-1)

# 2. Time-domain signal: Complex exponential x[n] = exp(j * 2*pi*k0*n / N)
n = np.arange(N)
x = np.exp(1j * 2 * np.pi * k0 * n / N)

# 3. Discrete Fourier Transform (DFT) computation based on the analytical solution
# The solution proves that X[k] = N * delta[k - k0]
k = np.arange(N)
X_k = np.zeros(N, dtype=complex)
X_k[k0] = N  # Impulse at k = k0

# Clean phase values where magnitude is essentially zero to avoid float noise
phase_cleaned = np.angle(X_k)
phase_cleaned[np.abs(X_k) < 1e-10] = 0.0

# 4. Plotting: Time Domain, DFT Magnitude, and DFT Phase
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(16, 4.5))
fig.suptitle(rf'Complex DFT of a Harmonic Signal: $x[n] = e^{{j 2\pi ({k0}) n / {N}}}$ ($N={N}$)', fontsize=13, fontweight='bold')

# --- Plot 1: Time Domain Signal ---
# Plotting both Real (cosine) and Imaginary (sine) parts of the complex signal
ax1.stem(n, np.real(x), basefmt="k-", linefmt='C0-', markerfmt='C0o', label=r'Real Part ($\cos$)')
ax1.stem(n, np.imag(x), basefmt="k-", linefmt='C1--', markerfmt='C1x', label=r'Imag Part ($\sin$)')
ax1.set_title(r'Time Domain: $x[n]$', fontsize=11, fontweight='bold')
ax1.set_xlabel(r'Time Index $n$', fontsize=10)
ax1.set_ylabel(r'Amplitude', fontsize=10)
ax1.set_xlim(-0.5, N - 0.5)
ax1.set_ylim(-1.3, 1.8) # Extra space at the top for the legend
ax1.grid(True, linestyle='--', alpha=0.5)
ax1.legend(loc='upper right', fontsize=9)

# --- Plot 2: DFT Magnitude ---
ax2.stem(k, np.abs(X_k), basefmt="k-", linefmt='C3-', markerfmt='C3o')
ax2.set_title(r'Frequency Domain: Magnitude $|\mathcal{X}[k]|$', fontsize=11, fontweight='bold')
ax2.set_xlabel(r'Frequency Index $k$', fontsize=10)
ax2.set_ylabel(r'Magnitude', fontsize=10)
ax2.set_xlim(-0.5, N - 0.5)
ax2.set_ylim(-1, N + 2)
ax2.grid(True, linestyle='--', alpha=0.5)

# --- Plot 3: DFT Phase (Cleaned) ---
ax3.stem(k, phase_cleaned, basefmt="k-", linefmt='C2-', markerfmt='C2o')
ax3.set_title(r'Frequency Domain: Phase $\angle \mathcal{X}[k]$ (rad)', fontsize=11, fontweight='bold')
ax3.set_xlabel(r'Frequency Index $k$', fontsize=10)
ax3.set_ylabel(r'Phase (radians)', fontsize=10)
ax3.set_xlim(-0.5, N - 0.5)
ax3.set_ylim(-0.5, 0.5)
ax3.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
