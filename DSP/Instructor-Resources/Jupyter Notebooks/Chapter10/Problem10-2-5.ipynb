```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, HTML

display(HTML("""
<script>
    if (window.MathJax) {
        MathJax.Hub.Config({
            "HTML-CSS": { scale: 130 },
            SVG: { scale: 130 }
        });
    }
</script>
<style>
    .rendered_html math { font-size: 1.3em !important; }
</style>
"""))

sp.init_printing(use_latex='mathjax')

def plot_spectra_and_parseval():
    A_val = 1
    L_val = 5
    N_val = 10
    
    n, k = sp.symbols('n k', integer=True)
    
    a_0 = A_val * L_val / N_val
    a_k_nonzero = (A_val / N_val) * sp.exp(-sp.I * sp.pi * k * (L_val - 1) / N_val) * (
        sp.sin(sp.pi * k * L_val / N_val) / sp.sin(sp.pi * k / N_val)
    )
    
    a_k_final = sp.Piecewise(
        (a_0, sp.Eq(k, 0)),
        (a_k_nonzero, True)
    )

    k_vals = np.arange(N_val)
    ak_vals = np.array([complex(a_k_final.subs(k, int(r)).evalf()) for r in k_vals])
    
    amplitudes = np.abs(ak_vals)
    phases = np.angle(ak_vals)
    power_density = amplitudes**2

    display(HTML(f"<h3 style='color: #2c3e50;'>1. Numerical DTFS Coefficients for $A={A_val}, L={L_val}, N={N_val}$:</h3>"))
    
    # Μειωμένο ύψος γραφημάτων (από 10 σε 5.5) για να μην είναι τεράστια
    fig, axes = plt.subplots(3, 1, figsize=(8, 5.5), sharex=True)
    
    # Amplitude Spectrum
    axes[0].stem(k_vals, amplitudes, basefmt="none")
    axes[0].set_ylabel(r'$|a_k|$', fontsize=10)
    axes[0].set_title('Amplitude Spectrum', fontsize=11)
    axes[0].grid(True, linestyle='--', alpha=0.6)
    
    # Phase Spectrum
    axes[1].stem(k_vals, phases, basefmt="none")
    axes[1].set_ylabel(r'$\angle a_k$ (rad)', fontsize=10)
    axes[1].set_title('Phase Spectrum', fontsize=11)
    axes[1].grid(True, linestyle='--', alpha=0.6)
    
    # Power Density Spectrum
    axes[2].stem(k_vals, power_density, basefmt="none")
    axes[2].set_xlabel('Frequency Index $k$', fontsize=10)
    axes[2].set_ylabel(r'$|a_k|^2$', fontsize=10)
    axes[2].set_title('Power Density Spectrum', fontsize=11)
    axes[2].grid(True, linestyle='--', alpha=0.6)
    
    plt.xticks(k_vals)
    plt.tight_layout()
    plt.show()

    lhs_signal_energy = sum((A_val**2 if n_val < L_val else 0) for n_val in range(N_val)) / N_val
    rhs_spectrum_energy = np.sum(power_density)

    display(HTML("<h3 style='color: #2c3e50;'>2. Parseval's Identity Verification:</h3>"))
    display(HTML(f"<b>Time-Domain Energy Expression $\\left(\\frac{{1}}{{N}}\\sum |x[n]|^2\\right)$:</b> {lhs_signal_energy}"))
    display(HTML(f"<b>Frequency-Domain Energy Expression $\\left(\\sum |a_k|^2\\right)$:</b> {rhs_spectrum_energy}"))

    
plot_spectra_and_parseval()
```
