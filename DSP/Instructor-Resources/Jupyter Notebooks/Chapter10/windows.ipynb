```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import windows
from ipywidgets import Checkbox, VBox, HBox, Layout, interactive_output
from IPython.display import display

# Fixed window length parameter
M = 128

def plot_windows_normalized(rect, hamm, hann, black, kais, bart):
    plt.close('all') # Close previous figure for a clean refresh
    plt.figure(figsize=(10, 5))
    
    # Normalized x-axis from 0 to 1 for each window
    x_norm = np.linspace(0, 1, M)
    
    # Dictionary of window functions with user selection states
    win_list = {
        'Rectangular':  (rect, windows.boxcar(M)),
        'Hamming':      (hamm, windows.hamming(M)),
        'von Hann':     (hann, windows.hann(M)),
        'Blackman':     (black, windows.blackman(M)),
        'Kaiser (β=5)': (kais, windows.kaiser(M, beta=5.0)),
        'Bartlett':     (bart, windows.bartlett(M))
    }
    
    any_active = False
    
    # Plotting only the selected windows
    for name, (is_active, win) in win_list.items():
        if is_active:
            any_active = True
            plt.plot(x_norm, win, label=name, lw=2)
        
    plt.title(f'Normalized Comparison of Window Functions (Length M = {M})', fontsize=12, fontweight='bold')
    plt.xlabel('Normalized Time ($n/M$)', fontsize=10)
    plt.ylabel('Magnitude', fontsize=10)
    plt.grid(True, linestyle='--', alpha=0.7)
    
    # Fixed scaling
    plt.xlim(0, 1)
    plt.ylim(0, 1.1)
    
    if any_active:
        # Legend outside the plot on the right
        plt.legend(bbox_to_anchor=(1.02, 1), loc='upper left', borderaxespad=0.)
    else:
        plt.text(0.5, 0.5, 'No window selected.', 
                horizontalalignment='center', verticalalignment='center', 
                fontsize=10, color='#666666', bbox=dict(boxstyle='round,pad=0.5', facecolor='#f8f9fa'))
    
    plt.tight_layout()
    plt.show()

# Checkboxes with tightly compressed horizontal spacing
cb_rect  = Checkbox(value=True, description='Rectangular', layout=Layout(width='auto', margin='0px -12px 0px 0px'))
cb_hamm  = Checkbox(value=True, description='Hamming', layout=Layout(width='auto', margin='0px -12px 0px 0px'))
cb_hann  = Checkbox(value=True, description='von Hann', layout=Layout(width='auto', margin='0px -12px 0px 0px'))
cb_black = Checkbox(value=True, description='Blackman', layout=Layout(width='auto', margin='0px -12px 0px 0px'))
cb_kais  = Checkbox(value=True, description='Kaiser (β=5)', layout=Layout(width='auto', margin='0px -12px 0px 0px'))
cb_bart  = Checkbox(value=True, description='Bartlett', layout=Layout(width='auto', margin='0px 0px 0px 0px'))

# Horizontal bar containing all checkboxes placed between the plot and nothing else
checkbox_bar = HBox([cb_rect, cb_hamm, cb_hann, cb_black, cb_kais, cb_bart], layout=Layout(margin='5px 0px 5px 0px'))

# Interactive output binding widgets to the plotting function
out = interactive_output(
    plot_windows_normalized, 
    {
        'rect': cb_rect, 'hamm': cb_hamm, 'hann': cb_hann, 
        'black': cb_black, 'kais': cb_kais, 'bart': cb_bart
    }
)

# Final dashboard layout: Plot -> Checkboxes
display(VBox([out, checkbox_bar]))
```
