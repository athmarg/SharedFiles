```python
import sympy as sp
from IPython.display import display, HTML

display(HTML("""
<script>
    if (window.MathJax) {
        MathJax.Hub.Config({
            "HTML-CSS": { scale: 130 },
            SVG: { scale: 130 }
        });
    }
</script>
<style>
    .rendered_html math { font-size: 1.3em !important; }
</style>
"""))

sp.init_printing(use_latex='mathjax')

def simplified_symbolic_periodic_rectangular():
    n, k, N, L, A = sp.symbols('n k N L A', integer=True, positive=True)
    j = sp.I

    term = A * sp.exp(-j * 2 * sp.pi * k * n / N)
    summation_expr = sp.Sum(term, (n, 0, L - 1))
    a_k = summation_expr / N
    evaluated = a_k.doit()

    a_0 = sp.simplify(a_k.doit().subs(k, 0))
    if a_0.has(sp.nan, sp.zoo, sp.oo) or a_0 == sp.nan:
        a_0 = sp.simplify((A / N) * sp.Sum(1, (n, 0, L - 1)).doit())

    a_k_nonzero_simplified = sp.trigsimp(sp.cancel(evaluated))

    a_k_final = sp.Piecewise(
        (a_0, sp.Eq(k, 0)),
        (a_k_nonzero_simplified, True)
    )

    display(HTML("<h3 style='color: #2c3e50;'>1. General Symbolic DTFS Coefficient ($a_k$):</h3>"))
    display(sp.Eq(sp.Symbol('a_k'), a_k_final))

    L_val, N_val = 5, 10
    a_k_510 = sp.trigsimp(a_k_final.subs({L: L_val, N: N_val}))

    display(HTML(f"<h3 style='color: #2c3e50;'>2. DTFS Coefficients for $L={L_val}$, $N={N_val}$:</h3>"))
    display(sp.Eq(sp.Symbol('a_k'), a_k_510))

    display(HTML("""
    <div style="border: 1px solid #bdc3c7; padding: 12px; border-radius: 6px; background-color: #f8f9fa; font-family: sans-serif; font-size: 14px; color: #2c3e50;">
        <strong>Summary:</strong> Αυτή είναι η
    </div>
    """))

simplified_symbolic_periodic_rectangular()
```
