```python
# ==============================================================================
# Output of an LTI System Using the Eigenfunction Property
#
# Given:
# h[n] = (1/2)^n cos(pi*n/2) u[n]
# x[n] = cos(pi*n/2)
#
# Method: Complex Exponential Eigenfunctions
# Target Publication: Springer
# ==============================================================================

import sympy as sp
from IPython.display import display, Markdown

sp.init_printing(use_unicode=True)

# ------------------------------------------------------------------------------
# 1. Symbolic Definitions
# ------------------------------------------------------------------------------

omega = sp.Symbol('omega', real=True)
n = sp.Symbol('n', integer=True)
I = sp.I

H = sp.Function('H')
x = sp.Function('x')
y = sp.Function('y')

# ------------------------------------------------------------------------------
# 2. Given Impulse Response
# ------------------------------------------------------------------------------

h_n = (sp.Rational(1, 2))**n * sp.cos(sp.pi * n / 2)

# ------------------------------------------------------------------------------
# 3. Convert the Cosine Symbolically to Exponential Form
# ------------------------------------------------------------------------------

h_exp = sp.expand(h_n.rewrite(sp.exp))

# ------------------------------------------------------------------------------
# 4. Symbolic DTFT of the Impulse Response
#
# The two geometric series are obtained from h[n] itself.
# Nothing from the textbook solution is supplied.
# ------------------------------------------------------------------------------

h_terms = sp.Add.make_args(h_exp)

H_omega = 0

for term in h_terms:
    H_omega += sp.summation(term * sp.exp(-I * omega * n), (n, 0, sp.oo))

H_omega = sp.factor(sp.simplify(H_omega))

display(Markdown("### **Frequency Response**"))
display(sp.Eq(H(sp.exp(I * omega)), H_omega))

# ------------------------------------------------------------------------------
# 5. Input Signal in Exponential Form
# ------------------------------------------------------------------------------

x_n = sp.cos(sp.pi * n / 2)
x_exp = sp.expand(x_n.rewrite(sp.exp))
x_exp = sp.collect(x_exp, sp.exp(I * sp.pi * n / 2))

display(Markdown("### **Input Signal in Exponential Form**"))
display(sp.Eq(x(n), x_exp))

# ------------------------------------------------------------------------------
# 6. Define the Two Eigenfunctions
# ------------------------------------------------------------------------------

omega_1 = sp.pi / 2
omega_2 = -sp.pi / 2

x_1 = sp.exp(I * omega_1 * n)
x_2 = sp.exp(I * omega_2 * n)

# ------------------------------------------------------------------------------
# 7. Calculate the Two Eigenvalues Symbolically
# ------------------------------------------------------------------------------

lambda_1 = sp.simplify(H_omega.subs(omega, omega_1))
lambda_2 = sp.simplify(H_omega.subs(omega, omega_2))

# ------------------------------------------------------------------------------
# 8. Responses to the Two Eigenfunctions
# ------------------------------------------------------------------------------

y_1 = sp.simplify(lambda_1 * x_1)
y_2 = sp.simplify(lambda_2 * x_2)

# ------------------------------------------------------------------------------
# 9. Superposition
# ------------------------------------------------------------------------------

y_exp = sp.simplify((y_1 + y_2) / 2)

# ------------------------------------------------------------------------------
# 10. Convert the Conjugate Exponential Pair to Cosine
# ------------------------------------------------------------------------------

theta = sp.pi * n / 2

y_cos = sp.simplify((lambda_1 + lambda_2) / 2 * sp.cos(theta))

display(Markdown("### **Output Signal**"))
display(sp.Eq(y(n), y_cos))
```
