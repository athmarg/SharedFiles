```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, FloatSlider, HBox, VBox, Layout, HTML, GridBox
from IPython.display import display

# ============================================================
# INTERACTIVE FUNCTION
# ============================================================

def plot_rc_white_noise(N0=2.0, RC=1.0):

    # --------------------------------------------------------
    # TIME AXES
    # --------------------------------------------------------

    t = np.linspace(0.0, 8.0, 1200)
    tau = np.linspace(-8.0, 8.0, 2000)

    # ========================================================
    # IMPULSE RESPONSE
    #
    # h(t) = (1/RC) exp(-t/RC) u(t)
    # ========================================================

    h = (1.0 / RC) * np.exp(-t / RC)

    # ========================================================
    # CROSS-CORRELATION Rxy(tau)
    #
    # Rxy(tau) = (N0/2) h(tau), tau >= 0
    # ========================================================

    Rxy = np.zeros_like(tau)

    positive_tau = tau >= 0.0

    Rxy[positive_tau] = (N0 / (2.0 * RC)) * np.exp(-tau[positive_tau] / RC)

    # ========================================================
    # CROSS-CORRELATION Ryx(tau)
    #
    # Ryx(tau) = (N0/2) h(-tau), tau <= 0
    # ========================================================

    Ryx = np.zeros_like(tau)

    negative_tau = tau <= 0.0

    Ryx[negative_tau] = (N0 / (2.0 * RC)) * np.exp(tau[negative_tau] / RC)

    # ========================================================
    # OUTPUT AUTOCORRELATION
    #
    # Ryy(tau) = N0/(4RC) exp(-|tau|/RC)
    # ========================================================

    Ryy = (N0 / (4.0 * RC)) * np.exp(-np.abs(tau) / RC)

    # ========================================================
    # OUTPUT MEAN-SQUARE VALUE
    #
    # y^2_bar = Ryy(0) = N0/(4RC)
    # ========================================================

    mean_square_theoretical = N0 / (4.0 * RC)

    # ========================================================
    # FIGURE
    # ========================================================

    fig, axes = plt.subplots(2, 2, figsize=(10.8, 6.8))

    ax1 = axes[0, 0]
    ax2 = axes[0, 1]
    ax3 = axes[1, 0]
    ax4 = axes[1, 1]

    # ========================================================
    # GRAPH 1:
    # IMPULSE RESPONSE
    # ========================================================

    ax1.plot(t, h, linewidth=2)

    ax1.axhline(0, color='k', linewidth=0.8)

    ax1.set_xlim(0, 8)

    ax1.set_ylim(0, max(1.15 / RC, 1.15 * np.max(h)))

    ax1.set_xlabel('Time t', fontsize=11)

    ax1.set_ylabel('h(t)', fontsize=11)

    ax1.set_title('RC Filter Impulse Response', fontsize=12, pad=8)

    ax1.tick_params(axis='both', labelsize=9)

    ax1.grid(True, linestyle=':', alpha=0.5)

    # ========================================================
    # GRAPH 2:
    # Rxy(tau)
    # ========================================================

    ax2.plot(tau, Rxy, linewidth=2)

    ax2.axhline(0, color='k', linewidth=0.8)

    ax2.axvline(0, color='k', linestyle=':', linewidth=0.8)

    ax2.set_xlim(-8, 8)

    ax2.set_ylim(-0.05 * max(1.0, np.max(Rxy)), 1.15 * max(1.0, np.max(Rxy)))

    ax2.set_xlabel('Lag τ', fontsize=11)

    ax2.set_ylabel('Rₓᵧ(τ)', fontsize=11)

    ax2.set_title('Input-Output Cross-Correlation', fontsize=12, pad=8)

    ax2.tick_params(axis='both', labelsize=9)

    ax2.grid(True, linestyle=':', alpha=0.5)

    # ========================================================
    # GRAPH 3:
    # Ryx(tau)
    # ========================================================

    ax3.plot(tau, Ryx, linewidth=2)

    ax3.axhline(0, color='k', linewidth=0.8)

    ax3.axvline(0, color='k', linestyle=':', linewidth=0.8)

    ax3.set_xlim(-8, 8)

    ax3.set_ylim(-0.05 * max(1.0, np.max(Ryx)), 1.15 * max(1.0, np.max(Ryx)))

    ax3.set_xlabel('Lag τ', fontsize=11)

    ax3.set_ylabel('Rᵧₓ(τ)', fontsize=11)

    ax3.set_title('Output-Input Cross-Correlation', fontsize=12, pad=8)

    ax3.tick_params(axis='both', labelsize=9)

    ax3.grid(True, linestyle=':', alpha=0.5)

    # ========================================================
    # GRAPH 4:
    # Ryy(tau)
    # ========================================================

    ax4.plot(tau, Ryy, linewidth=2)

    ax4.axhline(0, color='k', linewidth=0.8)

    ax4.axvline(0, color='k', linestyle=':', linewidth=0.8)

    ax4.axhline(mean_square_theoretical, linestyle='--', linewidth=1.4, label=f'Rᵧᵧ(0) = {mean_square_theoretical:.3f}')

    ax4.set_xlim(-8, 8)

    ax4.set_ylim(0, 1.15 * max(1.0, np.max(Ryy)))

    ax4.set_xlabel('Lag τ', fontsize=11)

    ax4.set_ylabel('Rᵧᵧ(τ)', fontsize=11)

    ax4.set_title('Output Autocorrelation', fontsize=12, pad=8)

    ax4.tick_params(axis='both', labelsize=9)

    ax4.grid(True, linestyle=':', alpha=0.5)

    ax4.legend(loc='upper center', bbox_to_anchor=(0.5, -0.20), ncol=1, fontsize=8, frameon=True)

    # ========================================================
    # FIGURE SPACING
    # ========================================================

    plt.subplots_adjust(left=0.08, right=0.97, top=0.92, bottom=0.13, hspace=0.48, wspace=0.30)

    plt.show()

# ============================================================
# SLIDERS
# ============================================================

slider_style = {'description_width': '0px'}

slider_layout = Layout(width='150px')

N0_slider = FloatSlider(min=0.5, max=10.0, step=0.5, value=2.0, description=' ', continuous_update=True, readout=False, style=slider_style, layout=slider_layout)

RC_slider = FloatSlider(min=0.2, max=4.0, step=0.1, value=1.0, description=' ', continuous_update=True, readout=False, style=slider_style, layout=slider_layout)

# ============================================================
# CURRENT VALUE LABELS
# ============================================================

N0_value = HTML('<div style="font-family:Arial; font-size:14px; font-weight:bold; color:#0b3d91;">2.0</div>')

RC_value = HTML('<div style="font-family:Arial; font-size:14px; font-weight:bold; color:#0b3d91;">1.0</div>')

# ============================================================
# UPDATE CURRENT VALUES
# ============================================================

def update_N0_value(change):

    N0_value.value = f'<div style="font-family:Arial; font-size:14px; font-weight:bold; color:#0b3d91;">{N0_slider.value:.1f}</div>'

def update_RC_value(change):

    RC_value.value = f'<div style="font-family:Arial; font-size:14px; font-weight:bold; color:#0b3d91;">{RC_slider.value:.1f}</div>'

N0_slider.observe(update_N0_value, names='value')

RC_slider.observe(update_RC_value, names='value')

# ============================================================
# INTERACTIVE OBJECT
# ============================================================

widget_plot = interactive(plot_rc_white_noise, N0=N0_slider, RC=RC_slider)

# ============================================================
# DOCUMENTATION
# ============================================================

theory_html = HTML("""
<div style="
    font-family:Arial, sans-serif;
    font-size:16px;
    line-height:1.40;
    width:650px;
    padding:14px 16px;
    border:1px solid #b8cbe8;
    border-radius:8px;
    background:#fbfdff;
    box-sizing:border-box;
">

<div style="
    font-size:22px;
    font-weight:bold;
    color:#12388c;
    margin-bottom:9px;
">
White Noise Through an RC Low-Pass Filter
</div>

<div style="margin-bottom:6px;">
<b>Input process:</b> stationary white noise with
Rₓₓ(τ) = (N₀/2)δ(τ).
</div>

<div style="margin-bottom:6px;">
<b>RC filter:</b>
h(t) = (1/RC)e<sup>-t/RC</sup>u(t).
</div>

<div style="margin-bottom:6px;">
<b>Cross-correlations:</b>
Rₓᵧ(τ) = (N₀/2)h(τ) and
Rᵧₓ(τ) = (N₀/2)h(-τ).
</div>

<div style="margin-bottom:6px;">
<b>Output autocorrelation:</b>
Rᵧᵧ(τ) = [N₀/(4RC)]e<sup>-|τ|/RC</sup>.
</div>

<div>
<b>This notebook:</b> verifies the statistical input-output relations for white noise passing through a first-order RC low-pass system.
</div>

</div>
""")

# ============================================================
# CONTROL LABELS
# ============================================================

N0_label = HTML('<div style="font-family:Arial; font-size:14px; font-weight:bold; white-space:nowrap;">Noise level N₀:</div>')

RC_label = HTML('<div style="font-family:Arial; font-size:14px; font-weight:bold; white-space:nowrap;">Time constant RC:</div>')

# ============================================================
# CONTROLS GRID
# LABEL | SLIDER | CURRENT VALUE
# ============================================================

controls_grid = GridBox(
    children=[
        N0_label, N0_slider, N0_value,
        RC_label, RC_slider, RC_value
    ],
    layout=Layout(
        width='430px',
        grid_template_columns='145px 150px 55px',
        grid_template_rows='34px 34px',
        grid_gap='4px 6px',
        align_items='center',
        overflow='hidden'
    )
)

# ============================================================
# CONTROLS CARD
# ============================================================

controls_card = VBox(
    [
        HTML("""
        <div style="
            font-family:Arial;
            font-size:17px;
            font-weight:bold;
            color:#12388c;
            margin-bottom:6px;
        ">
        Parameters
        </div>
        """),
        controls_grid
    ],
    layout=Layout(
        width='450px',
        min_width='450px',
        padding='14px 14px',
        border='1px solid #d2d2d2',
        overflow='hidden'
    )
)

# ============================================================
# TOP AREA
# ============================================================

top_layout = HBox(
    [
        theory_html,
        controls_card
    ],
    layout=Layout(
        width='1120px',
        align_items='stretch',
        justify_content='space-between'
    )
)

# ============================================================
# MEAN-SQUARE RESULT
# ============================================================

result_html = HTML()

def update_result(change=None):

    value = N0_slider.value / (4.0 * RC_slider.value)

    result_html.value = f"""
    <div style="
        font-family:Arial, sans-serif;
        font-size:15px;
        line-height:1.40;
        width:1120px;
        padding:11px 16px;
        border:1px solid #d7c38d;
        background:#fffbed;
        box-sizing:border-box;
    ">

    <span style="
        font-size:18px;
        font-weight:bold;
        color:#775500;
    ">
    Mean-square verification:
    </span>

    &nbsp;&nbsp;

    <span style="
        font-family:serif;
        font-size:20px;
        font-style:italic;
    ">
    Rᵧᵧ(0) = y² = N₀/(4RC) = {value:.4f}
    </span>

    </div>
    """

N0_slider.observe(update_result, names='value')

RC_slider.observe(update_result, names='value')

update_result()

# ============================================================
# INTERPRETATION
# ============================================================

interpretation_html = HTML("""
<div style="
    font-family:Arial, sans-serif;
    font-size:15px;
    line-height:1.42;
    width:1120px;
    padding:12px 16px;
    border:1px solid #c8dfce;
    background:#f8fcf9;
    box-sizing:border-box;
">

<div style="
    font-size:18px;
    font-weight:bold;
    color:#197b35;
    margin-bottom:7px;
">
Interpretation of the Results
</div>

<div style="margin-bottom:5px;">
<b>Impulse response:</b> increasing RC makes the RC filter respond more slowly and increases its memory.
</div>

<div style="margin-bottom:5px;">
<b>Cross-correlations:</b> Rₓᵧ(τ) follows the causal impulse response h(τ), whereas Rᵧₓ(τ) is its time-reversed counterpart.
</div>

<div style="margin-bottom:5px;">
<b>Output autocorrelation:</b> although the input white noise is uncorrelated for nonzero lags, the RC filter produces an exponentially correlated output.
</div>

<div>
<b>Mean-square value:</b> the peak Rᵧᵧ(0) equals the average output power N₀/(4RC).
</div>

</div>
""")

# ============================================================
# COMPLETE NOTEBOOK
# ============================================================

main_layout = VBox(
    [
        top_layout,
        widget_plot.children[-1],
        result_html,
        interpretation_html
    ],
    layout=Layout(
        width='1120px',
        overflow='hidden'
    )
)

display(main_layout)
```
