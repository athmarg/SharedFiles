```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, FloatSlider, RadioButtons, HBox, VBox, Layout, HTML, GridBox
from IPython.display import display

# ============================================================
# CSS FOR HORIZONTAL RADIO BUTTONS
# ============================================================

display(HTML("""
<style>
.horizontal-radio .widget-radio-box {
    display: flex !important;
    flex-direction: row !important;
    flex-wrap: nowrap !important;
    align-items: center !important;
    gap: 18px !important;
}
.horizontal-radio .widget-radio-box label {
    margin: 0 !important;
    white-space: nowrap !important;
}
.horizontal-radio > label {
    display: none !important;
}
</style>
"""))

# ============================================================
# RANDOM PROCESS PARAMETERS
# ============================================================

np.random.seed(91)

M_max = 100
N_max = 800

W = np.random.randn(M_max, N_max)

# ============================================================
# INTERACTIVE FUNCTION
# ============================================================

def plot_mean_propagation(system_type='Low-pass', mu_x=1.0, M=30, N=400, alpha=0.85):

    X = mu_x + W[:M, :N]

    Y = np.zeros_like(X)

    # --------------------------------------------------------
    # LOW-PASS SYSTEM
    # --------------------------------------------------------

    if system_type == 'Low-pass':

        for m in range(M):

            Y[m, 0] = X[m, 0]

            for n in range(1, N):

                Y[m, n] = (1.0 - alpha) * X[m, n] + alpha * Y[m, n - 1]

        H0 = 1.0

        system_name = 'First-Order Low-Pass Filter'

    # --------------------------------------------------------
    # HIGH-PASS SYSTEM
    # --------------------------------------------------------

    else:

        for m in range(M):

            Y[m, 0] = 0.0

            for n in range(1, N):

                Y[m, n] = X[m, n] - X[m, n - 1]

        H0 = 0.0

        system_name = 'First-Difference High-Pass Filter'

    # --------------------------------------------------------
    # ENSEMBLE MEANS
    # --------------------------------------------------------

    mean_input = np.mean(X, axis=0)

    mean_output = np.mean(Y, axis=0)

    theoretical_output_mean = mu_x * H0

    # --------------------------------------------------------
    # FIGURE
    # --------------------------------------------------------

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(11.0, 4.8))

    number_to_show = min(12, M)

    time_axis = np.arange(N)

    # ========================================================
    # INPUT RANDOM PROCESS
    # ========================================================

    for m in range(number_to_show):

        ax1.plot(time_axis, X[m, :], linewidth=0.65, alpha=0.25)

    ax1.plot(time_axis, mean_input, linewidth=2.4, color='gold', label='Estimated ensemble mean')

    ax1.axhline(mu_x, color='tab:blue', linestyle='--', linewidth=1.8, label=f'Theoretical mean = {mu_x:.2f}')

    ax1.set_xlim(0, N - 1)

    ax1.set_ylim(-4.0, 4.0)

    ax1.set_xlabel('Time index n', fontsize=12)

    ax1.set_ylabel('x[n]', fontsize=12)

    ax1.set_title('Input Random Process', fontsize=15, fontweight='bold', pad=12)

    ax1.tick_params(axis='both', labelsize=10)

    ax1.grid(True, linestyle=':', alpha=0.45)

    ax1.legend(loc='upper center', bbox_to_anchor=(0.5, -0.18), ncol=1, fontsize=9, frameon=True)

    # ========================================================
    # OUTPUT RANDOM PROCESS
    # ========================================================

    for m in range(number_to_show):

        ax2.plot(time_axis, Y[m, :], linewidth=0.65, alpha=0.25)

    ax2.plot(time_axis, mean_output, linewidth=2.4, color='gold', label='Estimated ensemble mean')

    ax2.axhline(theoretical_output_mean, color='tab:blue', linestyle='--', linewidth=1.8, label=f'Theoretical mean = {theoretical_output_mean:.2f}')

    ax2.set_xlim(0, N - 1)

    ax2.set_ylim(-4.0, 4.0)

    ax2.set_xlabel('Time index n', fontsize=12)

    ax2.set_ylabel('y[n]', fontsize=12)

    ax2.set_title(system_name, fontsize=15, fontweight='bold', pad=12)

    ax2.tick_params(axis='both', labelsize=10)

    ax2.grid(True, linestyle=':', alpha=0.45)

    ax2.legend(loc='upper center', bbox_to_anchor=(0.5, -0.18), ncol=1, fontsize=9, frameon=True)

    plt.subplots_adjust(left=0.07, right=0.98, top=0.90, bottom=0.27, wspace=0.26)

    plt.show()

# ============================================================
# RADIO BUTTONS
# ============================================================

system_selector = RadioButtons(
    options=['Low-pass', 'High-pass'],
    value='Low-pass',
    description='',
    layout=Layout(width='180px')
)

system_selector.add_class('horizontal-radio')

# ============================================================
# SLIDERS
# ============================================================

slider_style = {
    'description_width': '0px'
}

slider_layout = Layout(
    width='180px'
)

mu_slider = FloatSlider(
    min=-2.0,
    max=2.0,
    step=0.1,
    value=1.0,
    description=' ',
    continuous_update=True,
    readout=False,
    style=slider_style,
    layout=slider_layout
)

M_slider = IntSlider(
    min=10,
    max=100,
    step=10,
    value=30,
    description=' ',
    continuous_update=True,
    readout=False,
    style=slider_style,
    layout=slider_layout
)

N_slider = IntSlider(
    min=100,
    max=800,
    step=50,
    value=400,
    description=' ',
    continuous_update=True,
    readout=False,
    style=slider_style,
    layout=slider_layout
)

alpha_slider = FloatSlider(
    min=0.10,
    max=0.95,
    step=0.05,
    value=0.85,
    description=' ',
    continuous_update=True,
    readout=False,
    style=slider_style,
    layout=slider_layout
)

# ============================================================
# CURRENT VALUE LABELS
# ============================================================

mu_value = HTML(
    '<div style="font-family:Arial; font-size:15px; font-weight:bold; color:#0b3d91;">1.00</div>'
)

M_value = HTML(
    '<div style="font-family:Arial; font-size:15px; font-weight:bold; color:#0b3d91;">30</div>'
)

N_value = HTML(
    '<div style="font-family:Arial; font-size:15px; font-weight:bold; color:#0b3d91;">400</div>'
)

alpha_value = HTML(
    '<div style="font-family:Arial; font-size:15px; font-weight:bold; color:#0b3d91;">0.85</div>'
)

empty_value = HTML(
    '<div></div>'
)

# ============================================================
# UPDATE CURRENT VALUES
# ============================================================

def update_mu_value(change):

    mu_value.value = f'<div style="font-family:Arial; font-size:15px; font-weight:bold; color:#0b3d91;">{mu_slider.value:.2f}</div>'

def update_M_value(change):

    M_value.value = f'<div style="font-family:Arial; font-size:15px; font-weight:bold; color:#0b3d91;">{M_slider.value}</div>'

def update_N_value(change):

    N_value.value = f'<div style="font-family:Arial; font-size:15px; font-weight:bold; color:#0b3d91;">{N_slider.value}</div>'

def update_alpha_value(change):

    alpha_value.value = f'<div style="font-family:Arial; font-size:15px; font-weight:bold; color:#0b3d91;">{alpha_slider.value:.2f}</div>'

mu_slider.observe(update_mu_value, names='value')

M_slider.observe(update_M_value, names='value')

N_slider.observe(update_N_value, names='value')

alpha_slider.observe(update_alpha_value, names='value')

# ============================================================
# ENABLE / DISABLE ALPHA
# ============================================================

def update_controls(change):

    alpha_slider.disabled = system_selector.value != 'Low-pass'

system_selector.observe(update_controls, names='value')

# ============================================================
# INTERACTIVE OBJECT
# ============================================================

widget_plot = interactive(
    plot_mean_propagation,
    system_type=system_selector,
    mu_x=mu_slider,
    M=M_slider,
    N=N_slider,
    alpha=alpha_slider
)

# ============================================================
# THEORY CARD
# ============================================================

theory_html = HTML("""
<div style="
    font-family:Arial, sans-serif;
    font-size:16px;
    line-height:1.45;
    width:610px;
    min-height:205px;
    padding:16px 18px;
    border:1px solid #b8cbe8;
    border-radius:9px;
    background:#fbfdff;
    box-sizing:border-box;
">

<div style="
    font-size:23px;
    font-weight:bold;
    color:#12388c;
    margin-bottom:15px;
">
Propagation of the Mean Through an LTI System
</div>

<div style="margin-bottom:14px;">
<span style="font-weight:bold; color:#12388c;">Input process:</span>
&nbsp; a WSS random process with constant mean μₓ.
</div>

<div style="margin-bottom:14px;">
<span style="font-weight:bold; color:#12388c;">LTI property:</span>
&nbsp; the mean of the output is the system response to the mean of the input.
</div>

<div style="
    display:flex;
    align-items:center;
    margin-bottom:14px;
">
<span style="font-weight:bold; color:#12388c;">
For a WSS input:
</span>

<span style="
    margin-left:18px;
    padding:7px 18px;
    border:1px solid #eadfa5;
    border-radius:7px;
    background:#fffbe6;
    font-family:serif;
    font-size:21px;
    font-style:italic;
">
μᵧ = μₓ H(0)
</span>
</div>

<div>
<span style="font-weight:bold; color:#12388c;">This notebook:</span>
&nbsp; compares a low-pass system with H(0) = 1 and a high-pass system with H(0) = 0.
</div>

</div>
""")

# ============================================================
# CONTROL LABELS
# ============================================================

system_label = HTML(
    '<div style="font-family:Arial; font-size:14px; font-weight:bold;">System:</div>'
)

mu_label = HTML(
    '<div style="font-family:Arial; font-size:14px; font-weight:bold;">Input mean μₓ:</div>'
)

M_label = HTML(
    '<div style="font-family:Arial; font-size:14px; font-weight:bold;">Realizations M:</div>'
)

N_label = HTML(
    '<div style="font-family:Arial; font-size:14px; font-weight:bold;">Samples N:</div>'
)

alpha_label = HTML(
    '<div style="font-family:Arial; font-size:14px; font-weight:bold;">LP parameter α:</div>'
)

# ============================================================
# COMPLETE CONTROL GRID
#
# IMPORTANT:
# SYSTEM RADIO BUTTONS ARE IN THE SAME SECOND COLUMN
# AS ALL SLIDER CONTROLS
# ============================================================

controls_grid = GridBox(
    children=[
        system_label, system_selector, empty_value,
        mu_label, mu_slider, mu_value,
        M_label, M_slider, M_value,
        N_label, N_slider, N_value,
        alpha_label, alpha_slider, alpha_value
    ],
    layout=Layout(
        width='455px',
        grid_template_columns='125px 180px 60px',
        grid_template_rows='34px 34px 34px 34px 34px',
        grid_gap='5px 8px',
        align_items='center',
        overflow='hidden'
    )
)

# ============================================================
# CONTROLS CARD
# ============================================================

controls_card = VBox(
    [
        controls_grid,
        HTML(
            '<div style="font-family:Arial; font-size:12px; margin-top:3px; margin-left:133px;">'
            '(Used only for the low-pass system.)'
            '</div>'
        )
    ],
    layout=Layout(
        width='475px',
        min_width='475px',
        padding='14px 15px',
        border='1px solid #d2d2d2',
        margin='0px',
        overflow='hidden'
    )
)

# ============================================================
# TOP AREA
# ============================================================

top_layout = HBox(
    [
        theory_html,
        controls_card
    ],
    layout=Layout(
        width='1100px',
        align_items='stretch',
        justify_content='space-between'
    )
)

# ============================================================
# PLOT CARD
# ============================================================

plot_card = VBox(
    [
        widget_plot.children[-1]
    ],
    layout=Layout(
        width='1100px',
        padding='10px 4px 2px 4px',
        border='1px solid #dedede',
        margin='12px 0px 0px 0px'
    )
)

# ============================================================
# INTERPRETATION CARD
# ============================================================

interpretation_text = HTML("""
<div style="
    font-family:Arial, sans-serif;
    font-size:15px;
    line-height:1.45;
">

<div style="
    font-size:19px;
    font-weight:bold;
    color:#197b35;
    margin-bottom:7px;
">
Interpretation of the Results
</div>

<div style="margin-bottom:5px;">
• For the <b>low-pass system used in this notebook</b>, H(0) = 1, so the input mean passes unchanged to the output.
</div>

<div style="margin-bottom:5px;">
• For the <b>high-pass system used in this notebook</b>, H(0) = 0, so the DC component (and the input mean) is removed.
</div>

<div>
These conclusions apply to the <b> LTI systems used in this notebook</b>. In general, the output mean is determined by μᵧ = μₓ H(0).
</div>

</div>
""")

interpretation_formula = HTML("""
<div style="
    font-family:serif;
    font-size:27px;
    font-style:italic;
    color:#16813b;
    text-align:center;
    padding:14px 20px;
    border:1px solid #cde3d2;
    border-radius:7px;
    background:#f7fcf8;
">
μᵧ = μₓ H(0)
</div>
""")

interpretation_card = HBox(
    [
        interpretation_text,
        interpretation_formula
    ],
    layout=Layout(
        width='1100px',
        padding='14px 20px',
        margin='12px 0px 0px 0px',
        border='1px solid #c8dfce',
        align_items='center',
        justify_content='space-between'
    )
)

# ============================================================
# COMPLETE NOTEBOOK
# ============================================================

main_layout = VBox(
    [
        top_layout,
        plot_card,
        interpretation_card
    ],
    layout=Layout(
        width='1100px',
        overflow='hidden'
    )
)

display(main_layout)
```
