```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, RadioButtons, HBox, Layout, VBox, HTML
from IPython.display import display

# ============================================================
# RANDOM PROCESSES
# ============================================================

np.random.seed(12)

N = 400
n = np.arange(N)

# Stationary process: constant mean and variance
x_stationary = 0.55 * np.random.randn(N)

# Non-stationary process: time-varying mean
mean_component = 1.2 * np.sin(2 * np.pi * n / 260)
x_mean = mean_component + 0.55 * np.random.randn(N)

# Non-stationary process: time-varying variance
sigma_component = 0.25 + 1.0 * n / (N - 1)
x_variance = sigma_component * np.random.randn(N)

# ============================================================
# INTERACTIVE FUNCTION
# ============================================================

def plot_stationarity(process_type='Stationary', n0=200, L=80):

    if process_type == 'Stationary':
        x = x_stationary
        process_name = 'Stationary Process'

    elif process_type == 'Time-varying mean':
        x = x_mean
        process_name = 'Non-Stationary Process: Time-Varying Mean'

    else:
        x = x_variance
        process_name = 'Non-Stationary Process: Time-Varying Variance'

    # --------------------------------------------------------
    # WINDOW LIMITS
    # --------------------------------------------------------

    half_L = L // 2

    start = max(0, n0 - half_L)
    stop = min(N, n0 + half_L)

    x_window = x[start:stop]

    local_mean = np.mean(x_window)
    local_variance = np.var(x_window)
    local_std = np.sqrt(local_variance)

    # --------------------------------------------------------
    # FIGURE
    # --------------------------------------------------------

    fig, ax = plt.subplots(figsize=(10, 5))

    ax.plot(n, x, linewidth=1.1)

    ax.axvspan(start, stop, alpha=0.15)

    ax.hlines(local_mean, start, stop, linewidth=2.0, linestyle='--', label=f'Local mean = {local_mean:.3f}')

    ax.hlines(local_mean + local_std, start, stop, linewidth=1.2, linestyle=':')
    ax.hlines(local_mean - local_std, start, stop, linewidth=1.2, linestyle=':')

    ax.axvline(n0, color='k', linestyle='--', linewidth=1.0)

    # --------------------------------------------------------
    # AXES
    # --------------------------------------------------------

    ax.set_xlim(0, N - 1)
    ax.set_ylim(-3.2, 3.2)

    ax.set_xlabel('Time index n', fontsize=14)
    ax.set_ylabel('x[n]', fontsize=14)

    ax.set_title(
        f'{process_name}\n'
        f'Local mean = {local_mean:.3f}     Local variance = {local_variance:.3f}',
        fontsize=13,
        pad=12
    )

    ax.tick_params(axis='both', labelsize=11)

    ax.grid(True, linestyle=':', alpha=0.6)

    ax.legend(fontsize=10, loc='upper right')

    plt.show()

# ============================================================
# CONTROLS
# ============================================================

control_layout = Layout(width='250px')

style_opts = {
    'description_width': '100px'
}

process_selector = RadioButtons(
    options=[
        'Stationary',
        'Time-varying mean',
        'Time-varying variance'
    ],
    value='Stationary',
    description='Process:',
    style=style_opts,
    layout=Layout(width='250px')
)

n0_slider = IntSlider(
    min=40,
    max=360,
    step=5,
    value=200,
    description='Window center:',
    continuous_update=True,
    style=style_opts,
    layout=control_layout
)

L_slider = IntSlider(
    min=20,
    max=160,
    step=10,
    value=80,
    description='Window length:',
    continuous_update=True,
    style=style_opts,
    layout=control_layout
)

# ============================================================
# INTERACTIVE OBJECT
# ============================================================

widget_plot = interactive(
    plot_stationarity,
    process_type=process_selector,
    n0=n0_slider,
    L=L_slider
)

# ============================================================
# THEORY
# ============================================================

theory_html = HTML("""
<div style="
    font-family: Arial, sans-serif;
    font-size: 16px;
    line-height: 1.5;
    width: 1050px;
">

<div style="
    font-size: 21px;
    font-weight: bold;
    margin-bottom: 10px;
">
Stationary and Non-Stationary Random Processes
</div>

A random process is considered <b>stationary</b> when its statistical
properties do not change significantly with time.

<br><br>

In this demonstration, a movable observation window is used to estimate
the local statistics of the process.

<br><br>

<b>Stationary:</b> approximately constant mean and variance.<br>

<b>Time-varying mean:</b> the mean changes with time.<br>

<b>Time-varying variance:</b> the variance changes with time.

<br><br>

The shaded region indicates the current observation window.<br>

The dashed horizontal line indicates the <b>local mean</b>.<br>

The dotted horizontal lines indicate approximately
<b>local mean ± local standard deviation</b>.

</div>
""")

# ============================================================
# CONTROLS TO THE RIGHT OF THE GRAPH
# ============================================================

controls = VBox(
    [
        process_selector,
        n0_slider,
        L_slider
    ],
    layout=Layout(
        width='270px',
        justify_content='center',
        margin='0px 0px 0px 10px'
    )
)

# ============================================================
# GRAPH LEFT - CONTROLS RIGHT
# ============================================================

graph_and_controls = HBox(
    [
        widget_plot.children[-1],
        controls
    ],
    layout=Layout(
        width='1100px',
        align_items='center',
        justify_content='flex-start'
    )
)

# ============================================================
# COMPLETE NOTEBOOK LAYOUT
# ============================================================

main_layout = VBox(
    [
        theory_html,
        graph_and_controls
    ],
    layout=Layout(
        width='1100px'
    )
)

display(main_layout)
```
