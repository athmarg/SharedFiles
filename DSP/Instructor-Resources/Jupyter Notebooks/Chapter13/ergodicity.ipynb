```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, HBox, Layout, VBox, HTML, GridBox
from IPython.display import display

# ============================================================
# ERGODIC RANDOM PROCESS
# ============================================================

np.random.seed(21)

M_max = 200
N_max = 1000

sigma = 1.0

X = sigma * np.random.randn(M_max, N_max)

# ============================================================
# INTERACTIVE FUNCTION
# ============================================================

def plot_ergodicity(M=50, N=300, realization_index=0):

    # --------------------------------------------------------
    # SELECT DATA
    # --------------------------------------------------------

    ensemble = X[:M, :N]

    realization_index = min(realization_index, M - 1)

    realization = ensemble[realization_index, :]

    # --------------------------------------------------------
    # ENSEMBLE MEAN
    # --------------------------------------------------------

    ensemble_mean = np.mean(ensemble[:, -1])

    # --------------------------------------------------------
    # TIME MEAN
    # --------------------------------------------------------

    running_time_mean = np.cumsum(realization) / np.arange(1, N + 1)

    final_time_mean = running_time_mean[-1]

    # --------------------------------------------------------
    # FIGURE
    # --------------------------------------------------------

    fig, ax = plt.subplots(figsize=(8.0, 5.0))

    ax.plot(
        np.arange(N),
        running_time_mean,
        linewidth=2,
        label='Running time mean'
    )

    ax.axhline(
        ensemble_mean,
        color='k',
        linestyle='--',
        linewidth=1.5,
        label=f'Ensemble mean = {ensemble_mean:.3f}'
    )

    ax.axhline(
        0,
        linestyle=':',
        linewidth=1.2,
        label='Theoretical mean = 0'
    )

    ax.set_xlim(0, N - 1)

    ax.set_ylim(-1.0, 1.0)

    ax.set_xlabel('Number of samples N', fontsize=14)

    ax.set_ylabel('Mean value', fontsize=14)

    ax.set_title(
        f'Ergodicity Demonstration\n'
        f'Time mean = {final_time_mean:.3f}     Ensemble mean = {ensemble_mean:.3f}',
        fontsize=13,
        pad=12
    )

    ax.tick_params(axis='both', labelsize=11)

    ax.grid(True, linestyle=':', alpha=0.6)

    ax.legend(fontsize=10)

    plt.tight_layout()

    plt.show()

# ============================================================
# SLIDERS
# ============================================================

slider_style = {
    'description_width': '0px'
}

slider_layout = Layout(
    width='100px'
)

M_slider = IntSlider(
    min=10,
    max=200,
    step=10,
    value=50,
    description=' ',
    continuous_update=True,
    readout=False,
    style=slider_style,
    layout=slider_layout
)

N_slider = IntSlider(
    min=50,
    max=1000,
    step=50,
    value=300,
    description=' ',
    continuous_update=True,
    readout=False,
    style=slider_style,
    layout=slider_layout
)

realization_slider = IntSlider(
    min=0,
    max=49,
    step=1,
    value=0,
    description=' ',
    continuous_update=True,
    readout=False,
    style=slider_style,
    layout=slider_layout
)

# ============================================================
# MAXIMUM VALUE LABELS
# ============================================================

M_max_label = HTML(
    '<div style="font-family:Arial; font-size:14px;">200</div>'
)

N_max_label = HTML(
    '<div style="font-family:Arial; font-size:14px;">1000</div>'
)

realization_max_label = HTML(
    '<div style="font-family:Arial; font-size:14px;">49</div>'
)

# ============================================================
# KEEP REALIZATION INDEX CONSISTENT WITH M
# ============================================================

def update_realization_range(change):

    realization_slider.max = M_slider.value - 1

    if realization_slider.value > realization_slider.max:
        realization_slider.value = realization_slider.max

    realization_max_label.value = f'<div style="font-family:Arial; font-size:14px;">{realization_slider.max}</div>'

M_slider.observe(update_realization_range, names='value')

# ============================================================
# INTERACTIVE OBJECT
# ============================================================

widget_plot = interactive(
    plot_ergodicity,
    M=M_slider,
    N=N_slider,
    realization_index=realization_slider
)

# ============================================================
# THEORY
# ============================================================

theory_html = HTML("""
<div style="
    font-family: Arial, sans-serif;
    font-size: 16px;
    line-height: 1.5;
    width: 1050px;
">

<div style="
    font-size: 21px;
    font-weight: bold;
    margin-bottom: 10px;
">
Ergodic Random Processes
</div>

A stationary process is called <b>ergodic</b> when the statistical
properties of the complete ensemble can be estimated from a single
sufficiently long realization.

<br><br>

In this demonstration:

<br><br>

<b>Ensemble mean:</b> the average over many realizations at one fixed
time instant.

<br>

<b>Time mean:</b> the average over time using one realization.

<br><br>

For an ergodic process, as the observation interval becomes longer,
the time mean approaches the ensemble mean.

<br><br>

The solid curve shows the <b>running time mean</b> of one realization.

<br>

The dashed horizontal line shows the <b>ensemble mean</b>.

<br>

The dotted horizontal line shows the theoretical mean, which is zero.

</div>
""")

# ============================================================
# LEFT-ALIGNED LABELS
# ============================================================

M_label = HTML(
    '<div style="font-family:Arial; font-size:14px;">Realizations M:</div>'
)

N_label = HTML(
    '<div style="font-family:Arial; font-size:14px;">Samples N:</div>'
)

realization_label = HTML(
    '<div style="font-family:Arial; font-size:14px;">Realization:</div>'
)

# ============================================================
# FIXED THREE-COLUMN GRID
# LABEL | SLIDER | MAXIMUM VALUE
# ============================================================

slider_grid = GridBox(
    children=[
        M_label, M_slider, M_max_label,
        N_label, N_slider, N_max_label,
        realization_label, realization_slider, realization_max_label
    ],
    layout=Layout(
        width='265px',
        grid_template_columns='110px 100px 40px',
        grid_template_rows='30px 30px 30px',
        grid_gap='2px 6px',
        align_items='center',
        overflow='hidden'
    )
)

# ============================================================
# CONTROLS TO THE RIGHT OF THE FIGURE
# ============================================================

controls = VBox(
    [
        slider_grid
    ],
    layout=Layout(
        width='275px',
        min_width='275px',
        align_items='flex-start',
        justify_content='center',
        margin='0px 0px 0px 10px',
        overflow='hidden'
    )
)

# ============================================================
# FIGURE LEFT - CONTROLS RIGHT
# ============================================================

graph_and_controls = HBox(
    [
        widget_plot.children[-1],
        controls
    ],
    layout=Layout(
        width='1100px',
        align_items='center',
        justify_content='flex-start',
        overflow='hidden'
    )
)

# ============================================================
# COMPLETE LAYOUT
# ============================================================

main_layout = VBox(
    [
        theory_html,
        graph_and_controls
    ],
    layout=Layout(
        width='1100px',
        overflow='hidden'
    )
)

display(main_layout)
```
