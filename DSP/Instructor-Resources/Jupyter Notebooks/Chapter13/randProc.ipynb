```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import IntSlider, HBox, Layout, VBox, HTML, GridBox, interactive_output
from IPython.display import display

# ============================================================
# RANDOM PROCESS DATA
# ============================================================

np.random.seed(7)

N = 120
M_max = 100

A = 1.0
omega0 = 0.12 * np.pi
sigma = 0.25

n = np.arange(N)

phases = np.random.uniform(0, 2*np.pi, M_max)
noise = sigma * np.random.randn(M_max, N)

X = A * np.cos(omega0 * n[None, :] + phases[:, None]) + noise


# ============================================================
# FIGURE 1: ENSEMBLE OF REALIZATIONS
# ============================================================

def plot_ensemble(M=30, n0=40):

    fig, ax = plt.subplots(figsize=(8.0, 3.8))

    values = X[:M, n0]

    for i in range(M):
        ax.plot(n, X[i, :], linewidth=0.8, alpha=0.30)

    ax.axvline(n0, color='k', linestyle='--', linewidth=1.5)

    ax.scatter(np.full(M, n0), values, s=25, zorder=5)

    ax.set_xlim(0, N - 1)
    ax.set_ylim(-1.8, 1.8)

    ax.set_xlabel('Time index n', fontsize=14)
    ax.set_ylabel(r'$x[n,\xi_i]$', fontsize=14)

    ax.set_title('Ensemble of Realizations', fontsize=14, pad=12)

    ax.tick_params(axis='both', labelsize=11)

    ax.grid(True, linestyle=':', alpha=0.6)

    plt.show()
    plt.close(fig)


# ============================================================
# FIGURE 2: RANDOM VARIABLE AT FIXED TIME
# ============================================================

def plot_random_variable(M=30, n0=40):

    fig, ax = plt.subplots(figsize=(8.0, 3.5))

    values = X[:M, n0]

    ensemble_mean = np.mean(values)
    ensemble_variance = np.var(values)

    ax.hist(
        values,
        bins=18,
        range=(-1.8, 1.8),
        density=True,
        alpha=0.75
    )

    ax.axvline(
        ensemble_mean,
        color='k',
        linestyle='--',
        linewidth=1.5,
        label=f'Ensemble mean = {ensemble_mean:.3f}'
    )

    ax.set_xlim(-1.8, 1.8)
    ax.set_ylim(0, 1.4)

    ax.set_xlabel(f'X[{n0}]', fontsize=14)
    ax.set_ylabel('Estimated probability density', fontsize=14)

    ax.set_title(
        f'Random Variable X[{n0}] Across the Ensemble',
        fontsize=14,
        pad=12
    )

    ax.tick_params(axis='both', labelsize=11)

    ax.grid(True, linestyle=':', alpha=0.6)

    ax.legend(fontsize=10)

    plt.show()
    plt.close(fig)

    print(f'Number of realizations M = {M}')
    print(f'Fixed time index n0 = {n0}')
    print(f'Ensemble mean at n0 = {ensemble_mean:.4f}')
    print(f'Ensemble variance at n0 = {ensemble_variance:.4f}')


# ============================================================
# SLIDERS
# NO DESCRIPTION - NO READOUT
# ============================================================

slider_style = {
    'description_width': '0px'
}

slider_layout = Layout(
    width='175px'
)

M_slider = IntSlider(
    min=5,
    max=100,
    step=5,
    value=30,
    description=' ',
    continuous_update=True,
    readout=False,
    style=slider_style,
    layout=slider_layout
)

n0_slider = IntSlider(
    min=0,
    max=N - 1,
    step=1,
    value=40,
    description=' ',
    continuous_update=True,
    readout=False,
    style=slider_style,
    layout=slider_layout
)


# ============================================================
# TWO INDEPENDENT OUTPUTS
# ============================================================

output_ensemble = interactive_output(
    plot_ensemble,
    {
        'M': M_slider,
        'n0': n0_slider
    }
)

output_random_variable = interactive_output(
    plot_random_variable,
    {
        'M': M_slider,
        'n0': n0_slider
    }
)


# ============================================================
# THEORY
# ============================================================

theory_html = HTML("""
<div style="
    font-family: Arial, sans-serif;
    font-size: 16px;
    line-height: 1.55;
    width: 1050px;
">

<div style="
    font-size: 21px;
    font-weight: bold;
    margin-bottom: 10px;
">
Random Process, Ensemble, Realization and Random Variable
</div>

Consider the stochastic process

<div style="
    font-size: 18px;
    font-weight: bold;
    margin-top: 8px;
    margin-bottom: 8px;
">
X[n] = A cos(ω₀ n + Φ) + W[n]
</div>

where <b>Φ</b> is a random phase and <b>W[n]</b> is Gaussian noise.

<br><br>

<b>Random process:</b> the complete ensemble of possible signals.<br>

<b>Realization:</b> one particular signal x[n, ξᵢ].<br>

<b>Random variable:</b> for fixed n = n₀, the values obtained from all realizations.<br>

<b>Sample value:</b> one numerical value for fixed n = n₀ and ξ = ξᵢ.

<br><br>

<b>Basic interpretation:</b><br><br>

ξ = ξᵢ, n varies
&nbsp;&nbsp; → &nbsp;&nbsp;
one realization x[n, ξᵢ]<br>

n = n₀, ξ varies
&nbsp;&nbsp; → &nbsp;&nbsp;
one random variable X[n₀]<br>

n = n₀, ξ = ξᵢ
&nbsp;&nbsp; → &nbsp;&nbsp;
one numerical value x[n₀, ξᵢ]

</div>
""")


# ============================================================
# LEFT-ALIGNED LABELS
# ============================================================

M_label = HTML(
    '<div style="font-family:Arial; font-size:14px;">Realizations M:</div>'
)

n0_label = HTML(
    '<div style="font-family:Arial; font-size:14px;">Time index n₀:</div>'
)


# ============================================================
# FIXED TWO-COLUMN GRID
# ============================================================

slider_grid = GridBox(
    children=[
        M_label, M_slider,
        n0_label, n0_slider
    ],
    layout=Layout(
        width='290px',
        grid_template_columns='110px 175px',
        grid_template_rows='30px 30px',
        grid_gap='2px 5px',
        align_items='center',
        overflow='hidden'
    )
)


# ============================================================
# CONTROLS
# ============================================================

controls = VBox(
    [
        slider_grid
    ],
    layout=Layout(
        width='300px',
        min_width='300px',
        align_items='flex-start',
        justify_content='center',
        margin='0px 0px 0px 10px',
        overflow='hidden'
    )
)


# ============================================================
# FIRST GRAPH LEFT - CONTROLS RIGHT
# ============================================================

first_row = HBox(
    [
        output_ensemble,
        controls
    ],
    layout=Layout(
        width='1100px',
        align_items='center',
        justify_content='flex-start',
        overflow='hidden'
    )
)


# ============================================================
# COMPLETE LAYOUT
# ============================================================

main_layout = VBox(
    [
        theory_html,
        first_row,
        output_random_variable
    ],
    layout=Layout(
        width='1100px',
        overflow='hidden'
    )
)

display(main_layout)


# ============================================================
# FORCE INITIAL DRAWING OF BOTH FIGURES
# ============================================================

output_ensemble.clear_output(wait=True)

with output_ensemble:
    plot_ensemble(
        M_slider.value,
        n0_slider.value
    )

output_random_variable.clear_output(wait=True)

with output_random_variable:
    plot_random_variable(
        M_slider.value,
        n0_slider.value
    )
```
