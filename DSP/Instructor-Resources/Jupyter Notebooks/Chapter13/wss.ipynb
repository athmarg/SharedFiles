```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, RadioButtons, HBox, Layout, VBox, HTML, GridBox
from IPython.display import display

# ============================================================
# RANDOM PROCESS PARAMETERS
# ============================================================

np.random.seed(53)

M_max = 200
N_max = 600

n_full = np.arange(N_max)

# Base Gaussian random ensemble
W = np.random.randn(M_max, N_max)

# ============================================================
# INTERACTIVE FUNCTION
# ============================================================

def plot_stationarity_types(process_type='WSS', M=100, N=400):

    n = np.arange(N)
    W_use = W[:M, :N]

    # --------------------------------------------------------
    # WSS PROCESS
    # --------------------------------------------------------

    if process_type == 'WSS':

        X = W_use.copy()
        process_name = 'Wide-Sense Stationary Process'
        theoretical_mean = np.zeros(N)
        theoretical_variance = np.ones(N)

    # --------------------------------------------------------
    # CYCLOSTATIONARY PROCESS
    # --------------------------------------------------------

    elif process_type == 'Cyclostationary':

        T = 100

        periodic_mean = 0.8 * np.sin(2 * np.pi * n / T)
        periodic_sigma = 0.75 + 0.35 * np.cos(2 * np.pi * n / T)

        X = periodic_mean[None, :] + periodic_sigma[None, :] * W_use

        process_name = 'Cyclostationary Process'
        theoretical_mean = periodic_mean
        theoretical_variance = periodic_sigma**2

    # --------------------------------------------------------
    # NONSTATIONARY PROCESS
    # --------------------------------------------------------

    else:

        time_scale = n / max(N - 1, 1)

        varying_mean = -1.0 + 2.0 * time_scale
        varying_sigma = 0.45 + 0.85 * time_scale

        X = varying_mean[None, :] + varying_sigma[None, :] * W_use

        process_name = 'Nonstationary Process'
        theoretical_mean = varying_mean
        theoretical_variance = varying_sigma**2

    # --------------------------------------------------------
    # ENSEMBLE STATISTICS
    # --------------------------------------------------------

    ensemble_mean = np.mean(X, axis=0)
    ensemble_variance = np.var(X, axis=0)

    # --------------------------------------------------------
    # FIGURE
    # --------------------------------------------------------

    fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(8.0, 7.0))

    # ========================================================
    # UPPER GRAPH: SAMPLE REALIZATIONS
    # ========================================================

    number_to_show = min(8, M)

    for i in range(number_to_show):
        ax1.plot(n, X[i, :], linewidth=0.8, alpha=0.65)

    ax1.set_xlim(0, N - 1)
    ax1.set_ylim(-4.0, 4.0)

    ax1.set_xlabel('Time index n', fontsize=11)
    ax1.set_ylabel('x[n]', fontsize=11)
    ax1.set_title(process_name, fontsize=13, pad=8)

    ax1.tick_params(axis='both', labelsize=9)
    ax1.grid(True, linestyle=':', alpha=0.6)

    # ========================================================
    # MIDDLE GRAPH: ENSEMBLE MEAN
    # ========================================================

    ax2.plot(n, ensemble_mean, linewidth=1.8, label='Estimated ensemble mean')
    ax2.plot(n, theoretical_mean, linestyle='--', linewidth=1.4, label='Theoretical mean')

    ax2.set_xlim(0, N - 1)
    ax2.set_ylim(-1.8, 1.8)

    ax2.set_xlabel('Time index n', fontsize=11)
    ax2.set_ylabel('Mean', fontsize=11)
    ax2.set_title('Ensemble Mean versus Time', fontsize=12, pad=7)

    ax2.tick_params(axis='both', labelsize=9)
    ax2.grid(True, linestyle=':', alpha=0.6)

    ax2.legend(
        loc='upper center',
        bbox_to_anchor=(0.5, -0.38),
        ncol=2,
        fontsize=8,
        borderaxespad=0.0
    )

    # ========================================================
    # LOWER GRAPH: ENSEMBLE VARIANCE
    # ========================================================

    ax3.plot(n, ensemble_variance, linewidth=1.8, label='Estimated ensemble variance')
    ax3.plot(n, theoretical_variance, linestyle='--', linewidth=1.4, label='Theoretical variance')

    ax3.set_xlim(0, N - 1)
    ax3.set_ylim(0, 2.5)

    ax3.set_xlabel('Time index n', fontsize=11)
    ax3.set_ylabel('Variance', fontsize=11)
    ax3.set_title('Ensemble Variance versus Time', fontsize=12, pad=7)

    ax3.tick_params(axis='both', labelsize=9)
    ax3.grid(True, linestyle=':', alpha=0.6)

    ax3.legend(
        loc='upper center',
        bbox_to_anchor=(0.5, -0.38),
        ncol=2,
        fontsize=8,
        borderaxespad=0.0
    )

    # ========================================================
    # VERTICAL SPACING
    # ========================================================

    plt.subplots_adjust(
        left=0.12,
        right=0.97,
        top=0.96,
        bottom=0.14,
        hspace=0.88
    )

    plt.show()

# ============================================================
# RADIO BUTTONS
# ============================================================

process_selector = RadioButtons(
    options=[
        'WSS',
        'Cyclostationary',
        'Nonstationary'
    ],
    value='WSS',
    description='Process:',
    style={'description_width': 'initial'},
    layout=Layout(width='270px')
)

# ============================================================
# SLIDERS
# ============================================================

slider_style = {
    'description_width': '0px'
}

slider_layout = Layout(
    width='100px'
)

M_slider = IntSlider(
    min=20,
    max=200,
    step=20,
    value=100,
    description=' ',
    continuous_update=True,
    readout=False,
    style=slider_style,
    layout=slider_layout
)

N_slider = IntSlider(
    min=100,
    max=600,
    step=50,
    value=400,
    description=' ',
    continuous_update=True,
    readout=False,
    style=slider_style,
    layout=slider_layout
)

# ============================================================
# MAXIMUM VALUE LABELS
# ============================================================

M_max_label = HTML(
    '<div style="font-family:Arial; font-size:14px;">200</div>'
)

N_max_label = HTML(
    '<div style="font-family:Arial; font-size:14px;">600</div>'
)

# ============================================================
# INTERACTIVE OBJECT
# ============================================================

widget_plot = interactive(
    plot_stationarity_types,
    process_type=process_selector,
    M=M_slider,
    N=N_slider
)

# ============================================================
# COMPACT THEORY / DOCUMENTATION
# ============================================================

theory_html = HTML("""
<div style="
    font-family: Arial, sans-serif;
    font-size: 16px;
    line-height: 1.30;
    width: 1050px;
">

<div style="
    font-size: 21px;
    font-weight: bold;
    margin-bottom: 6px;
">
WSS, SSS, Cyclostationary and Nonstationary Processes
</div>

<div style="margin-bottom:5px;">
<b>WSS:</b> the mean is constant and the autocorrelation depends only on the time lag.
</div>

<div style="margin-bottom:5px;">
<b>SSS:</b> all joint probability distributions remain unchanged under a common time shift.
</div>

<div style="margin-bottom:5px;">
<b>Cyclostationary:</b> the statistical properties vary with time but repeat periodically.
</div>

<div style="margin-bottom:5px;">
<b>Nonstationary:</b> one or more statistical properties change with time without the stationarity condition being satisfied.
</div>

<div>
<b>This notebook:</b> compares the ensemble mean and ensemble variance of WSS, cyclostationary and nonstationary processes as functions of time.
</div>

</div>
""")

# ============================================================
# LEFT-ALIGNED LABELS
# ============================================================

M_label = HTML(
    '<div style="font-family:Arial; font-size:14px;">Realizations M:</div>'
)

N_label = HTML(
    '<div style="font-family:Arial; font-size:14px;">Samples N:</div>'
)

# ============================================================
# FIXED THREE-COLUMN GRID
# LABEL | SLIDER | MAXIMUM VALUE
# ============================================================

slider_grid = GridBox(
    children=[
        M_label, M_slider, M_max_label,
        N_label, N_slider, N_max_label
    ],
    layout=Layout(
        width='265px',
        grid_template_columns='110px 100px 40px',
        grid_template_rows='30px 30px',
        grid_gap='2px 6px',
        align_items='center',
        overflow='hidden'
    )
)

# ============================================================
# SPACE BETWEEN RADIO BUTTONS AND SLIDER GROUP
# ============================================================

slider_group = VBox(
    [
        slider_grid
    ],
    layout=Layout(
        margin='12px 0px 0px 0px'
    )
)

# ============================================================
# CONTROLS
# ============================================================

controls = VBox(
    [
        process_selector,
        slider_group
    ],
    layout=Layout(
        width='285px',
        min_width='285px',
        align_items='flex-start',
        justify_content='center',
        margin='0px 0px 0px 10px',
        overflow='hidden'
    )
)

# ============================================================
# FIGURE LEFT - CONTROLS RIGHT
# ============================================================

graph_and_controls = HBox(
    [
        widget_plot.children[-1],
        controls
    ],
    layout=Layout(
        width='1100px',
        align_items='center',
        justify_content='flex-start',
        overflow='hidden'
    )
)

# ============================================================
# COMPLETE LAYOUT
# ============================================================

main_layout = VBox(
    [
        theory_html,
        graph_and_controls
    ],
    layout=Layout(
        width='1100px',
        overflow='hidden'
    )
)

display(main_layout)
```
