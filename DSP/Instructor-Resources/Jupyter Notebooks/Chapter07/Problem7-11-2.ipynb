```python
import sympy as sp

# Definition of symbolic variables and functions for the 3rd order system
n = sp.Symbol("n", integer=True)
q1 = sp.Function("q1")(n)
q2 = sp.Function("q2")(n)
q3 = sp.Function("q3")(n)
q1_next = sp.Function("q1")(n + 1)
q2_next = sp.Function("q2")(n + 1)
q3_next = sp.Function("q3")(n + 1)
x = sp.Function("x")(n)
y = sp.Function("y")(n)

# State vectors q[n] and q[n+1] for 3rd order
q_n = sp.Matrix([[q1], [q2], [q3]])
q_next = sp.Matrix([[q1_next], [q2_next], [q3_next]])

print("=== 1. NORMAL FORM (3RD ORDER SYSTEM) ===")
A_norm = sp.Matrix([[0, 1, 0], [0, 0, 1], [1, -2, -3]])
B_norm = sp.Matrix([[0], [0], [1]])
C_norm = sp.Matrix([[5, 2, -19]])
D_norm = sp.Matrix([[5]])

print("State Equation:")
display(sp.Eq(q_next, A_norm * q_n + B_norm * x))

print("Output Equation:")
display(sp.Eq(y, (C_norm * q_n + D_norm * x)[0, 0]))

print("\n" + "=" * 50 + "\n")

print("=== 2. NON-NORMAL FORM (3RD ORDER SYSTEM) ===")
A_non = sp.Matrix([[0, 0, 1], [1, 0, -2], [0, 1, -3]])
B_non = sp.Matrix([[5], [2], [-19]])
C_non = sp.Matrix([[0, 0, 1]])
D_non = sp.Matrix([[5]])

print("State Equation:")
display(sp.Eq(q_next, A_non * q_n + B_non * x))

print("Output Equation:")
display(sp.Eq(y, (C_non * q_n + D_non * x)[0, 0]))
```
