```python
import ipywidgets as widgets
from IPython.display import display
import matplotlib.pyplot as plt
import numpy as np


def plot_noncausal_stability(alpha, beta):
  # Define range for n including negative indices for the non-causal system
  n = np.arange(-20, 20)

  # Compute h[n]: alpha^n for n >= 0, beta^n for n < 0
  h = np.where(n >= 0, alpha ** n, beta ** n)

  # Compute partial sums for absolute values
  n_neg = n[n < 0]
  h_neg = beta ** n_neg

  n_pos = n[n >= 0]
  h_pos = alpha ** n_pos
  partial_sum_pos = np.cumsum(np.abs(h_pos))

  fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

  # 1. Impulse Response Plot
  ax1.stem(n, h, basefmt=" ", linefmt="b-", markerfmt="bo")
  ax1.set_title( f"Non-Causal Impulse Response h[n]\n(alpha = {alpha:.2f}, beta =" f" {beta:.2f})"
  )
  ax1.set_xlabel("n")
  ax1.set_ylabel("h[n]")
  ax1.grid(True)
  ax1.set_ylim(-2, 2.5)
  ax1.axvline(0, color="gray", linestyle="--", alpha=0.7)

  # Check stability conditions: |alpha| < 1 and |beta| > 1
  stable_alpha = abs(alpha) < 1
  stable_beta = abs(beta) > 1

  if stable_alpha and stable_beta:
    status = "STABLE (|alpha| < 1 AND |beta| > 1)"
    color = "green"
  else:
    status = "UNSTABLE"
    color = "red"

  # Place status text below the first subplot to avoid overlapping with the curve
  ax1.text( 0.5, -0.25, f"Status: {status}", transform=ax1.transAxes, ha="center", color=color, fontweight="bold",
      fontsize=11, bbox=dict(facecolor="white", alpha=0.9, edgecolor=color, boxstyle="round,pad=0.5"), )

  # 2. Absolute Values & Convergence Behaviour on the second axis
  ax2.plot( n_pos, partial_sum_pos, "r-o", label="Sum (n >= 0): converges if |alpha| < 1", )
  ax2.plot( n_neg, np.abs(h_neg), "g-s", label="Terms (n < 0): converges if |beta| > 1", )

  ax2.set_title("Absolute Values & Convergence Behaviour", fontweight="bold")
  ax2.set_xlabel("n")
  ax2.set_ylabel("Magnitude / Partial Sum")
  ax2.grid(True)
  ax2.legend(loc="upper left")
  ax2.set_ylim(0, 5)
  plt.tight_layout()
  plt.show()

# Create Interactive Sliders for alpha and beta
alpha_slider = widgets.FloatSlider( value=0.5, min=-1.5, max=1.5, step=0.05, description="Parameter alpha:", continuous_update=False, )
beta_slider = widgets.FloatSlider( value=1.5, min=-2.5, max=2.5, step=0.05, description="Parameter beta:", continuous_update=False, )
widgets.interactive( plot_noncausal_stability, alpha=alpha_slider, beta=beta_slider )
```
