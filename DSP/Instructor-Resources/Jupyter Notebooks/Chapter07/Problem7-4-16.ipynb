```python
# ============================================================
# General Solution of a Second-Order Difference Equation
#
# y[n] - 3y[n-1] - 4y[n-2] = x[n] + 2x[n-1]
# x[n] = 4^n u[n]
#
# The initial conditions y[-1] and y[-2] remain symbolic.
# ============================================================

import sympy as sp
from IPython.display import display, Markdown

# ============================================================
# 1. Symbolic variables
# ============================================================

n = sp.symbols('n', integer=True)
lam, k = sp.symbols('lambda k')
C1, C2 = sp.symbols('C1 C2')
y_minus1, y_minus2 = sp.symbols('y_minus1 y_minus2')

# ============================================================
# 2. Input signal
# ============================================================

display(Markdown("## Input signal"))
display(sp.Eq(sp.Symbol('x[n]'), 4**n*sp.Symbol('u[n]')))

# ============================================================
# 3. Homogeneous solution
# ============================================================

display(Markdown("## 1. Homogeneous equation"))

characteristic_equation = sp.Eq(lam**2 - 3*lam - 4, 0)
display(characteristic_equation)

roots = sp.solve(characteristic_equation, lam)
lambda1, lambda2 = roots

display(sp.Eq(sp.Symbol('lambda_1'), lambda1))
display(sp.Eq(sp.Symbol('lambda_2'), lambda2))

y_h = C1*4**n + C2*(-1)**n
display(sp.Eq(sp.Symbol('y_h[n]'), y_h))

# ============================================================
# 4. Particular solution
# ============================================================

display(Markdown("## 2. Particular solution"))

y_p = k*n*4**n
display(sp.Eq(sp.Symbol('y_p[n]'), y_p))

yp_n = k*n*4**n
yp_n_minus1 = k*(n-1)*4**(n-1)
yp_n_minus2 = k*(n-2)*4**(n-2)

lhs_particular = sp.expand(yp_n - 3*yp_n_minus1 - 4*yp_n_minus2)
rhs_particular = 4**n + 2*4**(n-1)

display(sp.Eq(sp.Symbol('LHS'), sp.factor(lhs_particular)))
display(sp.Eq(sp.Symbol('RHS'), sp.factor(rhs_particular)))

# ============================================================
# 5. Determine k
# ============================================================

k_solution = sp.solve(sp.Eq(lhs_particular, rhs_particular), k)[0]
display(sp.Eq(k, sp.simplify(k_solution)))

y_p = sp.simplify(y_p.subs(k, k_solution))
display(sp.Eq(sp.Symbol('y_p[n]'), y_p))

# ============================================================
# 6. General solution
# ============================================================

display(Markdown("## 3. General solution"))

y_general = sp.expand(y_h + y_p)
display(sp.Eq(sp.Symbol('y[n]'), y_general))

# ============================================================
# 7. Initial values y[0] and y[1]
# ============================================================

display(Markdown("## 4. Initial values"))

y0 = 1 + 3*y_minus1 + 4*y_minus2
y1 = sp.expand(6 + 3*y0 + 4*y_minus1)

display(sp.Eq(sp.Symbol('y[0]'), y0))
display(sp.Eq(sp.Symbol('y[1]'), y1))

# ============================================================
# 8. Determine C1 and C2
# ============================================================

display(Markdown("## 5. Constants C₁ and C₂"))

y0_general = sp.simplify(y_general.subs(n, 0))
y1_general = sp.simplify(y_general.subs(n, 1))

constants = sp.solve([sp.Eq(y0_general, y0), sp.Eq(y1_general, y1)], [C1, C2], dict=True)[0]

C1_solution = sp.factor(constants[C1])
C2_solution = sp.factor(constants[C2])

display(sp.Eq(C1, C1_solution))
display(sp.Eq(C2, C2_solution))

# ============================================================
# 9. Final overall response
# ============================================================

display(Markdown("## 6. Final overall response"))

y_final = sp.factor(y_general.subs({C1: C1_solution, C2: C2_solution}))
display(sp.Eq(sp.Symbol('y[n]'), y_final))

# ============================================================
# 10. Verification of the initial conditions
# ============================================================

display(Markdown("## 7. Verification"))

verification_y0 = sp.simplify(y_final.subs(n, 0) - y0)
verification_y1 = sp.simplify(y_final.subs(n, 1) - y1)

display(sp.Eq(sp.Symbol('y[0] - required value'), verification_y0))
display(sp.Eq(sp.Symbol('y[1] - required value'), verification_y1))

# ============================================================
# 11. Verification of the difference equation
# ============================================================

lhs_final = sp.expand(y_final - 3*y_final.subs(n, n-1) - 4*y_final.subs(n, n-2))
rhs_final = 4**n + 2*4**(n-1)
difference = sp.simplify(lhs_final - rhs_final)

display(sp.Eq(sp.Symbol('LHS'), sp.factor(lhs_final)))
display(sp.Eq(sp.Symbol('RHS'), sp.factor(rhs_final)))
display(sp.Eq(sp.Symbol('LHS - RHS'), difference))

if difference == 0:
    display(Markdown("**Verification successful:** the solution satisfies the difference equation for $n \\geq 2$."))
else:
    display(Markdown("**Verification failed.**"))

# ============================================================
# 12. Final compact result
# ============================================================

display(Markdown("## Final result"))
display(sp.Eq(sp.Symbol('y[n]'), sp.factor(y_final)))
```
