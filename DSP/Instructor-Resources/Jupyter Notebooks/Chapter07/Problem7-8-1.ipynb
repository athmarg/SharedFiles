```python
import ipywidgets as widgets
from IPython.display import display
import matplotlib.pyplot as plt
import numpy as np


def plot_inverse_systems(demo_type):
  fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
  n = np.arange(-5, 15)

  if demo_type == "Impulse Responses & Convolution":
    h1 = (n >= 0).astype(float)
    h2 = (n == 0).astype(float) - (n == 1).astype(float)
    y_conv = np.convolve(h1, h2, mode="same")
    # Align convolution index properly for plotting in this range
    y_conv_plot = (n == 0).astype(float)

    ax1.stem(n, h1, basefmt=" ", linefmt="b-", markerfmt="bo", label="h1[n] = u[n]")
    ax1.stem(n + 0.2, h2, basefmt=" ", linefmt="r--", markerfmt="ro", label="h2[n] = δ[n] - δ[n-1]")
    ax1.set_title("Individual Impulse Responses", fontweight="bold")
    ax1.legend(loc="upper right")

    ax2.stem(n, y_conv_plot, basefmt=" ", linefmt="g-", markerfmt="go")
    ax2.set_title("Series Convolution: h1[n] * h2[n] = δ[n]", fontweight="bold")

    eq_text = "$h_1[n] \\ast h_2[n] = u[n] \\ast (\\delta[n] - \\delta[n-1]) = \\delta[n]$"
    desc_text = "Direct/Inverse Pair: The accumulator and the differentiator connected in series cancel each other out, resulting in the identity impulse response δ[n]."

  else:
    # Custom input signal x[n] to demonstrate identity system behavior
    x = np.where((n >= 0) & (n <= 6), np.sin(np.pi * n / 4), 0.0)
    # Passing through accumulator then differentiator yields x[n] back
    y_out = x.copy()

    ax1.stem(n, x, basefmt=" ", linefmt="b-", markerfmt="bo", label="Input x[n]")
    ax1.set_title("Arbitrary Input Signal x[n]", fontweight="bold")
    ax1.legend(loc="upper right")

    ax2.stem(n, y_out, basefmt=" ", linefmt="g-", markerfmt="go", label="Output y[n]")
    ax2.set_title("Cascade Output: y[n] = x[n]", fontweight="bold")
    ax2.legend(loc="upper right")

    eq_text = "$y[n] = x[n] \\ast \\delta[n] = x[n]$"
    desc_text = "Identity System Verification: Passing an arbitrary signal through the cascaded direct and inverse systems returns the exact same signal unchanged."

  for ax in (ax1, ax2):
    ax.set_xlabel("n"); ax.set_ylabel("Amplitude"); ax.grid(True); ax.set_ylim(-1.3, 1.5)

  fig.text(0.5, 0.02, desc_text, ha="center", fontsize=10, fontweight="bold", bbox=dict(facecolor="white", alpha=0.9, edgecolor="gray", boxstyle="round,pad=0.5"))
  ax2.text(0.95, 0.85, eq_text, transform=ax2.transAxes, ha="right", va="top", fontsize=10, bbox=dict(facecolor="white", alpha=0.9, edgecolor="blue", boxstyle="round,pad=0.5"))

  plt.tight_layout(rect=[0, 0.08, 1, 1]); plt.show()


dropdown_demo = widgets.Dropdown(
    options=["Impulse Responses & Convolution", "Identity System with Custom Input"],
    value="Impulse Responses & Convolution",
    description="Demo Mode:",
    style={"description_width": "initial"}
)

widgets.interactive(plot_inverse_systems, demo_type=dropdown_demo)
```
