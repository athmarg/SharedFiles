```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
from IPython.display import display, Markdown, clear_output
from ipywidgets import Output, VBox, Layout

text_out = Output(layout=Layout(width='99%', overflow='hidden'))
fig_out = Output(layout=Layout(width='99%', overflow='hidden'))

def update_visualization():
    # --- Analytical step-by-step derivation using SymPy ---
    n = sp.Symbol('n', integer=True)
    C1, C2 = sp.symbols('C1 C2')
    
    # 1. Given initial conditions
    y_minus1 = sp.Rational(1, 8)
    y_minus2 = sp.Rational(1, 8)
    
    # 2. Difference equation evaluation for n = 0 and n = 1:
    # y[n] - 8y[n-1] + 16y[n-2] = 0 => y[n] = 8y[n-1] - 16y[n-2]
    y_0 = 8 * y_minus1 - 16 * y_minus2  # y[0] = -1
    
    # For n = 1: y[1] = 8y[0] - 16y[-1] = 8*(-1) - 16*(1/8) = -8 - 2 = -10
    # Alternatively, using the book's direct formula: y[1] = 48*y[-1] - 128*y[-2] = 48(1/8) - 128(1/8) = 6 - 16 = -10
    y_1 = 48 * y_minus1 - 128 * y_minus2  # y[1] = -10

    # 3. General solution form from the characteristic equation (double root lambda = 4):
    # y[n] = (C1 + C2 * n) * (4**n)
    # For n = 0: y[0] = C1 * (4**0) = C1
    # For n = 1: y[1] = (C1 + C2 * 1) * (4**1) = 4 * (C1 + C2)

    # 4. Formulate the system of equations based on the book's exact derivation:
    # C1 = 8*y[-1] - 16*y[-2]
    # C2 = 16*y[-1] - 64*y[-2]
    c1_val = 8 * y_minus1 - 16 * y_minus2
    c2_val = 16 * y_minus1 - 64 * y_minus2

    # Analytical symbolic output expression
    y_zi_sym = (c1_val + c2_val * n) * (4**n)

    with text_out:
        clear_output(wait=True)
        display(Markdown("### Natural Response Analysis"))
        display(Markdown(r"**Homogeneous Difference Equation:** $y[n] - 8y[n-1] + 16y[n-2] = 0$"))
        display(Markdown(r"**Initial Conditions:** $y[-1] = \frac{1}{8},\ y[-2] = \frac{1}{8}$"))
        display(Markdown(r"**Step 1: Characteristic Equation:** $Q(\lambda) = \lambda^2 - 8\lambda + 16 = (\lambda - 4)^2 = 0 \implies \lambda_1 = \lambda_2 = 4$ (Double root)"))
        display(Markdown(r"**Step 2: General Solution Form:** $y[n] = (C_1 + C_2 n)(4)^n$"))
        display(Markdown(r"**Step 3: State Evaluation for $n=0$ and $n=1$:**"))
        display(Markdown(f"- $y[0] = 8y[-1] - 16y[-2] = 8\\left(\\frac{{1}}{{8}}\\right) - 16\\left(\\frac{{1}}{{8}}\\right) = {y_0}$"))
        display(Markdown(f"- $y[1] = 48y[-1] - 128y[-2] = 48\\left(\\frac{{1}}{{8}}\\right) - 128\\left(\\frac{{1}}{{8}}\\right) = {y_1}$"))
        display(Markdown(r"**Step 4: Analytical Solution for Constants $C_1$ and $C_2$:**"))
        display(Markdown(f"- $C_1 = 8y[-1] - 16y[-2] = {c1_val}$"))
        display(Markdown(f"- $C_2 = 16y[-1] - 64y[-2] = {c2_val}$"))
        display(Markdown(r"**Analytical Natural Response:** $y_{\text{zi}}[n] = " + sp.latex(y_zi_sym) + r" = -(1 + 6n)(4)^n$"))
        display(Markdown(r"*Note: We restrict the domain of $n$ to a small range (up to $n=6$) because the output grows rapidly due to the dominant exponential term $4^n$.*"))

    with fig_out:
        clear_output(wait=True)
        
        n_vals = np.arange(0, 7)
        f_lambdified = sp.lambdify(n, y_zi_sym, 'numpy')
        y_vals = f_lambdified(n_vals)

        fig, ax = plt.subplots(figsize=(9.5, 4.5))
        ax.stem(n_vals, y_vals, linefmt='r-', markerfmt='ro', basefmt='k-', label=r'$y_{\text{zi}}[n]$')
        ax.set_title(r'Natural Response $y_{\text{zi}}[n]$', fontsize=11, fontweight='bold', color='darkred')
        ax.set_xlabel('Time index n', fontsize=10)
        ax.set_ylabel('Amplitude', fontsize=10)
        ax.grid(True, linestyle='--', alpha=0.6)
        ax.legend(loc='lower left')

        plt.tight_layout()
        plt.show()

update_visualization()
ui = VBox([text_out, fig_out], layout=Layout(align_items='flex-start', width='99%', overflow='hidden'))
display(ui)
```
