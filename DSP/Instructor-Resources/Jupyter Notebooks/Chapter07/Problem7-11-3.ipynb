```python
import sympy as sp

# Definition of symbolic variables
n = sp.Symbol("n", integer=True)

print("=== CALCULATION OF A^n FOR THE 2x2 SYSTEM ===")
# Define matrix A
A = sp.Matrix([[sp.Rational(1, 2), 0], [sp.Rational(1, 4), sp.Rational(1, 4)]])

print("Matrix A:")
display(A)

# Method 1: Using Cayley-Hamilton Theorem & Eigenvalues
print("\n--- Method 1: Cayley-Hamilton Theorem ---")
eigenvals = A.eigenvals()
print("Eigenvalues and their algebraic multiplicities:")
display(eigenvals)

# Diagonalization method for A^n = P * Lambda^n * P^-1
print("\n--- Method 2: Diagonalization Method ---")
P, D = A.diagonalize()
Lambda_n = D.applyfunc(lambda i: i**n)
A_n = P * Lambda_n * P.inv()

print("Diagonalization Matrix P:")
display(P)
print("Inverse Matrix P^-1:")
display(P.inv())
print("State Transition Matrix A^n:")
display(sp.simplify(A_n))
```
