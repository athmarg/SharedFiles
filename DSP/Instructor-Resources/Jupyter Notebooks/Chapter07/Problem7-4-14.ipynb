```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp
from IPython.display import display, Markdown, clear_output
from ipywidgets import Output, VBox, Layout

text_out = Output(layout=Layout(width='99%', overflow='hidden'))
fig_out = Output(layout=Layout(width='99%', overflow='hidden'))

def update_visualization():
    n = sp.Symbol('n', integer=True)
    
    y_minus1 = 1
    y_minus2 = 0
    
    y_0 = 2 * y_minus1 - 5 * y_minus2  # 2
    y_1 = 2 * y_0 - 5 * y_minus1      # -1

    theta_deg = 63.4343032488097
    theta_rad = sp.rad(theta_deg)

    A_val = y_0
    cos_t = np.cos(np.radians(theta_deg))
    sin_t = np.sin(np.radians(theta_deg))
    B_val = (y_1 / np.sqrt(5.0) - A_val * cos_t) / sin_t

    with text_out:
        clear_output(wait=True)
        display(Markdown("### Natural Response Analysis - Exercise 3 (Complex Conjugate Roots)"))
        display(Markdown(r"**Homogeneous Difference Equation:** $y[n] - 2y[n-1] + 5y[n-2] = 0$"))
        display(Markdown(r"**Initial Conditions:** $y[-1] = 1,\ y[-2] = 0$"))
        display(Markdown(r"**Step 1: Characteristic Equation:** $Q(\lambda) = \lambda^2 - 2\lambda + 5 = 0 \implies \lambda_{1,2} = 1 \pm 2j = \sqrt{5}e^{\pm j 63.434^\circ}$"))
        display(Markdown(r"**Step 2: General Trigonometric Solution Form:** $y[n] = (\sqrt{5})^n \left( A \cos(n \cdot 63.434^\circ) + B \sin(n \cdot 63.434^\circ) \right)$"))
        display(Markdown(r"**Step 3: State Evaluation for $n=0$ and $n=1$:**"))
        display(Markdown(r"- $y[0] = 2y[-1] - 5y[-2] = 2$"))
        display(Markdown(r"- $y[1] = 2y[0] - 5y[-1] = -1$"))
        display(Markdown(r"**Step 4: Analytical Solution for Constants $A$ and $B$:**"))
        display(Markdown(r"- $A = 2$"))
        display(Markdown(r"- $B \approx -1.1180$"))
        display(Markdown(r"**Analytical Natural Response:** $y_{\text{zi}}[n] = (\sqrt{5})^n \left( 2 \cos(n \cdot 63.434^\circ) - 1.1180 \sin(n \cdot 63.434^\circ) \right)$"))
        display(Markdown(r"*Note: Since the characteristic roots are complex conjugates, the natural response exhibits a growing oscillatory behavior due to $r = \sqrt{5} > 1$.*"))

    with fig_out:
        clear_output(wait=True)
        
        n_vals = np.arange(0, 11)
        y_vals = (np.sqrt(5.0)**n_vals) * (A_val * np.cos(n_vals * np.radians(theta_deg)) + B_val * np.sin(n_vals * np.radians(theta_deg)))

        fig, ax = plt.subplots(figsize=(9.5, 4.5))
        ax.stem(n_vals, y_vals, linefmt='b-', markerfmt='bo', basefmt='k-', label=r'$y_{\text{zi}}[n]$')
        ax.set_title(r'Natural Response with Complex Roots $y_{\text{zi}}[n]$', fontsize=11, fontweight='bold', color='darkblue')
        ax.set_xlabel('Time index n', fontsize=10)
        ax.set_ylabel('Amplitude', fontsize=10)
        ax.grid(True, linestyle='--', alpha=0.6)
        ax.legend(loc='upper left')

        plt.tight_layout()
        plt.show()

update_visualization()
ui = VBox([text_out, fig_out], layout=Layout(align_items='flex-start', width='99%', overflow='hidden'))
display(ui)
```
