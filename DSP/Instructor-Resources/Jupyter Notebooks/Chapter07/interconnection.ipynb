```python
import ipywidgets as widgets
from IPython.display import display
import matplotlib.pyplot as plt
import numpy as np


def plot_interconnections(conn_type):
  fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
  n = np.arange(0, 15)

  if conn_type == "Series Connection":
    h1 = 0.5 ** n * (n >= 0)
    h2 = (n == 0) + (n == 1).astype(float)
    y = np.convolve(h1, h2)[: len(n)]
    
    ax1.stem(n, h1, basefmt=" ", linefmt="b-", markerfmt="bo", label="h1[n] = (0.5)^n u[n]")
    ax1.stem(n + 0.2, h2, basefmt=" ", linefmt="r--", markerfmt="ro", label="h2[n] = delta[n] + delta[n-1]")
    ax1.set_title("Individual Impulse Responses", fontweight="bold")
    ax1.legend(loc="upper right")

    ax2.stem(n, y, basefmt=" ", linefmt="g-", markerfmt="go")
    ax2.set_title("Equivalent System: h[n] = h1[n] * h2[n]", fontweight="bold")
    
    eq_text = "$h[n] = h_1[n] \\ast h_2[n]$\nOutput equals the discrete convolution\nof the two cascaded systems."
    desc_text = "Series Connection: The output of the first system becomes the input of the second. The total impulse response is given by the convolution of the individual impulse responses."

  elif conn_type == "Parallel Connection":
    h1 = 0.8 ** n * (n >= 0)
    h2 = (-0.4) ** n * (n >= 0)
    y = h1 + h2

    ax1.stem(n, h1, basefmt=" ", linefmt="b-", markerfmt="bo", label="h1[n] = (0.8)^n u[n]")
    ax1.stem(n + 0.2, h2, basefmt=" ", linefmt="r--", markerfmt="ro", label="h2[n] = (-0.4)^n u[n]")
    ax1.set_title("Individual Impulse Responses", fontweight="bold")
    ax1.legend(loc="upper right")

    ax2.stem(n, y, basefmt=" ", linefmt="g-", markerfmt="go")
    ax2.set_title("Equivalent System: h[n] = h1[n] + h2[n]", fontweight="bold")

    eq_text = "$h[n] = h_1[n] + h_2[n]$\nOutput equals the sum of\nthe individual responses."
    desc_text = "Parallel Connection: Both systems receive the same input in parallel without interacting, and their separate outputs are added together to form the total response."

  else:
    sig = 0.6 ** n * (n >= 0)
    imp = (n == 0) + 0.5 * (n == 1).astype(float)
    y = np.convolve(sig, imp)[: len(n)]

    ax1.stem(n, sig, basefmt=" ", linefmt="b-", markerfmt="bo", label="Input x[n] = (0.6)^n u[n]")
    ax1.stem(n + 0.2, imp, basefmt=" ", linefmt="r--", markerfmt="ro", label="Impulse h[n] = delta[n] + 0.5delta[n-1]")
    ax1.set_title("Signals x[n] and h[n]", fontweight="bold")
    ax1.legend(loc="upper right")

    ax2.stem(n, y, basefmt=" ", linefmt="g-", markerfmt="go")
    ax2.set_title("Convolution Output: x[n] * h[n] = h[n] * x[n]", fontweight="bold")

    eq_text = "$x[n] \\ast h[n] = h[n] \\ast x[n]$\nDemonstrating the commutative\nproperty of LTI convolution."
    desc_text = "Commutative Property: A discrete LTI system with impulse response h[n] receiving input x[n] is fully equivalent to a system with input h[n] and impulse response x[n]."

  for ax in (ax1, ax2):
    ax.set_xlabel("n"); ax.set_ylabel("Amplitude"); ax.grid(True); ax.set_ylim(-0.6, 1.5)

  fig.text(0.5, 0.02, desc_text, ha="center", fontsize=10, fontweight="bold", bbox=dict(facecolor="white", alpha=0.9, edgecolor="gray", boxstyle="round,pad=0.5"))
  ax2.text(0.95, 0.85, eq_text, transform=ax2.transAxes, ha="right", va="top", fontsize=10, bbox=dict(facecolor="white", alpha=0.9, edgecolor="blue", boxstyle="round,pad=0.5"))

  plt.tight_layout(rect=[0, 0.08, 1, 1]); plt.show()


dropdown_conn = widgets.Dropdown(
    options=["Series Connection", "Parallel Connection", "Commutative Property"],
    value="Series Connection",
    description="Connection Type:",
    style={"description_width": "initial"}
)

widgets.interactive(plot_interconnections, conn_type=dropdown_conn)
```
