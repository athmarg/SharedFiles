```python
import sympy as sp

# Definition of symbolic variables
n = sp.Symbol("n", integer=True)

print("=== SOLUTION OF THE STATE EQUATION AND SYSTEM OUTPUT (EXERCISE 7.29) ===")

# 1. Given matrices and vectors
A = sp.Matrix([[0, 1], [sp.Rational(-1, 8), sp.Rational(3, 4)]])
B = sp.Matrix([[0], [1]])
C = sp.Matrix([[sp.Rational(-1, 8), sp.Rational(3, 4)]])
D = sp.Matrix([[1]])
q0 = sp.Matrix([[4], [2]])

print("Matrix A:")
display(A)
print("Initial State Vector q[0]:")
display(q0)

# 2. State Transition Matrix A^n using Diagonalization
P, Lambda = A.diagonalize()
Lambda_n = Lambda.applyfunc(lambda i: i**n)
A_n = sp.simplify(P * Lambda_n * P.inv())

print("State Transition Matrix A^n:")
display(A_n)

# 3. Zero-Input State Response: q_zi[n] = A^n * q[0]
q_zi = sp.simplify(A_n * q0)
print("Zero-Input State Vector q_zi[n]:")
display(q_zi)

# 4. Zero-Input System Output: y_zi[n] = C * q_zi[n]
y_zi = sp.simplify((C * q_zi)[0, 0])
print("Zero-Input Output y_zi[n]:")
display(y_zi)

print("\n" + "=" * 50 + "\n")

# 5. Symbolic verification of final total state vector and total output
# Based on the analytical solution derived from the problem description:
q_n = sp.Matrix([
    [8 * n * sp.Rational(1, 2) ** n - 12 * sp.Rational(1, 2) ** n + 16 * sp.Rational(1, 4) ** n],
    [4 * n * sp.Rational(1, 2) ** n - 2 * sp.Rational(1, 2) ** n + 4 * sp.Rational(1, 4) ** n]
])

print("Total State Vector q[n]:")
display(q_n)

# Total output y[n] = C * q[n] + D * x[n] (where x[n] = (1/2)^n for n >= 0)
x_n = sp.Rational(1, 2) ** n
y_total = sp.simplify((C * q_n + D * x_n)[0, 0])
print("Total System Output y[n]:")
display(y_total)
```
