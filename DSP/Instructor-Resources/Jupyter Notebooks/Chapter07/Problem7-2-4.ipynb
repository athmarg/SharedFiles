```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, Markdown

# Define π for clean usage
π = np.pi

# --- Fix output height for Jupyter notebooks ---
display(Markdown("<style>.output_scroll { max-height: none !important; }</style>"))

display(Markdown("### System Properties Explorer (Linearity, Causality, Stability, Time-Invariance)"))
display(Markdown(r"Explore the behavior of the two systems from the lecture exercise: <br>"
                 r"**(a)** $y[n] = \cos(\pi n)x[n]$ <br>"
                 r"**(b)** $y[n] = \sum_{k=n-1}^{\infty}x[k]$"))

# Define input signal x[n] (e.g., a pulse/step or custom function)
n = np.arange(-5, 11)

def system_analysis(sys_choice, input_type):
    # Generate input signal based on user selection
    if input_type == "Unit Step u[n]":
        x = (n >= 0).astype(float)
        input_label = "$x[n] = u[n]$ (Bounded)"
    elif input_type == "Ramp n*u[n]":
        x = n * (n >= 0).astype(float)
        input_label = "$x[n] = n \\cdot u[n]$ (Unbounded)"
    else:
        x = np.where((n >= -2) & (n <= 2), 1.0, 0.0)
        input_label = "$x[n]$ (Rectangular Pulse)"

    # Compute output based on system choice
    if sys_choice == "System (a): y[n] = cos(π*n)*x[n]":
        y = np.cos(π * n) * x
        sys_desc = (
            r"System (a): $y[n] = \cos(\pi n)x[n]$ <br>"
            r"• **Linear:** Yes<br>"
            r"• **Time-Invariant:** No (due to time-varying multiplier $\cos(\pi n)$)<br>"
            r"• **Causal:** Yes<br>"
            r"• **Stable:** Yes"
        )
    else:
        # y[n] = sum_{k=n-1}^{inf} x[k]
        y = np.zeros_like(x, dtype=float)
        for idx, val in enumerate(n):
            mask = (n >= val - 1)
            y[idx] = np.sum(x[mask])
        sys_desc = (
            r"System (b): $y[n] = \sum_{k=n-1}^{\infty}x[k]$ <br>"
            r"• **Linear:** Yes<br>"
            r"• **Time-Invariant:** Yes<br>"
            r"• **Causal:** No (depends on future inputs like $x[n]$ and $x[n+1]$)<br>"
            r"• **Stable:** No (accumulator-like sum towards $+\infty$ can diverge for bounded inputs)"
        )

    fig, axes = plt.subplots(2, 1, figsize=(10, 6), sharex=True)
    
    axes[0].stem(n, x, linefmt='b-', markerfmt='bo', basefmt='k-')
    axes[0].set_title(f'Input Signal: {input_label}', fontsize=9.5, fontweight='bold', color='darkblue')
    axes[0].set_ylabel('Amplitude', fontsize=8.5)
    axes[0].grid(True, linestyle='--', alpha=0.6)
    
    axes[1].stem(n, y, linefmt='r-', markerfmt='ro', basefmt='k-')
    axes[1].set_title('Output Response $y[n]$', fontsize=9.5, fontweight='bold', color='darkred')
    axes[1].set_xlabel('Index $n$', fontsize=8.5)
    axes[1].set_ylabel('Amplitude', fontsize=8.5)
    axes[1].grid(True, linestyle='--', alpha=0.6)
    
    plt.tight_layout()
    plt.show()
    
    display(Markdown(f"<div style='background-color: #f8f9fa; padding: 10px; border-left: 4px solid #007bff;'>{sys_desc}</div>"))

interactive_analysis = widgets.interactive(
    system_analysis,
    sys_choice=widgets.Dropdown(
        options=['System (a): y[n] = cos(π*n)*x[n]', 'System (b): y[n] = sum(x[k]) from k=n-1 to inf'],
        value='System (a): y[n] = cos(π*n)*x[n]',
        description='System:',
        style={'description_width': 'initial'}
    ),
    input_type=widgets.Dropdown(
        options=['Unit Step u[n]', 'Ramp n*u[n]', 'Rectangular Pulse'],
        value='Unit Step u[n]',
        description='Input Signal:',
        style={'description_width': 'initial'}
    )
)
display(interactive_analysis)
```
