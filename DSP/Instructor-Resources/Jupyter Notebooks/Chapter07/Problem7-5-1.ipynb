```python
import ipywidgets as widgets
from IPython.display import display
import matplotlib.pyplot as plt
import numpy as np


def plot_stability(alpha):
  n = np.arange(0, 40)
  h = (alpha ** n) * (n >= 0)
  partial_sum = np.cumsum(np.abs(h))

  fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4))

  # Impulse Response
  ax1.stem(n, h, basefmt=" ", linefmt="b-", markerfmt="bo")
  ax1.set_title(f"Impulse Response h[n] (alpha = {alpha:.2f})")
  ax1.set_xlabel("n")
  ax1.set_ylabel("h[n]")
  ax1.grid(True)
  ax1.set_ylim(-1.5, 1.5)

  # Sum of Absolute Values (Sh)
  ax2.plot(n, partial_sum, "r-o", label="Partial Sum Sh")
  if abs(alpha) < 1:
    theoretical_limit = 1 / (1 - abs(alpha))
    ax2.axhline(
        theoretical_limit,
        color="g",
        linestyle="--",
        label=f"Limit: {theoretical_limit:.2f}",
    )
    status = "STABLE (|alpha| < 1)"
    color = "green"
  else:
    status = "UNSTABLE (|alpha| >= 1)"
    color = "red"

  ax2.set_title(f"Status: {status}", color=color, fontweight="bold")
  ax2.set_xlabel("n")
  ax2.set_ylabel("Sum |h[k]|")
  ax2.grid(True)
  ax2.legend()
  ax2.set_ylim(0, 15)

  plt.tight_layout()
  plt.show()


# Create Interactive Slider
alpha_slider = widgets.FloatSlider(
    value=0.5,
    min=-1.5,
    max=1.5,
    step=0.05,
    description="Parameter alpha:",
    continuous_update=False,
)

widgets.interactive(plot_stability, alpha=alpha_slider)
```
