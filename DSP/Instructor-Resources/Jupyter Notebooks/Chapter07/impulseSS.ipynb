```python
import sympy as sp

# Definition of symbolic variables
n = sp.Symbol('n', integer=True)
k = sp.Symbol('k', integer=True)

# 1. Given matrices from Problem 071105
A = sp.Matrix([[0, 1], [sp.Rational(-1, 6), sp.Rational(5, 6)]])
B = sp.Matrix([[0], [1]])
C = sp.Matrix([[-1, 5]])
D = sp.Matrix([[0]])

# 2. State Transition Matrix A^n using Diagonalization
P, Lambda = A.diagonalize()
Lambda_n = Lambda.applyfunc(lambda i: i**n)
A_n = sp.simplify(P * Lambda_n * P.inv())

# 3. Calculation of C * A^(n-1) * B (for the expression inside the brackets)
# Αντικαθιστούμε το n με n-1 στον πίνακα A^n
A_n_minus_1 = A_n.subs(n, n - 1)
term_matrix = sp.simplify((C * A_n_minus_1 * B)[0, 0])

u_shift = sp.Function('u')(n - 1)
h_n_final = term_matrix * u_shift

print('Total Impulse Response h[n]:')
display(h_n_final)
```
