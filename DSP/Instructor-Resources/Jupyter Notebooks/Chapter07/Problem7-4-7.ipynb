```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, Markdown
import sympy as sp
from ipywidgets import interact, IntSlider, Layout

@interact(N=IntSlider(min=1, max=10, step=1, value=3, description='Delay N:', layout=Layout(width='400px')))
def plot_ideal_time_delay(N):
    A = 2.5
    omega_0 = np.pi / 4.0
    phi = np.pi / 6.0

    display(Markdown(f"### Ideal Time Delay Subsystem Analysis ($N = {N}$)"))
    display(Markdown(r"**Input Signal:** $x[n] = " + f"{A}" + r"\cos(" + f"{omega_0:.3f}" + r"n + " + f"{phi:.3f}" + r")$"))

    omega_vals = np.linspace(-np.pi, np.pi, 1000)
    mag_H_vals = np.ones_like(omega_vals)
    phase_H_vals = -omega_vals * N

    mag_H_val = 1.0
    phase_H_val = -omega_0 * N

    display(Markdown(r"**Calculated Values at $\omega_0$:** $|\mathcal{H}(e^{j\omega_0})| = 1.000$, $\angle\mathcal{H}(e^{j\omega_0}) = " + f"{phase_H_val:.3f}" + r" \text{ rad}$"))

    n = sp.symbols('n', integer=True)
    y_symbolic = A * sp.cos(sp.Float(omega_0, 3) * (n - N) + sp.Float(phi, 3))

    display(Markdown(r"**System Output Expression $y[n] = x[n - N]$:**"))
    display(sp.Eq(sp.Symbol('y[n]'), y_symbolic))

    n_vals = np.arange(0, 41)
    x = A * np.cos(omega_0 * n_vals + phi)
    y = A * np.cos(omega_0 * (n_vals - N) + phi)

    fig, axes = plt.subplots(3, 1, figsize=(10, 7.5))

    axes[0].stem(n_vals, x, linefmt='b-', markerfmt='bo', basefmt='k-', label='Input $x[n]$')
    axes[0].stem(n_vals, y, linefmt='r-', markerfmt='ro', basefmt='k-', label=f'Output $y[n] = x[n-{N}]$')
    axes[0].set_title('Input and Output Signals (Time Shifted by N)', fontsize=9.5, fontweight='bold', color='darkblue')
    axes[0].set_ylabel('Amplitude', fontsize=8.5)
    axes[0].grid(True, linestyle='--', alpha=0.6)
    axes[0].legend(loc='upper right')

    axes[1].plot(omega_vals, mag_H_vals, 'b-', linewidth=1.5, label='Magnitude')
    axes[1].axvline(x=omega_0, color='g', linestyle='--', label='Input Frequency')
    axes[1].scatter([omega_0], [mag_H_val], color='g', zorder=5)
    axes[1].text(omega_0 + 0.1, mag_H_val + 0.35, 'H(e^{j\\omega_0}) = 1.000', color='g', fontweight='bold', va='bottom')
    axes[1].set_title('Magnitude Response |H(e^{j\\omega})|', fontsize=9.5, fontweight='bold', color='darkblue')
    axes[1].set_ylabel('Magnitude', fontsize=8.5)
    axes[1].set_ylim(-0.2, 1.8)
    axes[1].grid(True, linestyle='--', alpha=0.6)
    axes[1].legend(loc='upper right')

    axes[2].plot(omega_vals, phase_H_vals, 'r-', linewidth=1.5, label='Phase')
    axes[2].axvline(x=omega_0, color='g', linestyle='--', label='Input Frequency')
    axes[2].scatter([omega_0], [phase_H_val], color='g', zorder=5)
    axes[2].text(omega_0 + 0.1, phase_H_val + 0.35, f'Phase ~ {phase_H_val:.3f} rad', color='g', fontweight='bold', va='bottom')
    axes[2].set_title('Phase Response (Linear Phase)', fontsize=9.5, fontweight='bold', color='darkred')
    axes[2].set_xlabel('Frequency \\omega (rad/sample)', fontsize=8.5)
    axes[2].set_ylabel('Phase (rad)', fontsize=8.5)
    axes[2].grid(True, linestyle='--', alpha=0.6)
    axes[2].legend(loc='upper right')

    plt.tight_layout()
    plt.show()
```
