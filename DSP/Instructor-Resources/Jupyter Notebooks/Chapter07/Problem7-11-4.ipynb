```python
import sympy as sp

# Definition of symbolic variables
n = sp.Symbol("n", integer=True)

print("=== CALCULATION OF A^n FOR THE 3x3 SYSTEM ===")
# Define matrix A (3x3)
A = sp.Matrix([[3, 0, 0], [0, -2, 1], [0, 4, 1]])

print("Matrix A:")
display(A)

# Diagonalization method with advanced algebraic simplification
P, D = A.diagonalize()
Lambda_n = D.applyfunc(lambda i: i**n)

# Compute A^n and apply full algebraic simplification
A_n = sp.simplify(P * Lambda_n * P.inv())

print("Simplified State Transition Matrix A^n:")
display(A_n)
```
