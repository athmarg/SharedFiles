```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, FloatSlider, VBox, HBox
from IPython.display import display

t = np.linspace(0, 1, 2000)
rng = np.random.default_rng()

A0, f0, phi0 = rng.uniform(0.5, 2.0, 3), rng.uniform(1, 10, 3), rng.uniform(0, 2*np.pi, 3)

# Δημιουργία των 9 sliders
s = [
    [
        FloatSlider(value=A0[0], min=0, max=3, step=0.05, description="A1", continuous_update=True),
        FloatSlider(value=f0[0], min=0.5, max=20, step=0.1, description="f1", continuous_update=True),
        FloatSlider(value=phi0[0], min=0, max=2*np.pi, step=0.05, description="φ1", continuous_update=True)
    ],
    [
        FloatSlider(value=A0[1], min=0, max=3, step=0.05, description="A2", continuous_update=True),
        FloatSlider(value=f0[1], min=0.5, max=20, step=0.1, description="f2", continuous_update=True),
        FloatSlider(value=phi0[1], min=0, max=2*np.pi, step=0.05, description="φ2", continuous_update=True)
    ],
    [
        FloatSlider(value=A0[2], min=0, max=3, step=0.05, description="A3", continuous_update=True),
        FloatSlider(value=f0[2], min=0.5, max=20, step=0.1, description="f3", continuous_update=True),
        FloatSlider(value=phi0[2], min=0, max=2*np.pi, step=0.05, description="φ3", continuous_update=True)
    ]
]

def plot_signals(A1, f1, phi1, A2, f2, phi2, A3, f3, phi3):
    sig1 = A1 * np.sin(2 * np.pi * f1 * t + phi1)
    sig2 = A2 * np.sin(2 * np.pi * f2 * t + phi2)
    sig3 = A3 * np.sin(2 * np.pi * f3 * t + phi3)
    total = sig1 + sig2 + sig3
    
    fig, axes = plt.subplots(4, 1, figsize=(9, 6), sharex=True)
    signals = [sig1, sig2, sig3, total]
    colors = ['b', 'b', 'b', 'red']
    labels = [r"$x_1(t)$", r"$x_2(t)$", r"$x_3(t)$", r"$x(t)=\sum x_i$"]
    
    for k in range(4):
        axes[k].plot(t, signals[k], color=colors[k], linewidth=1.5 if k < 3 else 2)
        axes[k].set_ylabel(labels[k])
        axes[k].grid(True)
        
    axes[3].set_xlabel("Time")
    plt.tight_layout()
    plt.show()
    plt.close(fig)

# Χρήση του interactive αντί για το @interact για να ελέγχουμε εμείς τα widgets
interactive_plot = interactive(
    plot_signals,
    A1=s[0][0], f1=s[0][1], phi1=s[0][2],
    A2=s[1][0], f2=s[1][1], phi2=s[1][2],
    A3=s[2][0], f3=s[2][1], phi3=s[2][2]
)

# Οργάνωση: Οι 3 γραμμές με τα sliders πάνω, και από κάτω μόνο το γράφημα (χωρίς τα διπλά sliders)
ui = VBox([
    HBox(s[0]),
    HBox(s[1]),
    HBox(s[2]),
    interactive_plot.children[-1]  # Εμφανίζει μόνο το Output (το γράφημα)
])

display(ui)
```
