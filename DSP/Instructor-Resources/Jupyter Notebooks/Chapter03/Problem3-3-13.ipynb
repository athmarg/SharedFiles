```python
# Notebook: Frequency Shift Property Visualization
# F{1} = 2*pi*delta(omega)  -->  F{e^(j*alpha*t)} = 2*pi*delta(omega - alpha)

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

# ==========================
# Plotting Function with Interact
# ==========================
def plot_frequency_shift(alpha=3.0):
    # Frequency range for omega
    omega = np.linspace(-10.0, 10.0, 1000)
    
    # Create figure
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 6), sharex=True)
    
    # 1. Fourier Transform of x(t) = 1  -->  2*pi*delta(omega)
    # Δείχνουμε την κρουστική στο omega = 0 με ένα βέλος
    ax1.axvline(0, color='b', linewidth=2, ymin=0, ymax=0.9)
    ax1.plot(0, 2 * np.pi, 'bo', markersize=8) # "Κορυφή" Dirac
    ax1.set_ylabel(r"$\mathcal{F}\{1\}$", fontsize=11)
    ax1.set_title(r"Frequency Shift Property: $\mathcal{F}\{e^{j\alpha t} x(t)\} = \mathcal{X}(j(\omega - \alpha))$", fontsize=12)
    ax1.grid(True, linestyle=":", alpha=0.7)
    ax1.set_xlim(-10.0, 10.0)
    ax1.set_ylim(-1.0, 8.0)
    ax1.text(0.3, 5.0, r"$2\pi\delta(\omega)$", color='blue', fontsize=12)
    
    # 2. Fourier Transform of x(t) = e^(j*alpha*t) * 1  -->  2*pi*delta(omega - alpha)
    # Η κρουστική μετατοπίζεται στη θέση omega = alpha
    ax1_color = 'r'
    ax2.axvline(alpha, color=ax1_color, linewidth=2, ymin=0, ymax=0.9)
    ax2.plot(alpha, 2 * np.pi, 'ro', markersize=8) # "Κορυφή" Dirac στη νέα θέση
    ax2.set_xlabel(r"Frequency $\omega$ (rad/s)", fontsize=11)
    ax2.set_ylabel(r"$\mathcal{F}\{e^{j\alpha t}\}$", fontsize=11)
    ax2.grid(True, linestyle=":", alpha=0.7)
    ax2.set_xlim(-10.0, 10.0)
    ax2.set_ylim(-1.0, 8.0)
    ax2.axhline(0, color='black', linewidth=0.8, linestyle='--')
    ax2.text(alpha + 0.3, 5.0, r"$2\pi\delta(\omega - \alpha)$", color='red', fontsize=12)
    
    plt.tight_layout()
    plt.show()
    
    # Εκτύπωση παρατήρησης
    print(f"Observation (alpha = {alpha:.1f}):")
    print("-> Multiplying signal x(t) = 1 by the complex exponential e^(j*alpha*t) shifts its spectrum")
    print(f"   from omega = 0 to the new frequency omega = {alpha:.1f} (Frequency Shift Property).")

# ==========================
# Interactive Widget Display
# ==========================
interact(
    plot_frequency_shift, 
    alpha=FloatSlider(value=3.0, min=-5.0, max=5.0, step=0.5, description="Shift alpha", style={'description_width': 'initial'})
);
```
