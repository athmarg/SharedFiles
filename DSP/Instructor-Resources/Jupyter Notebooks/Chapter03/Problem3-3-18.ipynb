```python
# Notebook: Fourier Transform of a Triangular Signal via Properties
# Signal: x(t) = Lambda_alpha(t) (Triangular pulse)
# Transform: X(omega) = (4 / (alpha * omega^2)) * sin^2(omega * alpha / 2)

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

# ==========================
# Συνάρτηση Οπτικοποίησης
# ==========================
def plot_triangular_transform(alpha=2.0):
    # 1. Περιοχή χρόνου t για το τριγωνικό σήμα Lambda_alpha(t)
    t = np.linspace(-5.0, 5.0, 1000)
    xt = np.maximum(0, 1 - np.abs(t) / alpha)
    xt[np.abs(t) > alpha] = 0
    
    # 2. Περιοχή συχνοτήτων omega για το φάσμα X(omega)
    omega = np.linspace(-15.0, 15.0, 1000)
    X_omega = np.zeros_like(omega)
    non_zero = omega != 0
    X_omega[non_zero] = (4.0 / (alpha * omega[non_zero]**2)) * np.sin(omega[non_zero] * alpha / 2.0)**2
    X_omega[~non_zero] = alpha  # Το όριο καθώς omega -> 0 είναι alpha
    
    # 3. Δημιουργία διαγραμμάτων (Subplots)
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 7))
    
    # --- Time Domain Plot ---
    ax1.plot(t, xt, 'b-', linewidth=2.5, label=r'$\Lambda_\alpha(t)$')
    ax1.set_title(r"Triangular Signal and its Fourier Transform ($\alpha$ = %.1f)" % alpha, fontsize=12)
    ax1.set_xlabel(r"Time $t$", fontsize=11)
    ax1.set_ylabel(r"$x(t)$", fontsize=11)
    ax1.grid(True, linestyle=":", alpha=0.7)
    ax1.set_xlim(-5.0, 5.0)
    ax1.set_ylim(-0.2, 1.2)
    ax1.legend(fontsize=11)
    
    # --- Frequency Domain Plot ---
    ax1_color = 'r'
    ax2.plot(omega, X_omega, color=ax1_color, linewidth=2.5, label=r'$\mathcal{X}(j\omega) = \frac{4}{\alpha\omega^2}\sin^2\left(\frac{\omega\alpha}{2}\right)$')
    ax2.set_xlabel(r"Frequency $\omega$ (rad/s)", fontsize=11)
    ax2.set_ylabel(r"$\mathcal{X}(j\omega)$", fontsize=11)
    ax2.grid(True, linestyle=":", alpha=0.7)
    ax2.set_xlim(-15.0, 15.0)
    ax2.legend(fontsize=11)
    
    plt.tight_layout()
    plt.show()
    
    # Εκτύπωση επεξηγηματικών παρατηρήσεων
    print(f"Observation (alpha = {alpha:.1f}):")
    print("-> (a) Convolution Property: Lambda_alpha(t) results from the convolution of two rectangular pulses.")
    print("-> (b) Differentiation & Time Shift: The 2nd derivative yields impulses at -alpha, 0, and +alpha.")
    print("-> Both approaches analytically yield the squared sinc-like spectrum shown above.")

# ==========================
# Εφαρμογή Interactive Widget
# ==========================
interact(
    plot_triangular_transform, 
    alpha=FloatSlider(value=2.0, min=0.5, max=4.0, step=0.5, description="Width alpha", style={'description_width': 'initial'})
);
```
