```python
# Notebook: Fourier Transform Duality in Normal Distribution
# Time Domain: x(t) = (1 / sqrt(2*pi*sigma^2)) * exp(-t^2 / (2*sigma^2))
# Frequency Domain: X(j\omega) = exp(-omega^2 * sigma^2 / 2)

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

# ==========================
# Plotting Function with Interact
# ==========================
def plot_fourier_transform(sigma=1.0):
    # Time domain range for t
    t = np.linspace(-10.0, 10.0, 1000)
    # Frequency range for omega
    omega = np.linspace(-10.0, 10.0, 1000)
    
    # Time domain signal: Normal distribution
    x_t = (1.0 / np.sqrt(2.0 * np.pi * sigma**2)) * np.exp(-(t**2) / (2.0 * sigma**2))
    
    # Frequency domain calculations
    magnitude = np.exp(-(omega**2) * (sigma**2) / 2.0)
    phase = np.zeros_like(omega) # Phase is identically zero
    
    # Create figure with 3 subplots (Time Domain, Magnitude, Phase)
    fig, (ax0, ax1, ax2) = plt.subplots(3, 1, figsize=(10, 10))
    
    # 0. Time Domain Signal
    time_label = r"$x(t) = \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left(-\frac{t^2}{2\sigma^2}\right)$"
    ax0.plot(t, x_t, 'g-', linewidth=2, label=time_label)
    ax0.set_ylabel(r"Time Signal $x(t)$", fontsize=11)
    ax0.set_title(r"Duality Property of Gaussian Signal ($\sigma$ = %.1f)" % sigma, fontsize=12)
    ax0.grid(True, linestyle=":", alpha=0.7)
    ax0.set_xlim(-10.0, 10.0)
    ax0.set_ylim(0.0, 0.9)  # Σταθερό όριο Y για τον χρόνο
    ax0.legend(loc="upper right", frameon=True)
    
    # 1. Magnitude Spectrum
    mag_label = r"$|\mathcal{X}(j\omega)| = \exp\left(-\frac{\omega^2 \sigma^2}{2}\right)$"
    ax1.plot(omega, magnitude, 'b-', linewidth=2, label=mag_label)
    ax1.set_ylabel(r"Magnitude $|\mathcal{X}(j\omega)|$", fontsize=11)
    ax1.grid(True, linestyle=":", alpha=0.7)
    ax1.set_xlim(-10.0, 10.0)
    ax1.set_ylim(0.0, 1.1)  # Σταθερό όριο Y πλάτους
    ax1.legend(loc="upper right", frameon=True)
    
    # 2. Phase Spectrum
    phase_label = r"$\angle \mathcal{X}(j\omega) = 0$ (Real & Even)"
    ax2.plot(omega, phase, 'r-', linewidth=2, label=phase_label)
    ax2.set_xlabel(r"Time $t$ / Frequency $\omega$ (rad/s)", fontsize=11)
    ax2.set_ylabel(r"Phase (rad)", fontsize=11)
    ax2.grid(True, linestyle=":", alpha=0.7)
    ax2.set_xlim(-10.0, 10.0)
    ax2.set_ylim(-0.5, 0.5)  # Σταθερό όριο Y φάσης
    ax2.axhline(0, color='black', linewidth=0.8, linestyle='--')
    ax2.legend(loc="upper right", frameon=True)
    
    plt.tight_layout()
    plt.show()
    
    # Εκτύπωση σχολίου στα αγγλικά κάτω από το γράφημα
    print(f"Observation (sigma = {sigma:.1f}):")
    print("-> The Fourier transform of a normal distribution is itself a normal distribution (Gaussian spectrum).")
    print("-> Since the Gaussian signal is real and even, its spectrum is entirely real and its phase is zero.")
    print("-> As sigma increases in time, the time-domain pulse widens, while the frequency spectrum becomes narrower (dual property).")

# ==========================
# Interactive Widget Display
# ==========================
interact(
    plot_fourier_transform, 
    sigma=FloatSlider(value=1.0, min=0.5, max=3.0, step=0.1, description="Parameter sigma", style={'description_width': 'initial'})
);
```
