```python
# Notebook: Interactive High-Pass Filter Analysis (Ideal & Real)
# Goal: Dynamically explore high-pass filter frequency responses with adjustable cutoff frequency \omega_c.

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

def plot_interactive_high_pass(omega_c):
    omega_vals = np.linspace(-25, 25, 1200)
    
    # 1. Ideal High-Pass Magnitude Response: 1 for |\omega| >= \omega_c, else 0
    ideal_hp = np.where(np.abs(omega_vals) >= omega_c, 1.0, 0.0)
    
    # 2. Real High-Pass Magnitude Response with smooth transition band and ripples
    real_hp = np.zeros_like(omega_vals)
    transition_width = 3.5
    
    for i, w in enumerate(np.abs(omega_vals)):
        if w < omega_c - transition_width:
            # Stopband with small attenuation ripples near zero
            real_hp[i] = 0.04 * np.sin(2 * w) * (w / (omega_c - transition_width) if omega_c > transition_width else 1.0)
        elif omega_c - transition_width <= w < omega_c:
            # Transition band rising smoothly up towards 0.7071 at \omega_c
            fraction = (w - (omega_c - transition_width)) / transition_width
            real_hp[i] = 0.5 * (1.0 - np.cos(np.pi * fraction)) + 0.04 * np.sin(2 * w)
        else:
            # Passband for |\omega| >= \omega_c with minor passband ripples
            real_hp[i] = 1.0 + 0.04 * np.cos(3 * w)

    # Create figure with two subplots: Ideal HP vs Real HP
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(11, 8), sharex=True)
    
    # Plot Ideal High-Pass Filter
    ax1.plot(omega_vals, ideal_hp, 'b-', linewidth=2.5, label=r"Ideal HP $|\mathcal{H}_{hp}(j\omega)|$")
    ax1.axvline(omega_c, color='red', linestyle='--', linewidth=1.5, label=r"Cutoff $\omega_c = %.1f$" % omega_c)
    ax1.axvline(-omega_c, color='red', linestyle='--', linewidth=1.5)
    ax1.set_title(r"Interactive Ideal and Real High-Pass Filter Responses", fontsize=13)
    ax1.set_ylabel(r"Amplitude $|\mathcal{H}(j\omega)|$", fontsize=11)
    ax1.grid(True, linestyle=":", alpha=0.7)
    ax1.axhline(0, color='black', linewidth=0.8, linestyle='--')
    ax1.set_ylim(-0.15, 1.25)
    # Legend placed outside upper right
    ax1.legend(loc='upper left', bbox_to_anchor=(1.02, 1), fontsize=10)
    
    # Plot Real High-Pass Filter
    ax2.plot(omega_vals, real_hp, 'g-', linewidth=2, label=r"Real HP (ripples & transition band)")
    ax2.axhline(0.7071, color='orange', linestyle=':', linewidth=1.5, label=r"-3 dB point ($0.7071$)")
    ax2.axvline(omega_c, color='red', linestyle='--', linewidth=1.5, label=r"Cutoff $\omega_c$")
    ax2.axvline(-omega_c, color='red', linestyle='--', linewidth=1.5)
    ax2.set_xlabel(r"Frequency $\omega$ (rad/s)", fontsize=11)
    ax2.set_ylabel(r"Amplitude $|\mathcal{H}(j\omega)|$", fontsize=11)
    ax2.grid(True, linestyle=":", alpha=0.7)
    ax2.axhline(0, color='black', linewidth=0.8, linestyle='--')
    ax2.set_ylim(-0.15, 1.25)
    # Legend placed outside upper right
    ax2.legend(loc='upper left', bbox_to_anchor=(1.02, 1), fontsize=10)
    
    plt.tight_layout()
    plt.show()

# Interactive widget slider for cutoff frequency omega_c
widgets.interact(
    plot_interactive_high_pass, 
    omega_c=widgets.FloatSlider(value=5.0, min=1.0, max=15.0, step=0.5, description=r'$\omega_c$ (rad/s):', style={'description_width': 'initial'})
);
```
