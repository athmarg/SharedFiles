```python
# Notebook: Fourier Transform Magnitude and Phase Example
# Signal: x(t) = A * exp(-alpha * t) * u(t)
# Transform: X(j\omega) = A / (alpha + j\omega)

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

# ==========================
# Plotting Function with Interact
# ==========================
def plot_fourier_transform(A=1.0, alpha=1.0):
    # Frequency range for omega
    omega = np.linspace(-20.0, 20.0, 1000)
    
    # Magnitude and Phase calculations
    magnitude = np.abs(A / (alpha + 1j * omega))
    phase = np.angle(A / (alpha + 1j * omega)) # in radians
    
    # Create figure with 2 subplots (Magnitude and Phase)
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 7), sharex=True)
    
    # 1. Magnitude Spectrum
    ax1.plot(omega, magnitude, 'b-', linewidth=2, label=r"$|\mathcal{X}(j\omega)| = \frac{A}{\sqrt{\alpha^2 + \omega^2}}$")
    ax1.set_ylabel(r"Magnitude $|\mathcal{X}(j\omega)|$", fontsize=11)
    ax1.set_title(r"Fourier Transform of $x(t) = A e^{-\alpha t}u(t)$  (A = %.1f, $\alpha$ = %.1f)" % (A, alpha), fontsize=12)
    ax1.grid(True, linestyle=":", alpha=0.7)
    ax1.set_xlim(-20.0, 20.0)
    ax1.set_ylim(0.0, 5.0)  # Σταθερό όριο Y πλάτους από 0 έως 3
    ax1.legend(loc="upper right", frameon=True)
    
    # 2. Phase Spectrum
    ax2.plot(omega, phase, 'r-', linewidth=2, label=r"$\angle \mathcal{X}(j\omega) = -\arctan\left(\frac{\omega}{\alpha}\right)$")
    ax2.set_xlabel(r"Frequency $\omega$ (rad/s)", fontsize=11)
    ax2.set_ylabel(r"Phase $\angle \mathcal{X}(j\omega)$ (rad)", fontsize=11)
    ax2.grid(True, linestyle=":", alpha=0.7)
    ax2.set_xlim(-20.0, 20.0)
    ax2.set_ylim(-2.0, 2.0)  # Σταθερό όριο Y φάσης από -2 έως 2
    ax2.axhline(0, color='black', linewidth=0.8, linestyle='--')
    ax2.legend(loc="upper right", frameon=True)
    
    plt.tight_layout()
    plt.show()

# ==========================
# Interactive Widget Display
# ==========================
interact(
    plot_fourier_transform, 
    A=FloatSlider(value=1.0, min=0.1, max=5.0, step=0.1, description="Amplitude A", style={'description_width': 'initial'}),
    alpha=FloatSlider(value=1.0, min=0.2, max=5.0, step=0.2, description="Parameter alpha", style={'description_width': 'initial'})
);
```
