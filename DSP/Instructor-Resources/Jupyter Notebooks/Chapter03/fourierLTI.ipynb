```python
# Notebook: LTI System Response to a Periodic Signal via Fourier Expansion
# Author: Adapted for Springer-style presentation
# 
# Python functions used in this script:
# - numpy: For numerical arrays, trigonometric functions, and exponential evaluations.
# - scipy.integrate.quad: For numerical integration to compute the exponential Fourier coefficients.

import numpy as np
from scipy.integrate import quad

# ==========================
# 1. Define the Periodic Signal x(t)
# ==========================
T = 8.0  # Period
omega_0 = 2 * np.pi / T

def x_signal(t):
    # Square wave: +1 for 0 <= t < 4, -1 for 4 <= t < 8 (over one period shifted or modulo T)
    t_mod = t % T
    return np.where((t_mod >= 0) & (t_mod < 4), 1.0, -1.0)

# ==========================
# 2. Define the LTI System Frequency Response H(j*omega)
# ==========================
# H(j*omega) = sin(4*omega) / omega
def H_frequency_response(omega):
    # Avoid division by zero at omega = 0 using a safe limit or small check
    if np.isclose(omega, 0.0):
        return 4.0  # limit as omega -> 0 of sin(4*omega)/omega is 4
    return np.sin(4 * omega) / omega


# ==========================
# 3. Compute Fourier Coefficients & System Output Harmonics
# ==========================
N_harmonics = 5  # Number of positive/negative odd harmonics to check (m from -N to N)

print("=" * 70)
print(" LTI SYSTEM RESPONSE TO A PERIODIC SIGNAL VIA FOURIER EXPANSION")
print("=" * 70)
print(f"Input Signal: Square wave with Period T = {T}")
print(f"System Transfer Function: H(j*omega) = sin(4*omega) / omega\n")

print(f"{'Harmonic (m)':<15} | {'Frequency omega_m':<20} | {'Input Coeff alpha_m':<22} | {'H(j*omega_m)':<15} | {'Output Coeff':<15}")
print("-" * 93)

# For odd harmonics: omega_m = (2m+1) * pi / 4
for m in range(-N_harmonics, N_harmonics + 1):
    # Harmonic frequency
    omega_m = (2 * m + 1) * np.pi / 4
    
    # Theoretical input coefficient alpha_m for this square wave
    # alpha_m = 2 / (j * (2m+1) * pi) for odd m
    alpha_m = 2 / (1j * (2 * m + 1) * np.pi)
    
    # System response at this specific frequency
    H_val = H_frequency_response(omega_m)
    
    # Output coefficient = alpha_m * H(j*omega_m)
    output_coeff = alpha_m * H_val
    
    print(f"{m:<15} | {omega_m:<20.4f} | {alpha_m.imag:+.4f}j              | {H_val:<15.4f} | {output_coeff.real:+.4f} {output_coeff.imag:+.4f}j")

print("-" * 93)
print("THEORETICAL INSIGHT:")
print("- Notice that H(j*omega_m) evaluates strictly to 0 for all odd harmonics")
print("  because sin((2m+1)*pi) = 0.")
print("- Consequently, every output harmonic coefficient becomes 0, verifying")
print("  that the total system output is identically zero: y(t) = 0.")
print("=" * 70)
```
