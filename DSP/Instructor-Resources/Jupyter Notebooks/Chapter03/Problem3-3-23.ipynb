```python
# Notebook: Fourier Transform of the Autocorrelation Function for a Periodic Signal
# Signal: x(t) = Periodic signal (e.g., sum of cosines)
# Autocorrelation Transform: R(omega) = 2pi * sum(|alpha_n|^2 * delta(omega - n*omega_0))

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, IntSlider

# ==========================
# Συνάρτηση Οπτικοποίησης
# ==========================
def plot_periodic_autocorrelation_transform(harmonics=3):
    T = 2.0  # Περίοδος
    omega_0 = 2 * np.pi / T
    
    # Ορίζουμε τα amplitudes (συντελεστές Fourier alpha_n) για μερικές αρμονικές
    n_vals = np.arange(-harmonics, harmonics + 1)
    
    # Αποφυγή RuntimeWarning με ασφαλή διαίρεση όπου n_vals != 0
    alpha_n = np.ones_like(n_vals, dtype=float)
    non_zero = n_vals != 0
    alpha_n[non_zero] = 1.0 / np.abs(n_vals[non_zero])
    alpha_n[~non_zero] = 1.0  # Για n = 0
    
    alpha_n = alpha_n / np.sum(alpha_n)  # Κανονικοποίηση
    
    # Τα πλάτη στο φάσμα της αυτοσυσχέτισης R(omega) είναι 2*pi * |alpha_n|^2
    R_amplitudes = 2 * np.pi * (np.abs(alpha_n)**2)
    
    # Θέσεις συχνοτήτων των κρουστικών: n * omega_0
    omega_positions = n_vals * omega_0
    
    # Δημιουργία διαγράμματος
    fig, ax = plt.subplots(figsize=(10, 5))
    
    # Σχεδίαση κρουστικών (stem plot) για το φάσμα R(omega)
    markerline, stemlines, baseline = ax.stem(
        omega_positions, R_amplitudes, linefmt='purple', markerfmt='mo', basefmt=" "
    )
    plt.setp(stemlines, linewidth=2.5)
    
    ax.set_title(r"Fourier Transform of Autocorrelation for a Periodic Signal ($\mathcal{R}(\omega)$)", fontsize=12)
    ax.set_xlabel(r"Frequency $\omega$ (rad/s)", fontsize=11)
    ax.set_ylabel(r"$\mathcal{R}(\omega) = 2\pi \sum |\alpha_n|^2 \delta(\omega - n\omega_0)$", fontsize=11)
    ax.grid(True, linestyle=":", alpha=0.7)
    
    max_w = (harmonics + 1) * omega_0
    ax.set_xlim(-max_w, max_w)
    ax.set_ylim(0, max(R_amplitudes) * 1.15)
    ax.axhline(0, color='black', linewidth=0.8, linestyle='--')
    
    plt.tight_layout()
    plt.show()
    
    # Εκτύπωση παρατηρήσεων στα αγγλικά
    print(f"Observation (Number of harmonics = {harmonics}):")
    print("-> The autocorrelation spectrum R(omega) consists of impulses located at multiples of omega_0.")
    print("-> The height of each impulse is proportional to the POWER (|alpha_n|^2) of each harmonic component.")

# ==========================
# Εφαρμογή Interactive Widget
# ==========================
interact(
    plot_periodic_autocorrelation_transform, 
    harmonics=IntSlider(value=3, min=1, max=7, step=1, description="Harmonics", style={'description_width': 'initial'})
);
```
