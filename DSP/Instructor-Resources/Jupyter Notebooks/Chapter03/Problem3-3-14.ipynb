```python
# Notebook: Fourier Transform Modulation & Frequency Shift Properties (Cosine & Sine)
# Signals: x1(t) = cos(alpha * t) and x2(t) = sin(alpha * t)

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

# ==========================
# Plotting Function with Interact
# ==========================
def plot_modulation_property(alpha=4.0):
    omega = np.linspace(-10.0, 10.0, 1000)
    
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 6), sharex=True)
    
    peak_val = 2.5
    
    # 1. Cosine Spectrum (Χρήση stem για τέλεια ευθυγράμμιση γραμμής και τελείας)
    # Δημιουργούμε τα δεδομένα για τα stems στις θέσεις -alpha και alpha
    markerline1, stemlines1, baseline1 = ax1.stem(
        [-alpha, alpha], [peak_val, peak_val], linefmt='b-', markerfmt='bo', basefmt=" "
    )
    plt.setp(stemlines1, linewidth=2)
    
    ax1.set_ylabel(r"Magnitude $|\mathcal{F}\{\cos(\alpha t)\}|$ ", fontsize=11)
    ax1.set_title(r"Modulation Property: Spectra of $\cos(\alpha t)$ and $\sin(\alpha t)$ ($\alpha$ = %.1f)" % alpha, fontsize=12)
    ax1.grid(True, linestyle=":", alpha=0.7)
    ax1.set_xlim(-10.0, 10.0)
    ax1.set_ylim(-1.0, 5.0)
    
    ax1.text(-alpha - 0.4, peak_val, r"$\pi\delta(\omega+\alpha)$", color='blue', fontsize=11, ha='right', va='center')
    ax1.text(alpha + 0.4, peak_val, r"$\pi\delta(\omega-\alpha)$", color='blue', fontsize=11, ha='left', va='center')
    
    # 2. Sine Spectrum (Αριστερά θετικό στο +peak_val, Δεξιά αρνητικό στο -peak_val)
    markerline2, stemlines2, baseline2 = ax2.stem(
        [-alpha, alpha], [peak_val, -peak_val], linefmt='r-', markerfmt='ro', basefmt=" "
    )
    plt.setp(stemlines2, linewidth=2)
    
    ax2.set_xlabel(r"Frequency $\omega$ (rad/s)", fontsize=11)
    ax2.set_ylabel(r"Imag. Part $\mathcal{F}\{\sin(\alpha t)\}/j$", fontsize=11)
    ax2.grid(True, linestyle=":", alpha=0.7)
    ax2.set_xlim(-10.0, 10.0)
    ax2.set_ylim(-5.0, 5.0)
    ax2.axhline(0, color='black', linewidth=0.8, linestyle='--')
    
    ax2.text(-alpha - 0.4, peak_val, r"$j\pi\delta(\omega+\alpha)$", color='red', fontsize=11, ha='right', va='center')
    ax2.text(alpha + 0.4, -peak_val, r"$-j\pi\delta(\omega-\alpha)$", color='red', fontsize=11, ha='left', va='center')
    
    plt.tight_layout()
    plt.show()

# ==========================
# Interactive Widget Display
# ==========================
interact(
    plot_modulation_property, 
    alpha=FloatSlider(value=4.0, min=1.0, max=8.0, step=0.5, description="Frequency alpha", style={'description_width': 'initial'})
);
```
