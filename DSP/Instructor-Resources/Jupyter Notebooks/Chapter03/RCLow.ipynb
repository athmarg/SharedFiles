```python
# Notebook: First-Order RC Low-Pass Filter (Horizontal Side-by-Side Layout)
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

def plot_rc_low_pass_side_by_side(R, C):
    tau = R * C
    cutoff_freq = 1.0 / tau if tau > 0 else 1.0
    
    omega = np.logspace(-1, 2, 1000)
    H_mag = 1.0 / np.sqrt(1.0 + (omega * tau)**2)
    H_mag_dB = 20.0 * np.log10(H_mag)
    H_phase = -np.arctan(omega * tau)
    
    t = np.linspace(0, 5 * tau, 1000) if tau > 0 else np.linspace(0, 5, 1000)
    h_t = (1.0 / tau) * np.exp(-t / tau) if tau > 0 else np.zeros_like(t)
    s_t = (1.0 - np.exp(-t / tau)) if tau > 0 else np.ones_like(t)

    # Create figure with 1 row, 3 columns side by side
    fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(16, 4.5))
    
    # 1. Magnitude Response in dB
    ax1.semilogx(omega, H_mag_dB, 'b-', linewidth=2, label=r"Magnitude $|\mathcal{H}|$ (dB)")
    ax1.axhline(-3.01, color='orange', linestyle=':', linewidth=1.5, label=r"-3 dB ($0.707$)")
    ax1.axvline(cutoff_freq, color='red', linestyle='--', linewidth=1.5, label=r"$\omega_c = %.2f$" % cutoff_freq)
    ax1.set_title(r"Magnitude Response", fontsize=11)
    ax1.set_ylabel(r"dB", fontsize=10)
    ax1.set_xlabel(r"$\omega$ (rad/s)", fontsize=10)
    ax1.grid(True, which="both", linestyle=":", alpha=0.7)
    ax1.legend(loc='upper center', bbox_to_anchor=(0.5, -0.2), fontsize=9)
    
    # 2. Phase Response
    ax2.semilogx(omega, np.rad2deg(H_phase), 'g-', linewidth=2, label=r"Phase $\angle \mathcal{H}$")
    ax2.axvline(cutoff_freq, color='red', linestyle='--', linewidth=1.5, label=r"$\omega_c$")
    ax2.set_title(r"Phase Response", fontsize=11)
    ax2.set_ylabel(r"Degrees", fontsize=10)
    ax2.set_xlabel(r"$\omega$ (rad/s)", fontsize=10)
    ax2.grid(True, which="both", linestyle=":", alpha=0.7)
    ax2.legend(loc='upper center', bbox_to_anchor=(0.5, -0.2), fontsize=9)
    
    # 3. Time Responses
    ax3.plot(t, h_t, 'm-', linewidth=2, label=r"Impulse $h(t)$")
    ax3.plot(t, s_t, 'c-', linewidth=2, label=r"Step $s(t)$")
    ax3.set_title(r"Time Responses", fontsize=11)
    ax3.set_ylabel(r"Amplitude", fontsize=10)
    ax3.set_xlabel(r"Time $t$ (s)", fontsize=10)
    ax3.grid(True, linestyle=":", alpha=0.7)
    ax3.legend(loc='upper center', bbox_to_anchor=(0.5, -0.2), fontsize=9)
    
    fig.suptitle(r"First-Order RC Low-Pass Filter ($R=%.1f\,\Omega,\, C=%.2f\,\text{F}$)" % (R, C), fontsize=13, y=1.03)
    plt.tight_layout()
    plt.show()

widgets.interact(
    plot_rc_low_pass_side_by_side, 
    R=widgets.FloatSlider(value=1.0, min=0.1, max=10.0, step=0.1, description=r'$R$:', style={'description_width': 'initial'}),
    C=widgets.FloatSlider(value=1.0, min=0.1, max=10.0, step=0.1, description=r'$C$:', style={'description_width': 'initial'})
);
```
