```python
# Notebook: Symbolic Derivation of Fourier Coefficients (Square Pulse Sequence)
# Author: Adapted for Springer-style presentation
# 
# Python functions used in this script:
# - sympy.symbols: Used to define mathematical symbols (t, A, tau, T, n).
# - sympy.integrate: Used for symbolic definite integration to find Fourier coefficients.
# - sympy.simplify: Used to simplify algebraic expressions.
# - display: Used in Jupyter to nicely render SymPy mathematical formulas.

import sympy as sp

# ==========================
# Symbolic Derivation (Paper & Pencil equivalent)
# ==========================

# 1. Define symbolic variables
t, A_sym, tau_sym, T_sym = sp.symbols('t A tau T', real=True, positive=True)
n = sp.symbols('n', integer=True, positive=True)

omega = 2 * sp.pi / T_sym

# 2. Define the square pulse over one period (-tau/2 to tau/2)
# Inside this interval the signal is A_sym, outside it is 0.
# 3. Compute A0 coefficient (DC term evaluation: average value)
a0 = (1 / T_sym) * sp.integrate(A_sym, (t, -tau_sym/2, tau_sym/2))
a0 = sp.simplify(a0)

# 4. Compute an coefficients (Cosine coefficients evaluation)
# an = (2 / T) * int_{-tau/2}^{tau/2} A * cos(n * omega * t) dt
an_integral = sp.integrate(A_sym * sp.cos(n * omega * t), (t, -tau_sym/2, tau_sym/2))
an = (2 / T_sym) * an_integral
an = sp.simplify(an)

# ==========================
# Presentation of Results & Insights
# ==========================

print("=" * 60)
print(" SYMBOLIC CALCULATION OF FOURIER COEFFICIENTS (SQUARE PULSE)")
print("=" * 60)

print("\n1. Mean Value / DC Coefficient (a_0):")
display(a0)

print("\n2. Cosine Coefficients (a_n):")
display(an)

print("-" * 60)
print("THEORETICAL REMARKS:")
print("- The square pulse signal is even (symmetric with respect to the y-axis),")
print("  therefore the sine coefficients b_n are zero (b_n = 0).")
print("- The coefficient a_n contains the sinc function: ")
print("  a_n = (A * tau / T) * sinc(n * omega * tau / 2).")
print("- Due to the nature of the sinc function, the spectrum exhibits zeros (nulls),")
print("  meaning the harmonic power is not strictly monotonically decreasing.")
print("=" * 60)
```
