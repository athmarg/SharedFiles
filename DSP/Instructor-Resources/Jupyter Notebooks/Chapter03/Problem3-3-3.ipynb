```python
# Notebook: Fourier Transform Magnitude and Phase Example
# Signal: x(t) = exp(-alpha * t) * cos(omega_0 * t) * u(t)
# Transform: X(j\omega) = (alpha + j\omega) / (alpha^2 + 2j\omega\alpha + (omega_0^2 - omega^2))

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

# ==========================
# Plotting Function with Interact
# ==========================
def plot_fourier_transform(alpha=1.0, omega_0=5.0):
    # Frequency range for omega
    omega = np.linspace(-20.0, 20.0, 1000)
    
    # Fourier Transform Calculation using the derived formula
    numerator = alpha + 1j * omega
    denominator = alpha**2 + 2j * omega * alpha + (omega_0**2 - omega**2)
    X = numerator / denominator
    
    # Magnitude and Phase calculations
    magnitude = np.abs(X)
    phase = np.angle(X) # in radians
    
    # Create figure with 2 subplots (Magnitude and Phase)
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 7), sharex=True)
    
    # 1. Magnitude Spectrum
    mag_label = r"$|\mathcal{X}(j\omega)| = \left| \frac{\alpha + j\omega}{\alpha^2 - \omega^2 + \omega_0^2 + j2\alpha\omega} \right|$"
    ax1.plot(omega, magnitude, 'b-', linewidth=2, label=mag_label)
    ax1.set_ylabel(r"Magnitude $|\mathcal{X}(j\omega)|$", fontsize=11)
    ax1.set_title(r"Fourier Transform of $x(t) = e^{-\alpha t}\cos(\omega_0 t)u(t)$  ($\alpha$ = %.1f, $\omega_0$ = %.1f)" % (alpha, omega_0), fontsize=12)
    ax1.grid(True, linestyle=":", alpha=0.7)
    ax1.set_xlim(-20.0, 20.0)
    ax1.set_ylim(0.0, 3.0)  # Σταθερό όριο Y πλάτους από 0 έως 3
    ax1.legend(loc="upper right", frameon=True)
    
    # 2. Phase Spectrum
    phase_label = r"$\angle \mathcal{X}(j\omega) = \arctan\left(\frac{\omega}{\alpha}\right) - \arctan\left(\frac{2\alpha\omega}{\alpha^2 - \omega^2 + \omega_0^2}\right)$"
    ax2.plot(omega, phase, 'r-', linewidth=2, label=phase_label)
    ax2.set_xlabel(r"Frequency $\omega$ (rad/s)", fontsize=11)
    ax2.set_ylabel(r"Phase $\angle \mathcal{X}(j\omega)$ (rad)", fontsize=11)
    ax2.grid(True, linestyle=":", alpha=0.7)
    ax2.set_xlim(-20.0, 20.0)
    ax2.set_ylim(-2.0, 2.0)  # Σταθερό όριο Y φάσης από -2 έως 2
    ax2.axhline(0, color='black', linewidth=0.8, linestyle='--')
    ax2.legend(loc="upper right", frameon=True)
    
    plt.tight_layout()
    plt.show()
    
    # Εκτύπωση σχολίου στα αγγλικά κάτω από το γράφημα
    print(f"Observation (alpha = {alpha:.1f}, omega_0 = {omega_0:.1f}):")
    print("-> The two peaks correspond to the resonance frequencies +omega_0 and -omega_0.")
    print("-> As alpha increases, the exponential damping exp(-alpha*t) becomes stronger in time,")
    print("   causing the spectral peaks to flatten, widen, and decrease in magnitude.")

# ==========================
# Interactive Widget Display
# ==========================
interact(
    plot_fourier_transform, 
    alpha=FloatSlider(value=1.0, min=0.18, max=5.0, step=0.2, description="Parameter alpha", style={'description_width': 'initial'}),
    omega_0=FloatSlider(value=5.0, min=1.0, max=10.0, step=0.5, description="Frequency omega_0", style={'description_width': 'initial'})
);
```
