```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import FloatSlider, VBox, HBox, Layout, HTML, HTMLMath
from IPython.display import display

plt.ioff()

# ============================================================
# JUPYTER DISPLAY SETTINGS
# Remove scrollbars generated by the notebook output container.
# ============================================================

display(HTML("""
<style>

/* JupyterLab output area */
.jp-OutputArea,
.jp-OutputArea-child,
.jp-OutputArea-output {
    overflow: visible !important;
    max-height: none !important;
    height: auto !important;
}

/* Classic Notebook output area */
.output,
.output_area,
.output_subarea,
.output_scroll {
    overflow: visible !important;
    max-height: none !important;
    height: auto !important;
}

/* Widget containers */
.jupyter-widgets,
.widget-box,
.widget-html,
.widget-html-content {
    overflow: visible !important;
    max-height: none !important;
}

/* Prevent an unnecessary horizontal scrollbar */
.jp-Cell-outputWrapper {
    overflow: visible !important;
}

</style>
"""))

# ============================================================
# NOTEBOOK DESCRIPTION
#
# Red solid curves:
#     Exact real and imaginary components calculated directly
#     from the analytical frequency response H(jω).
#
# Black dashed curves:
#     Components reconstructed numerically from the other
#     component using the principal-value Hilbert-transform
#     integrals.
#
# Ideally, the red and black curves coincide.
#
# Small differences are numerical and result mainly from
# replacing the infinite interval (-∞,+∞) with the finite
# numerical interval [-Ωmax,+Ωmax].
#
# Increasing Ωmax generally improves the approximation in the
# central frequency region.
# ============================================================

description = HTML("""
<div style="
    border:1px solid #9ec9f5;
    border-radius:7px;
    padding:7px 10px;
    margin:0px 0px 7px 0px;
    font-size:13px;
    line-height:1.42;
    background-color:#f7fbff;
    width:700px;
    max-width:700px;
    box-sizing:border-box;
">
<b>Purpose:</b>
Demonstrate the Hilbert-transform relation between the real and imaginary parts of the frequency response of a real causal system.
<br>
<b>Interpretation:</b>
The solid curves show the exact real and imaginary parts of H(jω), while the dashed curves reconstruct each component <b>using only the other component</b> through the principal-value Hilbert-transform integrals.
Their close agreement demonstrates that, for a causal system, the real and imaginary parts of H(jω) are not independent.
Small differences are due to the finite numerical integration range [−Ωmax,+Ωmax].
</div>
""", layout=Layout(width='710px', max_width='710px'))

# ============================================================
# EQUATIONS
# ============================================================

formula_1 = HTMLMath(
    value=r"""
    \[
    H(s)=\frac{1}{s+a},
    \qquad
    H(j\omega)=R(\omega)+jI(\omega)
    \]
    """,
    layout=Layout(width='700px', max_width='700px')
)

formula_2 = HTMLMath(
    value=r"""
    \[
    I(\omega)=\frac{1}{\pi}\mathcal{P}
    \int_{-\infty}^{+\infty}\frac{R(\eta)}{\eta-\omega}\,d\eta,
    \qquad
    R(\omega)=-\frac{1}{\pi}\mathcal{P}
    \int_{-\infty}^{+\infty}\frac{I(\eta)}{\eta-\omega}\,d\eta
    \]
    """,
    layout=Layout(width='700px', max_width='700px')
)

# ============================================================
# CONTROLS
# ============================================================

slider_layout = Layout(width='250px')
style_opts = {'description_width':'70px'}

a_slider = FloatSlider(
    min=0.5,
    max=5.0,
    step=0.1,
    value=1.0,
    description='a:',
    continuous_update=True,
    readout_format='.1f',
    style=style_opts,
    layout=slider_layout
)

omega_max_slider = FloatSlider(
    min=10.0,
    max=80.0,
    step=5.0,
    value=40.0,
    description='Ωmax:',
    continuous_update=True,
    readout_format='.0f',
    style=style_opts,
    layout=slider_layout
)

parameter_title = HTML("""
<div style="
    font-size:14px;
    font-weight:bold;
    margin-top:3px;
    margin-bottom:5px;
">
Parameters:
</div>
""")

info_html = HTML(layout=Layout(width='265px', max_width='265px'))

# ============================================================
# REAL-PART FIGURE
# ============================================================

fig_real, ax_real = plt.subplots(figsize=(6.5, 2.55))

real_exact_line, = ax_real.plot([], [], 'r-', linewidth=2.0, label='Exact R(ω)')
real_hilbert_line, = ax_real.plot([], [], 'k--', linewidth=1.6, label='Reconstructed from I(ω)')

ax_real.set_xlabel('Angular Frequency ω (rad/s)', fontsize=10)
ax_real.set_ylabel('Real Part', fontsize=10)
ax_real.set_title('Real Part of H(jω)', fontsize=12, fontweight='bold', pad=5)
ax_real.tick_params(axis='both', labelsize=9)
ax_real.grid(True, linestyle=':', alpha=0.5)
ax_real.axhline(0.0, color='gray', linewidth=0.8)
ax_real.legend(loc='center left', bbox_to_anchor=(1.01, 0.5), fontsize=9)

fig_real.subplots_adjust(left=0.12, right=0.75, bottom=0.22, top=0.84)

fig_real.canvas.header_visible = False
fig_real.canvas.toolbar_visible = False
fig_real.canvas.resizable = False
fig_real.canvas.layout.width = '650px'
fig_real.canvas.layout.height = '260px'
fig_real.canvas.layout.margin = '0px 0px -6px 0px'

# ============================================================
# IMAGINARY-PART FIGURE
# ============================================================

fig_imag, ax_imag = plt.subplots(figsize=(6.5, 2.55))

imag_exact_line, = ax_imag.plot([], [], 'r-', linewidth=2.0, label='Exact I(ω)')
imag_hilbert_line, = ax_imag.plot([], [], 'k--', linewidth=1.6, label='Reconstructed from R(ω)')

ax_imag.set_xlabel('Angular Frequency ω (rad/s)', fontsize=10)
ax_imag.set_ylabel('Imaginary Part', fontsize=10)
ax_imag.set_title('Imaginary Part of H(jω)', fontsize=12, fontweight='bold', pad=5)
ax_imag.tick_params(axis='both', labelsize=9)
ax_imag.grid(True, linestyle=':', alpha=0.5)
ax_imag.axhline(0.0, color='gray', linewidth=0.8)
ax_imag.legend(loc='center left', bbox_to_anchor=(1.01, 0.5), fontsize=9)

fig_imag.subplots_adjust(left=0.12, right=0.75, bottom=0.22, top=0.84)

fig_imag.canvas.header_visible = False
fig_imag.canvas.toolbar_visible = False
fig_imag.canvas.resizable = False
fig_imag.canvas.layout.width = '650px'
fig_imag.canvas.layout.height = '260px'
fig_imag.canvas.layout.margin = '0px'

# ============================================================
# PRINCIPAL-VALUE HILBERT TRANSFORM
# ============================================================

def principal_value_kernel(omega):
    denominator = omega[None, :] - omega[:, None]
    kernel = np.zeros_like(denominator)
    mask = ~np.eye(len(omega), dtype=bool)
    kernel[mask] = 1.0 / denominator[mask]
    return kernel

def hilbert_imag_from_real(omega, real_part):
    d_omega = omega[1] - omega[0]
    weights = np.ones_like(omega)
    weights[0] = 0.5
    weights[-1] = 0.5
    kernel = principal_value_kernel(omega)
    return (d_omega / np.pi) * (kernel @ (weights * real_part))

def hilbert_real_from_imag(omega, imag_part):
    d_omega = omega[1] - omega[0]
    weights = np.ones_like(omega)
    weights[0] = 0.5
    weights[-1] = 0.5
    kernel = principal_value_kernel(omega)
    return -(d_omega / np.pi) * (kernel @ (weights * imag_part))

# ============================================================
# UPDATE FUNCTION
# ============================================================

def update_plots(change=None):

    a = a_slider.value
    omega_max = omega_max_slider.value
    N = 1201

    omega_full = np.linspace(-omega_max, omega_max, N)

    real_exact = a / (a**2 + omega_full**2)
    imag_exact = -omega_full / (a**2 + omega_full**2)

    imag_reconstructed = hilbert_imag_from_real(omega_full, real_exact)
    real_reconstructed = hilbert_real_from_imag(omega_full, imag_exact)

    display_limit = 0.5 * omega_max
    display_mask = np.abs(omega_full) <= display_limit

    omega = omega_full[display_mask]

    real_exact_display = real_exact[display_mask]
    imag_exact_display = imag_exact[display_mask]

    real_reconstructed_display = real_reconstructed[display_mask]
    imag_reconstructed_display = imag_reconstructed[display_mask]

    real_exact_line.set_data(omega, real_exact_display)
    real_hilbert_line.set_data(omega, real_reconstructed_display)

    imag_exact_line.set_data(omega, imag_exact_display)
    imag_hilbert_line.set_data(omega, imag_reconstructed_display)

    real_min = min(np.min(real_exact_display), np.min(real_reconstructed_display))
    real_max = max(np.max(real_exact_display), np.max(real_reconstructed_display))

    imag_min = min(np.min(imag_exact_display), np.min(imag_reconstructed_display))
    imag_max = max(np.max(imag_exact_display), np.max(imag_reconstructed_display))

    real_margin = 0.08 * max(real_max - real_min, 1.0 / a)
    imag_margin = 0.08 * max(imag_max - imag_min, 1.0 / a)

    ax_real.set_xlim(-display_limit, display_limit)
    ax_real.set_ylim(real_min - real_margin, real_max + real_margin)

    ax_imag.set_xlim(-display_limit, display_limit)
    ax_imag.set_ylim(imag_min - imag_margin, imag_max + imag_margin)

    real_error = np.max(np.abs(real_exact_display - real_reconstructed_display))
    imag_error = np.max(np.abs(imag_exact_display - imag_reconstructed_display))

    info_html.value = f"""
    <div style="
        border:1px solid #cccccc;
        border-radius:7px;
        padding:8px 9px;
        margin-top:9px;
        font-size:12px;
        line-height:1.65;
        background:white;
        width:260px;
        box-sizing:border-box;
    ">

    <div>
        <b>System:</b>
        <span style="color:#0066cc;">H(s) = 1/(s+a)</span>
    </div>

    <div>
        <b>a:</b>
        <span style="color:#0066cc;">{a:.2f}</span>
    </div>

    <div>
        <b>Integration range:</b>
        <span style="color:#0066cc;">[−{omega_max:.0f}, +{omega_max:.0f}] rad/s</span>
    </div>

    <div>
        <b>Displayed range:</b>
        <span style="color:#0066cc;">[−{display_limit:.0f}, +{display_limit:.0f}] rad/s</span>
    </div>

    <div style="
        margin-top:6px;
        padding-top:6px;
        border-top:1px solid #eeeeee;
    ">
        <b>Exact components:</b><br>
        <span style="color:#0066cc;">
        R(ω) = a/(a²+ω²)<br>
        I(ω) = −ω/(a²+ω²)
        </span>
    </div>

    <div style="
        margin-top:6px;
        padding-top:6px;
        border-top:1px solid #eeeeee;
    ">
        <b>Maximum reconstruction error:</b><br>
        R from I:
        <span style="color:#0066cc;">{real_error:.4e}</span><br>
        I from R:
        <span style="color:#0066cc;">{imag_error:.4e}</span>
    </div>

    <div style="
        margin-top:6px;
        padding-top:6px;
        border-top:1px solid #eeeeee;
    ">
        <b>Observation:</b><br>
        The dashed Hilbert-transform reconstruction follows the exact solid curve.
        Small discrepancies are caused by replacing the infinite principal-value integral by a finite numerical interval.
    </div>

    </div>
    """

    fig_real.canvas.draw_idle()
    fig_imag.canvas.draw_idle()

# ============================================================
# CALLBACKS
# ============================================================

a_slider.observe(update_plots, names='value')
omega_max_slider.observe(update_plots, names='value')

# ============================================================
# LAYOUT
# ============================================================

controls = VBox(
    [parameter_title, a_slider, omega_max_slider, info_html],
    layout=Layout(
        width='275px',
        min_width='275px',
        max_width='275px',
        flex='0 0 275px',
        align_items='flex-start'
    )
)

plots = VBox(
    [fig_real.canvas, fig_imag.canvas],
    layout=Layout(
        width='660px',
        min_width='660px',
        max_width='660px',
        align_items='flex-start'
    )
)

main_layout = HBox(
    [controls, plots],
    layout=Layout(
        width='935px',
        min_width='935px',
        max_width='935px',
        align_items='flex-start',
        justify_content='flex-start'
    )
)

# ============================================================
# INITIAL DISPLAY
# ============================================================

update_plots()

display(description)
display(formula_1)
display(formula_2)
display(main_layout)
```
