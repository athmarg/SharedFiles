```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import FloatSlider, RadioButtons, HBox, VBox, Layout, HTML, Output
from IPython.display import display, Math

plt.ioff()

# ------------------------------------------------------------
# 1. CSS
# ------------------------------------------------------------

radio_style = HTML("""
<style>

.horizontal-radio .widget-radio-box {
    display: flex !important;
    flex-direction: row !important;
    flex-wrap: nowrap !important;
    gap: 18px !important;
}

.horizontal-radio .widget-radio-box label {
    margin: 0 !important;
    white-space: nowrap !important;
    font-size: 13px !important;
    font-weight: bold !important;
}

.horizontal-radio > label {
    display: none !important;
}

.jp-OutputArea,
.jp-OutputArea-output,
.jp-OutputArea-child,
.output_area,
.output_subarea {
    max-width: 100% !important;
    overflow-x: hidden !important;
}

.jupyter-widgets {
    max-width: 100% !important;
}

</style>
""")

# ------------------------------------------------------------
# 2. DESCRIPTION
# ------------------------------------------------------------

description = HTML("""
<div style="
    border:1px solid #b9d7f5;
    border-radius:8px;
    padding:8px 12px;
    margin-bottom:4px;
    font-size:13px;
    line-height:1.45;
    background-color:#f7fbff;
    width:100%;
    max-width:100%;
    box-sizing:border-box;
    overflow:hidden;
">

<div>
<b>Purpose:</b>
Illustrate the Paley-Wiener realizability condition using representative magnitude responses.
</div>

<div>
<b>Interpretation:</b>
Convergence of I(Ω) as Ω → ∞ is compatible with the Paley-Wiener condition, while divergence violates it.
</div>

<div style="margin-top:7px; margin-left:18px;">
• <span style="color:green;"><b>Convergent:</b></span>
<b> Rational LP</b> response
&nbsp; → &nbsp;
<span style="color:green;"><b>Paley-Wiener condition satisfied.</b></span>
</div>

<div style="margin-top:4px; margin-left:18px;">
• <span style="color:red;"><b>Divergent:</b></span>
<b> Exponential</b> and <b>Gaussian</b> responses
&nbsp; (I(Ω) → ∞ as Ω → ∞)
&nbsp; → &nbsp;
<span style="color:red;"><b>Paley-Wiener condition violated.</b></span>
</div>

<div style="margin-top:4px; margin-left:18px;">
• <span style="color:red;"><b>Divergent:</b></span>
<b> Ideal LP</b> response
&nbsp; (I(Ω) = ∞ for any Ω > ωc)
&nbsp; → &nbsp;
<span style="color:red;"><b>Paley-Wiener condition violated.</b></span>
</div>

</div>
""", layout=Layout(width='100%', max_width='100%', overflow='hidden'))

# ------------------------------------------------------------
# 3. FORMULA
# ------------------------------------------------------------

formula_output = Output(layout=Layout(width='100%', max_width='100%', overflow='hidden'))

with formula_output:
    display(Math(r'I(\Omega)=\int_{-\Omega}^{+\Omega}\frac{\left|\ln\left|H(j\omega)\right|\right|}{1+\omega^2}\,d\omega'))

# ------------------------------------------------------------
# 4. CONTROLS
# ------------------------------------------------------------

slider_layout = Layout(width='250px', max_width='250px')
style_opts = {'description_width':'60px'}

response_selector = RadioButtons(options=['Rational LP', 'Exponential', 'Gaussian', 'Ideal LP'], value='Rational LP', description='', layout=Layout(width='480px', max_width='480px'))
response_selector.add_class('horizontal-radio')

wc_slider = FloatSlider(min=0.5, max=5.0, step=0.1, value=1.0, description='ωc:', continuous_update=True, readout_format='.1f', style=style_opts, layout=slider_layout)

a_slider = FloatSlider(min=0.1, max=2.0, step=0.05, value=0.5, description='a:', continuous_update=True, readout_format='.2f', style=style_opts, layout=slider_layout)

omega_max_slider = FloatSlider(min=10.0, max=100.0, step=5.0, value=40.0, description='Ωmax:', continuous_update=True, readout_format='.0f', style=style_opts, layout=slider_layout)

# ------------------------------------------------------------
# 5. INFORMATION FRAME
# ------------------------------------------------------------

info_html = HTML(layout=Layout(width='260px', max_width='260px', overflow='hidden'))

parameter_label = HTML("""
<div style="
    font-size:14px;
    font-weight:bold;
    margin-top:8px;
    margin-bottom:4px;
">
Parameters:
</div>
""")

# ------------------------------------------------------------
# 6. RESPONSE SELECTOR ROW
# ------------------------------------------------------------

response_label = HTML("""
<div style="
    font-size:13px;
    font-weight:bold;
    white-space:nowrap;
    margin-right:14px;
">
Response:
</div>
""")

response_row = HBox([response_label, response_selector], layout=Layout(width='100%', max_width='100%', align_items='center', justify_content='flex-start', margin='0px 0px 2px 0px', overflow='hidden'))

# ------------------------------------------------------------
# 7. MAGNITUDE FIGURE
# ------------------------------------------------------------

fig_mag, ax_mag = plt.subplots(figsize=(6.4, 2.45))

mag_line, = ax_mag.plot([], [], 'r-', linewidth=2.0)

wc_left_line = ax_mag.axvline(-1.0, color='black', linestyle=':', linewidth=1.2)
wc_right_line = ax_mag.axvline(1.0, color='black', linestyle=':', linewidth=1.2)

ax_mag.set_xlim(-40.0, 40.0)
ax_mag.set_ylim(-0.05, 1.08)

ax_mag.set_xlabel('Angular Frequency ω (rad/s)', fontsize=10)
ax_mag.set_ylabel('Magnitude |H(jω)|', fontsize=10)

ax_mag.set_title('Magnitude Response Used in the Paley-Wiener Test', fontsize=12, fontweight='bold', pad=5)

ax_mag.tick_params(axis='both', labelsize=9)
ax_mag.grid(True, linestyle=':', alpha=0.5)

fig_mag.subplots_adjust(left=0.13, right=0.97, bottom=0.23, top=0.83)

fig_mag.canvas.header_visible = False
fig_mag.canvas.toolbar_visible = False
fig_mag.canvas.resizable = False

fig_mag.canvas.layout.width = '100%'
fig_mag.canvas.layout.max_width = '650px'
fig_mag.canvas.layout.height = '250px'
fig_mag.canvas.layout.overflow = 'hidden'
fig_mag.canvas.layout.margin = '0px 0px -6px 0px'

# ------------------------------------------------------------
# 8. INTEGRAL FIGURE
# ------------------------------------------------------------

fig_int, ax_int = plt.subplots(figsize=(6.4, 2.45))

int_line, = ax_int.plot([], [], 'r-', linewidth=2.0)

wc_integral_line = ax_int.axvline(1.0, color='black', linestyle=':', linewidth=1.2)

ax_int.set_xlim(0.0, 40.0)
ax_int.set_ylim(0.0, 2.0)

ax_int.set_xlabel('Truncation Frequency Ω (rad/s)', fontsize=10)
ax_int.set_ylabel('I(Ω)', fontsize=10)

ax_int.set_title('Truncated Paley-Wiener Integral', fontsize=12, fontweight='bold', pad=5)

ax_int.tick_params(axis='both', labelsize=9)
ax_int.grid(True, linestyle=':', alpha=0.5)

fig_int.subplots_adjust(left=0.13, right=0.97, bottom=0.23, top=0.83)

fig_int.canvas.header_visible = False
fig_int.canvas.toolbar_visible = False
fig_int.canvas.resizable = False

fig_int.canvas.layout.width = '100%'
fig_int.canvas.layout.max_width = '650px'
fig_int.canvas.layout.height = '250px'
fig_int.canvas.layout.overflow = 'hidden'
fig_int.canvas.layout.margin = '0px'

# ------------------------------------------------------------
# 9. TEXT FOR IDEAL FILTER
# ------------------------------------------------------------

ideal_text_1 = ax_int.text(0.0, 0.0, '', color='red', fontsize=13, fontweight='bold', visible=False)
ideal_text_2 = ax_int.text(0.0, 0.0, '', color='red', fontsize=11, visible=False)
ideal_text_3 = ax_int.text(0.0, 0.0, '', fontsize=11, ha='center', visible=False)

# ------------------------------------------------------------
# 10. NUMERICAL INTEGRATION
# ------------------------------------------------------------

def cumulative_trapezoid_np(y, x):
    dx = np.diff(x)
    areas = 0.5 * (y[:-1] + y[1:]) * dx
    return np.concatenate(([0.0], np.cumsum(areas)))

# ------------------------------------------------------------
# 11. UPDATE FUNCTION
# ------------------------------------------------------------

def update_plots(change=None):

    response_type = response_selector.value
    wc = wc_slider.value
    a = a_slider.value
    omega_max = omega_max_slider.value

    # --------------------------------------------------------
    # ENABLE / DISABLE PARAMETERS
    # --------------------------------------------------------

    if response_type in ['Rational LP', 'Ideal LP']:
        wc_slider.disabled = False
        a_slider.disabled = True
    else:
        wc_slider.disabled = True
        a_slider.disabled = False

    omega_max_slider.disabled = False

    # --------------------------------------------------------
    # FREQUENCY AXES
    # --------------------------------------------------------

    omega = np.linspace(-omega_max, omega_max, 5001)
    Omega = np.linspace(0.0, omega_max, 6001)

    ideal_mode = False

    # --------------------------------------------------------
    # RATIONAL LOW-PASS
    # --------------------------------------------------------

    if response_type == 'Rational LP':

        H_mag = 1.0 / np.sqrt(1.0 + (omega / wc)**2)
        H_pos = 1.0 / np.sqrt(1.0 + (Omega / wc)**2)

        integrand = np.abs(np.log(H_pos)) / (1.0 + Omega**2)
        I_Omega = 2.0 * cumulative_trapezoid_np(integrand, Omega)

        status = 'Convergent'
        condition = 'Paley-Wiener condition satisfied'
        expression = '|H(jω)| = 1 / √(1 + (ω/ωc)²)'
        behavior = '|ln|H||/(1+ω²) ~ ln|ω|/ω²'
        integral_value = f'{I_Omega[-1]:.4f}'

        wc_visible = True

    # --------------------------------------------------------
    # EXPONENTIAL
    # --------------------------------------------------------

    elif response_type == 'Exponential':

        H_mag = np.exp(-a * np.abs(omega))

        integrand = (a * Omega) / (1.0 + Omega**2)
        I_Omega = 2.0 * cumulative_trapezoid_np(integrand, Omega)

        status = 'Divergent'
        condition = 'Paley-Wiener condition violated'
        expression = '|H(jω)| = exp(−a|ω|)'
        behavior = 'I(Ω) = a ln(1+Ω²) → ∞'
        integral_value = f'{I_Omega[-1]:.4f} and increasing'

        wc_visible = False

    # --------------------------------------------------------
    # GAUSSIAN
    # --------------------------------------------------------

    elif response_type == 'Gaussian':

        H_mag = np.exp(-a * omega**2)

        integrand = (a * Omega**2) / (1.0 + Omega**2)
        I_Omega = 2.0 * cumulative_trapezoid_np(integrand, Omega)

        status = 'Divergent'
        condition = 'Paley-Wiener condition violated'
        expression = '|H(jω)| = exp(−aω²)'
        behavior = 'I(Ω) = 2a[Ω−arctan(Ω)] → ∞'
        integral_value = f'{I_Omega[-1]:.4f} and increasing'

        wc_visible = False

    # --------------------------------------------------------
    # IDEAL LOW-PASS
    # --------------------------------------------------------

    else:

        H_mag = np.where(np.abs(omega) <= wc, 1.0, 0.0)

        Omega_ideal = Omega[Omega <= wc]
        I_Omega = np.zeros_like(Omega_ideal)

        status = 'Infinite'
        condition = 'Paley-Wiener condition violated'
        expression = '|H(jω)| = 1 for |ω| ≤ ωc and 0 otherwise'
        behavior = 'ln|H| = −∞ over a nonzero frequency interval'
        integral_value = '∞ for Ω > ωc'

        wc_visible = True
        ideal_mode = True

    # --------------------------------------------------------
    # MAGNITUDE UPDATE
    # --------------------------------------------------------

    mag_line.set_data(omega, H_mag)

    ax_mag.set_xlim(-omega_max, omega_max)
    ax_mag.set_ylim(-0.05, 1.08)

    wc_left_line.set_xdata([-wc, -wc])
    wc_right_line.set_xdata([wc, wc])

    wc_left_line.set_visible(wc_visible)
    wc_right_line.set_visible(wc_visible)

    # --------------------------------------------------------
    # INTEGRAL UPDATE
    # --------------------------------------------------------

    if not ideal_mode:

        int_line.set_data(Omega, I_Omega)

        ymax = np.max(I_Omega)

        if ymax <= 0.0:
            ymax = 1.0

        ax_int.set_xlim(0.0, omega_max)
        ax_int.set_ylim(0.0, 1.12 * ymax)

        ideal_text_1.set_visible(False)
        ideal_text_2.set_visible(False)
        ideal_text_3.set_visible(False)

    else:

        int_line.set_data(Omega_ideal, I_Omega)

        ax_int.set_xlim(0.0, omega_max)
        ax_int.set_ylim(0.0, 1.0)

        ideal_text_1.set_position((wc + 0.04 * omega_max, 0.56))
        ideal_text_1.set_text('I(Ω) = ∞')
        ideal_text_1.set_visible(True)

        ideal_text_2.set_position((wc + 0.04 * omega_max, 0.42))
        ideal_text_2.set_text('for Ω > ωc')
        ideal_text_2.set_visible(True)

        ideal_text_3.set_position((0.5 * wc, 0.08))
        ideal_text_3.set_text('I(Ω) = 0')
        ideal_text_3.set_visible(True)

    wc_integral_line.set_xdata([wc, wc])
    wc_integral_line.set_visible(wc_visible)

    # --------------------------------------------------------
    # INFORMATION FRAME
    # --------------------------------------------------------

    if response_type in ['Rational LP', 'Ideal LP']:

        variable_parameter = f"""
        <div><b>ωc:</b> <span style="color:#0066cc;">{wc:.2f} rad/s</span></div>
        """

    else:

        variable_parameter = f"""
        <div><b>a:</b> <span style="color:#0066cc;">{a:.2f}</span></div>
        """

    info_html.value = f"""
    <div style="
        border:1px solid #cccccc;
        border-radius:7px;
        padding:7px 9px;
        margin-top:8px;
        font-size:12px;
        line-height:1.7;
        background:white;
        width:255px;
        max-width:255px;
        box-sizing:border-box;
        overflow:hidden;
    ">

        <div>
            <b>Response:</b>
            <span style="color:#0066cc;">{response_type}</span>
        </div>

        {variable_parameter}

        <div>
            <b>Ωmax:</b>
            <span style="color:#0066cc;">{omega_max:.0f} rad/s</span>
        </div>

        <div>
            <b>I(Ωmax):</b>
            <span style="color:#0066cc;">{integral_value}</span>
        </div>

        <div>
            <b>Behavior:</b>
            <span style="color:#0066cc;">{status}</span>
        </div>

        <div>
            <b>Conclusion:</b>
            <span style="color:#0066cc;">{condition}</span>
        </div>

        <div style="
            margin-top:5px;
            padding-top:5px;
            border-top:1px solid #eeeeee;
        ">
            <b>Magnitude model:</b><br>
            <span style="color:#0066cc;">{expression}</span>
        </div>

        <div style="
            margin-top:5px;
            padding-top:5px;
            border-top:1px solid #eeeeee;
        ">
            <b>Asymptotic behavior:</b><br>
            <span style="color:#0066cc;">{behavior}</span>
        </div>

    </div>
    """

    # --------------------------------------------------------
    # REDRAW EXISTING FIGURES
    # --------------------------------------------------------

    fig_mag.canvas.draw_idle()
    fig_int.canvas.draw_idle()

# ------------------------------------------------------------
# 12. CONNECT CONTROLS
# ------------------------------------------------------------

response_selector.observe(update_plots, names='value')
wc_slider.observe(update_plots, names='value')
a_slider.observe(update_plots, names='value')
omega_max_slider.observe(update_plots, names='value')

# ------------------------------------------------------------
# 13. LEFT COLUMN
# ------------------------------------------------------------

controls = VBox(
    [parameter_label, wc_slider, a_slider, omega_max_slider, info_html],
    layout=Layout(
        width='270px',
        min_width='270px',
        max_width='270px',
        flex='0 0 270px',
        align_items='flex-start',
        overflow='hidden'
    )
)

# ------------------------------------------------------------
# 14. RIGHT COLUMN
# ------------------------------------------------------------

plot_column = VBox(
    [response_row, fig_mag.canvas, fig_int.canvas],
    layout=Layout(
        width='calc(100% - 280px)',
        min_width='0px',
        max_width='650px',
        flex='1 1 auto',
        align_items='flex-start',
        overflow='hidden'
    )
)

# ------------------------------------------------------------
# 15. MAIN LAYOUT
# ------------------------------------------------------------

main_layout = HBox(
    [controls, plot_column],
    layout=Layout(
        width='100%',
        max_width='100%',
        min_width='0px',
        align_items='flex-start',
        justify_content='flex-start',
        overflow='hidden'
    )
)

# ------------------------------------------------------------
# 16. COMPLETE NOTEBOOK
# ------------------------------------------------------------

notebook_layout = VBox(
    [description, formula_output, main_layout],
    layout=Layout(
        width='100%',
        max_width='100%',
        min_width='0px',
        overflow='hidden'
    )
)

# ------------------------------------------------------------
# 17. INITIALIZE
# ------------------------------------------------------------

update_plots()

# ------------------------------------------------------------
# 18. DISPLAY
# ------------------------------------------------------------

display(radio_style)
display(notebook_layout)
```
