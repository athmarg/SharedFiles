```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import IntSlider, FloatSlider, VBox, HBox, Layout, HTML, HTMLMath
from IPython.display import display

plt.ioff()

# ==============================================================================
# USAGE
#
# This notebook illustrates two quantitative measures used to characterize
# how closely the magnitude response of a real low-pass filter approaches
# that of an ideal low-pass filter:
#
#       1. Selectivity
#       2. Spectral efficiency factor
#
# A generic monotonic low-pass magnitude curve is used:
#
#       G(ω) = 1 / sqrt(1 + (ω/ωc)^(2N)).
#
# This curve is used only as a convenient didactic model for controlling
# transition sharpness; no particular classical approximation is being
# introduced at this stage.
#
# SELECTIVITY
#
# The selectivity is
#
#       Fs = -dG(ω)/dω |_(ω=ωc).
#
# The tangent to the magnitude response at ωc is displayed in the first plot.
# Increasing N makes the transition sharper and increases Fs.
#
# SPECTRAL EFFICIENCY FACTOR
#
# For attenuation levels α and β, with β > α, the corresponding frequencies
# BWα and BWβ are determined from the magnitude response. The spectral
# efficiency factor is
#
#       Sαβ = BWβ / BWα.
#
# The two bandwidths are marked in the second plot.
#
# The student should observe that a better approximation to an ideal low-pass
# filter is associated with:
#
#       larger selectivity:              Fs -> infinity
#
#       spectral efficiency closer to:   Sαβ -> 1
#
# The sliders α and β control the attenuation levels used for the spectral
# efficiency calculation. The condition β > α is imposed automatically.
# ==============================================================================

# ==============================================================================
# JUPYTER DISPLAY SETTINGS
# ==============================================================================

display(HTML("""
<style>
.jp-OutputArea,
.jp-OutputArea-child,
.jp-OutputArea-output {
    overflow: visible !important;
    max-height: none !important;
    height: auto !important;
}

.output,
.output_area,
.output_subarea,
.output_scroll {
    overflow: visible !important;
    max-height: none !important;
    height: auto !important;
}

.jupyter-widgets,
.widget-box,
.widget-html,
.widget-html-content {
    overflow: visible !important;
    max-height: none !important;
}

.jp-Cell-outputWrapper {
    overflow: visible !important;
}
</style>
"""))

# ==============================================================================
# DESCRIPTION
# ==============================================================================

description = HTML("""
<div style="
    border:1px solid #9ec9f5;
    border-radius:7px;
    padding:8px 10px;
    margin:0px 0px 7px 0px;
    font-size:13px;
    line-height:1.45;
    background-color:#f7fbff;
    width:750px;
    max-width:750px;
    box-sizing:border-box;
">
<b>Purpose:</b>
Illustrate the selectivity and spectral efficiency factor of a real low-pass filter.
<br>
<b>Interpretation:</b>
A sharper transition produces a larger selectivity F<sub>s</sub> and brings the spectral efficiency factor S<sub>α</sub><sup>β</sup> closer to unity. In the limiting case of an ideal low-pass filter, F<sub>s</sub> → ∞ and S<sub>α</sub><sup>β</sup> → 1.
</div>
""", layout=Layout(width='760px', max_width='760px'))

# ==============================================================================
# EQUATIONS
# ==============================================================================

formula_1 = HTMLMath(value=r"""
\[
G(\omega)=\frac{1}{\sqrt{1+(\omega/\omega_c)^{2N}}},
\qquad
F_s=-\left.\frac{dG(\omega)}{d\omega}\right|_{\omega=\omega_c}
\]
""", layout=Layout(width='750px', max_width='750px'))

formula_2 = HTMLMath(value=r"""
\[
S_{\alpha}^{\beta}=\frac{\mathrm{BW}_{\beta}}{\mathrm{BW}_{\alpha}},
\qquad \beta>\alpha
\]
""", layout=Layout(width='750px', max_width='750px'))

# ==============================================================================
# CONTROLS
# ==============================================================================

slider_layout = Layout(width='250px')
style_opts = {'description_width':'70px'}

N_slider = IntSlider(value=3, min=1, max=10, step=1, description='N:', continuous_update=True, style=style_opts, layout=slider_layout)
wc_slider = FloatSlider(value=1.0, min=0.5, max=3.0, step=0.1, description='ωc:', continuous_update=True, readout_format='.1f', style=style_opts, layout=slider_layout)
alpha_slider = FloatSlider(value=3.0, min=1.0, max=20.0, step=1.0, description='α (dB):', continuous_update=True, readout_format='.0f', style=style_opts, layout=slider_layout)
beta_slider = FloatSlider(value=20.0, min=5.0, max=60.0, step=1.0, description='β (dB):', continuous_update=True, readout_format='.0f', style=style_opts, layout=slider_layout)

parameter_title = HTML("""
<div style="font-size:14px; font-weight:bold; margin-top:3px; margin-bottom:5px;">
Parameters:
</div>
""")

info_html = HTML(layout=Layout(width='265px', max_width='265px'))

# ==============================================================================
# FIGURE 1: SELECTIVITY
# ==============================================================================

fig_sel, ax_sel = plt.subplots(figsize=(6.5, 2.6))

gain_sel_line, = ax_sel.plot([], [], 'r-', linewidth=2.0, label='G(ω)')
tangent_line, = ax_sel.plot([], [], 'k--', linewidth=1.5, label='Tangent at ωc')
wc_line_1 = ax_sel.axvline(1.0, color='gray', linestyle=':', linewidth=1.2, label='ωc')

ax_sel.set_xlabel('Angular Frequency ω (rad/s)', fontsize=10)
ax_sel.set_ylabel('G(ω)', fontsize=10)
ax_sel.set_title('Filter Selectivity', fontsize=12, fontweight='bold', pad=5)
ax_sel.tick_params(axis='both', labelsize=9)
ax_sel.grid(True, linestyle=':', alpha=0.5)
ax_sel.legend(loc='center left', bbox_to_anchor=(1.01, 0.5), fontsize=9)

# Fixed axes
ax_sel.set_xlim(0.0, 15.0)
ax_sel.set_ylim(0.0, 1.08)

fig_sel.subplots_adjust(left=0.12, right=0.75, bottom=0.22, top=0.84)
fig_sel.canvas.header_visible = False
fig_sel.canvas.toolbar_visible = False
fig_sel.canvas.resizable = False
fig_sel.canvas.layout.width = '650px'
fig_sel.canvas.layout.height = '265px'
fig_sel.canvas.layout.margin = '0px 0px -6px 0px'

# ==============================================================================
# FIGURE 2: SPECTRAL EFFICIENCY
# ==============================================================================

fig_spec, ax_spec = plt.subplots(figsize=(6.5, 2.6))

gain_spec_line, = ax_spec.plot([], [], 'r-', linewidth=2.0, label='G(ω)')
alpha_horizontal = ax_spec.axhline(0.0, color='black', linestyle='--', linewidth=1.0, label='α level')
beta_horizontal = ax_spec.axhline(0.0, color='gray', linestyle='--', linewidth=1.0, label='β level')
alpha_vertical = ax_spec.axvline(0.0, color='black', linestyle=':', linewidth=1.2, label='BWα')
beta_vertical = ax_spec.axvline(0.0, color='gray', linestyle=':', linewidth=1.2, label='BWβ')

ax_spec.set_xlabel('Angular Frequency ω (rad/s)', fontsize=10)
ax_spec.set_ylabel('G(ω)', fontsize=10)
ax_spec.set_title('Spectral Efficiency Factor', fontsize=12, fontweight='bold', pad=5)
ax_spec.tick_params(axis='both', labelsize=9)
ax_spec.grid(True, linestyle=':', alpha=0.5)
ax_spec.legend(loc='center left', bbox_to_anchor=(1.01, 0.5), fontsize=9)

# Fixed axes
ax_spec.set_xlim(0.0, 15.0)
ax_spec.set_ylim(0.0, 1.08)

fig_spec.subplots_adjust(left=0.12, right=0.75, bottom=0.22, top=0.84)
fig_spec.canvas.header_visible = False
fig_spec.canvas.toolbar_visible = False
fig_spec.canvas.resizable = False
fig_spec.canvas.layout.width = '650px'
fig_spec.canvas.layout.height = '265px'

# ==============================================================================
# GENERIC LOW-PASS MODEL
# ==============================================================================

def gain_response(omega, wc, N):
    return 1.0 / np.sqrt(1.0 + (omega / wc)**(2 * N))

def attenuation_to_frequency(A_dB, wc, N):
    G_level = 10.0**(-A_dB / 20.0)
    omega_level = wc * (1.0 / G_level**2 - 1.0)**(1.0 / (2.0 * N))
    return G_level, omega_level

# ==============================================================================
# UPDATE FUNCTION
# ==============================================================================

def update_plots(change=None):

    N = N_slider.value
    wc = wc_slider.value
    alpha = alpha_slider.value
    beta = beta_slider.value

    # Enforce β > α.
    if beta <= alpha:
        beta = alpha + 1.0
        beta_slider.value = beta
        return

    # --------------------------------------------------------------------------
    # Frequency response
    #
    # The numerical frequency range is kept wide enough for all slider values,
    # whereas the displayed axes remain fixed.
    # --------------------------------------------------------------------------

    omega = np.linspace(0.0, 15.0, 2400)
    G = gain_response(omega, wc, N)

    # --------------------------------------------------------------------------
    # Selectivity
    # --------------------------------------------------------------------------

    Gc = 1.0 / np.sqrt(2.0)
    Fs = N / (2.0 * np.sqrt(2.0) * wc)

    tangent = Gc - Fs * (omega - wc)

    # --------------------------------------------------------------------------
    # Spectral efficiency
    # --------------------------------------------------------------------------

    G_alpha, BW_alpha = attenuation_to_frequency(alpha, wc, N)
    G_beta, BW_beta = attenuation_to_frequency(beta, wc, N)

    S = BW_beta / BW_alpha

    # --------------------------------------------------------------------------
    # Plot 1
    # --------------------------------------------------------------------------

    gain_sel_line.set_data(omega, G)
    tangent_line.set_data(omega, tangent)
    wc_line_1.set_xdata([wc, wc])

    # Axes remain fixed.
    ax_sel.set_xlim(0.0, 15.0)
    ax_sel.set_ylim(0.0, 1.08)

    # --------------------------------------------------------------------------
    # Plot 2
    # --------------------------------------------------------------------------

    gain_spec_line.set_data(omega, G)

    alpha_horizontal.set_ydata([G_alpha, G_alpha])
    beta_horizontal.set_ydata([G_beta, G_beta])

    alpha_vertical.set_xdata([BW_alpha, BW_alpha])
    beta_vertical.set_xdata([BW_beta, BW_beta])

    # Axes remain fixed.
    ax_spec.set_xlim(0.0, 15.0)
    ax_spec.set_ylim(0.0, 1.08)

    # --------------------------------------------------------------------------
    # Information panel
    # --------------------------------------------------------------------------

    info_html.value = f"""
    <div style="
        border:1px solid #cccccc;
        border-radius:7px;
        padding:8px 9px;
        margin-top:9px;
        font-size:12px;
        line-height:1.65;
        background:white;
        width:260px;
        box-sizing:border-box;
    ">

    <div>
        <b>Model order:</b>
        <span style="color:#0066cc;">N = {N}</span>
    </div>

    <div>
        <b>Cutoff frequency:</b>
        <span style="color:#0066cc;">ωc = {wc:.2f} rad/s</span>
    </div>

    <div style="
        margin-top:6px;
        padding-top:6px;
        border-top:1px solid #eeeeee;
    ">
        <b>Selectivity:</b><br>
        <span style="color:#0066cc;">
        Fs = {Fs:.4f}
        </span>
    </div>

    <div style="
        margin-top:6px;
        padding-top:6px;
        border-top:1px solid #eeeeee;
    ">
        <b>Attenuation levels:</b><br>
        α = <span style="color:#0066cc;">{alpha:.0f} dB</span><br>
        β = <span style="color:#0066cc;">{beta:.0f} dB</span>
    </div>

    <div style="
        margin-top:6px;
        padding-top:6px;
        border-top:1px solid #eeeeee;
    ">
        <b>Bandwidths:</b><br>
        BWα =
        <span style="color:#0066cc;">{BW_alpha:.4f} rad/s</span><br>
        BWβ =
        <span style="color:#0066cc;">{BW_beta:.4f} rad/s</span>
    </div>

    <div style="
        margin-top:6px;
        padding-top:6px;
        border-top:1px solid #eeeeee;
    ">
        <b>Spectral efficiency:</b><br>
        <span style="color:#0066cc;">
        S<sub>α</sub><sup>β</sup> = {S:.4f}
        </span>
    </div>

    <div style="
        margin-top:6px;
        padding-top:6px;
        border-top:1px solid #eeeeee;
    ">
        <b>Ideal-filter limit:</b><br>
        F<sub>s</sub> → ∞<br>
        S<sub>α</sub><sup>β</sup> → 1
    </div>

    </div>
    """

    fig_sel.canvas.draw_idle()
    fig_spec.canvas.draw_idle()

# ==============================================================================
# CALLBACKS
# ==============================================================================

N_slider.observe(update_plots, names='value')
wc_slider.observe(update_plots, names='value')
alpha_slider.observe(update_plots, names='value')
beta_slider.observe(update_plots, names='value')

# ==============================================================================
# LAYOUT
# ==============================================================================

controls = VBox([parameter_title, N_slider, wc_slider, alpha_slider, beta_slider, info_html], layout=Layout(width='275px', min_width='275px', max_width='275px', flex='0 0 275px', align_items='flex-start'))

plots = VBox([fig_sel.canvas, fig_spec.canvas], layout=Layout(width='660px', min_width='660px', max_width='660px', align_items='flex-start'))

main_layout = HBox([controls, plots], layout=Layout(width='935px', min_width='935px', max_width='935px', align_items='flex-start', justify_content='flex-start'))

# ==============================================================================
# INITIAL DISPLAY
# ==============================================================================

update_plots()

display(description)
display(formula_1)
display(formula_2)
display(main_layout)
```
