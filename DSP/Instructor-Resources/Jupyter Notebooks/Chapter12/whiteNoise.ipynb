```python
# ============================================================
# QUANTIZATION NOISE AND THE WHITE-NOISE MODEL
# ============================================================
#
# This notebook demonstrates how the quantization error
#
#               e[n] = Q(x[n]) - x[n]
#
# may behave approximately as white noise under suitable
# conditions.
#
# HOW TO USE THE NOTEBOOK
#
# 1. Select the signal condition:
#
#       - Rich / non-periodic signal
#       - Simple periodic signal
#
# 2. Change the number of quantization bits.
#
# 3. Change the number of samples.
#
# 4. Observe:
#
#       - the original and quantized signals,
#       - the quantization error,
#       - the histogram of the error,
#       - the sample autocorrelation of the error,
#       - the measured error variance,
#       - the theoretical value Delta^2 / 12.
#
# IMPORTANT
#
# The white-noise description is a MODEL and is not an exact
# property of quantization error for every possible signal.
#
# The approximation is generally more appropriate when:
#
#       - the signal is sufficiently rich or non-periodic,
#       - many quantization levels are used,
#       - the quantization error is weakly correlated with
#         the signal.
#
# For simple periodic signals, the error may exhibit an organized
# structure and the white-noise approximation may become poor.
#
#
# ============================================================
# INTERPRETATION OF THE WHITE-NOISE MODEL
# ============================================================
#
# The purpose of the white-noise model is to replace the actual
# deterministic quantization error by a simpler statistical model.
#
# For a rounding quantizer with step size Delta, the idealized
# model assumes approximately:
#
#                   mean{e[n]} ≈ 0
#
#                   variance{e[n]} ≈ Delta^2 / 12
#
# and
#
#                   r_ee[m] ≈ 0,    m != 0
#
# while
#
#                   r_ee[0] = variance{e[n]}.
#
# Equivalently, the autocorrelation of ideal white quantization
# noise has approximately the form
#
#                   r_ee[m] ≈ sigma_e^2 delta[m].
#
#
# ------------------------------------------------------------
# WHEN IS THE WHITE-NOISE MODEL A GOOD APPROXIMATION?
# ------------------------------------------------------------
#
# The model tends to work well when the input signal moves through
# many quantization intervals in a sufficiently complicated way.
#
# Typical favorable conditions include:
#
#   1. The signal is sufficiently rich or non-periodic.
#
#   2. The signal visits many different quantization levels.
#
#   3. The quantization step Delta is small relative to the
#      variations of the signal.
#
#   4. The quantization error does not exhibit a strong repeating
#      pattern.
#
#   5. The error is weakly correlated from one sample to another.
#
#   6. The error is weakly correlated with the input signal.
#
# Under these conditions, the sequence e[n] often appears
# irregular and noise-like.
#
#
# ------------------------------------------------------------
# WHY DOES A RICH / NON-PERIODIC SIGNAL HELP?
# ------------------------------------------------------------
#
# A rich signal passes through the quantizer thresholds in a
# complicated and usually non-repeating manner.
#
# Consequently, successive quantization errors are less likely to
# repeat the same pattern.
#
# In this case:
#
#   - e[n] tends to look irregular,
#   - its histogram tends to fill the interval
#
#             -Delta/2 <= e <= Delta/2
#
#     approximately uniformly,
#
#   - its measured variance tends to approach
#
#             Delta^2 / 12,
#
#   - and its autocorrelation tends to contain one dominant value
#     at zero lag and much smaller values at non-zero lags.
#
# These observations support the white-noise approximation.
#
#
# ------------------------------------------------------------
# WHY CAN A SIMPLE PERIODIC SIGNAL BE A POOR CASE?
# ------------------------------------------------------------
#
# A simple periodic input repeatedly visits the same positions of
# the quantizer.
#
# Therefore, the quantization error may also repeat periodically.
#
# For example, an error sequence may behave approximately like
#
#     e[0], e[1], ..., e[P-1],
#     e[0], e[1], ..., e[P-1], ...
#
# for some period P.
#
# Such an error sequence is highly structured and should not be
# interpreted as white noise.
#
# In this case:
#
#   - visible patterns may appear in e[n],
#   - the histogram may contain preferred error values instead of
#     filling the interval uniformly,
#   - the autocorrelation may contain significant peaks at
#     non-zero lags,
#   - and the error may remain correlated with the input signal.
#
# Thus, the comparison between the rich signal and the simple
# periodic signal is important: it demonstrates that quantization
# error is not automatically white noise.
#
#
# ------------------------------------------------------------
# WHAT SHOULD BE OBSERVED IN THE FOUR DIAGRAMS?
# ------------------------------------------------------------
#
# DIAGRAM 1: ORIGINAL AND QUANTIZED SIGNALS
#
# This diagram shows how the error is generated:
#
#             x[n]  --->  Q(x[n])
#
# The vertical difference between the two signals is
#
#             e[n] = Q(x[n]) - x[n].
#
#
# DIAGRAM 2: QUANTIZATION ERROR
#
# This diagram shows the actual sequence e[n].
#
# For a good white-noise approximation, the error should appear
# irregular and should not display an obvious repeating structure.
#
# For a poor approximation, periodic or organized patterns may be
# visible.
#
#
# DIAGRAM 3: QUANTIZATION ERROR HISTOGRAM
#
# For the ideal rounding-noise model, the error is approximately
# uniformly distributed in
#
#             [-Delta/2, Delta/2].
#
# Therefore, a histogram that is reasonably flat over this
# interval supports the uniform-error assumption.
#
# Strong concentrations at particular error values indicate that
# the assumption is less accurate.
#
#
# DIAGRAM 4: ERROR AUTOCORRELATION
#
# This is one of the most important tests of the white-noise
# interpretation.
#
# For white noise,
#
#             r_ee[m] ≈ 0,   m != 0.
#
# Therefore, the normalized autocorrelation should ideally show
# one dominant peak at m = 0 and values close to zero elsewhere.
#
# Significant peaks at non-zero lags indicate that the error has
# memory or periodic structure and is therefore not very
# white-noise-like.
#
#
# ------------------------------------------------------------
# VARIANCE TEST
# ------------------------------------------------------------
#
# For a uniform rounding error in
#
#             [-Delta/2, Delta/2],
#
# the theoretical variance is
#
#             sigma_e^2 = Delta^2 / 12.
#
# The notebook compares the measured variance with this theoretical
# prediction.
#
# A variance ratio
#
#             measured variance
#             -----------------  ≈ 1
#               Delta^2 / 12
#
# supports the model.
#
# However, agreement of the variance alone is not sufficient.
#
# An error sequence may have approximately the correct variance
# while still being strongly correlated or periodic.
#
# Therefore, the variance and autocorrelation should be examined
# together.
#
#
# ------------------------------------------------------------
# SIGNAL-ERROR CORRELATION
# ------------------------------------------------------------
#
# Another assumption commonly associated with the quantization
# noise model is that the error is approximately uncorrelated with
# the input signal.
#
# Ideally,
#
#             r_xe[m] ≈ 0.
#
# If the input signal and the quantization error are strongly
# correlated, treating the error as an independent noise source
# becomes less accurate.
#
#
# ------------------------------------------------------------
# PEDAGOGICAL CONCLUSION
# ------------------------------------------------------------
#
# The important conclusion is NOT:
#
#             "Quantization error is white noise."
#
# The correct conclusion is:
#
#             "Under suitable conditions, quantization error can
#              often be usefully modeled as approximately white
#              noise."
#
# The quality of this approximation depends on the relationship
# between the input signal and the quantizer.
#
# The rich/non-periodic and simple-periodic examples are included
# specifically to demonstrate this dependence.
#
# ============================================================


%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, RadioButtons, VBox, HBox, HTML, Layout
from IPython.display import display


# ------------------------------------------------------------
# Quantization function
# ------------------------------------------------------------

def uniform_rounding_quantizer(x, bits):
    xmin = -1.0
    xmax = 1.0
    levels = 2**bits
    delta = (xmax - xmin) / levels
    q = np.floor((x - xmin) / delta + 0.5) * delta + xmin
    q = np.clip(q, xmin, xmax - delta)

    return q, delta


# ------------------------------------------------------------
# Signal generation
# ------------------------------------------------------------

def generate_signal(signal_condition, N):
    n = np.arange(N)

    if signal_condition == 'Rich / non-periodic signal':
        x = 0.48 * np.sin(2.0 * np.pi * 0.071 * n) + 0.27 * np.sin(2.0 * np.pi * 0.173 * n + 0.6) + 0.15 * np.sin(2.0 * np.pi * 0.0137 * n * n / N)
    else:
        x = 0.75 * np.sin(2.0 * np.pi * 0.0625 * n)

    max_abs = np.max(np.abs(x))

    if max_abs > 0:
        x = 0.85 * x / max_abs

    return n, x


# ------------------------------------------------------------
# Sample autocorrelation
# ------------------------------------------------------------

def sample_autocorrelation(x, max_lag):
    x = np.asarray(x)
    x = x - np.mean(x)

    N = len(x)
    lags = np.arange(-max_lag, max_lag + 1)
    r = []

    for lag in lags:
        if lag >= 0:
            value = np.sum(x[:N-lag] * x[lag:]) / N
        else:
            k = -lag
            value = np.sum(x[k:] * x[:N-k]) / N

        r.append(value)

    return lags, np.array(r)


# ------------------------------------------------------------
# Style
# ------------------------------------------------------------

style_html = HTML("""
<style>

.qn-root {
    font-family: monospace;
    width: 960px;
    max-width: 960px;
}

.qn-description {
    font-size: 13px;
    line-height: 1.45;
    padding: 9px 12px;
    border: 1px solid #bfc7d5;
    border-left: 6px solid #4a6fa5;
    background: #f7f9fc;
    border-radius: 8px;
    margin-bottom: 8px;
    box-sizing: border-box;
}

.qn-box {
    border: 1px solid #c8d0dc;
    border-radius: 9px;
    padding: 9px 12px;
    box-sizing: border-box;
}

.qn-title {
    font-size: 16px;
    font-weight: bold;
    color: #243447;
    margin-bottom: 6px;
}

.qn-info {
    font-size: 13px;
    line-height: 1.50;
}

.qn-label {
    display: inline-block;
    min-width: 205px;
    font-weight: bold;
}

.qn-value {
    font-size: 15px;
    font-weight: bold;
}

.qn-good {
    color: #176b34;
    font-weight: bold;
}

.qn-warning {
    color: #a85b00;
    font-weight: bold;
}

.qn-model {
    margin-top: 6px;
    padding: 5px 8px;
    border-radius: 6px;
    background: #f2f4f7;
    font-size: 12px;
    line-height: 1.35;
}


/* ---------------------------------------------------------
   Binder/Jupyter output behavior

   Allow the matplotlib figure to extend beyond the normal
   output width without creating a horizontal scrollbar.
   --------------------------------------------------------- */

.jupyter-widgets-output-area,
.widget-output,
.output_area,
.output_subarea {
    overflow-x: visible !important;
    max-width: none !important;
}

.jupyter-widgets-output-area .output,
.widget-output .output {
    overflow: visible !important;
}

</style>
""")


# ------------------------------------------------------------
# Title
# ------------------------------------------------------------

title_html = HTML("""
<div class="qn-root">

    <div style="
        font-family:monospace;
        font-size:22px;
        font-weight:bold;
        margin-bottom:8px;
    ">
        Quantization Noise and the White-Noise Model
    </div>

</div>
""")


# ------------------------------------------------------------
# Description
# ------------------------------------------------------------

description_html = HTML("""
<div class="qn-root">

    <div class="qn-description">

        Quantization generates the error <b>e[n] = Q(x[n]) - x[n]</b>.
        Although this error is deterministic for a given input signal,
        under suitable operating conditions it is often modeled as an
        approximately white random process.<br>

        The white-noise approximation predicts an approximately uniform
        error distribution in the interval <b>-Δ/2 ≤ e ≤ Δ/2</b> and an
        error variance approximately equal to <b>Δ²/12</b>.<br>

        Compare a rich non-periodic signal with a simple periodic signal
        and observe when the white-noise model becomes a good or poor approximation.

    </div>

</div>
""")


# ------------------------------------------------------------
# Dynamic summary
# ------------------------------------------------------------

summary_html = HTML()

summary_html.layout = Layout(width='610px', min_width='610px', overflow='visible')


# ------------------------------------------------------------
# Main plotting function
# ------------------------------------------------------------

def plot_quantization_noise(signal_condition='Rich / non-periodic signal', bits=6, N=2048):

    n, x = generate_signal(signal_condition, N)

    xq, delta = uniform_rounding_quantizer(x, bits)

    error = xq - x


    # --------------------------------------------------------
    # Statistical quantities
    # --------------------------------------------------------

    measured_mean = np.mean(error)

    measured_variance = np.var(error)

    theoretical_variance = delta**2 / 12.0

    variance_ratio = measured_variance / theoretical_variance


    # --------------------------------------------------------
    # Error autocorrelation
    # --------------------------------------------------------

    max_lag = 30

    lags, r = sample_autocorrelation(error, max_lag)

    r0 = r[max_lag]

    if abs(r0) > 1e-15:
        normalized_r = r / r0
    else:
        normalized_r = r

    nonzero_lags = np.delete(normalized_r, max_lag)

    maximum_offzero_correlation = np.max(np.abs(nonzero_lags))


    # --------------------------------------------------------
    # Qualitative assessment
    # --------------------------------------------------------

    if maximum_offzero_correlation < 0.10 and 0.75 < variance_ratio < 1.25:

        assessment = "<span class='qn-good'>GOOD WHITE-NOISE APPROXIMATION</span>"

        explanation = "The quantization error has relatively small correlation at non-zero lags and its measured variance is close to Δ²/12."

    else:

        assessment = "<span class='qn-warning'>WHITE-NOISE APPROXIMATION IS LIMITED</span>"

        explanation = "The error contains noticeable structure or its variance differs substantially from the Δ²/12 prediction."


    # --------------------------------------------------------
    # Update summary
    # --------------------------------------------------------

    summary_html.value = f"""
    <div class="qn-box">

        <div class="qn-title">
            Quantization Noise Summary
        </div>

        <div class="qn-info">

            <span class="qn-label">Signal condition</span>
            {signal_condition}
            <br>

            <span class="qn-label">Quantization bits</span>
            B = {bits}
            <br>

            <span class="qn-label">Number of samples</span>
            N = {N}
            <br>

            <span class="qn-label">Quantization step</span>
            Δ = {delta:.8f}
            <br>

            <span class="qn-label">Mean error</span>
            {measured_mean:.8e}
            <br>

            <span class="qn-label">Measured variance</span>
            <span class="qn-value">{measured_variance:.8e}</span>
            <br>

            <span class="qn-label">Theoretical Δ²/12</span>
            <span class="qn-value">{theoretical_variance:.8e}</span>
            <br>

            <span class="qn-label">Variance ratio</span>
            {variance_ratio:.4f}
            <br>

            <span class="qn-label">Max |ρee[m]|, m ≠ 0</span>
            {maximum_offzero_correlation:.4f}
            <br>

            <span class="qn-label">Model assessment</span>
            {assessment}

        </div>

        <div class="qn-model">
            {explanation}
            The white-noise description is a statistical model rather
            than an exact identity.
        </div>

    </div>
    """


    # --------------------------------------------------------
    # Display window
    # --------------------------------------------------------

    display_samples = min(160, N)


    # --------------------------------------------------------
    # Wide and compact 2 x 2 figure
    # --------------------------------------------------------

    fig, axes = plt.subplots(2, 2, figsize=(14.0, 5.6))

    ax1 = axes[0, 0]
    ax2 = axes[0, 1]
    ax3 = axes[1, 0]
    ax4 = axes[1, 1]


    # --------------------------------------------------------
    # Plot 1: Original and quantized signals
    # --------------------------------------------------------

    ax1.plot(n[:display_samples], x[:display_samples], linewidth=1.4, label='Original x[n]')

    ax1.step(n[:display_samples], xq[:display_samples], where='mid', linewidth=1.1, label='Quantized Q(x[n])')

    ax1.set_xlim(0, display_samples - 1)

    ax1.set_ylim(-1.0, 1.0)

    ax1.set_xlabel('Sample index n')

    ax1.set_ylabel('Amplitude')

    ax1.set_title('Original and Quantized Signals', fontsize=11)

    ax1.grid(True, linestyle=':', alpha=0.6)

    ax1.legend(loc='upper center', bbox_to_anchor=(0.5, -0.25), ncol=2, fontsize=8, frameon=False)


    # --------------------------------------------------------
    # Plot 2: Quantization error
    # --------------------------------------------------------

    ax2.plot(n[:display_samples], error[:display_samples], linewidth=1.2, label='e[n]')

    ax2.axhline(delta / 2, linestyle='--', linewidth=1.0, label='+Δ/2')

    ax2.axhline(-delta / 2, linestyle='--', linewidth=1.0, label='-Δ/2')

    ax2.axhline(0.0, linewidth=0.8)

    ax2.set_xlim(0, display_samples - 1)

    ax2.set_ylim(-0.55 * delta, 0.55 * delta)

    ax2.set_xlabel('Sample index n')

    ax2.set_ylabel('e[n]')

    ax2.set_title('Quantization Error', fontsize=11)

    ax2.grid(True, linestyle=':', alpha=0.6)

    ax2.legend(loc='upper center', bbox_to_anchor=(0.5, -0.25), ncol=3, fontsize=8, frameon=False)


    # --------------------------------------------------------
    # Plot 3: Quantization error histogram
    # --------------------------------------------------------

    ax3.hist(error, bins=31, density=True, alpha=0.75, label='Measured distribution')

    theoretical_pdf = 1.0 / delta

    ax3.hlines(theoretical_pdf, -delta / 2, delta / 2, linestyle='--', linewidth=1.5, label='Ideal uniform model')

    ax3.set_xlim(-0.60 * delta, 0.60 * delta)

    ax3.set_xlabel('Quantization error e')

    ax3.set_ylabel('Probability density')

    ax3.set_title('Quantization Error Histogram', fontsize=11)

    ax3.grid(True, linestyle=':', alpha=0.6)

    ax3.legend(loc='upper center', bbox_to_anchor=(0.5, -0.25), ncol=2, fontsize=8, frameon=False)


    # --------------------------------------------------------
    # Plot 4: Normalized autocorrelation
    # --------------------------------------------------------

    ax4.stem(lags, normalized_r)

    ax4.axhline(0.0, linewidth=0.8)

    ax4.set_xlim(-max_lag - 1, max_lag + 1)

    ax4.set_ylim(-1.1, 1.1)

    ax4.set_xlabel('Lag m')

    ax4.set_ylabel(r'Normalized $r_{ee}[m]$')

    ax4.set_title('Error Autocorrelation', fontsize=11)

    ax4.grid(True, linestyle=':', alpha=0.6)


    # --------------------------------------------------------
    # Space for the horizontal legends
    # --------------------------------------------------------

    plt.subplots_adjust(left=0.07, right=0.98, top=0.93, bottom=0.15, wspace=0.22, hspace=0.62)

    plt.show()

    plt.close(fig)


# ------------------------------------------------------------
# Controls
# ------------------------------------------------------------

slider_layout = Layout(width='310px')

slider_style = {'description_width': '125px'}


signal_selector = RadioButtons(
    options=[
        'Rich / non-periodic signal',
        'Simple periodic signal'
    ],
    value='Rich / non-periodic signal',
    description='Signal:',
    style={'description_width': '60px'},
    layout=Layout(width='310px')
)


bits_slider = IntSlider(
    value=6,
    min=3,
    max=10,
    step=1,
    description='Quantization bits:',
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)


samples_slider = IntSlider(
    value=2048,
    min=512,
    max=8192,
    step=512,
    description='Samples N:',
    continuous_update=True,
    style=slider_style,
    layout=slider_layout
)


# ------------------------------------------------------------
# Interactive object
# ------------------------------------------------------------

widget_plot = interactive(
    plot_quantization_noise,
    signal_condition=signal_selector,
    bits=bits_slider,
    N=samples_slider
)


# ------------------------------------------------------------
# Separate controls from plot output
# ------------------------------------------------------------

controls = VBox(
    [
        HTML("<div class='qn-title'>Controls</div>"),
        signal_selector,
        bits_slider,
        samples_slider
    ],
    layout=Layout(
        width='330px',
        min_width='330px',
        overflow='visible',
        justify_content='center'
    )
)


# ------------------------------------------------------------
# Summary + controls
# ------------------------------------------------------------

top_row = HBox(
    [
        summary_html,
        controls
    ],
    layout=Layout(
        width='960px',
        max_width='960px',
        overflow='visible',
        align_items='flex-start',
        justify_content='space-between'
    )
)


# ------------------------------------------------------------
# Plot output
# ------------------------------------------------------------

plot_output = widget_plot.children[-1]

plot_output.layout = Layout(
    width='auto',
    overflow='visible'
)


# ------------------------------------------------------------
# Final notebook layout
# ------------------------------------------------------------

main_layout = VBox(
    [
        description_html,
        top_row,
        plot_output
    ],
    layout=Layout(
        width='960px',
        overflow='visible',
        align_items='flex-start'
    )
)


# ------------------------------------------------------------
# Display
# ------------------------------------------------------------

display(style_html)

display(title_html)

display(main_layout)
```
