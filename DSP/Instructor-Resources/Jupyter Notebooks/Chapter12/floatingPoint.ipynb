```python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interactive, IntSlider, HBox, Layout, VBox, HTML
from IPython.display import display

table_html = HTML()


def plot_number_distribution(coefficient_bits=4, max_cycle=4):
    minimum_coefficient = 2**(coefficient_bits - 1)
    maximum_coefficient = 2**coefficient_bits - 1

    floating_values = []

    for exponent in range(max_cycle + 1):
        values = np.arange(minimum_coefficient, maximum_coefficient + 1) * 2**exponent
        floating_values.extend(values)

    floating_values = np.array(floating_values)

    maximum_value = np.max(floating_values)

    fixed_values = np.arange(0, maximum_value + 1, 1)

    fig, ax = plt.subplots(figsize=(9, 4.5))

    ax.scatter(fixed_values, np.ones_like(fixed_values), s=10, label='Fixed-point values')

    ax.scatter(floating_values, np.zeros_like(floating_values), s=30, label='Floating-point values')

    ax.set_xlim(0, maximum_value * 1.03)

    ax.set_ylim(-0.6, 1.6)

    ax.set_yticks([0, 1])

    ax.set_yticklabels(['Floating point', 'Fixed point'], fontsize=12)

    ax.set_xlabel('Representable numerical values', fontsize=13)

    ax.grid(True, axis='x', linestyle=':', alpha=0.6)

    ax.legend(loc='upper left', bbox_to_anchor=(1.02, 1.0), borderaxespad=0, ncol=1)

    plt.title('Fixed-Point versus Floating-Point Number Distribution', fontsize=13, pad=15)

    plt.show()

    rows = ""

    for exponent in range(max_cycle + 1):
        spacing = 2**exponent
        first_value = minimum_coefficient * 2**exponent
        last_value = maximum_coefficient * 2**exponent

        rows += f"""
        <tr>
        <td style="padding: 3px 18px; text-align:center;">{exponent}</td>
        <td style="padding: 3px 18px; text-align:center;">{first_value}</td>
        <td style="padding: 3px 18px; text-align:center;">{last_value}</td>
        <td style="padding: 3px 18px; text-align:center;">{spacing}</td>
        </tr>
        """

    table_html.value = f"""
    <div style="font-family: monospace; font-size: 13px; line-height: 1.45;">

    <b>Floating-point cycles</b><br><br>

    <table style="border-collapse:collapse;">

    <tr>
    <th style="padding: 3px 18px; text-align:center;">Exponent<br>E</th>
    <th style="padding: 3px 18px; text-align:center;">First<br>value</th>
    <th style="padding: 3px 18px; text-align:center;">Last<br>value</th>
    <th style="padding: 3px 18px; text-align:center;">Spacing</th>
    </tr>

    {rows}

    </table>

    </div>
    """


slider_layout = Layout(width='300px')

style_opts = {'description_width': '110px'}

coefficient_slider = IntSlider(min=3, max=6, step=1, value=4, description='Coefficient bits:', style=style_opts, layout=slider_layout, continuous_update=True)

cycle_slider = IntSlider(min=1, max=5, step=1, value=4, description='Maximum cycle:', style=style_opts, layout=slider_layout, continuous_update=True)


widget_plot = interactive(plot_number_distribution, coefficient_bits=coefficient_slider, max_cycle=cycle_slider)


theory_html = HTML("""
<div style="font-family: monospace; font-size: 13px; line-height: 1.55; margin-bottom: 10px;">

<b>Fixed point:</b> The distance between consecutive representable numbers is constant.<br>
Therefore, fixed-point representation provides uniform resolution over the complete numerical range.<br><br>

<b>Floating point:</b> The distance between consecutive representable numbers increases as the exponent increases.<br>
Small numerical values are represented more densely, whereas larger values are represented with progressively greater spacing.<br><br>

<b>Coefficient bits:</b> Determine the number of representable coefficient values inside every exponent cycle.<br>
Increasing the number of coefficient bits produces more representable values inside each cycle and therefore increases numerical precision.<br><br>

<b>Maximum cycle:</b> Determines the largest exponent region displayed in the graph.<br>
Increasing this value extends the dynamic range and makes the progressive increase in floating-point spacing more evident.<br><br>

<b>Blue points:</b> Fixed-point representable values.<br>
<b>Orange points:</b> Floating-point representable values.

</div>
""")


table_html.layout = Layout(width='390px', min_width='390px')


controls = VBox(
    [coefficient_slider, cycle_slider],
    layout=Layout(
        width='310px',
        min_width='310px',
        overflow='visible',
        justify_content='flex-start',
        margin='70px 0 0 10px'
    )
)


bottom_layout = HBox(
    [table_html, controls],
    layout=Layout(
        width='710px',
        align_items='flex-start',
        justify_content='flex-start',
        overflow='visible'
    )
)


main_layout = VBox(
    [widget_plot.children[-1], bottom_layout],
    layout=Layout(
        width='900px',
        align_items='flex-start',
        overflow='visible'
    )
)


display(
    VBox(
        [theory_html, main_layout],
        layout=Layout(
            width='100%',
            align_items='flex-start',
            overflow='visible'
        )
    )
)
```
