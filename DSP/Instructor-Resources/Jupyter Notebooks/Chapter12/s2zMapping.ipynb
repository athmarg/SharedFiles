```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import FloatSlider, VBox, HBox, HTML, Layout
from IPython.display import display

# ============================================================
# BACKWARD-DIFFERENCE MAPPING: s-PLANE -> z-PLANE
# ============================================================

plt.ioff()

CONTENT_WIDTH = '900px'

plt.rcParams.update({'font.size':11,'axes.titlesize':12.5,'axes.labelsize':11,'xtick.labelsize':9.5,'ytick.labelsize':9.5,'legend.fontsize':9})

# ============================================================
# STYLE
# ============================================================

display(HTML("""
<style>

.bd-root{
    width:900px;
    max-width:900px;
    font-family:Arial,sans-serif;
}

.bd-header{
    background:linear-gradient(90deg,#1565c0,#1976d2);
    color:white;
    padding:11px 15px;
    border-radius:8px 8px 0 0;
    font-size:18px;
    font-weight:bold;
}

.bd-doc{
    background:#f6f9fd;
    border:1px solid #b9cce5;
    border-top:none;
    padding:11px 14px;
    border-radius:0 0 8px 8px;
    font-size:14.5px;
    line-height:1.55;
    margin-bottom:9px;
}

.bd-box{
    width:100%;
    box-sizing:border-box;
    border:1px solid #b9cce5;
    border-radius:7px;
    padding:9px 11px;
    margin-bottom:8px;
    font-size:14px;
    line-height:1.50;
}

.bd-title{
    font-weight:bold;
    color:#0d47a1;
    font-size:15px;
    margin-bottom:6px;
}

.bd-cols{
    display:flex;
    gap:20px;
    align-items:flex-start;
}

.bd-col{
    flex:1;
    min-width:0;
}

.widget-label{
    font-size:13.5px !important;
}

.jupyter-widgets input{
    font-size:13px !important;
}

.jupyter-widgets-output-area,
.widget-output,
.output_area,
.output_subarea,
.jp-OutputArea-output,
.jp-OutputArea-child{
    overflow-x:visible !important;
    overflow-y:visible !important;
    max-width:none !important;
}

.jp-OutputArea,
.output_wrapper,
.widget-box,
.jupyter-widgets{
    overflow:visible !important;
}

</style>
"""))

# ============================================================
# DOCUMENTATION
# ============================================================

display(HTML("""
<div class="bd-root">

<div class="bd-header">
Backward-Difference Mapping from the s-Plane to the z-Plane
</div>

<div class="bd-doc">

<b>Transformation.</b>
Approximating a continuous derivative by the backward difference leads to

<br><br>

<div style="text-align:center;font-size:16px;">
<b>s = (1-z<sup>-1</sup>)/T</b>
&nbsp;&nbsp;&Longleftrightarrow;&nbsp;&nbsp;
<b>z = 1/(1-sT)</b>.
</div>

<br>

For points on the imaginary axis, s=jΩ, the image in the z-plane lies on a circle centered at
(1/2,0) with radius 1/2, rather than on the unit circle. A point in the left half of the s-plane is mapped inside the unit circle, illustrating that the
backward-difference transformation preserves stability.

</div>

</div>
"""))

# ============================================================
# CONTROLS
# ============================================================

T_slider = FloatSlider(value=0.10,min=0.02,max=0.40,step=0.01,description='T:',continuous_update=True,readout_format='.2f',style={'description_width':'45px'},layout=Layout(width='270px'))

alpha_slider = FloatSlider(value=-1.0,min=-4.0,max=0.8,step=0.05,description='α:',continuous_update=True,readout_format='.2f',style={'description_width':'45px'},layout=Layout(width='270px'))

beta_slider = FloatSlider(value=2.0,min=-8.0,max=8.0,step=0.10,description='β:',continuous_update=True,readout_format='.2f',style={'description_width':'45px'},layout=Layout(width='270px'))

controls = VBox([
    HTML('<div class="bd-title">Mapping parameters</div>'),
    HBox([T_slider,alpha_slider,beta_slider])
],layout=Layout(width=CONTENT_WIDTH,border='1px solid #b9cce5',padding='9px 12px',margin='0 0 8px 0'))

info = HTML(layout=Layout(width=CONTENT_WIDTH,margin='0 0 8px 0'))

# ============================================================
# FIGURE — CREATED ONCE
# ============================================================

fig,(ax_s,ax_z) = plt.subplots(1,2,figsize=(9.0,4.4))

fig.canvas.toolbar_visible = False
fig.canvas.header_visible = False
fig.canvas.footer_visible = False

# ============================================================
# s-PLANE
# ============================================================

ax_s.axhline(0,color='black',linewidth=0.8)
ax_s.axvline(0,color='black',linewidth=0.8)

ax_s.axvspan(-5,0,alpha=0.05,label='Stable half-plane')

imag_axis_s, = ax_s.plot([0,0],[-10,10],'--',linewidth=1.2,label=r'Imaginary axis $s=j\Omega$')
point_s, = ax_s.plot([],[],'ro',markersize=7,label=r'$s=\alpha+j\beta$')

ax_s.set_xlim(-5,2)
ax_s.set_ylim(-10,10)
ax_s.set_aspect('equal',adjustable='box')

ax_s.set_title('s-Plane')
ax_s.set_xlabel(r'$\Re\{s\}$')
ax_s.set_ylabel(r'$\Im\{s\}$')

ax_s.grid(True,linestyle=':',alpha=0.25)
ax_s.legend(loc='upper center',bbox_to_anchor=(0.5,-0.16),ncol=2,frameon=False)

# ============================================================
# z-PLANE
# ============================================================

theta = np.linspace(0,2*np.pi,1000)

unit_x = np.cos(theta)
unit_y = np.sin(theta)

bd_x = 0.5+0.5*np.cos(theta)
bd_y = 0.5*np.sin(theta)

ax_z.axhline(0,color='black',linewidth=0.8)
ax_z.axvline(0,color='black',linewidth=0.8)

unit_circle, = ax_z.plot(unit_x,unit_y,'--',linewidth=1.2,label='Unit circle')
bd_circle, = ax_z.plot(bd_x,bd_y,linewidth=1.3,label='Backward-difference circle')

mapped_axis, = ax_z.plot([],[],color='red',linewidth=1.7,label=r'Image of $s=j\Omega$')
point_z, = ax_z.plot([],[],'ro',markersize=7,label='Mapped point')

ax_z.set_xlim(-1.2,1.2)
ax_z.set_ylim(-1.2,1.2)
ax_z.set_aspect('equal',adjustable='box')

ax_z.set_title('z-Plane')
ax_z.set_xlabel(r'$\Re\{z\}$')
ax_z.set_ylabel(r'$\Im\{z\}$')

ax_z.grid(True,linestyle=':',alpha=0.25)
ax_z.legend(loc='upper center',bbox_to_anchor=(0.5,-0.16),ncol=2,frameon=False)

plt.subplots_adjust(left=0.08,right=0.98,top=0.92,bottom=0.24,wspace=0.28)

# ============================================================
# UPDATE
# ============================================================

def update_mapping(change=None):

    T = T_slider.value
    alpha = alpha_slider.value
    beta = beta_slider.value

    # --------------------------------------------------------
    # Selected point in the s-plane
    # --------------------------------------------------------

    s = alpha+1j*beta

    # --------------------------------------------------------
    # Backward-difference transformation
    # --------------------------------------------------------

    z = 1.0/(1.0-s*T)

    point_s.set_data([alpha],[beta])
    point_z.set_data([np.real(z)],[np.imag(z)])

    # --------------------------------------------------------
    # Image of the imaginary axis s = j*Omega
    # --------------------------------------------------------

    Omega = np.linspace(-40,40,4000)

    s_axis = 1j*Omega
    z_axis = 1.0/(1.0-s_axis*T)

    mapped_axis.set_data(np.real(z_axis),np.imag(z_axis))

    # --------------------------------------------------------
    # Numerical information
    # --------------------------------------------------------

    stable_s = alpha < 0
    stable_z = np.abs(z) < 1

    info.value = f"""
    <div class="bd-root">

    <div class="bd-box">

    <div class="bd-title">Current mapping</div>

    <div class="bd-cols">

    <div class="bd-col">
    Sampling period: <b>T = {T:.2f}</b><br>
    s-plane point:
    <b>s = {alpha:.2f} {beta:+.2f}j</b>
    </div>

    <div class="bd-col">
    z-plane image:
    <b>z = {np.real(z):.4f} {np.imag(z):+.4f}j</b><br>
    Magnitude:
    <b>|z| = {np.abs(z):.4f}</b>
    </div>

    <div class="bd-col">
    s-plane stability:
    <b>{"STABLE" if stable_s else "NOT STABLE"}</b><br>
    z-plane stability:
    <b>{"STABLE" if stable_z else "NOT STABLE"}</b>
    </div>

    </div>

    </div>

    </div>
    """

    fig.canvas.draw_idle()

# ============================================================
# EVENTS
# ============================================================

T_slider.observe(update_mapping,names='value')
alpha_slider.observe(update_mapping,names='value')
beta_slider.observe(update_mapping,names='value')

# ============================================================
# DISPLAY
# ============================================================

display(controls)
display(info)
display(fig.canvas)

update_mapping()
```
