```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import freqz
from ipywidgets import HTML
from IPython.display import display

# ============================================================
# POLE-ZERO PLACEMENT
# FIRST-ORDER LOW-PASS AND HIGH-PASS EXAMPLES
# ============================================================

plt.ioff()

CONTENT_WIDTH = '900px'

plt.rcParams.update({'font.size':10.5,'axes.titlesize':12,'axes.labelsize':10.5,'xtick.labelsize':9,'ytick.labelsize':9,'legend.fontsize':8.5})

# ============================================================
# STYLE
# ============================================================

display(HTML("""
<style>

.pz-root{
    width:900px;
    max-width:900px;
    font-family:Arial,sans-serif;
}

.pz-header{
    background:linear-gradient(90deg,#1565c0,#1976d2);
    color:white;
    padding:10px 14px;
    border-radius:8px 8px 0 0;
    font-size:19px;
    font-weight:bold;
}

.pz-doc{
    background:#f6f9fd;
    border:1px solid #b9cce5;
    border-top:none;
    padding:9px 13px;
    border-radius:0 0 8px 8px;
    font-size:14.5px;
    line-height:1.45;
    margin-bottom:7px;
}

.pz-box{
    width:100%;
    box-sizing:border-box;
    border:1px solid #b9cce5;
    border-radius:7px;
    padding:8px 11px;
    margin-bottom:7px;
    font-size:14px;
    line-height:1.45;
}

.pz-title{
    font-weight:bold;
    color:#0d47a1;
    font-size:14.5px;
    margin-bottom:5px;
}

.pz-cols{
    display:flex;
    gap:18px;
    align-items:flex-start;
    flex-wrap:nowrap;
}

.pz-col{
    flex:1;
    min-width:0;
}

.pz-note{
    background:#fff9e8;
    border:1px solid #d9c477;
}

.jupyter-widgets-output-area,
.widget-output,
.output_area,
.output_subarea,
.jp-OutputArea-output,
.jp-OutputArea-child{
    overflow-x:visible !important;
    overflow-y:visible !important;
    max-width:none !important;
}

.jp-OutputArea,
.output_wrapper,
.widget-box,
.jupyter-widgets{
    overflow:visible !important;
}

</style>
"""))

# ============================================================
# DOCUMENTATION
# ============================================================

display(HTML("""
<div class="pz-root">

<div class="pz-header">
First-Order Low-Pass and High-Pass Filters by Pole-Zero Placement
</div>

<div class="pz-doc">

This notebook reproduces the two first-order pole-zero placement examples.

For the <b>low-pass filter</b>, the zero is fixed at <b>z = -1</b>, forcing the response
to zero at the Nyquist frequency. The real pole is placed at <b>z = α</b>, with

<div style="text-align:center;font-size:15.5px;margin:6px 0;">
<b>
α ≈ 1 - 2πf<sub>c</sub>/f<sub>s</sub>,
&nbsp;&nbsp;
H₀ = (1-α)/2.
</b>
</div>

For the <b>high-pass filter</b>, the zero is fixed at <b>z = +1</b>, forcing the response
to zero at DC. The real pole is again placed at <b>z = α</b>, but for the present
high-frequency example

<div style="text-align:center;font-size:15.5px;margin:6px 0;">
<b>
α ≈ -(1-π+2πf<sub>c</sub>/f<sub>s</sub>),
&nbsp;&nbsp;
H₀ = (1+α)/2.
</b>
</div>

The pole-zero diagrams and magnitude responses are shown side by side so that the
effect of the zero position and the real pole location can be observed directly.

</div>

</div>
"""))

# ============================================================
# LOW-PASS EXAMPLE
#
# fc = 110 Hz
# fs = 9000 Hz
# ============================================================

fc_lp = 110.0
fs_lp = 9000.0

alpha_lp = 1.0-2.0*np.pi*fc_lp/fs_lp

H0_lp = (1.0-alpha_lp)/2.0

b_lp = np.array([H0_lp,H0_lp])

a_lp = np.array([1.0,-alpha_lp])

zero_lp = -1.0

pole_lp = alpha_lp

# ============================================================
# HIGH-PASS EXAMPLE
#
# fc = 3900 Hz
# fs = 8000 Hz
# ============================================================

fc_hp = 3900.0
fs_hp = 8000.0

alpha_hp = -(1.0-np.pi+2.0*np.pi*fc_hp/fs_hp)

H0_hp = (1.0+alpha_hp)/2.0

b_hp = np.array([H0_hp,-H0_hp])

a_hp = np.array([1.0,-alpha_hp])

zero_hp = 1.0

pole_hp = alpha_hp

# ============================================================
# FREQUENCY RESPONSES
# ============================================================

omega_lp,H_lp = freqz(b_lp,a_lp,worN=32768)

omega_hp,H_hp = freqz(b_hp,a_hp,worN=32768)

fn_lp = omega_lp/np.pi

fn_hp = omega_hp/np.pi

mag_lp = np.abs(H_lp)

mag_hp = np.abs(H_hp)

# ============================================================
# CHECK NORMALIZATION
# ============================================================

H_lp_dc = (b_lp[0]+b_lp[1])/(1.0+a_lp[1])

H_hp_nyquist = (b_hp[0]-b_hp[1])/(1.0-a_hp[1])

# ============================================================
# NUMERICAL RESULTS
# ============================================================

display(HTML(f"""
<div class="pz-root">

<div class="pz-box">

<div class="pz-title">Numerical results</div>

<div class="pz-cols">

<div class="pz-col">

<b>Low-pass filter</b><br>

f<sub>c</sub> = <b>{fc_lp:.0f} Hz</b>,
&nbsp;
f<sub>s</sub> = <b>{fs_lp:.0f} Hz</b><br>

α = <b>{alpha_lp:.9f}</b><br>

H₀ = <b>{H0_lp:.9f}</b><br>

Zero: <b>z = -1</b><br>

Pole: <b>z = {pole_lp:.9f}</b>

</div>

<div class="pz-col">

<b>High-pass filter</b><br>

f<sub>c</sub> = <b>{fc_hp:.0f} Hz</b>,
&nbsp;
f<sub>s</sub> = <b>{fs_hp:.0f} Hz</b><br>

α = <b>{alpha_hp:.9f}</b><br>

H₀ = <b>{H0_hp:.9f}</b><br>

Zero: <b>z = +1</b><br>

Pole: <b>z = {pole_hp:.9f}</b>

</div>

</div>

</div>

<div class="pz-box pz-note">

<div class="pz-title">Transfer functions</div>

<div class="pz-cols">

<div class="pz-col">

<b>Low-pass</b>

<div style="text-align:center;font-size:14.5px;margin:5px 0;">
H(z) =
<b>
({b_lp[0]:.6f} + {b_lp[1]:.6f}z<sup>-1</sup>)
/
(1 {a_lp[1]:+.6f}z<sup>-1</sup>)
</b>
</div>

Normalization check:
<b>H(1) = {H_lp_dc:.6f}</b>

</div>

<div class="pz-col">

<b>High-pass</b>

<div style="text-align:center;font-size:14.5px;margin:5px 0;">
H(z) =
<b>
({b_hp[0]:.6f} {b_hp[1]:+.6f}z<sup>-1</sup>)
/
(1 {a_hp[1]:+.6f}z<sup>-1</sup>)
</b>
</div>

Normalization check:
<b>H(-1) = {H_hp_nyquist:.6f}</b>

</div>

</div>

</div>

</div>
"""))

# ============================================================
# FIGURE — 2 x 2 GRID
# ============================================================

fig,axes = plt.subplots(2,2,figsize=(9.0,6.4))

ax1,ax2,ax3,ax4 = axes.flat

fig.canvas.toolbar_visible = False
fig.canvas.header_visible = False
fig.canvas.footer_visible = False

theta = np.linspace(0,2*np.pi,1000)

unit_x = np.cos(theta)
unit_y = np.sin(theta)

# ============================================================
# 1. LOW-PASS POLE-ZERO PLOT
# ============================================================

ax1.axhline(0,color='black',linewidth=0.8)

ax1.axvline(0,color='black',linewidth=0.8)

ax1.plot(unit_x,unit_y,'--',linewidth=1.0,label='Unit circle')

ax1.plot([pole_lp],[0],'rx',markersize=8,markeredgewidth=1.8,label='Pole')

ax1.plot([zero_lp],[0],'bo',markersize=7,markerfacecolor='none',markeredgewidth=1.6,label='Zero')

ax1.set_xlim(-1.2,1.2)

ax1.set_ylim(-1.2,1.2)

ax1.set_aspect('equal',adjustable='box')

ax1.set_title('Low-Pass Pole-Zero Plot')

ax1.set_xlabel(r'$\Re\{z\}$')

ax1.set_ylabel(r'$\Im\{z\}$')

ax1.grid(True,linestyle=':',alpha=0.25)

ax1.legend(loc='upper center',bbox_to_anchor=(0.5,-0.16),ncol=3,frameon=False)

# ============================================================
# 2. LOW-PASS MAGNITUDE RESPONSE
# ============================================================

ax2.plot(fn_lp,mag_lp,color='red',linewidth=1.5,label='Low-pass response')

ax2.axvline(2.0*fc_lp/fs_lp,linestyle='--',linewidth=1.0,label=r'$f_c$')

ax2.set_xlim(0,1)

ax2.set_ylim(0,1.08)

ax2.set_title('Low-Pass Magnitude Response')

ax2.set_xlabel(r'Normalized frequency $\omega/\pi$')

ax2.set_ylabel(r'$|H(e^{j\omega})|$')

ax2.grid(True,linestyle=':',alpha=0.25)

ax2.legend(loc='upper center',bbox_to_anchor=(0.5,-0.16),ncol=2,frameon=False)

# ============================================================
# 3. HIGH-PASS POLE-ZERO PLOT
# ============================================================

ax3.axhline(0,color='black',linewidth=0.8)

ax3.axvline(0,color='black',linewidth=0.8)

ax3.plot(unit_x,unit_y,'--',linewidth=1.0,label='Unit circle')

ax3.plot([pole_hp],[0],'rx',markersize=8,markeredgewidth=1.8,label='Pole')

ax3.plot([zero_hp],[0],'bo',markersize=7,markerfacecolor='none',markeredgewidth=1.6,label='Zero')

ax3.set_xlim(-1.2,1.2)

ax3.set_ylim(-1.2,1.2)

ax3.set_aspect('equal',adjustable='box')

ax3.set_title('High-Pass Pole-Zero Plot')

ax3.set_xlabel(r'$\Re\{z\}$')

ax3.set_ylabel(r'$\Im\{z\}$')

ax3.grid(True,linestyle=':',alpha=0.25)

ax3.legend(loc='upper center',bbox_to_anchor=(0.5,-0.16),ncol=3,frameon=False)

# ============================================================
# 4. HIGH-PASS MAGNITUDE RESPONSE
# ============================================================

ax4.plot(fn_hp,mag_hp,color='red',linewidth=1.5,label='High-pass response')

ax4.axvline(2.0*fc_hp/fs_hp,linestyle='--',linewidth=1.0,label=r'$f_c$')

ax4.set_xlim(0,1)

ax4.set_ylim(0,1.08)

ax4.set_title('High-Pass Magnitude Response')

ax4.set_xlabel(r'Normalized frequency $\omega/\pi$')

ax4.set_ylabel(r'$|H(e^{j\omega})|$')

ax4.grid(True,linestyle=':',alpha=0.25)

ax4.legend(loc='upper center',bbox_to_anchor=(0.5,-0.16),ncol=2,frameon=False)

# ============================================================
# LAYOUT
# ============================================================

plt.subplots_adjust(left=0.08,right=0.98,top=0.94,bottom=0.13,wspace=0.28,hspace=0.56)

# ============================================================
# DISPLAY
# ============================================================

display(fig.canvas)
```
