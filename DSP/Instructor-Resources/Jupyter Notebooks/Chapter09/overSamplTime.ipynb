```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML

# Disable auto-scroll in notebook output for smooth interaction
display(HTML("<style>.output_scroll { height: unset !important; }</style>"))

def time_domain_oversampling_simulation(L):
    clear_output(wait=True)
    
    # Continuous-time / high-resolution base axis for smooth envelope
    t_fine = np.linspace(0, 1.0, 1000)
    f1 = 2.0  # Hz
    x_envelope = np.cos(2 * np.pi * f1 * t_fine)
    
    # Base Ticks (Original coarse samples x[n])
    base_points_count = 12
    n_base = np.arange(base_points_count)
    t_base = n_base / float(base_points_count)
    x_base = np.cos(2 * np.pi * f1 * t_base)
    
    # Oversampled / Interpolated discrete grid (x_u[n])
    interpolated_points_count = base_points_count * L
    n_interp = np.arange(interpolated_points_count)
    t_interp = n_interp / float(interpolated_points_count)
    x_interp = np.cos(2 * np.pi * f1 * t_interp)

    fig, axes = plt.subplots(2, 1, figsize=(11, 5), sharex=True)
    
    # --- PLOT 1: Original Sparse Signal ---
    axes[0].stem(t_base, x_base, linefmt='tab:blue', markerfmt='bo', basefmt='k-', label='Original Sparse Samples $x[n]$')
    axes[0].plot(t_fine, x_envelope, color='tab:blue', alpha=0.3, linestyle='--', label='Continuous Envelope')
    axes[0].set_title(r'1. Original Sparse Discrete Signal', fontsize=10, fontweight='bold')
    axes[0].set_ylabel('Amplitude')
    axes[0].set_ylim(-1.5, 1.5)
    axes[0].grid(True, linestyle='--', alpha=0.6)
    axes[0].legend(loc='upper left', bbox_to_anchor=(1.02, 1), frameon=True)

    # --- PLOT 2: Oversampled / Interpolated Signal ---
    axes[1].stem(t_interp, x_interp, linefmt='tab:orange', markerfmt='ro', basefmt='k-', label=fr'Oversampled Signal $x_u[n]$ ($L = {L}$)')
    axes[1].plot(t_fine, x_envelope, color='tab:orange', alpha=0.3, linestyle='--', label='Interpolated Envelope')
    axes[1].set_title(fr'2. Oversampled Signal after Interpolation by Factor $L = {L}$', fontsize=10, fontweight='bold')
    axes[1].set_xlabel('Time ($s$)')
    axes[1].set_ylabel('Amplitude')
    axes[1].set_xlim(0, 1.0)
    axes[1].set_ylim(-1.5, 1.5)
    axes[1].grid(True, linestyle='--', alpha=0.6)
    axes[1].legend(loc='upper left', bbox_to_anchor=(1.02, 1), frameon=True)

    plt.tight_layout()
    plt.show()
    
    # Print information
    print(f"--- Time Domain Oversampling Info (L = {L}) ---")
    print(f"• Original sample count: {base_points_count}")
    print(f"• Oversampled sample count: {len(t_interp)}")
    print(f"• Sampling rate multiplication factor: {L}")

# Interactive widget slider for oversampling factor L
l_slider_time = widgets.IntSlider(
    value=2, min=1, max=8, step=1,
    description='Oversampling ($L$):',
    style={'description_width': 'initial'}
)

ui_time = widgets.VBox([l_slider_time])
display(ui_time)

out_time = widgets.interactive_output(time_domain_oversampling_simulation, {'L': l_slider_time})
display(out_time)
```
