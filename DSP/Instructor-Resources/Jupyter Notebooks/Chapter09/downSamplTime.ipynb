```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML

# Disable auto-scroll in notebook output for smooth interaction
display(HTML("<style>.output_scroll { height: unset !important; }</style>"))

def time_domain_decimation_simulation(M):
    clear_output(wait=True)
    
    # Continuous-time / high-resolution time axis
    t = np.linspace(0, 1.0, 1000)
    
    # Original signal consisting of two low-frequency sinusoids
    f1 = 3.0  # Hz
    f2 = 7.0  # Hz
    x_continuous = np.cos(2 * np.pi * f1 * t) + 0.5 * np.sin(2 * np.pi * f2 * t)
    
    # Decimated / Subsampled time grid and signal
    # Selecting every M-th sample from a finely sampled discrete base
    n_indices = np.arange(0, len(t), M)
    t_decimated = t[n_indices]
    x_decimated = x_continuous[n_indices]

    fig, axes = plt.subplots(2, 1, figsize=(10, 5), sharex=True)
    
    # --- PLOT 1: Original Signal (Time Domain) ---
    axes[0].plot(t, x_continuous, color='tab:blue', linewidth=2, label='Original Continuous/High-Res Signal $x(t)$')
    axes[0].set_title(r'1. Original Signal in the Time Domain', fontsize=10, fontweight='bold')
    axes[0].set_ylabel('Amplitude')
    axes[0].set_ylim(-1.8, 1.8)
    axes[0].grid(True, linestyle='--', alpha=0.6)
    axes[0].legend(loc='upper right', frameon=True)

    # --- PLOT 2: Decimated Signal (Time Domain) ---
    axes[1].stem(t_decimated, x_decimated, linefmt='tab:orange', markerfmt='ro', basefmt='k-', label=fr'Decimated Signal $x[n]$ ($M = {M}$)')
    # Overlay faint original signal for visual comparison
    axes[1].plot(t, x_continuous, color='tab:blue', alpha=0.3, linestyle='--', label='Original Envelope')
    axes[1].set_title(fr'2. Decimated Signal after Downsampling by Factor $M = {M}$', fontsize=10, fontweight='bold')
    axes[1].set_xlabel('Time ($s$)')
    axes[1].set_ylabel('Amplitude')
    axes[1].set_xlim(0, 1.0)
    axes[1].set_ylim(-1.8, 1.8)
    axes[1].grid(True, linestyle='--', alpha=0.6)
    axes[1].legend(loc='upper right', frameon=True)

    plt.tight_layout()
    plt.show()
    
    # Print information
    print(f"--- Time Domain Decimation Info (M = {M}) ---")
    print(f"• Original number of points: {len(t)}")
    print(f"• Decimated number of points: {len(t_decimated)}")
    print(f"• Effective sampling rate reduction factor: {M}")

# Interactive widget slider for decimation factor M
m_slider_time = widgets.IntSlider(
    value=4, min=1, max=25, step=1,
    description='Decimation ($M$):',
    style={'description_width': 'initial'}
)

ui_time = widgets.VBox([m_slider_time])
display(ui_time)

out_time = widgets.interactive_output(time_domain_decimation_simulation, {'M': m_slider_time})
display(out_time)
```
