```python
import sympy as sp
from IPython.display import display, Math

sp.init_printing(use_unicode=True)

z = sp.Symbol('z', complex=True)
n = sp.Symbol('n', integer=True, nonnegative=True)
k = sp.Symbol('k', integer=True, nonnegative=True)
alpha_1, beta_0, beta_1 = sp.symbols('alpha_1 beta_0 beta_1', real=True)
A = sp.Symbol('A', real=True, nonzero=True)
omega_0 = sp.Symbol('omega_0', real=True)


# =============================================================================
# Given difference equation
# =============================================================================

display("Exercise 3.47 — Symbolic Solution")
display("Given system")
display(Math(r"y[n]=-\alpha_1y[n-1]+\beta_0x[n]+\beta_1x[n-1]"))


# =============================================================================
# (a) Impulse response
# =============================================================================

display("(a) Impulse response")

Y, X = sp.symbols('Y(z) X(z)')
Z_equation = sp.Eq(Y, -alpha_1*z**(-1)*Y + beta_0*X + beta_1*z**(-1)*X)
H = sp.factor(sp.simplify(sp.solve(Z_equation, Y)[0] / X))

display(Math(r"\mathcal{H}(z)=\frac{\mathcal{Y}(z)}{\mathcal{X}(z)}=" + sp.latex(H)))

H_rational = sp.factor(sp.together(H))
display(Math(r"\mathcal{H}(z)=" + sp.latex(H_rational)))

poles_H = sp.solve(sp.Eq(sp.denom(H_rational), 0), z)
display(Math(r"\text{Poles: }" + ", ".join(sp.latex(p) for p in poles_H)))

H_decomposed = sp.apart(H, z)
display(Math(r"\mathcal{H}(z)=" + sp.latex(H_decomposed)))


# =============================================================================
# Impulse response
# =============================================================================

h_component_1 = (-alpha_1)**n
h_component_2 = (-alpha_1)**(n - 1)
h = sp.factor(beta_0*h_component_1 + beta_1*h_component_2)

display(Math(r"h[n]=" + sp.latex(beta_0*h_component_1) + r"u[n]+" + sp.latex(beta_1*h_component_2) + r"u[n-1]"))


# =============================================================================
# (b) Zero-state step response
# =============================================================================

display("(b) Zero-state step response")

display(Math(r"x[n]=u[n]"))
display(Math(r"y_{ZS}[n]=h[n]*x[n]"))
display(Math(r"y_{ZS}[n]=\sum_{k=-\infty}^{+\infty}h[k]x[n-k]"))
display(Math(r"y_{ZS}[n]=\beta_0\sum_{k=0}^{n}(-\alpha_1)^k+\beta_1\sum_{k=1}^{n}(-\alpha_1)^{k-1}"))

sum_1 = sp.factor((1 - (-alpha_1)**(n + 1)) / (1 + alpha_1))
sum_2 = sp.factor((1 - (-alpha_1)**n) / (1 + alpha_1))

display(Math(r"\sum_{k=0}^{n}(-\alpha_1)^k=" + sp.latex(sum_1)))
display(Math(r"\sum_{k=1}^{n}(-\alpha_1)^{k-1}=" + sp.latex(sum_2)))

y_zs = sp.factor(sp.simplify(beta_0*sum_1 + beta_1*sum_2))
display(Math(r"y_{ZS}[n]=" + sp.latex(y_zs) + r"\,u[n]"))


# =============================================================================
# (c) Step response with initial condition y[-1] = A
# =============================================================================

display("(c) Step response with initial condition")
display(Math(r"y[-1]=A,\qquad A\neq0"))

Y_plus = sp.Symbol('Y^+(z)')
Y_plus_shift = z**(-1)*Y_plus + A
display(Math(r"\mathcal{Z}^{+}\{y[n-1]\}=" + sp.latex(Y_plus_shift)))

ZI_equation = sp.Eq(Y_plus, -alpha_1*Y_plus_shift)
Y_zi = sp.factor(sp.solve(ZI_equation, Y_plus)[0])
display(Math(r"\mathcal{Y}_{ZI}(z)=" + sp.latex(Y_zi)))

poles_zi = sp.solve(sp.Eq(sp.denom(sp.together(Y_zi)), 0), z)
display(Math(r"\text{Poles: }" + ", ".join(sp.latex(p) for p in poles_zi)))

F_zi = sp.factor(Y_zi*z**(n - 1))
res_zi = sp.factor(sp.simplify(sp.residue(F_zi, z, poles_zi[0])))
y_zi = sp.factor(sp.simplify(res_zi))
display(Math(r"y_{ZI}[n]=" + sp.latex(y_zi) + r"\,u[n]"))


# =============================================================================
# Total response
# =============================================================================

y_total = sp.factor(sp.simplify(y_zs + y_zi))

display(Math(r"y[n]=y_{ZS}[n]+y_{ZI}[n]"))
display(Math(r"y[n]=" + sp.latex(y_total) + r"\,u[n]"))


# =============================================================================
# (d) Response to x[n] = cos(omega_0 n)u[n]
# =============================================================================

display("(d) Response to x[n] = cos(omega_0 n)u[n]")
display(Math(r"x[n]=\cos(\omega_0n)u[n]"))


# =============================================================================
# Z-transform of cosine input
# =============================================================================

X_cos = sp.Rational(1, 2) * (1 / (1 - sp.exp(sp.I*omega_0)*z**(-1)) + 1 / (1 - sp.exp(-sp.I*omega_0)*z**(-1)))
X_cos = sp.factor(sp.together(X_cos))

display(Math(r"\mathcal{X}(z)=" + sp.latex(X_cos)))

X_cos_real = sp.trigsimp(sp.simplify(X_cos.rewrite(sp.cos)))
display(Math(r"\mathcal{X}(z)=" + sp.latex(X_cos_real)))


# =============================================================================
# Output transform
# =============================================================================

Y_cos = sp.factor(sp.together(H*X_cos))
display(Math(r"\mathcal{Y}(z)=\mathcal{H}(z)\mathcal{X}(z)=" + sp.latex(Y_cos)))


# =============================================================================
# Poles of Y(z)
# =============================================================================

den_Y_cos = sp.factor(sp.denom(sp.together(Y_cos)))
poles_cos = sp.solve(sp.Eq(den_Y_cos, 0), z)

display(Math(r"\text{Poles: }" + ", ".join(sp.latex(pole) for pole in poles_cos)))


# =============================================================================
# Inverse Z-transform kernel
# =============================================================================

A_kernel = sp.factor(Y_cos*z**(n - 1))
display(Math(r"\mathcal{A}(z)=" + sp.latex(A_kernel)))


# =============================================================================
# Residues
# =============================================================================

display("Residues")

residues_cos = []

for pole in poles_cos:
    residue = sp.factor(sp.simplify(sp.residue(A_kernel, z, pole)))
    residues_cos.append(residue)
    display(Math(r"\operatorname{Res}_{z=" + sp.latex(pole) + r"}=" + sp.latex(residue)))


# =============================================================================
# Final response
# =============================================================================

y_cos = sp.factor(sp.trigsimp(sp.simplify(sp.cancel(sum(residues_cos)))))

display(Math(r"y[n]=" + sp.latex(y_cos) + r"\,u[n]"))
```
