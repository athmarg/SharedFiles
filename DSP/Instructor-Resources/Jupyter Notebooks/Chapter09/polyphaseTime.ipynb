```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, HTML

# Disable auto-scroll in notebook output
display(HTML("<style>.output_scroll { height: unset !important; }</style>"))

def polyphase_decomposition_demo():
    # 1. Define the original impulse response h[n] of length 12 (as in Example 1)
    h = np.array([10, 15, 8, 22, 14, 9, 18, 12, 7, 25, 11, 16])
    N = len(h)
    M = 3  # Polyphase factor
    
    # 2. Extract Polyphase Components e_k[n]
    # e_k[n] = h[n*M + k]
    e0 = h[0::3]  # indices 0, 3, 6, 9
    e1 = h[1::3]  # indices 1, 4, 7, 10
    e2 = h[2::3]  # indices 2, 5, 8, 11
    
    # 3. Construct the interpolated/delayed branches h_k[n] for reconstruction
    # h_0[n]: samples at 0, 3, 6, 9 (padded with zeros)
    h0_full = np.zeros(N)
    h0_full[0::3] = e0
    
    # h_1[n]: shifted by 1 with samples at 1, 4, 7, 10
    h1_full = np.zeros(N)
    h1_full[1::3] = e1
    
    # h_2[n]: shifted by 2 with samples at 2, 5, 8, 11
    h2_full = np.zeros(N)
    h2_full[2::3] = e2
    
    # Reconstruction by summation
    h_reconstructed = h0_full + h1_full + h2_full
    
    # 4. Plotting the results
    fig, axes = plt.subplots(5, 1, figsize=(10, 12), sharex=True)
    n_axis = np.arange(N)
    
    # Original signal
    axes[0].stem(n_axis, h, basefmt=" ", linefmt='tab:blue', markerfmt='bo')
    axes[0].set_title(r'Original Impulse Response $h[n]$ ($N=12$)', fontsize=10, fontweight='bold')
    axes[0].set_ylabel('Amplitude')
    axes[0].grid(True, linestyle='--', alpha=0.6)
    
    # Polyphase Branch 0
    axes[1].stem(n_axis, h0_full, basefmt=" ", linefmt='tab:orange', markerfmt='o')
    axes[1].set_title(r'Branch 0: $h_0[n]$ (derived from $e_0[n] = \{h[0], h[3], h[6], h[9]\}$)', fontsize=10, fontweight='bold')
    axes[1].set_ylabel('Amplitude')
    axes[1].grid(True, linestyle='--', alpha=0.6)
    
    # Polyphase Branch 1
    axes[2].stem(n_axis, h1_full, basefmt=" ", linefmt='tab:green', markerfmt='o')
    axes[2].set_title(r'Branch 1: $h_1[n]$ (derived from $e_1[n] = \{h[1], h[4], h[7], h[10]\}$ with 1-unit delay)', fontsize=10, fontweight='bold')
    axes[2].set_ylabel('Amplitude')
    axes[2].grid(True, linestyle='--', alpha=0.6)
    
    # Polyphase Branch 2
    axes[3].stem(n_axis, h2_full, basefmt=" ", linefmt='tab:purple', markerfmt='o')
    axes[3].set_title(r'Branch 2: $h_2[n]$ (derived from $e_2[n] = \{h[2], h[5], h[8], h[11]\}$ with 2-unit delay)', fontsize=10, fontweight='bold')
    axes[3].set_ylabel('Amplitude')
    axes[3].grid(True, linestyle='--', alpha=0.6)
    
    # Reconstructed signal
    axes[4].stem(n_axis, h_reconstructed, basefmt=" ", linefmt='tab:red', markerfmt='ro')
    axes[4].set_title(r'Reconstructed Signal $\sum h_k[n-k]$ (Matches Original $h[n]$)', fontsize=10, fontweight='bold')
    axes[4].set_xlabel('Sample Index ($n$)')
    axes[4].set_ylabel('Amplitude')
    axes[4].grid(True, linestyle='--', alpha=0.6)
    
    plt.tight_layout()
    plt.show()
    
    # Conclusion Box
    display(HTML("""
    <div style="border: 1px solid #ccc; padding: 12px; border-radius: 5px; background-color: #f9f9f9; font-family: sans-serif; font-size: 14px;">
        <strong>Conclusion:</strong> 
        This demonstration verifies the principles of <strong>polyphase decomposition</strong> (Example 1). 
        The original filter impulse response $h[n]$ is successfully broken down into $M=3$ parallel polyphase components ($e_0, e_1, e_2$). 
        After zero-insertion and appropriate time delays, their summation perfectly reconstructs the original sequence, illustrating the structural foundation for efficient multi-rate filter architectures.
    </div>
    """))

# Run the demonstration function
polyphase_decomposition_demo()
```
