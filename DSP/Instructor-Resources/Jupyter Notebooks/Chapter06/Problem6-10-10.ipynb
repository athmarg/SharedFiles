```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, Markdown

display(Markdown("### 7. Convolution of $x[n] = -n 3^n u[-n]$ and $y[n] = u[-n-1]$"))

n_base7 = np.arange(-10, 3)
n = n_base7

x7 = np.zeros_like(n, dtype=float)
y7 = np.zeros_like(n, dtype=float)
z7 = np.zeros_like(n, dtype=float)

for idx, val in enumerate(n):
    # x[n] = -n * 3^n for n <= 0
    if val <= 0:
        x7[idx] = -val * (3.0 ** val)
    # y[n] = 1 for n <= -1
    if val <= -1:
        y7[idx] = 1.0
    # Convolution analytical formula for n <= -2
    if val <= -2:
        z7[idx] = 0.75 + (2.0 * val - 1.0) * (3.0 ** (val + 1)) / 4.0
    elif val == -1:
        z7[idx] = 0.75 + (-3.0) * (3.0 ** 0) / 4.0 # evaluated at boundary limit
    else:
        z7[idx] = 0.0

fig, axes = plt.subplots(3, 1, figsize=(10, 7.5), sharex=True)
axes[0].stem(n, x7, linefmt='b-', markerfmt='bo', basefmt='k-')
axes[0].set_title(r'Signal $x[n] = -n 3^n u[-n]$', fontsize=9.5, fontweight='bold', color='darkblue')
axes[0].set_ylabel('Amplitude', fontsize=8.5)
axes[0].grid(True, linestyle='--', alpha=0.6)

axes[1].stem(n, y7, linefmt='g-', markerfmt='go', basefmt='k-')
axes[1].set_title(r'Signal $y[n] = u[-n-1]$', fontsize=9.5, fontweight='bold', color='darkgreen')
axes[1].set_ylabel('Amplitude', fontsize=8.5)
axes[1].grid(True, linestyle='--', alpha=0.6)

axes[2].stem(n, z7, linefmt='r-', markerfmt='ro', basefmt='k-')
axes[2].set_title(r'Convolution Result $z[n] = x[n] * y[n]$', fontsize=9.5, fontweight='bold', color='darkred')
axes[2].set_xlabel('Index $n$', fontsize=8.5)
axes[2].set_ylabel('Amplitude', fontsize=8.5)
axes[2].grid(True, linestyle='--', alpha=0.6)
plt.show()
```
