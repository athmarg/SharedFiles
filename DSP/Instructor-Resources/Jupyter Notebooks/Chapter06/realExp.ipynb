```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets

@widgets.interact(
    k=widgets.FloatSlider(value=0.8, min=-1.5, max=1.5, step=0.05, description='Parameter k:', style={'description_width': 'initial'}, layout=widgets.Layout(width='700px')),
    num_samples=widgets.IntSlider(value=15, min=5, max=30, step=1, description='Samples:', style={'description_width': 'initial'}, layout=widgets.Layout(width='700px'))
)
def plot_real_exponential(k, num_samples):
    n = np.arange(0, num_samples)
    x = k**n
    
    fig, ax = plt.subplots(figsize=(10, 5))
    ax.stem(n, x, linefmt='purple', markerfmt='o', basefmt='k-')
    
    # Categorize behavior for educational feedback
    behavior = ""
    if abs(k) < 1:
        behavior += "Decaying (Aenuated)"
    elif abs(k) == 1:
        behavior += "Constant amplitude"
    else:
        behavior += "Growing (Unbounded)"
        
    if k < 0:
        behavior += " with Sign Alternation"
    
    ax.set_title(rf'Real Exponential Signal: $x[n] = ({k:.2f})^n$ — Behavior: {behavior}', fontsize=11, fontweight='bold')
    ax.set_xlabel('Discrete Time index $n$')
    ax.set_ylabel('$x[n]$')
    ax.grid(True, linestyle=':', alpha=0.7)
    
    # Dynamic limits for clean visualization
    y_max = max(2.0, np.max(np.abs(x)) * 1.2) if len(x) > 0 else 2.0
    if k < -1 or k > 1:
        ax.set_ylim(-y_max, y_max)
    else:
        ax.set_ylim(-1.5, 1.5)
        
    plt.show()
```
