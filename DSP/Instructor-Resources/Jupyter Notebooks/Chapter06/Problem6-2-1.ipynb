```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Definition of the continuous signal x(t) = 5 * cos(200 * pi * t)
# Signal parameters
A = 5
F = 100  # derived from 2 * pi * F = 200 * pi -> F = 100 Hz
omega = 2 * np.pi * F

# Create a dense time axis 't' for smooth continuous signal plotting (40 ms duration)
t_continuous = np.linspace(0, 0.04, 1000)
x_continuous = A * np.cos(omega * t_continuous)

# 2. Sampling Frequencies (Changed Fs1 to 250 Hz for proper phase distribution)
Fs1 = 250   # Hz
Fs2 = 800   # Hz

Ts1 = 1 / Fs1
Ts2 = 1 / Fs2

# Create discrete time instants using true uniform sampling over 40 ms
t_samples1 = np.arange(0, 0.04, Ts1)
x_samples1 = A * np.cos(omega * t_samples1)

t_samples2 = np.arange(0, 0.04, Ts2)
x_samples2 = A * np.cos(omega * t_samples2)

# 3. Plotting the signals
plt.figure(figsize=(10, 6))

# Continuous-time signal (Blue)
plt.plot(t_continuous * 1000, x_continuous, label='Continuous Signal x(t)', color='blue', linewidth=1.5)

# Samples for Fs1 = 250 Hz (Green)
plt.stem(t_samples1 * 1000, x_samples1, linefmt='g-', markerfmt='go', basefmt='g--', 
         label=f'Samples at Fs1 = {Fs1} Hz')

# Samples for Fs2 = 800 Hz (Red)
plt.stem(t_samples2 * 1000, x_samples2, linefmt='r-', markerfmt='ro', basefmt='r--', 
         label=f'Samples at Fs2 = {Fs2} Hz')

# Formatting and labels
plt.title(r'Sampling of Continuous-Time Signal x(t) = 5cos(200πt)')
plt.xlabel('Time (ms)')
plt.ylabel('Amplitude (V)')
plt.axhline(0, color='black', linewidth=0.8, linestyle='--')
plt.grid(True, linestyle=':', alpha=0.7)

# Move the legend outside and below the plot
plt.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=3)
plt.tight_layout()
plt.show()
```
