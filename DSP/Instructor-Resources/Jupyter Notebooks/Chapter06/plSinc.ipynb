```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, Markdown

# --- Educational Note on Advanced Elementary Signals ---
display(Markdown(r"""
> **Educational Note:**
> This notebook displays three advanced elementary discrete-time signals:
> 1. **Rectangular Pulse ($\Pi_D[n]$):** Equals $1$ for $0 \leq n < D$ and $0$ otherwise.
> 2. **Triangular Pulse ($\Lambda_{D_t}[n]$):** A symmetrical or shaped finite triangular window sequence with an independent duration parameter.
> 3. **Discrete Sinc Function ($\text{sinc}(\omega n)$):** Defined safely using normalized sinc to prevent runtime division-by-zero warnings.
> 
> Use the sliders below to dynamically adjust the individual pulse durations, frequency, and visualization range.
"""))

@widgets.interact(
    D_rect=widgets.IntSlider(value=8, min=2, max=20, step=1, description='Rect. Duration D:', style={'description_width': 'initial'}, layout=widgets.Layout(width='700px')),
    D_tri=widgets.IntSlider(value=8, min=2, max=20, step=1, description='Tri. Duration D_t:', style={'description_width': 'initial'}, layout=widgets.Layout(width='700px')),
    omega=widgets.FloatSlider(value=0.25, min=0.05, max=1.0, step=0.05, description=r'Freq ω:', style={'description_width': 'initial'}, layout=widgets.Layout(width='700px')),
    n_range=widgets.IntSlider(value=15, min=5, max=30, step=1, description=r'Range ($\pm n$):', style={'description_width': 'initial'}, layout=widgets.Layout(width='700px'))
)
def plot_advanced_elementary_signals_custom(D_rect, D_tri, omega, n_range):
    n = np.arange(-n_range, n_range + 1)
    
    # 1. Rectangular Pulse using its own duration parameter D_rect
    rect_n = np.where((n >= 0) & (n < D_rect), 1.0, 0.0)
    
    # 2. Triangular Pulse using its own independent duration parameter D_tri
    cond = np.abs(n + 1 - D_tri) <= (D_tri - 1)
    tri_n = np.where(cond, 1.0 - (np.abs(n + 1 - D_tri) / D_tri), 0.0)
    
    # 3. Discrete Sinc Function
    sinc_n = np.sinc(omega * n)
    
    fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(16, 4.5))
    
    # --- Rectangular Pulse Plot ---
    markerline1, stemlines1, baseline1 = ax1.stem(n, rect_n, linefmt='b-', markerfmt='bo', basefmt='k-')
    plt.setp(stemlines1, linewidth=1.5)
    ax1.set_title(rf'Rectangular Pulse: $\Pi_{{{D_rect}}}[n]$', fontsize=11, fontweight='bold', color='darkblue')
    ax1.set_xlabel('Index $n$', fontsize=10)
    ax1.set_ylabel('Amplitude', fontsize=10)
    ax1.grid(True, linestyle='--', alpha=0.6)
    ax1.set_ylim(-0.2, 1.3)
    ax1.set_xlim(-n_range - 1, n_range + 1)
    
    # --- Triangular Pulse Plot ---
    markerline2, stemlines2, baseline2 = ax2.stem(n, tri_n, linefmt='g-', markerfmt='go', basefmt='k-')
    plt.setp(stemlines2, linewidth=1.5)
    ax2.set_title(rf'Triangular Pulse: $\Lambda_{{{D_tri}}}[n]$', fontsize=11, fontweight='bold', color='darkgreen')
    ax2.set_xlabel('Index $n$', fontsize=10)
    ax2.set_ylabel('Amplitude', fontsize=10)
    ax2.grid(True, linestyle='--', alpha=0.6)
    ax2.set_ylim(-0.2, 1.3)
    ax2.set_xlim(-n_range - 1, n_range + 1)
    
    # --- Discrete Sinc Function Plot ---
    markerline3, stemlines3, baseline3 = ax3.stem(n, sinc_n, linefmt='r-', markerfmt='ro', basefmt='k-')
    plt.setp(stemlines3, linewidth=1.5)
    ax3.set_title(rf'Sinc Function: $\text{{sinc}}({omega:.2f}\pi n)$', fontsize=11, fontweight='bold', color='darkred')
    ax3.set_xlabel('Index $n$', fontsize=10)
    ax3.set_ylabel('Amplitude', fontsize=10)
    ax3.grid(True, linestyle='--', alpha=0.6)
    ax3.set_ylim(-0.4, 1.25)
    ax3.set_xlim(-n_range - 1, n_range + 1)
    
    plt.tight_layout()
    plt.show()
```
