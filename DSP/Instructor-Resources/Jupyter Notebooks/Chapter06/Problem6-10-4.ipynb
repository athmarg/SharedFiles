```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, Markdown

# --- Fix output height for Binder ---
display(Markdown("<style>.output_scroll { max-height: none !important; }</style>"))

display(Markdown("### 1. Convolution of $x[n] = \\alpha^n u[n]$ and $y[n] = u[n]$"))

n_base1 = np.arange(-5, 15)

def update_prob1(alpha):
    n = n_base1
    u = (n >= 0).astype(float)
    x = (alpha ** n) * u
    y = u
    
    # Analytical convolution calculation
    z = np.zeros_like(n, dtype=float)
    for idx, val in enumerate(n):
        if val >= 0:
            z[idx] = (1.0 - alpha**(val + 1)) / (1.0 - alpha) if not np.isclose(alpha, 1.0) else (val + 1.0)
        else:
            z[idx] = 0.0
            
    fig, axes = plt.subplots(3, 1, figsize=(10, 7.5), sharex=True)
    
    axes[0].stem(n, x, linefmt='b-', markerfmt='bo', basefmt='k-')
    axes[0].set_title(r'Signal $x[n] = \alpha^n u[n]$ ($\alpha = %.2f$)' % alpha, fontsize=9.5, fontweight='bold', color='darkblue')
    axes[0].set_ylabel('Amplitude', fontsize=8.5)
    axes[0].grid(True, linestyle='--', alpha=0.6)
    axes[0].set_ylim(-0.2, 1.3)
    
    axes[1].stem(n, y, linefmt='g-', markerfmt='go', basefmt='k-')
    axes[1].set_title(r'Signal $y[n] = u[n]$', fontsize=9.5, fontweight='bold', color='darkgreen')
    axes[1].set_ylabel('Amplitude', fontsize=8.5)
    axes[1].grid(True, linestyle='--', alpha=0.6)
    axes[1].set_ylim(-0.2, 1.3)
    
    axes[2].stem(n, z, linefmt='r-', markerfmt='ro', basefmt='k-')
    axes[2].set_title(r'Convolution Result $z[n] = x[n] * y[n]$', fontsize=9.5, fontweight='bold', color='darkred')
    axes[2].set_xlabel('Index $n$', fontsize=8.5)
    axes[2].set_ylabel('Amplitude', fontsize=8.5)
    axes[2].grid(True, linestyle='--', alpha=0.6)
    
    plt.show()

interactive_p1 = widgets.interactive(
    update_prob1,
    alpha=widgets.FloatSlider(value=0.5, min=0.1, max=0.9, step=0.1, description='Alpha (\u03b1):', style={'description_width': 'initial'}, layout=widgets.Layout(width='550px'))
)
display(interactive_p1)
```
