```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, Markdown

display(Markdown("### 5. Convolution of Finite-Length Signals (Proakis & Manolakis 2.19)"))

n_base5 = np.arange(-5, 12)
n = n_base5
x5 = np.where((n >= 0) & (n <= 4), 1.0, 0.0)

def update_prob5(alpha):
    # y[n] = alpha^n for -3 <= n <= 5
    y5 = np.zeros_like(n, dtype=float)
    for idx, val in enumerate(n):
        if -3 <= val <= 5:
            y5[idx] = alpha ** val
            
    z5 = np.convolve(x5, y5, mode='full')
    # Match support range for n_base5 spanning from min_n to max_n
    n_z = np.arange(n[0] + n[0], n[-1] + n[-1] + 1) # convolution time index support mapping
    # To precisely align with n_base5 (-5 to 11, total 17 points, conv length 17*2-1 = 33):
    n_z = np.arange(n[0] + n[0], n[0] + n[0] + len(z5))
    
    fig, axes = plt.subplots(3, 1, figsize=(10, 7.5), sharex=True)
    
    axes[0].stem(n, x5, linefmt='b-', markerfmt='bo', basefmt='k-')
    axes[0].set_title(r'Finite Signal $x[n]$ (Rectangular window)', fontsize=9.5, fontweight='bold', color='darkblue')
    axes[0].set_ylabel('Amplitude', fontsize=8.5)
    axes[0].grid(True, linestyle='--', alpha=0.6)
    
    axes[1].stem(n, y5, linefmt='g-', markerfmt='go', basefmt='k-')
    axes[1].set_title(r'Finite Signal $y[n] = \alpha^n$ for $-3 \leq n \leq 5$ ($\alpha = %.2f$)' % alpha, fontsize=9.5, fontweight='bold', color='darkgreen')
    axes[1].set_ylabel('Amplitude', fontsize=8.5)
    axes[1].grid(True, linestyle='--', alpha=0.6)
    
    axes[2].stem(n_z, z5, linefmt='r-', markerfmt='ro', basefmt='k-')
    axes[2].set_title(r'Convolution Result $z[n] = x[n] * y[n]$', fontsize=9.5, fontweight='bold', color='darkred')
    axes[2].set_xlabel('Index $n$', fontsize=8.5)
    axes[2].set_ylabel('Amplitude', fontsize=8.5)
    axes[2].grid(True, linestyle='--', alpha=0.6)
    
    plt.xlim(-6, 12)
    plt.show()

interactive_p5 = widgets.interactive(
    update_prob5,
    alpha=widgets.FloatSlider(value=1.2, min=0.5, max=2.0, step=0.1, description='Alpha (\u03b1):', style={'description_width': 'initial'}, layout=widgets.Layout(width='550px'))
)
display(interactive_p5)
```
