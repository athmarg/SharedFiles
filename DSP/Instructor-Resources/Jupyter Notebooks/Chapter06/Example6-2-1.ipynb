```python
import numpy as np
import matplotlib.pyplot as plt

# 1. Definition of continuous signals and parameters
# F1 = 10 Hz -> x1(t) = cos(20 * pi * t)
# F2 = 50 Hz -> x2(t) = cos(100 * pi * t)
# Fs = 40 Hz -> Ts = 1 / 40 = 0.025 s

F_s = 40.0
Ts = 1.0 / F_s

# Time duration for visualization (e.g., 0.2 seconds to show a few periods)
t_continuous = np.linspace(0, 0.2, 1000)
x1_continuous = np.cos(20 * np.pi * t_continuous)
x2_continuous = np.cos(100 * np.pi * t_continuous)

# Discrete time indices n and sampling instants
n_indices = np.arange(0, int(0.2 * F_s) + 1)
t_samples = n_indices * Ts

# Discrete signals (both yield identical numerical values due to aliasing)
x1_samples = np.cos((np.pi / 2) * n_indices)
x2_samples = np.cos((5 * np.pi / 2) * n_indices)

# Print the mathematical expressions of the two discrete signals
print("--- Mathematical Expressions of Discrete-Time Signals ---")
print("x1[n] = cos((pi / 2) * n)")
print("x2[n] = cos((5 * np.pi / 2) * n) = cos((pi / 2) * n)")
print("Conclusion: x1[n] == x2[n] due to aliasing! Both continuous curves pass through the exact same sample points.")

# 2. Plotting both continuous signals overlaid with each discrete signal in separate subplots
fig, axes = plt.subplots(2, 1, figsize=(10, 8), sharex=True)

# Subplot 1: Both continuous signals overlaid with samples of x1[n]
axes[0].plot(t_continuous * 1000, x1_continuous, label=r'$x_1(t) = \cos(20\pi t)$ ($F_1 = 10$ Hz)', color='blue', linewidth=1.5)
axes[0].plot(t_continuous * 1000, x2_continuous, label=r'$x_2(t) = \cos(100\pi t)$ ($F_2 = 50$ Hz)', color='orange', linestyle='--', linewidth=1.5)
axes[0].stem(t_samples * 1000, x1_samples, linefmt='g-', markerfmt='go', basefmt='g--', 
             label=r'Samples $x_1[n]$')
axes[0].set_title(r'Aliasing Demonstration: Superposition of Continuous Signals with $x_1[n]$ ($F_s = 40$ Hz)')
axes[0].set_ylabel('Amplitude (V)')
axes[0].axhline(0, color='black', linewidth=0.8, linestyle='--')
axes[0].grid(True, linestyle=':', alpha=0.7)
axes[0].legend(loc='upper center', bbox_to_anchor=(0.5, 1.25), ncol=3)

# Subplot 2: Both continuous signals overlaid with samples of x2[n] (identical to x1[n])
axes[1].plot(t_continuous * 1000, x1_continuous, label=r'$x_1(t) = \cos(20\pi t)$ ($F_1 = 10$ Hz)', color='blue', linewidth=1.5)
axes[1].plot(t_continuous * 1000, x2_continuous, label=r'$x_2(t) = \cos(100\pi t)$ ($F_2 = 50$ Hz)', color='orange', linestyle='--', linewidth=1.5)
axes[1].stem(t_samples * 1000, x2_samples, linefmt='r-', markerfmt='ro', basefmt='r--', 
             label=r'Samples $x_2[n]$ (Identical)')
axes[1].set_title(r'Aliasing Demonstration: Superposition of Continuous Signals with $x_2[n]$ ($F_s = 40$ Hz)')
axes[1].set_xlabel('Time (ms)')
axes[1].set_ylabel('Amplitude (V)')
axes[1].axhline(0, color='black', linewidth=0.8, linestyle='--')
axes[1].grid(True, linestyle=':', alpha=0.7)
axes[1].legend(loc='upper center', bbox_to_anchor=(0.5, 1.25), ncol=3)

plt.tight_layout()
plt.show()
```
