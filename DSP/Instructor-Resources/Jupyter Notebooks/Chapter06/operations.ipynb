```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, Markdown

# --- Fix output height for Binder ---
display(Markdown("<style>.output_scroll { max-height: none !important; }</style>"))

# --- Educational Note ---
display(Markdown(r"""
> **Educational Note:**
> This notebook interactively demonstrates **fundamental operations on discrete-time signals**:
> * **Amplitude Scaling:** Scaling a signal $x[n]$ by a constant $A$ yielding $Ax[n]$.
> * **Addition & Subtraction:** Combining two time-aligned signals sample-by-sample, so that $z[n] = Ax[n] \pm y[n]$.
> * *Note:* All signals are vertically aligned over the same time axis to clearly show how corresponding samples match up across operations.
"""))

# Common time index domain spanning both signals
n_base = np.arange(-3, 5)

# Define two signals with different supports (time-aligned over n_base)
def get_signals(n_arr):
    x_vals = np.zeros_like(n_arr, dtype=float)
    y_vals = np.zeros_like(n_arr, dtype=float)
    
    x_dict = {-2: 5.0, -1: 2.0, 0: -1.0, 1: 4.0, 2: 6.0}
    y_dict = {-1: 2.0, 0: 1.0, 1: 3.0, 2: 5.0}
    
    for idx, val in enumerate(n_arr):
        if val in x_dict:
            x_vals[idx] = x_dict[val]
        if val in y_dict:
            y_vals[idx] = y_dict[val]
            
    return x_vals, y_vals

# Core update function
def update_operations(A):
    n = n_base
    x_orig, y = get_signals(n)
    
    # Amplitude scaling of the first signal
    x_scaled = A * x_orig
    
    # Addition and subtraction using the scaled first signal and y[n]
    z_add = x_scaled + y
    z_sub = x_scaled - y
    
    # 4 vertically stacked subplots sharing the exact same time axis (sharex=True)
    fig, axes = plt.subplots(4, 1, figsize=(10, 8.5), sharex=True)
    
    # 1. Scaled First Signal (Ax[n])
    axes[0].stem(n, x_scaled, linefmt='b-', markerfmt='bo', basefmt='k-')
    axes[0].set_title(r'Scaled First Signal: $A \cdot x[n]$ (A = %.1f)' % A, fontsize=9.5, fontweight='bold', color='darkblue')
    axes[0].set_ylabel('Amplitude', fontsize=8.5)
    axes[0].grid(True, linestyle='--', alpha=0.6)
    axes[0].set_ylim(-12, 16)
    
    # 2. Second Signal y[n]
    axes[1].stem(n, y, linefmt='g-', markerfmt='go', basefmt='k-')
    axes[1].set_title('Second Signal: $y[n]$', fontsize=9.5, fontweight='bold', color='darkgreen')
    axes[1].set_ylabel('Amplitude', fontsize=8.5)
    axes[1].grid(True, linestyle='--', alpha=0.6)
    axes[1].set_ylim(-12, 16)
    
    # 3. Signal Addition (z[n] = Ax[n] + y[n])
    axes[2].stem(n, z_add, linefmt='r-', markerfmt='ro', basefmt='k-')
    axes[2].set_title(r'Signal Addition: $z[n] = A \cdot x[n] + y[n]$', fontsize=9.5, fontweight='bold', color='darkred')
    axes[2].set_ylabel('Amplitude', fontsize=8.5)
    axes[2].grid(True, linestyle='--', alpha=0.6)
    axes[2].set_ylim(-12, 18)
    
    # 4. Signal Subtraction (z[n] = Ax[n] - y[n])
    axes[3].stem(n, z_sub, linefmt='m-', markerfmt='mo', basefmt='k-')
    axes[3].set_title(r'Signal Subtraction: $z[n] = A \cdot x[n] - y[n]$', fontsize=9.5, fontweight='bold', color='purple')
    axes[3].set_xlabel('Index $n$', fontsize=8.5)
    axes[3].set_ylabel('Amplitude', fontsize=8.5)
    axes[3].grid(True, linestyle='--', alpha=0.6)
    axes[3].set_ylim(-16, 16)
    
    plt.tight_layout()
    plt.show()

# Interactive binding container
interactive_plot = widgets.interactive(
    update_operations, 
    A=widgets.FloatSlider(
        value=1.0, min=-2.0, max=3.0, step=0.5,
        description='Scaling Factor (A):',
        style={'description_width': 'initial'},
        layout=widgets.Layout(width='550px')
    )
)

display(interactive_plot)
```
