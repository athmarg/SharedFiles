```python
import numpy as np
import matplotlib.pyplot as plt
from IPython.display import display, Markdown

display(Markdown("### 3. Convolution of $x[n] = (0.5)^n u[n]$ and $y[n] = n u[n]$"))

n_base3 = np.arange(-2, 10)
n = n_base3
u = (n >= 0).astype(float)
x3 = ((0.5) ** n) * u
y3 = n * u

z3 = np.zeros_like(n, dtype=float)
for idx, val in enumerate(n):
    if val >= 0:
        z3[idx] = 2.0 * (val + (0.5 ** val) - 1.0)
    else:
        z3[idx] = 0.0

fig, axes = plt.subplots(3, 1, figsize=(10, 7.5), sharex=True)
axes[0].stem(n, x3, linefmt='b-', markerfmt='bo', basefmt='k-')
axes[0].set_title(r'Signal $x[n] = (0.5)^n u[n]$', fontsize=9.5, fontweight='bold', color='darkblue')
axes[0].set_ylabel('Amplitude', fontsize=8.5)
axes[0].grid(True, linestyle='--', alpha=0.6)

axes[1].stem(n, y3, linefmt='g-', markerfmt='go', basefmt='k-')
axes[1].set_title(r'Signal $y[n] = n u[n]$', fontsize=9.5, fontweight='bold', color='darkgreen')
axes[1].set_ylabel('Amplitude', fontsize=8.5)
axes[1].grid(True, linestyle='--', alpha=0.6)

axes[2].stem(n, z3, linefmt='r-', markerfmt='ro', basefmt='k-')
axes[2].set_title(r'Convolution Result $z[n] = x[n] * y[n]$', fontsize=9.5, fontweight='bold', color='darkred')
axes[2].set_xlabel('Index $n$', fontsize=8.5)
axes[2].set_ylabel('Amplitude', fontsize=8.5)
axes[2].grid(True, linestyle='--', alpha=0.6)
plt.show()
```
