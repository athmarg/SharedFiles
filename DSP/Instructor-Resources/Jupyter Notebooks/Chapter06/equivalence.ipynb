```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, Markdown

# --- Educational Note on Frequency Equivalence and Aliasing ---
display(Markdown("""
> **Educational Note:**
> This notebook demonstrates the property of **frequency equivalence (aliasing)** in discrete-time signals. 
> Two sinusoidal signals whose frequencies differ by an integer multiple of 2π (i.e., ω2 = ω1 + 2πm) 
> produce **identical discrete samples** when evaluated at integer indices $n$. 
> Use the slider below to observe how adding 2π to the base frequency ω0 leaves the discrete samples completely unchanged.
"""))

@widgets.interact(
    omega_base=widgets.FloatSlider(value=np.pi/4, min=0.0, max=np.pi, step=np.pi/16, description='Base Freq ω0:', style={'description_width': 'initial'}, layout=widgets.Layout(width='700px')),
    m=widgets.IntSlider(value=1, min=0, max=4, step=1, description='Multiple m (2π m):', style={'description_width': 'initial'}, layout=widgets.Layout(width='700px')),
    num_samples=widgets.IntSlider(value=24, min=10, max=40, step=2, description='Samples N:', style={'description_width': 'initial'}, layout=widgets.Layout(width='700px'))
)
def plot_frequency_equivalence(omega_base, m, num_samples):
    n = np.arange(0, num_samples)
    
    omega_1 = omega_base
    omega_2 = omega_base + 2 * np.pi * m
    
    x1 = np.cos(omega_1 * n)
    x2 = np.cos(omega_2 * n)
    
    # Continuous high-resolution time for underlying envelope comparison
    t_cont = np.linspace(0, num_samples - 1, 500)
    c1 = np.cos(omega_1 * t_cont)
    c2 = np.cos(omega_2 * t_cont)
    
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(11, 7), sharex=True)
    
    # --- Signal 1 Plot ---
    ax1.plot(t_cont, c1, color='blue', alpha=0.3, linewidth=1.5, label=rf'Continuous $\cos({omega_1:.3f}t)$')
    markerline1, stemlines1, baseline1 = ax1.stem(n, x1, linefmt='b-', markerfmt='bo', basefmt='k-')
    plt.setp(stemlines1, linewidth=1.2)
    ax1.set_title(rf'Signal 1: $\omega_1 = {omega_1:.3f}$ rad ($x_1[n] = \cos(\omega_1 n)$)', fontsize=11, fontweight='bold', color='darkblue')
    ax1.set_ylabel('Amplitude', fontsize=10)
    ax1.legend(loc='upper right', fontsize=9)
    ax1.grid(True, linestyle='--', alpha=0.6)
    ax1.set_ylim(-1.3, 1.3)
    
    # --- Signal 2 Plot (with added 2\pi m) ---
    ax2.plot(t_cont, c2, color='red', alpha=0.3, linewidth=1.5, label=rf'Continuous $\cos({omega_2:.3f}t)$ [2\pi shifted]')
    markerline2, stemlines2, baseline2 = ax2.stem(n, x2, linefmt='r-', markerfmt='ro', basefmt='k-')
    plt.setp(stemlines2, linewidth=1.2)
    ax2.set_title(rf'Signal 2 (Alias): $\omega_2 = \omega_1 + {m}\cdot(2\pi) = {omega_2:.3f}$ rad — **Identical Samples!**', fontsize=11, fontweight='bold', color='darkred')
    ax2.set_xlabel('Discrete Time Index $n$', fontsize=11)
    ax2.set_ylabel('Amplitude', fontsize=10)
    ax2.legend(loc='upper right', fontsize=9)
    ax2.grid(True, linestyle='--', alpha=0.6)
    ax2.set_ylim(-1.3, 1.3)
    
    plt.tight_layout()
    plt.show()
```
