```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets

# Frequencies and time axis (defined once globally)
F1 = 1000.0
F2 = 3000.0
F3 = 6000.0
t = np.linspace(0, 0.004, 1000)

@widgets.interact(Fs=widgets.FloatSlider(value=5000.0, min=500.0, max=15000.0, step=100.0, description='Sampling $F_s$ (Hz):', style={'description_width': 'initial'}, layout=widgets.Layout(width='650px')))
def update_plot(Fs):
    print(f"New Sampling Frequency Value: {Fs} Hz")
    
    fig, axes = plt.subplots(4, 3, figsize=(14, 9))
    plt.subplots_adjust(bottom=0.12, hspace=0.30)
      
    Ts = 1.0 / Fs
    n_indices = np.arange(0, int(0.004 * Fs) + 1)
    t_samples = n_indices * Ts
    
    # --- Column 1: Original Analog Components ---
    x1 = 3 * np.cos(2 * np.pi * F1 * t)
    x2 = 5 * np.sin(2 * np.pi * F2 * t)
    x3 = 10 * np.cos(2 * np.pi * F3 * t)
    x_total = x1 + x2 + x3
    
    axes[0, 0].plot(t * 1000, x1, color='blue', linewidth=1.5)
    axes[0, 0].set_title(f'Original $x_1(t)$ [{int(F1)} Hz]', fontsize=10)
    axes[0, 0].grid(True, linestyle=':', alpha=0.7)
    
    axes[1, 0].plot(t * 1000, x2, color='blue', linewidth=1.5)
    axes[1, 0].set_title(f'Original $x_2(t)$ [{int(F2)} Hz]', fontsize=10)
    axes[1, 0].grid(True, linestyle=':', alpha=0.7)
    
    axes[2, 0].plot(t * 1000, x3, color='blue', linewidth=1.5)
    axes[2, 0].set_title(f'Original $x_3(t)$ [{int(F3)} Hz]', fontsize=10)
    axes[2, 0].grid(True, linestyle=':', alpha=0.7)
    
    axes[3, 0].plot(t * 1000, x_total, color='blue', linewidth=1.5)
    axes[3, 0].set_title('Total Original $x(t)$', fontsize=10)
    axes[3, 0].set_xlabel('Time (ms)', fontsize=9)
    axes[3, 0].grid(True, linestyle=':', alpha=0.7)
    
    # --- Column 2: Discrete Signals ---
    x1_samples = 3 * np.cos(2 * np.pi * F1 * t_samples)
    x2_samples = 5 * np.sin(2 * np.pi * F2 * t_samples)
    x3_samples = 10 * np.cos(2 * np.pi * F3 * t_samples)
    x_total_samples = x1_samples + x2_samples + x3_samples
    
    axes[0, 1].stem(t_samples * 1000, x1_samples, linefmt='b-', markerfmt='bo', basefmt='b--')
    axes[0, 1].set_title('Discrete $x_1[n]$', fontsize=10)
    axes[0, 1].grid(True, linestyle=':', alpha=0.7)
    
    axes[1, 1].stem(t_samples * 1000, x2_samples, linefmt='g-', markerfmt='go', basefmt='g--')
    axes[1, 1].set_title('Discrete $x_2[n]$', fontsize=10)
    axes[1, 1].grid(True, linestyle=':', alpha=0.7)
    
    axes[2, 1].stem(t_samples * 1000, x3_samples, linefmt='m-', markerfmt='mo', basefmt='m--')
    axes[2, 1].set_title('Discrete $x_3[n]$', fontsize=10)
    axes[2, 1].grid(True, linestyle=':', alpha=0.7)
    
    axes[3, 1].stem(t_samples * 1000, x_total_samples, linefmt='k-', markerfmt='ko', basefmt='k--')
    axes[3, 1].set_title('Total Discrete $x[n]$', fontsize=10)
    axes[3, 1].set_xlabel('Time (ms)', fontsize=9)
    axes[3, 1].grid(True, linestyle=':', alpha=0.7)
    
    # --- Column 3: Reconstructed Components ---
    nyquist_limit = Fs / 2.0
    
    f1_rec = F1 if F1 < nyquist_limit else abs(F1 - round(F1 / Fs) * Fs)
    col1 = 'red' if F1 >= nyquist_limit else 'blue'
    rec_x1 = 3 * np.cos(2 * np.pi * f1_rec * t)
    axes[0, 2].plot(t * 1000, rec_x1, color=col1, linewidth=1.5)
    axes[0, 2].set_title(f'Reconstructed (f={int(f1_rec)} Hz)', fontsize=10)
    axes[0, 2].grid(True, linestyle=':', alpha=0.7)
    
    f2_rec = F2 if F2 < nyquist_limit else abs(F2 - round(F2 / Fs) * Fs)
    col2 = 'red' if F2 >= nyquist_limit else 'blue'
    rec_x2 = 5 * np.sin(2 * np.pi * f2_rec * t) if col1 == 'blue' else -5 * np.sin(2 * np.pi * f2_rec * t)
    axes[1, 2].plot(t * 1000, rec_x2, color=col2, linewidth=1.5)
    axes[1, 2].set_title(f'Reconstructed (f={int(f2_rec)} Hz)', fontsize=10)
    axes[1, 2].grid(True, linestyle=':', alpha=0.7)
    
    f3_rec = F3 if F3 < nyquist_limit else abs(F3 - round(F3 / Fs) * Fs)
    col3 = 'red' if F3 >= nyquist_limit else 'blue'
    rec_x3 = 10 * np.cos(2 * np.pi * f3_rec * t)
    axes[2, 2].plot(t * 1000, rec_x3, color=col3, linewidth=1.5)
    axes[2, 2].set_title(f'Reconstructed (f={int(f3_rec)} Hz)', fontsize=10)
    axes[2, 2].grid(True, linestyle=':', alpha=0.7)
    
    any_aliasing = (F1 >= nyquist_limit) or (F2 >= nyquist_limit) or (F3 >= nyquist_limit)
    col_total = 'red' if any_aliasing else 'blue'
    rec_total = rec_x1 + rec_x2 + rec_x3
    axes[3, 2].plot(t * 1000, rec_total, color=col_total, linewidth=1.5)
    axes[3, 2].set_title('Total Reconstructed Signal', fontsize=10)
    axes[3, 2].set_xlabel('Time (ms)', fontsize=9)
    axes[3, 2].grid(True, linestyle=':', alpha=0.7)
    
    for ax_row in axes[:-1]:
        for ax in ax_row:
            ax.tick_params(labelbottom=False)
            
    plt.show()
```
