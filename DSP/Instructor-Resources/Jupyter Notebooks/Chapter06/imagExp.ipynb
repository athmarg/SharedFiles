```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets

@widgets.interact(
    r=widgets.FloatSlider(value=0.99, min=0.85, max=1.0, step=0.01, description='Radius r:', style={'description_width': 'initial'}, layout=widgets.Layout(width='700px')),
    omega_idx=widgets.FloatSlider(value=1.0, min=0.5, max=4.0, step=0.1, description='Freq Factor (π/64):', style={'description_width': 'initial'}, layout=widgets.Layout(width='700px')),
    num_samples=widgets.IntSlider(value=256, min=64, max=512, step=32, description='Samples N:', style={'description_width': 'initial'}, layout=widgets.Layout(width='700px'))
)
def plot_book_style_exponential(r, omega_idx, num_samples):
    n = np.arange(0, num_samples)
    omega = omega_idx * np.pi / 64
    
    envelope = r**n
    x_real = envelope * np.cos(omega * n)
    x_imag = envelope * np.sin(omega * n)
    
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(12, 7), sharex=True)
    
    # Custom styling to match the textbook aesthetic (red/white gradient feel with clean grid)
    for ax, data, title_str in zip([ax1, ax2], [x_real, x_imag], 
                                   [rf'$\Re\{{x[n]\}} = ({r:.2f})^n \cos\left(\dfrac{{\pi n}}{{64}}\right)$  $(n=0,1,2,\dots,{num_samples-1})$',
                                    rf'$\Im\{{x[n]\}} = ({r:.2f})^n \sin\left(\dfrac{{\pi n}}{{64}}\right)$  $(n=0,1,2,\dots,{num_samples-1})$']):
        
        # Draw stems where the tip defines the envelope directly
        markerline, stemlines, baseline = ax.stem(n, data, linefmt='r-', markerfmt='ro', basefmt='k-')
        plt.setp(stemlines, linewidth=1.2, alpha=0.8)
        plt.setp(markerline, markersize=3)
        plt.setp(baseline, linewidth=1.5)
        
        ax.set_title(title_str, fontsize=12, fontweight='bold', pad=10)
        ax.set_ylabel('Amplitude', fontsize=10)
        ax.grid(True, linestyle='--', color='gray', alpha=0.6)
        
        # Dynamic limits based on max envelope bounds
        max_val = np.max(envelope) * 1.1
        ax.set_ylim(-max_val, max_val)
        ax.set_xlim(-5, num_samples + 5)
        
    ax2.set_xlabel(r'Discrete Time Index $n$', fontsize=11)
    plt.tight_layout()
    plt.show()
```
