```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, Markdown

# --- Educational Note on Frequency Index and Period ---
display(Markdown("""
> **Note:**
> - **Frequency Index ($k$):** Expresses how many full periods (cycles) the discrete sine or cosine wave completes within the total visualization window.
> - **Period ($N$):** Determines the number of samples that make up a single full repetition period. The plot below constantly displays **two full periods** ($2N$ samples), allowing for direct observation of the signal's behavior as frequency changes.
"""))

@widgets.interact(
    freq_idx=widgets.IntSlider(value=1, min=0, max=15, step=1, description='Freq Index (k):', style={'description_width': 'initial'}, layout=widgets.Layout(width='700px')),
    N=widgets.IntSlider(value=16, min=4, max=64, step=2, description='Period N:', style={'description_width': 'initial'}, layout=widgets.Layout(width='700px'))
)
def plot_discrete_trig_two_periods(freq_idx, N):
    # Ensure freq_idx doesn't exceed N-1 for clean representation
    if freq_idx >= N:
        freq_idx = N - 1
        
    omega = 2 * np.pi * freq_idx / N
    
    # Discrete indices for 2 full periods (2 * N samples)
    n_discrete = np.arange(0, 2 * N)
    x_cos_d = np.cos(omega * n_discrete)
    x_sin_d = np.sin(omega * n_discrete)
    
    # Continuous time axis for smooth envelope curves
    t_cont = np.linspace(0, 2 * N - 1, 500)
    x_cos_c = np.cos(omega * t_cont)
    x_sin_c = np.sin(omega * t_cont)
    
    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(11, 6), sharex=True)
    
    # --- Cosine Plot ---
    ax1.plot(t_cont, x_cos_c, color='blue', alpha=0.3, linewidth=1.5, label=r'Continuous Envelope $\cos(\omega t)$')
    ax1.stem(n_discrete, x_cos_d, linefmt='b-', markerfmt='bo', basefmt='r-')
    ax1.set_title(rf'Discrete Cosine ($2$ Periods): $x[n] = \cos\left(\frac{{2\pi \cdot {freq_idx} \cdot n}}{{{N}}}\right)$', fontsize=11, fontweight='bold')
    ax1.set_ylabel('Amplitude')
    ax1.legend(loc='upper right', fontsize=9)
    ax1.grid(True, linestyle=':', alpha=0.7)
    ax1.set_ylim(-1.3, 1.3)
    
    # --- Sine Plot ---
    ax2.plot(t_cont, x_sin_c, color='green', alpha=0.3, linewidth=1.5, label=r'Continuous Envelope $\sin(\omega t)$')
    ax2.stem(n_discrete, x_sin_d, linefmt='g-', markerfmt='go', basefmt='r-')
    ax2.set_title(rf'Discrete Sine ($2$ Periods): $x[n] = \sin\left(\frac{{2\pi \cdot {freq_idx} \cdot n}}{{{N}}}\right)$', fontsize=11, fontweight='bold')
    ax2.set_xlabel('Discrete Time Index $n$')
    ax2.set_ylabel('Amplitude')
    ax2.legend(loc='upper right', fontsize=9)
    ax2.grid(True, linestyle=':', alpha=0.7)
    ax2.set_ylim(-1.3, 1.3)
    
    plt.tight_layout()
    plt.show()
```
