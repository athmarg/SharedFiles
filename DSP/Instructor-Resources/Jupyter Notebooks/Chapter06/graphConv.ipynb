```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, Markdown

# --- Fix output height for Jupyter notebooks ---
display(Markdown("<style>.output_scroll { max-height: none !important; }</style>"))

display(Markdown("### Graphical Calculation of Convolution Step-by-Step"))
display(Markdown(r"Here, the graphical/diagrammatic convolution method is implemented for the text signals: $x[n] = \{\mathbf{1}, 2, 3, 1\}$ and $y[n] = \{1, \mathbf{2}, 1, -1\}$."))

# Define signals and their relative time axes based on the text 
# x[n] has origin (n=0) at index 0 -> support: n = 0, 1, 2, 3
n_x = np.array([0, 1, 2, 3])
x_vals = np.array([1.0, 2.0, 3.0, 1.0])

# y[n] has origin (n=0) at index 1 -> values: {1, 2, 1, -1} corresponding to n = -1, 0, 1, 2
n_y_orig = np.array([-1, 0, 1, 2])
y_vals = np.array([1.0, 2.0, 1.0, -1.0])

def update_graphical_conv(n_val):
    # Common global display axis for k to show alignment clearly
    k_min, k_max = -3, 6
    k = np.arange(k_min, k_max + 1)
    
    # Map x[k] onto the common k axis
    x_full = np.zeros_like(k, dtype=float)
    for idx, val in enumerate(n_x):
        mask = (k == val)
        x_full[mask] = x_vals[idx]
        
    # Step 1 & 2: Inversion y[-k] and shift by n -> y[n - k]
    # Mathematically, if y has samples at n_y_orig, then y[n - k] means:
    # n - k = t_y => k = n - t_y
    y_shifted_full = np.zeros_like(k, dtype=float)
    for idx, t_y in enumerate(n_y_orig):
        k_pos = n_val - t_y
        mask = (k == k_pos)
        if np.any(mask):
            y_shifted_full[mask] = y_vals[idx]
            
    # Step 3: Sample-by-sample multiplication v_n[k] = x[k] * y[n - k]
    v_n = x_full * y_shifted_full
    
    # Step 4: Summation to find z[n]
    z_current = np.sum(v_n)
    
    # Plotting
    fig, axes = plt.subplots(3, 1, figsize=(10, 7.5), sharex=True)
    
    # Plot x[k]
    axes[0].stem(k, x_full, linefmt='b-', markerfmt='bo', basefmt='k-')
    axes[0].set_title(r'Original Signal $x[k]$', fontsize=9.5, fontweight='bold', color='darkblue')
    axes[0].set_ylabel('Amplitude', fontsize=8.5)
    axes[0].grid(True, linestyle='--', alpha=0.6)
    axes[0].set_ylim(-1.5, 3.5)
    
    # Plot shifted & reversed signal y[n - k]
    axes[1].stem(k, y_shifted_full, linefmt='g-', markerfmt='go', basefmt='k-')
    axes[1].set_title(r'Reversed & Shifted Signal $y[n-k]$ for $n = %d$' % n_val, fontsize=9.5, fontweight='bold', color='darkgreen')
    axes[1].set_ylabel('Amplitude', fontsize=8.5)
    axes[1].grid(True, linestyle='--', alpha=0.6)
    axes[1].set_ylim(-1.5, 2.5)
    
    # Plot product v_n[k]
    axes[2].stem(k, v_n, linefmt='r-', markerfmt='ro', basefmt='k-')
    axes[2].set_title(r'Product Sequence $\nu_{%d}[k] = x[k]y[%d-k]$  $\rightarrow$  Sum ($z[%d]$) = %.1f' % (n_val, n_val, n_val, z_current), fontsize=9.5, fontweight='bold', color='darkred')
    axes[2].set_xlabel('Index $k$', fontsize=8.5)
    axes[2].set_ylabel('Amplitude', fontsize=8.5)
    axes[2].grid(True, linestyle='--', alpha=0.6)
    axes[2].set_ylim(-2.5, 7.0)
    
    plt.tight_layout()
    plt.show()

interactive_conv = widgets.interactive(
    update_graphical_conv,
    n_val=widgets.IntSlider(value=0, min=-2, max=7, step=1, description='Time Shift (n):', style={'description_width': 'initial'}, layout=widgets.Layout(width='600px'))
)
display(interactive_conv)
```
