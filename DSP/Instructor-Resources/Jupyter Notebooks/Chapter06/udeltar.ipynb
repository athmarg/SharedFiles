```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, Markdown

# --- Educational Note on Elementary Discrete-Time Signals ---
display(Markdown(r"""
> **Educational Note:**
> This notebook displays three fundamental elementary discrete-time signals defined in the theory:
> 1. **Unit Step Function ($u[n]$):** Equals $1$ for $n \geq 0$ and $0$ for $n < 0$.
> 2. **Unit Impulse Function ($\delta[n]$):** Equals $1$ only at $n = 0$ and $0$ otherwise.
> 3. **Unit Ramp Function ($r[n]$):** Equals $n$ for $n \geq 0$ and $0$ for $n < 0$ (such that $r[n] = n \cdot u[n]$).
> 
> Use the sliders below to dynamically adjust the amplitude scaling factors for the step and ramp functions.
"""))

@widgets.interact(
    step_amp=widgets.FloatSlider(value=1.0, min=0.5, max=3.0, step=0.1, description='Step Amplitude:', style={'description_width': 'initial'}, layout=widgets.Layout(width='700px')),
    ramp_slope=widgets.FloatSlider(value=1.0, min=0.5, max=3.0, step=0.1, description='Ramp Slope:', style={'description_width': 'initial'}, layout=widgets.Layout(width='700px')),
    n_range=widgets.IntSlider(value=10, min=5, max=25, step=1, description=r'Range ($\pm n$):', style={'description_width': 'initial'}, layout=widgets.Layout(width='700px'))
)
def plot_elementary_signals(step_amp, ramp_slope, n_range):
    n = np.arange(-n_range, n_range + 1)
    
    # Signal definitions
    u_n = step_amp * (n >= 0).astype(float)
    delta_n = (n == 0).astype(float)
    r_n = ramp_slope * n * (n >= 0).astype(float)
    
    fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(16, 4.5), sharey=False)
    
    # --- Unit Step Function Plot ---
    markerline1, stemlines1, baseline1 = ax1.stem(n, u_n, linefmt='b-', markerfmt='bo', basefmt='k-')
    plt.setp(stemlines1, linewidth=1.5)
    ax1.set_title(rf'Unit Step: $u[n]$ (Amp: {step_amp:.1f})', fontsize=11, fontweight='bold', color='darkblue')
    ax1.set_xlabel('Index $n$', fontsize=10)
    ax1.set_ylabel('Amplitude', fontsize=10)
    ax1.grid(True, linestyle='--', alpha=0.6)
    ax1.set_ylim(-0.2, max(3.3, step_amp * 1.2))
    ax1.set_xlim(-n_range - 1, n_range + 1)
    
    # --- Unit Impulse Function Plot ---
    markerline2, stemlines2, baseline2 = ax2.stem(n, delta_n, linefmt='g-', markerfmt='go', basefmt='k-')
    plt.setp(stemlines2, linewidth=1.5)
    ax2.set_title(r'Unit Impulse: $\delta[n]$', fontsize=11, fontweight='bold', color='darkgreen')
    ax2.set_xlabel('Index $n$', fontsize=10)
    ax2.set_ylabel('Amplitude', fontsize=10)
    ax2.grid(True, linestyle='--', alpha=0.6)
    ax2.set_ylim(-0.2, 1.3)
    ax2.set_xlim(-n_range - 1, n_range + 1)
    
    # --- Unit Ramp Function Plot ---
    markerline3, stemlines3, baseline3 = ax3.stem(n, r_n, linefmt='r-', markerfmt='ro', basefmt='k-')
    plt.setp(stemlines3, linewidth=1.5)
    ax3.set_title(rf'Unit Ramp: $r[n]$ (Slope: {ramp_slope:.1f})', fontsize=11, fontweight='bold', color='darkred')
    ax3.set_xlabel('Index $n$', fontsize=10)
    ax3.set_ylabel('Amplitude', fontsize=10)
    ax3.grid(True, linestyle='--', alpha=0.6)
    ax3.set_ylim(-1.0, max(10.0, ramp_slope * n_range * 1.2))
    ax3.set_xlim(-n_range - 1, n_range + 1)
    
    plt.tight_layout()
    plt.show()
```
