```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, Markdown

# --- Fix output height for Binder ---
display(Markdown("<style>.output_scroll { max-height: none !important; }</style>"))

# --- Educational Note ---
display(Markdown(r"""
> **Educational Note:**
> This notebook interactively demonstrates the **convolution of two exponential signals**:
> * $x[n] = \alpha^n u[n]$ and $y[n] = \beta^n u[n]$
> * **Convolution Result ($z[n] = x[n] * y[n]$):** 
>   * Case $\alpha = \beta$: $z[n] = \beta^n(n+1)u[n]$
>   * Case $\alpha \neq \beta$: $z[n] = \frac{\beta^{n+1} - \alpha^{n+1}}{\beta - \alpha}u[n]$
> * Use the sliders to change parameters $\alpha$ and $\beta$ in real-time.
"""))

# Time domain index range for visualization
n_base = np.arange(0, 15)

# Analytical convolution computation function
def compute_convolution(n_arr, alpha, beta):
    u = (n_arr >= 0).astype(float)
    x = (alpha ** n_arr) * u
    y = (beta ** n_arr) * u
    
    z = np.zeros_like(n_arr, dtype=float)
    for idx, n in enumerate(n_arr):
        if n < 0:
            z[idx] = 0.0
        elif np.isclose(alpha, beta):
            z[idx] = (beta ** n) * (n + 1)
        else:
            z[idx] = (beta**(n + 1) - alpha**(n + 1)) / (beta - alpha)
            
    return x, y, z

# Core update function linked to widgets
def update_plot(alpha, beta):
    n = n_base
    x, y, z = compute_convolution(n, alpha, beta)
    
    # Vertically stacked subplots sharing the same time axis for clean alignment
    fig, axes = plt.subplots(3, 1, figsize=(10, 7.5), sharex=True)
    
    # 1. First Signal x[n] with adaptive y-limits
    axes[0].stem(n, x, linefmt='b-', markerfmt='bo', basefmt='k-')
    axes[0].set_title(r'First Signal: $x[n] = \alpha^n u[n]$ ($\alpha = %.2f$)' % alpha, fontsize=9.5, fontweight='bold', color='darkblue')
    axes[0].set_ylabel('Amplitude', fontsize=8.5)
    axes[0].grid(True, linestyle='--', alpha=0.6)
    axes[0].set_ylim(-0.1, 1.2)
    
    # 2. Second Signal y[n] with adaptive y-limits
    axes[1].stem(n, y, linefmt='g-', markerfmt='go', basefmt='k-')
    axes[1].set_title(r'Second Signal: $y[n] = \beta^n u[n]$ ($\beta = %.2f$)' % beta, fontsize=9.5, fontweight='bold', color='darkgreen')
    axes[1].set_ylabel('Amplitude', fontsize=8.5)
    axes[1].grid(True, linestyle='--', alpha=0.6)
    axes[1].set_ylim(-0.1, 1.2)
    
    # 3. Convolution Result z[n] = x[n] * y[n]
    axes[2].stem(n, z, linefmt='r-', markerfmt='ro', basefmt='k-')
    mode_str = "(\u03b1 = \u03b2)" if np.isclose(alpha, beta) else "(\u03b1 \u2260 \u03b2)"
    axes[2].set_title(r'Convolution Result: $z[n] = x[n] * y[n]$ ' + mode_str, fontsize=9.5, fontweight='bold', color='darkred')
    axes[2].set_xlabel('Index $n$', fontsize=8.5)
    axes[2].set_ylabel('Amplitude', fontsize=8.5)
    axes[2].grid(True, linestyle='--', alpha=0.6)
    
    plt.show()

# Interactive container with independent sliders for alpha and beta
interactive_plot = widgets.interactive(
    update_plot,
    alpha=widgets.FloatSlider(
        value=0.5, min=0.1, max=0.9, step=0.1,
        description='Alpha (\u03b1):',
        style={'description_width': 'initial'},
        layout=widgets.Layout(width='550px')
    ),
    beta=widgets.FloatSlider(
        value=0.7, min=0.1, max=0.9, step=0.1,
        description='Beta (\u03b2):',
        style={'description_width': 'initial'},
        layout=widgets.Layout(width='550px')
    )
)

display(interactive_plot)
```
