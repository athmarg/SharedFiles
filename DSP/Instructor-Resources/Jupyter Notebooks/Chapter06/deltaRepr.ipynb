```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display, Markdown

# --- Fix output height for Binder ---
display(Markdown("<style>.output_scroll { max-height: none !important; }</style>"))

# --- Educational Note ---
display(Markdown(r"""
> **Educational Note:**
> This notebook interactively demonstrates the representation of a discrete-time signal using the **elementary impulse function** $\delta[n-k]$, based on the sifting property:
> $$x[n] = \sum_{k=-\infty}^{\infty} x[k] \delta[n-k]$$
> * Drag the slider to change the time shift index $k$.
> * The plots update **automatically in real-time** as you move the slider to isolate individual samples via the product $x[k]\delta[n-k]$.
> * **Note:** To reconstruct the entire signal $x[n]$, one simply needs to sum up all these individual single-sample signals across all time shift values $k$.
"""))

n_base = np.arange(-10, 11)

# Define custom signal values for n in [-10, 10]
def get_x(n_arr):
    x_vals = np.zeros_like(n_arr, dtype=float)
    signal_dict = {-2: 2.0, -1: 4.0, 0: 6.0, 1: 5.0, 2: 4.0, 3: 3.0, 4: 2.0, 5: 1.0}
    for idx, val in enumerate(n_arr):
        if val in signal_dict:
            x_vals[idx] = signal_dict[val]
    return x_vals

# Helper function for correct mathematical sign formatting (e.g., n - -4 -> n + 4)
def format_shift_label(k):
    if k > 0:
        return f"n - {k}"
    elif k < 0:
        return f"n + {abs(k)}"
    else:
        return "n"

# Automatic interactive binding using @widgets.interact
@widgets.interact(
    k=widgets.IntSlider(
        value=1, min=-8, max=8, step=1,
        description='Shift (k):',
        style={'description_width': 'initial'},
        layout=widgets.Layout(width='550px')
    )
)
def update_plot(k):
    n = n_base
    x = get_x(n)
    
    # Impulse delta[n - k] is 1 when n == k else 0
    delta = (n == k).astype(float)
    
    # Sifting property term: x[k] * delta[n - k]
    x_k = get_x(np.array([k]))[0]
    sifting_term = x_k * delta
    
    shift_str = format_shift_label(k)
    
    fig, axes = plt.subplots(3, 1, figsize=(10, 8), sharex=True)
    
    # 1. Original Signal x[n]
    axes[0].stem(n, x, linefmt='b-', markerfmt='bo', basefmt='k-')
    axes[0].set_title(r'Original Signal: $x[n]$', fontsize=11, fontweight='bold', color='darkblue')
    axes[0].set_ylabel('Amplitude')
    axes[0].grid(True, linestyle='--', alpha=0.6)
    axes[0].set_ylim(-1, 7)
    
    # 2. Shifted Impulse delta[n - k]
    axes[1].stem(n, delta, linefmt='g-', markerfmt='go', basefmt='k-')
    axes[1].set_title(fr'Time-Shifted Impulse Function: $\delta[{shift_str}]$', fontsize=11, fontweight='bold', color='darkgreen')
    axes[1].set_ylabel('Amplitude')
    axes[1].grid(True, linestyle='--', alpha=0.6)
    axes[1].set_ylim(-0.2, 1.5)
    
    # 3. Product / Sifting Property x[k] * delta[n - k]
    axes[2].stem(n, sifting_term, linefmt='r-', markerfmt='ro', basefmt='k-')
    axes[2].set_title(fr'Isolated Sample via Sifting: $x[{k}] \cdot \delta[{shift_str}]$ (Value = {x_k})', fontsize=11, fontweight='bold', color='darkred')
    axes[2].set_xlabel('Index $n$')
    axes[2].set_ylabel('Amplitude')
    axes[2].grid(True, linestyle='--', alpha=0.6)
    axes[2].set_ylim(-1, 7)
    
    plt.tight_layout()
    plt.show()
```
