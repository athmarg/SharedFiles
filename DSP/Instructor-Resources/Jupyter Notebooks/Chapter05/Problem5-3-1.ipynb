```python
# Notebook: Symbolic System Analysis, Pole-Zero Plot & Zero-Input Response (Example 6)
import warnings
warnings.filterwarnings('ignore', category=UserWarning) 

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt
import control as ct

# 1. Define symbolic variable s
s = sp.Symbol('s')

# 2. Define numerator and denominator based on the solution: (s^2+4s+3)(s^2+2s+5)
num_expr = s**2 - 7*s + 12
den_expr = (s**2 + 4*s + 3) * (s**2 + 2*s + 5) # s^4 + 6s^3 + 14s^2 + 22s + 15

# 3. Symbolic computation of zeros and poles
zeros_sym = sp.solve(num_expr, s)
poles_sym = sp.solve(den_expr, s)

# 4. Polynomial coefficients for the control library
num_coeffs = [1, -7, 12]
den_coeffs = [1, 6, 14, 22, 15]
H = ct.tf(num_coeffs, den_coeffs)

# 5. Compute zero-input response (free response) using zero input U and initial conditions X0
t = np.linspace(0, 5, 500)
u_zero = np.zeros_like(t)
t_out, y_response = ct.forced_response(H, T=t, U=u_zero, X0=[1.0, 1.0, 1.0, 1.0])

# 6. Create subplots (1. Pole-Zero Map, 2. Zero-Input Response y(t))
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4.5))
fig.subplots_adjust(wspace=0.3)

# --- Plot 1: Pole-Zero Map ---
ct.pzmap(H, ax=ax1, title='')
ax1.set_title(r'Pole-Zero Map of $\mathcal{H}(s)$', fontsize=11)
ax1.grid(True, linestyle=':', alpha=0.6)

# --- Plot 2: Zero-Input Response y(t) ---
ax2.plot(t_out, y_response, color='darkblue', lw=2.5, label='Zero-Input Response $y(t)$')
ax2.axhline(0, color='black', linewidth=1)
ax2.axvline(0, color='black', linewidth=1)
ax2.set_xlabel('Time $t$ (s)', fontsize=11)
ax2.set_ylabel('Amplitude $y(t)$', fontsize=11)
ax2.set_title(r'System Zero-Input Response $y(t)$', fontsize=11)
ax2.grid(True, linestyle=':', alpha=0.6)
ax2.legend(loc='best', frameon=True, fontsize=9)

plt.show()

# 7. Print results to output
print("=== Symbolic System Analysis (Example 6) ===")
print(r"• Transfer Function H(s):")
print(f"  Numerator:   {num_expr}")
print(f"  Denominator: {den_expr}")
print("-" * 55)
print("• Computed Zeros (Symbolic):")
for z in zeros_sym:
    print(f"  z = {z}")

print("• Computed Poles (Symbolic):")
for p in poles_sym:
    print(f"  p = {p}")

is_stable = all(np.real(ct.poles(H)) < 0)
print("-" * 55)
print(f"• BIBO Stability Status: {'Stable' if is_stable else 'Unstable'}")
print("• Zero-Input Response Form:")
print("  y(t) = C1*e^{-t} + C2*e^{-3t} + C3*e^{-t}*sin(2t + phi)")
```
