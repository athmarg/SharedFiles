```python
# Notebook: Computation of the Matrix Exponential e^{At} via Laplace Transform
import sympy as sp

# 1. Definition of matrix A based on the exercise
s = sp.symbols('s')
t = sp.symbols('t', real=True, positive=True)
A = sp.Matrix([[5, -6], [4, -5]])
I = sp.eye(2)

print("--- System Matrix A ---")
display(A)

# 2. Computation of the Resolvent Matrix Φ(s) = (sI - A)^-1
# This matrix represents the Laplace transform of the matrix exponential
resolvent = (s * I - A).inv()
print("\n--- Resolvent Matrix Φ(s) = (sI - A)^-1 ---")
display(resolvent)

# 3. Computation of the Matrix Exponential e^{At} via inverse Laplace transform
# SymPy internally performs the partial fraction expansion
exp_At = sp.Matrix([
    [sp.inverse_laplace_transform(resolvent[0,0], s, t), sp.inverse_laplace_transform(resolvent[0,1], s, t)],
    [sp.inverse_laplace_transform(resolvent[1,0], s, t), sp.inverse_laplace_transform(resolvent[1,1], s, t)]
])

print("\n--- Matrix Exponential e^{At} ---")
display(exp_At)
```
