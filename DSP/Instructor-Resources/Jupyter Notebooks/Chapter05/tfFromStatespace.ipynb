```python
# Notebook: Transfer Function Calculation and Degree Analysis from State-Space Equations
import sympy as sp

# 1. Definition of State-Space Matrices based on the example
s = sp.symbols('s')
A = sp.Matrix([[-1, 0, 0], 
               [0, -3, 0], 
               [0, 0, -2]])
B = sp.Matrix([1, 1, 1])
C = sp.Matrix([[1, 2, 1]])
D_zero = sp.Matrix([[0]])
D_nonzero = sp.Matrix([[3]])

I = sp.eye(3)

print("--- System Matrices ---")
print("Matrix A:")
display(A)
print("Matrix B:")
display(B)
print("Matrix C:")
display(C)

# 2. Computation of sI - A and its Determinant (Denominator)
sI_minus_A = s * I - A
print("\n--- Resolvent Matrix sI - A ---")
display(sI_minus_A)

det_sI_A = sI_minus_A.det()
print("\n--- Denominator: det(sI - A) ---")
display(det_sI_A)
print("Degree of Denominator:", sp.degree(det_sI_A, s))

# 3. Computation of the Adjugate Matrix and Numerator Polynomial (for D = 0)
adj_sI_A = sI_minus_A.adjugate()
print("\n--- Adjugate Matrix Adj(sI - A) ---")
display(adj_sI_A)

numerator_expr = C * adj_sI_A * B
numerator_poly = sp.expand(numerator_expr[0, 0])
print("\n--- Numerator Polynomial: C * Adj(sI - A) * B ---")
display(numerator_poly)
print("Degree of Numerator:", sp.degree(numerator_poly, s))

# 4. Transfer Function for D = 0 (Strictly Proper)
H_s_zero = numerator_poly / det_sI_A
print("\n--- Transfer Function H(s) with D = 0 ---")
display(sp.cancel(H_s_zero))

# 5. Transfer Function for D != 0 (Proper)
H_s_nonzero = (numerator_poly / det_sI_A) + D_nonzero[0, 0]
print("\n--- Transfer Function H(s) with D = 3 ---")
display(sp.cancel(H_s_nonzero))
```
