```python
# Notebook: Transfer Function and Pole-Zero Analysis (Example 1)
import numpy as np
import matplotlib.pyplot as plt
import control as ct
import ipywidgets as widgets
from ipywidgets import interact

# Ορισμός των δύο συστημάτων του παραδείγματος
# Σύστημα (a): H1(s) = 1 / (s^2 - s - 2)
num1 = [1]
den1 = [1, -1, -2]
H1 = ct.tf(num1, den1)

# Σύστημα (b): H2(s) = 1 / (s^3 + 6s^2 + 11s + 6)
num2 = [1]
den2 = [1, 6, 11, 6]
H2 = ct.tf(num2, den2)

@interact(
    system_choice=widgets.Dropdown(
        options=[('System (a): s^2 - s - 2', 1), ('System (b): s^3 + 6s^2 + 11s + 6', 2)],
        value=1,
        description='System:'
    )
)
def analyze_system(system_choice):
    H = H1 if system_choice == 1 else H2
    sys_name = "System (a)" if system_choice == 1 else "System (b)"
    
    # Υπολογισμός πόλων και μηδενικών
    p = ct.poles(H)
    z = ct.zeros(H)
    is_stable = all(np.real(p) < 0)
    
    # Δημιουργία γραφημάτων (Pole-Zero map & Step Response)
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 4.5))
    fig.subplots_adjust(wspace=0.3)
    
    # --- 1. Pole-Zero Plot ---
    ct.pzmap(H, ax=ax1, title=f'Pole-Zero Map - {sys_name}')
    ax1.set_title(f'Pole-Zero Plot\n({sys_name})', fontsize=11)
    ax1.grid(True, linestyle=':', alpha=0.6)
    
    # --- 2. Step Response ---
    t_out, y_out = ct.step_response(H, T=np.linspace(0, 5, 500))
    ax2.plot(t_out, y_out, color='crimson', lw=2.5, label='Step Response $s(t)$')
    ax2.axhline(0, color='black', linewidth=1)
    ax2.axvline(0, color='black', linewidth=1)
    ax2.set_xlabel('Time $t$ (s)', fontsize=11)
    ax2.set_ylabel('Output Amplitude', fontsize=11)
    ax2.set_title(f'Step Response\n({sys_name})', fontsize=11)
    ax2.grid(True, linestyle=':', alpha=0.6)
    ax2.legend(loc='best', frameon=True, fontsize=9)
    
    plt.show()
    
    # Εκτύπωση ανάλυσης και ευστάθειας
    stability_text = "Stable (All poles in LHP)" if is_stable else "Unstable (At least one pole in RHP)"
    
    print(f"=== Transfer Function Analysis: {sys_name} ===")
    print(f"• Transfer Function H(s): {H}")
    print(f"• Poles: {np.round(p, 4)}")
    print(f"• Zeros: {z if len(z) > 0 else 'None'}")
    print(f"• BIBO Stability Status: {stability_text}")
```
