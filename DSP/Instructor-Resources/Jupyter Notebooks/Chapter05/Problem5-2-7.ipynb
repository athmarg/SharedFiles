```python
# Notebook: RLC Circuit Transfer Function & Interactive Response Analysis (Example 5)
import numpy as np
import matplotlib.pyplot as plt
import control as ct
import ipywidgets as widgets
from ipywidgets import interact

@interact(
    R = widgets.FloatSlider(value=10.0, min=0.1, max=100.0, step=0.5, description='R (Ohm):'),
    L = widgets.FloatSlider(value=1.0, min=0.01, max=5.0, step=0.05, description='L (H):'),
    C = widgets.FloatSlider(value=0.001, min=0.0001, max=0.01, step=0.0001, description='C (F):', readout_format='.4f')
)
def analyze_rlc_circuit(R, L, C):
    # Συνάρτηση μεταφοράς RLC: H(s) = (C * s) / (L*C*s^2 + R*C*s + 1)
    num = [C, 0]
    den = [L*C, R*C, 1]
    H = ct.tf(num, den)
    
    # Υπολογισμός πόλων, μηδενικών και ευστάθειας
    p = ct.poles(H)
    z = ct.zeros(H)
    is_stable = all(np.real(p) < 0)
    
    t = np.linspace(0, 0.5, 500)
    t_out, y_resp = ct.step_response(H, T=t)
    
    # Δημιουργία γραφημάτων (1. Pole-Zero Map, 2. Step Response)
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4.5))
    fig.subplots_adjust(wspace=0.3)
    
    # --- Γράφημα 1: Pole-Zero Plot (με στρογγυλοποίηση R, L, C στα 2 δεκαδικά) ---
    ct.pzmap(H, ax=ax1, title='')
    ax1.set_title(f'Pole-Zero Plot\n(R={R:.2f}, L={L:.2f}, C={C:.2f})', fontsize=11)
    ax1.grid(True, linestyle=':', alpha=0.6)
    
    # --- Γράφημα 2: Step Response (Απόκριση σε βήμα τάσης) ---
    ax2.plot(t_out, y_resp, color='teal', lw=2.5, label='Current Response $i(t)$')
    ax2.axhline(0, color='black', linewidth=1)
    ax2.axvline(0, color='black', linewidth=1)
    ax2.set_xlabel('Time $t$ (s)', fontsize=11)
    ax2.set_ylabel('Current $i(t)$ (A)', fontsize=11)
    ax2.set_title('Step Response (Current Output)', fontsize=11)
    ax2.grid(True, linestyle=':', alpha=0.6)
    ax2.legend(loc='best', frameon=True, fontsize=9)
    
    plt.show()
    
    # Εκτύπωση αποτελεσμάτων (με τη θεωρητική μορφή στην αρχή)
    print("=== RLC Circuit Analysis (Example 5) ===")
    print("• Theoretical Transfer Function:")
    print("  H(s) = (C * s) / (L * C * s^2 + R * C * s + 1)")
    print("-" * 50)
    print(f"• Current Parameters: R = {R:.2f} Ω, L = {L:.2f} H, C = {C:.4f} F")
    print(f"• Numerical Transfer Function H(s) = I(s)/V(s):")
    print(f"  {H}")
    print(f"• Poles: {np.round(p, 4)}")
    print(f"• Zeros: {np.round(z, 4) if len(z) > 0 else 'None'}")
    print(f"• BIBO Stability Status: {'Stable' if is_stable else 'Unstable'}")
```
