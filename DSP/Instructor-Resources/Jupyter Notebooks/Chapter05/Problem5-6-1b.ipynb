```python
# Notebook: Geometric Evaluation at s2 = -4 + j (Exact Textbook Formulas)
import warnings
warnings.filterwarnings('ignore', category=UserWarning)

import numpy as np
import matplotlib.pyplot as plt
import control as ct

def wrap_angle_deg(angle_deg):
    return (angle_deg + 180) % 360 - 180

# 1. Define Transfer Function and Evaluation Point s2 = -4 + j
num_coeffs = [1, -1, -6]
den_coeffs = [1, 8, 29, 52]
H = ct.tf(num_coeffs, den_coeffs)

sigma, omega = -4.0, 1.0
s_val = sigma + 1j * omega

z1, z2 = -2.0, 3.0
p1, p2, p3 = -2.0 + 3.0j, -2.0 - 3.0j, -4.0
K = num_coeffs[0] / den_coeffs[0]

print("=== Geometric Evaluation of H(s) at s2 = -4 + j ===")
print("-" * 50)

# 2. Calculation matching your exact analytical relations
r1 = np.sqrt((sigma - z1)**2 + omega**2)
vartheta_1 = np.rad2deg(np.arctan(-0.5))  # tan^{-1}(-1/2) = -26.565°

r2 = np.sqrt((sigma - z2)**2 + omega**2)
vartheta_2 = np.rad2deg(np.arctan(-1/7)) # tan^{-1}(-1/7) = -8.130°

q1 = np.sqrt((sigma - np.real(p1))**2 + (omega - np.imag(p1))**2)
varphi_1 = np.rad2deg(np.arctan(1.0))    # tan^{-1}(1) = 45°

q2 = np.sqrt((sigma - np.real(p2))**2 + (omega - np.imag(p2))**2)
varphi_2 = np.rad2deg(np.arctan(-2.0))   # tan^{-1}(-2) = -63.435°

q3 = np.sqrt((sigma - np.real(p3))**2 + (omega - np.imag(p3))**2)
varphi_3 = 90.0                        # tan^{-1}(inf) = 90°

# Calculate Magnitude and Phase geometrically
mag_geom = (r1 * r2) / (q1 * q2 * q3) * np.abs(K)
phase_geom = wrap_angle_deg((vartheta_1 + vartheta_2) - (varphi_1 + varphi_2 + varphi_3))

# Direct evaluation for comparison
H_direct = ct.evalfr(H, s_val)
mag_direct = np.abs(H_direct)
phase_direct = wrap_angle_deg(np.rad2deg(np.angle(H_direct)))

print(f"• Numerator (Zeros) Parameters:")
print(f"  r1 = {r1:.4f} (sqrt(5)),  vartheta_1 = {vartheta_1:.3f}°")
print(f"  r2 = {r2:.4f} (5*sqrt(2)), vartheta_2 = {vartheta_2:.3f}°")
print(f"• Denominator (Poles) Parameters:")
print(f"  q1 = {q1:.4f} (2*sqrt(2)), varphi_1 = {varphi_1:.3f}°")
print(f"  q2 = {q2:.4f} (2*sqrt(5)), varphi_2 = {varphi_2:.3f}°")
print(f"  q3 = {q3:.4f} (1.0000),    varphi_3 = {varphi_3:.3f}°")
print("-" * 50)
print(f"  => |H(s2)| (Geometric) = {mag_geom:.6f}  (Direct: {mag_direct:.6f})")
print(f"  => ∠H(s2) (Geometric) = {phase_geom:.3f}° (Direct: {phase_direct:.3f}°)")
print("-" * 50)

# 3. Visualization
plt.figure(figsize=(7, 6))
plt.scatter([z1, z2], [0, 0], marker='o', color='blue', s=100, label='Zeros ($z_m$)')
plt.scatter([np.real(p1), np.real(p2), np.real(p3)], [np.imag(p1), np.imag(p2), np.imag(p3)], marker='x', color='red', s=100, linewidths=2, label='Poles ($p_n$)')
plt.scatter([sigma], [omega], marker='*', color='purple', s=150, label='Evaluation Point $s_2 = -4+j$')

for z in [z1, z2]:
    plt.plot([z, sigma], [0, omega], 'b--', alpha=0.6)
for p in [p1, p2, p3]:
    plt.plot([np.real(p), sigma], [np.imag(p), omega], 'r--', alpha=0.6)

plt.axhline(0, color='black', linewidth=0.8)
plt.axvline(0, color='black', linewidth=0.8)
plt.title(r'Geometric Evaluation of $\mathcal{H}(s)$ at $s_2 = -4+j$', fontsize=11)
plt.xlabel(r'Real Axis ($\sigma$)', fontsize=10)
plt.ylabel(r'Imaginary Axis ($j\omega$)', fontsize=10)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='best', frameon=True, fontsize=9)
plt.axis('equal')
plt.show()
```
