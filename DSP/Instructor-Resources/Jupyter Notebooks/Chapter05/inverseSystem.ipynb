```python
# Notebook: Identity System Demonstration via Inverse System Cascade
import warnings
warnings.filterwarnings('ignore', category=UserWarning)

import numpy as np
import matplotlib.pyplot as plt
import control as ct

# 1. Define a proper transfer function H(s) with a 2nd-degree numerator and 3rd-degree denominator
# H(s) = (2s^2 + 4s + 3) / (s^3 + 6s^2 + 11s + 6)
num = [2.0, 4.0, 3.0]
den = [1.0, 6.0, 11.0, 6.0]
H = ct.tf(num, den)

# 2. Compute the inverse system G(s) = 1 / H(s)
H_inv = ct.tf(den, num)

# 3. Series interconnection of H(s) and its inverse H_inv(s) to form the identity system
H_identity = ct.series(H, H_inv)

print("Original System H(s):")
print(H)
print("\nInverse System H_inv(s):")
print(H_inv)
print("\nComposite (Identity) System H_identity(s) = H * H_inv:")
print(H_identity)

# 4. Demonstrate the identity property in practice using a custom input signal
t = np.linspace(0, 10, 1000)
u = np.sin(1.0 * t) + 0.5 * np.cos(3.0 * t)

# Pass the input signal through the composite system using forced response
t_out, y_out = ct.forced_response(H_identity, T=t, U=u)

# 5. Plotting the results with the legend positioned below the figure
plt.figure(figsize=(10, 6))
plt.plot(t, u, 'b-', linewidth=2.5, label='Input Signal x(t)')
plt.plot(t, y_out, 'r--', linewidth=1.5, label='Output Signal y(t) of Composite System')
plt.title(r'Demonstration of the Identity System: $\mathcal{H}_{identity}(s) = \mathcal{H}(s) \cdot \mathcal{H}^{-1}(s) = 1$', fontsize=12)
plt.xlabel('Time [s]', fontsize=11)
plt.ylabel('Amplitude', fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6)

# Place the legend outside/below the axes
plt.legend(loc='upper center', bbox_to_anchor=(0.5, -0.15), ncol=2, fontsize=10, frameon=True)

plt.tight_layout()
plt.show()
```
