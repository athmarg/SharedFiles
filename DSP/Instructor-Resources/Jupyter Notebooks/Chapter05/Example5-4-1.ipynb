```python
# Notebook: Routh-Hurwitz Stability Criterion Verification (Example)
import numpy as np
import control as ct

# 1. Define polynomial coefficients: P(s) = s^5 + 3s^4 + 2s^3 + s^2 + 7s + 12
# Coefficients ordered from highest to lowest power of s
coeffs = [1, 3, 2, 1, 7, 12]

print("=== Routh-Hurwitz Stability Analysis ===")
print(f"Polynomial Coefficients: {coeffs}")
print("-" * 50)

# 2. System poles calculation to verify stability
num = [1]
den = coeffs
sys = ct.tf(num, den)
poles = ct.poles(sys)

is_stable = all(np.real(poles) < 0)

print(f"• System Poles: {np.round(poles, 4)}")
print(f"• Stability Status: {'Stable' if is_stable else 'Unstable (Has poles with Re(s) >= 0)'}")
print("-" * 50)

# 3. Computation of Hurwitz Determinants (D1, D2, D3, D4, D5) for theoretical verification
# Definition of coefficients according to the problem statement: a5=1, a4=3, a3=2, a2=1, a1=7, a0=12
a5, a4, a3, a2, a1, a0 = 1, 3, 2, 1, 7, 12

D1 = a4
D2 = np.linalg.det([[a4, a5], 
                    [a2, a3]])
D3 = np.linalg.det([[a4, a5, 0], 
                    [a2, a3, a4], 
                    [a0, a1, a2]])
D4 = np.linalg.det([[a4, a5, 0,  0], 
                    [a2, a3, a4, a5], 
                    [a0, a1, a2, a3], 
                    [ 0,  0, a4, a5]])
D5 = a0 * D4

print("• Hurwitz Determinants Evaluation:")
print(f"  D1 = {round(D1, 3)}")
print(f"  D2 = {round(D2, 3)}")
print(f"  D3 = {round(D3, 3)}  <-- Negative! (Indicates instability immediately)")
print(f"  D4 = {round(D4, 3)}")
print(f"  D5 = {round(D5, 3)}")
```
