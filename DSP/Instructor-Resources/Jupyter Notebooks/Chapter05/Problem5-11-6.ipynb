```python
# Notebook: Υπολογισμός Εκθετικού Πίνακα e^{At} μέσω Μετασχηματισμού Laplace
import sympy as sp

# 1. Ορισμός πίνακα A βάσει του παραδείγματος
s = sp.symbols('s')
t = sp.symbols('t', real=True, positive=True)
A = sp.Matrix([[0, 2], [-1, -3]])
I = sp.eye(2)

print("--- Πίνακας Συστήματος A ---")
display(A)

# 2. Υπολογισμός Resolvent Matrix Φ(s) = (sI - A)^-1
# Αυτός ο πίνακας αποτελεί τον μετασχηματισμό Laplace του εκθετικού πίνακα
resolvent = (s * I - A).inv()
print("\n--- Resolvent Matrix Φ(s) = (sI - A)^-1 ---")
display(resolvent)

# 3. Υπολογισμός του Εκθετικού Πίνακα e^{At} μέσω αντίστροφου μετασχηματισμού Laplace
# Το SymPy εκτελεί εσωτερικά την ανάλυση σε απλά κλάσματα (partial fraction expansion)
exp_At = sp.Matrix([
    [sp.inverse_laplace_transform(resolvent[0,0], s, t), sp.inverse_laplace_transform(resolvent[0,1], s, t)],
    [sp.inverse_laplace_transform(resolvent[1,0], s, t), sp.inverse_laplace_transform(resolvent[1,1], s, t)]
])

print("\n--- Εκθετικός Πίνακας e^{At} ---")
display(exp_At)
```
