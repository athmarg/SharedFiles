```python
# Notebook: Symbolic Mason's Rule Implementation using SymPy
import sympy as sp

sp.init_printing(use_unicode=True)

# 1. Define branch gain symbols based on the signal flow graph
a12, a23, a32, a34, a43, a24, a44, a25, a45 = sp.symbols(
    'alpha_12 alpha_23 alpha_32 alpha_34 alpha_43 alpha_24 alpha_44 alpha_25 alpha_45'
)

# 2. Define the individual loop gains (L_i) present in the graph
L1 = a23 * a32
L2 = a34 * a43
L3 = a24 * a43 * a32
L4 = a44

# 3. Compute the graph determinant Delta = 1 - sum(L_i) + sum(nontouching pairs) - ...
# In this graph, L1 and L4 are the only nontouching pair of loops
sum_loops = L1 + L2 + L3 + L4
nontouching_pairs = L1 * L4
Delta = 1 - sum_loops + nontouching_pairs

# 4. Define the forward paths (P_k) and their corresponding path determinants (Delta_k)
# Path 1: y1 -> y2 -> y3 -> y4 -> y5
P1 = a12 * a23 * a34 * a45
Delta1 = 1  # All loops touch Path 1

# Path 2: y1 -> y2 -> y4 -> y5
P2 = a12 * a24 * a45
Delta2 = 1  # All loops touch Path 2

# Path 3: y1 -> y2 -> y5
P3 = a12 * a25
# Loops L2 and L4 do not touch Path 3, so Delta3 = 1 - (L2 + L4)
Delta3 = 1 - L2 - L4

# 5. Apply Mason's Gain Formula: H = (1 / Delta) * sum(P_k * Delta_k)
numerator = P1 * Delta1 + P2 * Delta2 + P3 * Delta3
H_mason = numerator / Delta

print("--- Transfer Function via Mason's Rule ---")
display(sp.cancel(H_mason))
```
