```python
# Notebook: Frequency Response & Sinusoidal Steady-State Output (Example 2)
import warnings
warnings.filterwarnings('ignore', category=UserWarning)

import numpy as np
import matplotlib.pyplot as plt
import control as ct

# 1. Define Transfer Function H(s) = (s - 2) / (s^2 + 5s + 6)
num_coeffs = [1, -2]
den_coeffs = [1, 5, 6]
H = ct.tf(num_coeffs, den_coeffs)

# 2. Frequency Response (Bode plots setup)
omega = np.logspace(-1, 2, 1000) # Frequency range from 0.1 to 100 rad/s
mag, phase, omega_out = ct.frequency_response(H, omega)

mag_abs = np.abs(mag)
phase_deg = np.rad2deg(phase)

# 3. Evaluate specific input frequency: omega = 15 rad/s, A = 12, phi = 34 degrees
omega_val = 15.0
A_in = 12.0
phi_in_deg = 34.0

H_at_15 = ct.evalfr(H, 1j * omega_val)
mag_at_15 = np.abs(H_at_15)
phase_at_15_deg = np.rad2deg(np.angle(H_at_15))

A_out = A_in * mag_at_15
phi_out_deg = phi_in_deg + phase_at_15_deg

print("=== Sinusoidal Steady-State Analysis ===")
print(f"• At frequency omega = {omega_val} rad/s:")
print(f"  Magnitude |H(j15)| = {mag_at_15:.6f}")
print(f"  Phase angle arg(H(j15)) = {phase_at_15_deg:.6f}°")
print(f"• Output Signal Parameters:")
print(f"  Amplitude = {A_out:.5f}")
print(f"  Phase = {phi_out_deg:.6f}°")
print(f"  y(t) = {A_out:.5f} * cos(15t {phi_out_deg:+.6f}°)")
print("-" * 50)

# 4. Combined Layout: Bode Plots on the Left, Input vs Output Response on the Right
fig = plt.figure(figsize=(13, 5))
gs = fig.add_gridspec(2, 2, width_ratios=[1.1, 1], hspace=0.3, wspace=0.3)

# --- Left Top: Amplitude Response ---
ax1 = fig.add_subplot(gs[0, 0])
ax1.plot(omega_out, mag_abs, color='darkblue', lw=1.8)
ax1.axvline(omega_val, color='red', linestyle='--', alpha=0.7, label=r'$\omega = 15$ rad/s')
ax1.scatter([omega_val], [mag_at_15], color='red', zorder=5)
ax1.set_xscale('log')
ax1.set_ylabel(r'$|\mathcal{H}(j\omega)|$', fontsize=10)
ax1.set_title(r'Frequency Response of $\mathcal{H}(s) = \dfrac{s-2}{s^2+5s+6}$', fontsize=10)
ax1.grid(True, which="both", linestyle=':', alpha=0.6)
ax1.legend(loc='upper right', frameon=True, fontsize=8)

# --- Left Bottom: Phase Response ---
ax2 = fig.add_subplot(gs[1, 0], sharex=ax1)
ax2.plot(omega_out, phase_deg, color='darkgreen', lw=1.8)
ax2.axvline(omega_val, color='red', linestyle='--', alpha=0.7)
ax2.scatter([omega_val], [phase_at_15_deg], color='red', zorder=5)
ax2.set_xscale('log')
ax2.set_xlabel(r'Frequency $\omega$ (rad/s)', fontsize=10)
ax2.set_ylabel(r'$\angle \mathcal{H}(j\omega)$ (deg)', fontsize=10)
ax2.grid(True, which="both", linestyle=':', alpha=0.6)

# --- Right Side: Input vs Output Sinusoidal Signals (spanning both rows) ---
t = np.linspace(0, 2*np.pi / omega_val * 3, 500) # 3 full periods
x_t = A_in * np.cos(omega_val * t + np.deg2rad(phi_in_deg))
y_t = A_out * np.cos(omega_val * t + np.deg2rad(phi_out_deg))

ax3 = fig.add_subplot(gs[:, 1])
ax3.plot(t, x_t, color='gray', linestyle='--', lw=1.5, label=r'Input $x(t) = 12\cos(15t + 34^\circ)$')
ax3.plot(t, y_t, color='darkblue', lw=2, label=r'Output $y(t) = 0.784\cos(15t - 29.5^\circ)$')
ax3.axhline(0, color='black', linewidth=0.8)
ax3.set_title('Sinusoidal Steady-State Response', fontsize=10)
ax3.set_xlabel(r'Time $t$ (s)', fontsize=10)
ax3.set_ylabel('Amplitude', fontsize=10)
ax3.grid(True, linestyle=':', alpha=0.6)

# Legend placed nicely at the bottom outside/below the plot area
ax3.legend(loc='upper center', bbox_to_anchor=(0.5, -0.18), frameon=True, fontsize=8)

plt.show()
```
