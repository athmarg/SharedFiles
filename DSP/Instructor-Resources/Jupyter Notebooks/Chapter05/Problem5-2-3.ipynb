```python
# Notebook: Transfer Function from Input-Output Signals & Impulse Response (Example 3)
import numpy as np
import matplotlib.pyplot as plt
import control as ct
import ipywidgets as widgets
from ipywidgets import interact

# Ορισμός της Συνάρτησης Μεταφοράς H(s) = 10 * (s^2 + 2s + 1) / (s^2 + 2s + 17)
# Αριθμητής (Numerator): [10, 20, 10] -> αντιστοιχεί σε 10s^2 + 20s + 10
# Παρονομαστής (Denominator): [1, 2, 17] -> αντιστοιχεί σε s^2 + 2s + 17
num = [10, 20, 10]
den = [1, 2, 17]
H = ct.tf(num, den)

def analyze_example3():
    # Υπολογισμός πόλων, μηδενικών και ευστάθειας
    p = ct.poles(H)
    z = ct.zeros(H)
    is_stable = all(np.real(p) < 0)
    
    # Δημιουργία γραφημάτων (Pole-Zero map & Impulse/Step Response)
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 4.5))
    fig.subplots_adjust(wspace=0.3)
    
    # --- 1. Pole-Zero Plot ---
    ct.pzmap(H, ax=ax1, title='Pole-Zero Map (Example 3)')
    ax1.set_title('Pole-Zero Map\n(Double Zero at $s=-1$, Complex Poles)', fontsize=11)
    ax1.grid(True, linestyle=':', alpha=0.6)
    
    # --- 2. Step Response (πιο οπτικά ομαλή απόκριση για συστήματα ίσης τάξης) ---
    t_out, y_out = ct.step_response(H, T=np.linspace(0, 5, 500))
    ax2.plot(t_out, y_out, color='teal', lw=2.5, label='Step Response $s(t)$')
    ax2.axhline(0, color='black', linewidth=1)
    ax2.axvline(0, color='black', linewidth=1)
    ax2.set_xlabel('Time $t$ (s)', fontsize=11)
    ax2.set_ylabel('Output Amplitude', fontsize=11)
    ax2.set_title('Step Response\n(System with Feedthrough)', fontsize=11)
    ax2.grid(True, linestyle=':', alpha=0.6)
    ax2.legend(loc='best', frameon=True, fontsize=9)
    
    plt.show()
    
    # Εκτύπωση αποτελεσμάτων
    stability_text = "Stable (All poles in LHP)" if is_stable else "Unstable"
    
    print("=== Transfer Function & Signal Analysis (Example 3) ===")
    print(f"• Transfer Function H(s): {H}")
    print(f"• Poles: {np.round(p, 4)}")
    print(f"• Zeros: {np.round(z, 4) if len(z) > 0 else 'None'}")
    print(f"• BIBO Stability Status: {stability_text}")

# Εκτέλεση ανάλυσης
analyze_example3()
```
